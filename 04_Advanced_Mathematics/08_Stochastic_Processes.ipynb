## 1. Introduction and Foundational Framework {#introduction}

Stochastic process theory provides the mathematical framework for modeling and analyzing random phenomena evolving over time or space. Developed through the foundational work of Kolmogorov, Wiener, Itô, Doob, and others in the 20th century, it has become indispensable for understanding uncertainty, randomness, and temporal dynamics across mathematics, physics, finance, biology, and machine learning.

### Historical Context and Motivation

The development of stochastic process theory emerged from several mathematical and practical needs:
- **Brownian Motion**: Einstein's 1905 analysis of particle motion in fluids
- **Financial Mathematics**: Bachelier's 1900 model of stock price fluctuations  
- **Statistical Mechanics**: Random walks and diffusion processes
- **Signal Processing**: Noise analysis and filtering theory
- **Population Dynamics**: Birth-death processes and epidemic models
- **Machine Learning**: Sequential decision making and online learning

### Mathematical Foundations

**Definition 1.1** (Stochastic Process): A **stochastic process** is a collection of random variables $\{X_t : t \in T\}$ defined on a probability space $(\Omega, \mathcal{F}, P)$, indexed by a parameter set $T$ (usually time).

**Key Components**:
1. **State Space**: $S$ (values that $X_t$ can take)
2. **Index Set**: $T$ (discrete: $\mathbb{N}$, continuous: $\mathbb{R}_+$)
3. **Sample Paths**: $t \mapsto X_t(\omega)$ for fixed $\omega \in \Omega$
4. **Finite-Dimensional Distributions**: $P(X_{t_1} \in B_1, \ldots, X_{t_n} \in B_n)$

**Definition 1.2** (Kolmogorov Consistency): A family of finite-dimensional distributions $\{\mu_{t_1,\ldots,t_n}\}$ satisfies the **consistency conditions** if:
1. **Symmetry**: Invariant under permutations of indices
2. **Projectivity**: Marginals are consistent with projections

**Theorem 1.1** (Kolmogorov Extension Theorem): Given consistent finite-dimensional distributions, there exists a unique probability measure on the product space making the coordinate projections into random variables with the given distributions.

### Classification of Stochastic Processes

**By Index Set**:
- **Discrete-Time**: $(X_n)_{n \geq 0}$ with $n \in \mathbb{N}$
- **Continuous-Time**: $(X_t)_{t \geq 0}$ with $t \in \mathbb{R}_+$

**By State Space**:
- **Discrete State**: $S$ finite or countable
- **Continuous State**: $S = \mathbb{R}^d$ or general metric space

**By Dependence Structure**:
- **Independent Increments**: $(X_{t+s} - X_t)$ independent of $\mathcal{F}_t$
- **Markov Processes**: $P(X_t | \mathcal{F}_s) = P(X_t | X_s)$ for $s < t$
- **Martingales**: $\mathbb{E}[X_t | \mathcal{F}_s] = X_s$ for $s < t$
- **Stationary Processes**: Distributions invariant under time shifts

### Fundamental Examples

**Example 1.1** (Random Walk): $S_n = \sum_{i=1}^n X_i$ where $X_i$ i.i.d.
- **Simple Random Walk**: $X_i \in \{-1, +1\}$ with equal probability
- **Scaling Limit**: $S_{\lfloor nt \rfloor}/\sqrt{n} \to B_t$ (Brownian motion)

**Example 1.2** (Poisson Process): Counting process $(N_t)_{t \geq 0}$ with:
- Independent increments
- $N_t - N_s \sim \text{Poisson}(\lambda(t-s))$
- Right-continuous sample paths with jumps of size 1

**Example 1.3** (Ornstein-Uhlenbeck Process): Solution to SDE:
$$dX_t = -\alpha X_t dt + \sigma dB_t$$
Models mean reversion with Gaussian stationary distribution.

### Probabilistic Foundations

**Definition 1.3** (Filtration): An increasing family $(\mathcal{F}_t)_{t \geq 0}$ of σ-algebras:
$$\mathcal{F}_s \subset \mathcal{F}_t \subset \mathcal{F} \quad \text{for } s \leq t$$

**Natural Filtration**: For process $(X_t)$:
$$\mathcal{F}_t^X = \sigma(X_s : s \leq t)$$

**Definition 1.4** (Adapted Process): $(X_t)$ is **adapted** to $(\mathcal{F}_t)$ if $X_t$ is $\mathcal{F}_t$-measurable for all $t$.

**Definition 1.5** (Stopping Time): Random time $\tau$ is a **stopping time** w.r.t. $(\mathcal{F}_t)$ if:
$$\{\tau \leq t\} \in \mathcal{F}_t \quad \text{for all } t \geq 0$$

### Path Properties and Regularity

**Definition 1.6** (Sample Path Properties):
- **Right-Continuous**: $X_t(\omega) = \lim_{s \downarrow t} X_s(\omega)$
- **Left-Continuous**: $X_{t-}(\omega) = \lim_{s \uparrow t} X_s(\omega)$
- **RCLL (Càdlàg)**: Right-continuous with left limits
- **Continuous**: No jumps or discontinuities

**Theorem 1.2** (Kolmogorov Continuity Theorem): If there exist constants $\alpha, \beta, K > 0$ such that:
$$\mathbb{E}[|X_t - X_s|^\alpha] \leq K|t - s|^{1+\beta}$$
then there exists a modification of $X$ with Hölder continuous paths.

**Definition 1.7** (Modifications and Versions):
- **Modification**: $(X_t)$ and $(Y_t)$ are modifications if $P(X_t = Y_t) = 1$ for each $t$
- **Indistinguishable**: $P(\{X_t = Y_t \text{ for all } t\}) = 1$

## 3. Markov Chains and Markov Decision Processes {#markov-chains}

### Abstract Markov Property

**Definition 3.1** (Markov Property): A stochastic process $(X_t)_{t \geq 0}$ satisfies the **Markov property** if:
$$P(X_t \in B | \mathcal{F}_s) = P(X_t \in B | X_s) \quad \text{for all } s < t, B \in \mathcal{B}(S)$$

This states that the future evolution depends only on the current state, not on the path taken to reach it.

**Strong Markov Property**: For stopping time $\tau$:
$$P(X_{\tau+t} \in B | \mathcal{F}_\tau) = P(X_{\tau+t} \in B | X_\tau) \quad \text{on } \{\tau < \infty\}$$

### Discrete-Time Markov Chains

**Definition 3.2** (Transition Matrix): For finite state space $S = \{1, 2, \ldots, n\}$:
$$P_{ij} = P(X_{n+1} = j | X_n = i)$$
with properties:
1. $P_{ij} \geq 0$ for all $i,j$
2. $\sum_{j=1}^n P_{ij} = 1$ for all $i$

**n-Step Transition Probabilities**: 
$$P_{ij}^{(n)} = P(X_n = j | X_0 = i) = (P^n)_{ij}$$

**Chapman-Kolmogorov Equations**:
$$P_{ij}^{(m+n)} = \sum_{k=1}^N P_{ik}^{(m)} P_{kj}^{(n)}$$

### Classification of States

**Definition 3.3** (Communication and Classes):
- State $j$ is **accessible** from $i$ if $P_{ij}^{(n)} > 0$ for some $n \geq 0$
- States $i$ and $j$ **communicate** if each is accessible from the other
- A **class** is an equivalence class under communication
- Chain is **irreducible** if all states communicate

**Definition 3.4** (Recurrence and Transience):
Define first passage time: $T_j = \inf\{n \geq 1 : X_n = j\}$

- State $j$ is **recurrent** if $P(T_j < \infty | X_0 = j) = 1$
- State $j$ is **transient** if $P(T_j < \infty | X_0 = j) < 1$

**Theorem 3.1** (Characterization of Recurrence): State $j$ is recurrent iff:
$$\sum_{n=0}^{\infty} P_{jj}^{(n)} = \infty$$

**Definition 3.5** (Periodicity): State $i$ has **period** $d = \gcd\{n \geq 1 : P_{ii}^{(n)} > 0\}$.
- If $d = 1$, state is **aperiodic**
- If all states are aperiodic, chain is **aperiodic**

### Stationary Distributions and Long-Run Behavior

**Definition 3.6** (Stationary Distribution): Probability vector $\pi$ is **stationary** if:
$$\pi P = \pi \quad \text{i.e., } \sum_{i} \pi_i P_{ij} = \pi_j \text{ for all } j$$

**Theorem 3.2** (Fundamental Limit Theorem): For finite, irreducible, aperiodic chain:
$$\lim_{n \to \infty} P_{ij}^{(n)} = \pi_j > 0$$
where $\pi$ is the unique stationary distribution.

**Theorem 3.3** (Convergence Rate): Under regularity conditions:
$$\left|P_{ij}^{(n)} - \pi_j\right| \leq C \rho^n$$
for constants $C > 0$ and $0 < \rho < 1$.

### Continuous-Time Markov Chains

**Definition 3.7** (Q-Matrix): For continuous-time chain with rates:
$$Q_{ij} = \lim_{h \to 0} \frac{P_{ij}(h) - \delta_{ij}}{h}$$
where $\delta_{ij}$ is the Kronecker delta.

Properties:
- $Q_{ij} \geq 0$ for $i \neq j$
- $Q_{ii} = -\sum_{j \neq i} Q_{ij}$ (row sums are zero)

**Forward and Backward Equations**:
$$\frac{d}{dt} P_{ij}(t) = \sum_k P_{ik}(t) Q_{kj} = \sum_k Q_{ik} P_{kj}(t)$$

**Stationary Distribution**: Satisfies $\pi Q = 0$, i.e.:
$$\sum_{i} \pi_i Q_{ij} = 0 \quad \text{for all } j$$

### Markov Decision Processes (MDPs)

**Definition 3.8** (MDP): Tuple $(S, A, P, R, \gamma)$ where:
- $S$: State space
- $A$: Action space  
- $P(s' | s, a)$: Transition probabilities
- $R(s, a, s')$: Reward function
- $\gamma \in [0, 1]$: Discount factor

**Policy**: Function $\pi: S \to A$ (deterministic) or $\pi: S \times A \to [0,1]$ (stochastic)

**Value Functions**:
$$V^\pi(s) = \mathbb{E}_\pi\left[\sum_{t=0}^{\infty} \gamma^t R_t \mid S_0 = s\right]$$
$$Q^\pi(s,a) = \mathbb{E}_\pi\left[\sum_{t=0}^{\infty} \gamma^t R_t \mid S_0 = s, A_0 = a\right]$$

**Bellman Equations**:
$$V^\pi(s) = \sum_a \pi(a|s) \sum_{s'} P(s'|s,a)[R(s,a,s') + \gamma V^\pi(s')]$$
$$Q^\pi(s,a) = \sum_{s'} P(s'|s,a)[R(s,a,s') + \gamma \sum_{a'} \pi(a'|s') Q^\pi(s',a')]$$

**Optimal Value Functions**:
$$V^*(s) = \max_\pi V^\pi(s), \quad Q^*(s,a) = \max_\pi Q^\pi(s,a)$$

**Bellman Optimality Equations**:
$$V^*(s) = \max_a \sum_{s'} P(s'|s,a)[R(s,a,s') + \gamma V^*(s')]$$

### Advanced Markov Theory

**Definition 3.9** (Ergodicity): Chain is **ergodic** if:
1. Irreducible
2. Aperiodic  
3. Positive recurrent (finite expected return times)

**Theorem 3.4** (Ergodic Theorem): For ergodic chain with stationary distribution $\pi$:
$$\lim_{n \to \infty} \frac{1}{n} \sum_{k=0}^{n-1} f(X_k) = \sum_i \pi_i f(i) \quad \text{a.s.}$$

**Definition 3.10** (Reversibility): Chain is **reversible** w.r.t. $\pi$ if:
$$\pi_i P_{ij} = \pi_j P_{ji} \quad \text{for all } i,j \quad \text{(detailed balance)}$$

**Theorem 3.5** (Perron-Frobenius): For irreducible, finite chain:
1. Spectral radius equals 1 (largest eigenvalue)
2. Corresponding eigenvector is stationary distribution
3. All other eigenvalues have modulus < 1

### Machine Learning Applications

**Reinforcement Learning Algorithms**:

1. **Value Iteration**:
   $$V_{k+1}(s) = \max_a \sum_{s'} P(s'|s,a)[R(s,a,s') + \gamma V_k(s')]$$

2. **Policy Iteration**:
   - **Policy Evaluation**: Solve $V^\pi = R^\pi + \gamma P^\pi V^\pi$
   - **Policy Improvement**: $\pi'(s) = \arg\max_a Q^\pi(s,a)$

3. **Q-Learning**: Model-free temporal difference learning:
   $$Q_{t+1}(s,a) = Q_t(s,a) + \alpha[R + \gamma \max_{a'} Q_t(s',a') - Q_t(s,a)]$$

4. **SARSA**: On-policy alternative:
   $$Q_{t+1}(s,a) = Q_t(s,a) + \alpha[R + \gamma Q_t(s',a') - Q_t(s,a)]$$

**Monte Carlo Methods**:

1. **Markov Chain Monte Carlo**: Sample from $\pi$ using Markov chain
   - **Metropolis-Hastings**: Accept/reject with ratio $\min(1, \frac{\pi(y)q(x|y)}{\pi(x)q(y|x)})$
   - **Gibbs Sampling**: Coordinate-wise sampling from conditionals

2. **Convergence Theory**: 
   - **Mixing Time**: Time to approach stationary distribution
   - **Total Variation Distance**: $\|P^n(x, \cdot) - \pi\|_{TV} \leq Ce^{-n/\tau}$

**Hidden Markov Models (HMMs)**:

**Structure**:
- Hidden states: $(X_t)$ Markov chain with transition matrix $P$
- Observations: $Y_t | X_t$ with emission probabilities $B_{ij} = P(Y_t = j | X_t = i)$

**Algorithms**:
1. **Forward-Backward**: Compute $P(X_t | Y_{1:T})$
2. **Viterbi**: Find $\arg\max P(X_{1:T} | Y_{1:T})$  
3. **Baum-Welch**: EM algorithm for parameter estimation

**PageRank Algorithm**:

Model web as Markov chain with transition matrix:
$$P = \frac{1-d}{n} \mathbf{1}\mathbf{1}^T + d M$$
where $M$ is link structure matrix and $d$ is damping factor.

**PageRank vector**: Stationary distribution $\pi P = \pi$

**Applications in Deep Learning**:

1. **Dropout**: Random removal of neurons as Markov process
2. **Attention Mechanisms**: Soft alignment as transition probabilities  
3. **Sequence Models**: RNNs, LSTMs as approximate Markov models
4. **Generative Models**: Markov chain sampling in GANs, VAEs

**Optimal Control and Stochastic Games**:

1. **Linear Quadratic Gaussian (LQG)**: 
   $$\min \mathbb{E}\left[\sum_{t=0}^T (x_t^T Q x_t + u_t^T R u_t)\right]$$
   subject to $x_{t+1} = A x_t + B u_t + w_t$

2. **Hamilton-Jacobi-Bellman Equation**: 
   $$\frac{\partial V}{\partial t} + \min_u \left[\mathcal{L} V + L(x,u)\right] = 0$$

3. **Multi-Agent Reinforcement Learning**: Nash equilibria in Markov games

## 2. Markov Chains: Discrete Time {#markov-chains}

Markov chains are stochastic processes with the Markov property: the future is independent of the past given the present.

**Key Concepts:**
- **Markov Property**: $P(X_{n+1} | X_0, ..., X_n) = P(X_{n+1} | X_n)$
- **Transition Matrix**: $P_{ij} = P(X_{n+1} = j | X_n = i)$
- **Chapman-Kolmogorov Equations**: $P^{(n+m)} = P^{(n)} \cdot P^{(m)}$
- **Stationary Distribution**: $\pi = \pi P$
- **State Classification**: recurrent vs transient, periodic vs aperiodic

**Applications to ML:**
- Hidden Markov Models (HMM)
- PageRank algorithm
- Markov Decision Processes (MDPs) in reinforcement learning

## 5. Poisson Processes {#poisson-processes}Poisson processes model random events occurring over time at a constant average rate.**Definition**: $\{N_t\}_{t \geq 0}$ is a Poisson process with rate $\lambda$ if:- $N_0 = 0$- Independent increments- $N_t - N_s \sim \text{Poisson}(\lambda(t-s))$- Right-continuous step function**Properties:**- Inter-arrival times: $T_i \sim \text{Exp}(\lambda)$ (independent)- Waiting time: $S_n = \sum_{i=1}^n T_i \sim \text{Gamma}(n, \lambda)$- Superposition: Sum of independent Poisson processes is Poisson- Thinning: Random deletion gives Poisson process**Applications:**- Queueing theory- Network traffic modeling- Insurance claims- Neuroscience (spike trains)

## 6. Stochastic Differential Equations and Itô's Lemma {#sde}SDEs extend ODEs to include randomness from Brownian motion.**General Form**: $$dX_t = \mu(X_t, t) dt + \sigma(X_t, t) dW_t$$where $\mu$ is the drift and $\sigma$ is the diffusion coefficient.**Itô's Lemma**: For $Y_t = f(X_t, t)$:$$dY_t = \left(\frac{\partial f}{\partial t} + \mu \frac{\partial f}{\partial x} + \frac{1}{2}\sigma^2 \frac{\partial^2 f}{\partial x^2}\right) dt + \sigma \frac{\partial f}{\partial x} dW_t$$**Famous SDEs:**- **Geometric Brownian Motion**: $dS_t = \mu S_t dt + \sigma S_t dW_t$ (stock prices)- **Ornstein-Uhlenbeck**: $dX_t = \theta(\mu - X_t) dt + \sigma dW_t$ (mean reversion)- **CIR Model**: $dr_t = \kappa(\theta - r_t) dt + \sigma \sqrt{r_t} dW_t$ (interest rates)

```python
# =============================================================================# 7. ML APPLICATIONS: MCMC AND GAUSSIAN PROCESSES# =============================================================================print("=" * 80)print("7. MACHINE LEARNING APPLICATIONS")print("=" * 80)# 7.1 Metropolis-Hastings MCMCprint("\n7.1 Metropolis-Hastings: Sampling from Complex Distributions")print("-" * 60)# Target: Mixture of Gaussiansdef target_density(x):    return 0.3 * stats.norm.pdf(x, -2, 0.5) + 0.7 * stats.norm.pdf(x, 3, 1.0)# Metropolis-Hastingsn_samples = 10000proposal_std = 1.0samples = np.zeros(n_samples)samples[0] = 0.0  # Initial staten_accepted = 0np.random.seed(42)for i in range(1, n_samples):    current = samples[i-1]        # Propose new state    proposal = current + np.random.normal(0, proposal_std)        # Acceptance ratio    alpha = min(1, target_density(proposal) / target_density(current))        # Accept or reject    if np.random.rand() < alpha:        samples[i] = proposal        n_accepted += 1    else:        samples[i] = currentacceptance_rate = n_accepted / n_samplesprint(f"Number of samples: {n_samples}")print(f"Acceptance rate: {acceptance_rate:.2%}")print(f"Sample mean: {np.mean(samples):.3f}")print(f"Sample std: {np.std(samples):.3f}")# 7.2 Gaussian Process Regressionprint("\n7.2 Gaussian Process Regression")print("-" * 60)# Training datanp.random.seed(42)X_train = np.array([-4, -3, -2, -1, 1, 2, 3, 4]).reshape(-1, 1)y_train = np.sin(X_train).ravel() + 0.1 * np.random.randn(len(X_train))# Test pointsX_test = np.linspace(-5, 5, 100).reshape(-1, 1)# Gaussian Processfrom sklearn.gaussian_process import GaussianProcessRegressorfrom sklearn.gaussian_process.kernels import RBF, WhiteKernelkernel = RBF(length_scale=1.0) + WhiteKernel(noise_level=0.01)gp = GaussianProcessRegressor(kernel=kernel, random_state=42)gp.fit(X_train, y_train)# Predictionsy_pred, y_std = gp.predict(X_test, return_std=True)print(f"Training points: {len(X_train)}")print(f"Test points: {len(X_test)}")print(f"Optimized kernel: {gp.kernel_}")# 7.3 Stochastic Gradient Descent Dynamicsprint("\n7.3 SGD as Stochastic Process")print("-" * 60)# Simple quadratic loss: L(w) = (w - w*)^T Q (w - w*)w_star = np.array([1.5, 2.0])Q = np.array([[2.0, 0.5], [0.5, 1.0]])def loss(w):    diff = w - w_star    return diff @ Q @ diffdef gradient(w):    return 2 * Q @ (w - w_star)# SGD with noisen_steps_sgd = 1000lr = 0.01noise_std = 0.1w_trajectory = np.zeros((n_steps_sgd + 1, 2))w_trajectory[0] = np.array([0.0, 0.0])for i in range(n_steps_sgd):    grad = gradient(w_trajectory[i])    noise = np.random.normal(0, noise_std, size=2)    w_trajectory[i+1] = w_trajectory[i] - lr * (grad + noise)final_loss = loss(w_trajectory[-1])print(f"Initial loss: {loss(w_trajectory[0]):.4f}")print(f"Final loss: {final_loss:.4f}")print(f"Final weights: {w_trajectory[-1]}")print(f"True optimum: {w_star}")# Visualizationsfig = plt.figure(figsize=(16, 11))gs = fig.add_gridspec(3, 3, hspace=0.35, wspace=0.3)# Plot 1: MCMC samples (trace plot)ax1 = fig.add_subplot(gs[0, :2])burn_in = 1000ax1.plot(samples[:burn_in], 'r-', alpha=0.3, linewidth=0.5, label='Burn-in')ax1.plot(range(burn_in, n_samples), samples[burn_in:], 'b-', alpha=0.5,        linewidth=0.5, label='Samples')ax1.set_xlabel('Iteration', fontsize=11)ax1.set_ylabel('Sample Value', fontsize=11)ax1.set_title('MCMC Trace Plot', fontsize=12, fontweight='bold')ax1.legend()ax1.grid(True, alpha=0.3)# Plot 2: MCMC histogram vs targetax2 = fig.add_subplot(gs[0, 2])ax2.hist(samples[burn_in:], bins=50, density=True, alpha=0.7,        edgecolor='black', color='steelblue', label='MCMC samples')x_range = np.linspace(-5, 6, 200)target_vals = [target_density(x) for x in x_range]ax2.plot(x_range, target_vals, 'r-', linewidth=2.5, label='Target density')ax2.set_xlabel('x', fontsize=11)ax2.set_ylabel('Density', fontsize=11)ax2.set_title('MCMC: Target vs Samples', fontsize=12, fontweight='bold')ax2.legend()ax2.grid(True, alpha=0.3)# Plot 3: Gaussian Process regressionax3 = fig.add_subplot(gs[1, :])ax3.plot(X_test, y_pred, 'b-', linewidth=2, label='GP mean')ax3.fill_between(X_test.ravel(),                y_pred - 1.96 * y_std,                y_pred + 1.96 * y_std,                alpha=0.3, color='blue', label='95% confidence')ax3.plot(X_train, y_train, 'ro', markersize=8, label='Training data')ax3.plot(X_test, np.sin(X_test), 'g--', linewidth=1.5, alpha=0.7,        label='True function')ax3.set_xlabel('x', fontsize=11)ax3.set_ylabel('y', fontsize=11)ax3.set_title('Gaussian Process Regression', fontsize=12, fontweight='bold')ax3.legend()ax3.grid(True, alpha=0.3)# Plot 4: GP samples from priorax4 = fig.add_subplot(gs[2, 0])n_gp_samples = 5from sklearn.gaussian_process.kernels import RBFX_gp = np.linspace(-5, 5, 100).reshape(-1, 1)kernel_prior = RBF(length_scale=1.0)K = kernel_prior(X_gp, X_gp)for i in range(n_gp_samples):    f_sample = np.random.multivariate_normal(np.zeros(len(X_gp)), K)    ax4.plot(X_gp, f_sample, alpha=0.7, linewidth=1.5)ax4.set_xlabel('x', fontsize=11)ax4.set_ylabel('f(x)', fontsize=11)ax4.set_title('GP Prior Samples', fontsize=12, fontweight='bold')ax4.grid(True, alpha=0.3)# Plot 5: SGD trajectoryax5 = fig.add_subplot(gs[2, 1])# Create contour plot of lossw1_range = np.linspace(-1, 3, 50)w2_range = np.linspace(-1, 4, 50)W1, W2 = np.meshgrid(w1_range, w2_range)Loss = np.zeros_like(W1)for i in range(len(w1_range)):    for j in range(len(w2_range)):        w = np.array([W1[j,i], W2[j,i]])        Loss[j,i] = loss(w)contour = ax5.contour(W1, W2, Loss, levels=20, alpha=0.6, cmap='viridis')ax5.plot(w_trajectory[:, 0], w_trajectory[:, 1], 'r-', linewidth=1.5,        alpha=0.7, label='SGD path')ax5.plot(w_trajectory[0, 0], w_trajectory[0, 1], 'go', markersize=10,        label='Start')ax5.plot(w_star[0], w_star[1], 'r*', markersize=15, label='Optimum')ax5.set_xlabel('w₁', fontsize=11)ax5.set_ylabel('w₂', fontsize=11)ax5.set_title('SGD Trajectory (Stochastic Dynamics)', fontsize=12, fontweight='bold')ax5.legend()ax5.grid(True, alpha=0.3)# Plot 6: Autocorrelation of MCMCax6 = fig.add_subplot(gs[2, 2])max_lag = 100samples_centered = samples[burn_in:] - np.mean(samples[burn_in:])autocorr = np.correlate(samples_centered, samples_centered, mode='full')autocorr = autocorr[len(autocorr)//2:]autocorr = autocorr[:max_lag] / autocorr[0]ax6.bar(range(max_lag), autocorr, alpha=0.7, edgecolor='black', color='teal')ax6.axhline(y=0, color='k', linestyle='-', linewidth=0.5)ax6.set_xlabel('Lag', fontsize=11)ax6.set_ylabel('Autocorrelation', fontsize=11)ax6.set_title('MCMC Autocorrelation', fontsize=12, fontweight='bold')ax6.grid(True, alpha=0.3)plt.show()print("\n" + "=" * 80)print("Key Insights:")print("- MCMC: Sample from complex distributions using Markov chains")print("- Gaussian Processes: Flexible non-parametric models with uncertainty")print("- SGD: Stochastic process converging to minimum")print("- Stochastic processes enable modern ML: MCMC, GPs, RL, diffusion models")print("=" * 80)
```
