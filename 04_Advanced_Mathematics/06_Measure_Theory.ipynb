## 1. Introduction to Measure Theory {#introduction}

Measure theory provides the rigorous mathematical foundation for modern analysis, probability theory, and statistical inference. Developed in the early 20th century by mathematicians including Lebesgue, Borel, and Kolmogorov, it revolutionized our understanding of integration, probability, and the foundations of mathematics itself.

### Historical Context and Motivation

The development of measure theory addressed fundamental limitations of classical analysis:
- **Riemann integration**: Limited to continuous or piecewise continuous functions
- **Set theory paradoxes**: Need for rigorous foundations of infinite sets
- **Probability theory**: Lack of rigorous mathematical foundations
- **Fourier analysis**: Need for more general integration theory

### Mathematical Foundations

**Definition 1.1** (Abstract Measure Space): A measure space is a triple $(X, \mathcal{F}, \mu)$ where:
- $X$ is a non-empty set (the **sample space**)
- $\mathcal{F}$ is a σ-algebra on $X$ (the **measurable sets**)
- $\mu: \mathcal{F} \to [0, \infty]$ is a measure (the **measure function**)

**Definition 1.2** (σ-Algebra): A collection $\mathcal{F}$ of subsets of $X$ is a σ-algebra if:
1. $X \in \mathcal{F}$ (contains the universal set)
2. If $A \in \mathcal{F}$, then $A^c \in \mathcal{F}$ (closed under complementation)
3. If $(A_n)_{n=1}^{\infty} \subset \mathcal{F}$, then $\bigcup_{n=1}^{\infty} A_n \in \mathcal{F}$ (closed under countable unions)

**Definition 1.3** (Measure): A function $\mu: \mathcal{F} \to [0, \infty]$ is a measure if:
1. $\mu(\emptyset) = 0$ (null empty set)
2. **σ-additivity**: For disjoint $(A_n)_{n=1}^{\infty} \subset \mathcal{F}$:
   $$\mu\left(\bigcup_{n=1}^{\infty} A_n\right) = \sum_{n=1}^{\infty} \mu(A_n)$$

### Fundamental Examples

**Counting Measure**: For any set $X$:
$$\mu(A) = \begin{cases} |A| & \text{if } A \text{ is finite} \\ \infty & \text{if } A \text{ is infinite} \end{cases}$$

**Lebesgue Measure**: On $\mathbb{R}^n$, extends the notion of length, area, and volume:
- $\lambda([a,b]) = b - a$ for intervals
- $\lambda(\text{point}) = 0$ for single points
- $\lambda(\mathbb{Q}) = 0$ for rationals (countable)

**Dirac Measure**: For fixed $x_0 \in X$:
$$\delta_{x_0}(A) = \begin{cases} 1 & \text{if } x_0 \in A \\ 0 & \text{if } x_0 \notin A \end{cases}$$

### Theoretical Importance

**Theorem 1.1** (Carathéodory Extension): Every pre-measure on a ring can be extended to a measure on the generated σ-algebra.

**Theorem 1.2** (Uniqueness of Extension): If two measures agree on a π-system generating a σ-algebra and at least one is σ-finite, then they are identical.

### Applications in Machine Learning and AI:

**Probability Theory Foundations**:
- **Kolmogorov Axioms**: Probability as a normalized measure
- **Random Variables**: Measurable functions on probability spaces
- **Expectation**: Lebesgue integration with respect to probability measures

**Information Theory**:

## 2. σ-Algebras and Measurable Spaces {#sigma-algebras}

### Abstract Framework

**Definition 2.1** (σ-Algebra): Let $X$ be a non-empty set. A collection $\mathcal{F}$ of subsets of $X$ is called a **σ-algebra** (or σ-field) if:

1. **Universal Set**: $X \in \mathcal{F}$
2. **Complement Closure**: If $A \in \mathcal{F}$, then $A^c = X \setminus A \in \mathcal{F}$
3. **Countable Union Closure**: If $(A_n)_{n=1}^{\infty} \subset \mathcal{F}$, then $\bigcup_{n=1}^{\infty} A_n \in \mathcal{F}$

**Theorem 2.1** (Properties of σ-Algebras): If $\mathcal{F}$ is a σ-algebra on $X$, then:
1. $\emptyset \in \mathcal{F}$ (since $\emptyset = X^c$)
2. $\mathcal{F}$ is closed under countable intersections: $\bigcap_{n=1}^{\infty} A_n \in \mathcal{F}$ when $A_n \in \mathcal{F}$
3. $\mathcal{F}$ is closed under finite unions and intersections
4. $\mathcal{F}$ is closed under set differences: $A \setminus B \in \mathcal{F}$ when $A, B \in \mathcal{F}$

**Proof**: 
(1) $\emptyset = X^c \in \mathcal{F}$ by complement closure.
(2) $\bigcap_{n=1}^{\infty} A_n = \left(\bigcup_{n=1}^{\infty} A_n^c\right)^c \in \mathcal{F}$ by De Morgan's laws.
(3) Follows from (2) with finite sequences.
(4) $A \setminus B = A \cap B^c \in \mathcal{F}$ since $B^c \in \mathcal{F}$ and intersections are closed. □

### Construction and Examples

**Definition 2.2** (Generated σ-Algebra): For any collection $\mathcal{C}$ of subsets of $X$, the **σ-algebra generated by $\mathcal{C}$**, denoted $\sigma(\mathcal{C})$, is the smallest σ-algebra containing $\mathcal{C}$:
$$\sigma(\mathcal{C}) = \bigcap \{\mathcal{F} : \mathcal{F} \text{ is a σ-algebra and } \mathcal{C} \subset \mathcal{F}\}$$

**Example 2.1** (Borel σ-Algebra): The **Borel σ-algebra** on $\mathbb{R}$ is:
$$\mathcal{B}(\mathbb{R}) = \sigma(\mathcal{T})$$
where $\mathcal{T}$ is the collection of all open sets in $\mathbb{R}$. Equivalently:
- $\mathcal{B}(\mathbb{R}) = \sigma\{(a,b) : a < b, a,b \in \mathbb{R}\}$ (open intervals)
- $\mathcal{B}(\mathbb{R}) = \sigma\{[a,b] : a \leq b, a,b \in \mathbb{R}\}$ (closed intervals)
- $\mathcal{B}(\mathbb{R}) = \sigma\{(-\infty, a] : a \in \mathbb{R}\}$ (left-closed rays)

**Example 2.2** (Discrete σ-Algebra): For any set $X$:
- **Power Set**: $\mathcal{F} = 2^X = \{A : A \subset X\}$ (all possible subsets)
- **Trivial σ-Algebra**: $\mathcal{F} = \{\emptyset, X\}$ (minimal σ-algebra)

**Example 2.3** (Product σ-Algebras): For measurable spaces $(X_1, \mathcal{F}_1)$ and $(X_2, \mathcal{F}_2)$:
$$\mathcal{F}_1 \otimes \mathcal{F}_2 = \sigma\{A_1 \times A_2 : A_1 \in \mathcal{F}_1, A_2 \in \mathcal{F}_2\}$$

### Advanced Constructions

**Definition 2.3** (π-System and λ-System): 
- A **π-system** is a collection $\mathcal{P}$ closed under finite intersections
- A **λ-system** (or Dynkin class) $\mathcal{D}$ satisfies:
  1. $X \in \mathcal{D}$
  2. If $A, B \in \mathcal{D}$ and $A \subset B$, then $B \setminus A \in \mathcal{D}$
  3. If $(A_n)$ is increasing and $A_n \in \mathcal{D}$, then $\bigcup A_n \in \mathcal{D}$

**Theorem 2.2** (π-λ Theorem): If $\mathcal{P}$ is a π-system and $\mathcal{D}$ is a λ-system with $\mathcal{P} \subset \mathcal{D}$, then $\sigma(\mathcal{P}) \subset \mathcal{D}$.

**Applications in Probability Theory**:

**Filtrations**: A **filtration** $(\mathcal{F}_t)_{t \geq 0}$ is an increasing family of σ-algebras:
$$\mathcal{F}_s \subset \mathcal{F}_t \subset \mathcal{F} \quad \text{for } s \leq t$$

**Natural Filtration**: For a stochastic process $(X_t)_{t \geq 0}$:
$$\mathcal{F}_t^X = \sigma(X_s : s \leq t)$$

**Stopping Times**: A random time $\tau$ is a **stopping time** with respect to $(\mathcal{F}_t)$ if:
$$\{\tau \leq t\} \in \mathcal{F}_t \quad \text{for all } t \geq 0$$

**Machine Learning Applications**:

**Feature Spaces**: In supervised learning, feature mappings $\phi: \mathcal{X} \to \mathbb{R}^d$ require:
- $\mathcal{X}$ equipped with σ-algebra $\mathcal{A}$
- $\phi$ is $(\mathcal{A}, \mathcal{B}(\mathbb{R}^d))$-measurable

**Information Filtrations**: In online learning:
- $\mathcal{F}_t = \sigma(x_1, \ldots, x_t, y_1, \ldots, y_{t-1})$ (information at time $t$)
- Regret bounds depend on measurability with respect to natural filtrations

**Reproducible Learning**: Algorithm randomness requires careful σ-algebra constructions to ensure:
- Measurable selection of hypotheses
- Well-defined empirical risk functionals
- Proper conditioning in Bayesian methods

**High-Dimensional Statistics**: For $p$-dimensional observations with $p \to \infty$:
- Coordinate σ-algebras: $\mathcal{F}_k = \sigma(X_1, \ldots, X_k)$
- Tail σ-algebras: $\mathcal{T} = \bigcap_{k=1}^{\infty} \sigma(X_k, X_{k+1}, \ldots)$
- Applications to concentration inequalities and dimension reduction

## 4. Measurable Functions and Integration Theory {#measurable-functions}

### Measurable Functions

**Definition 4.1** (Measurable Function): Let $(X, \mathcal{F})$ and $(Y, \mathcal{G})$ be measurable spaces. A function $f: X \to Y$ is **$(\mathcal{F}, \mathcal{G})$-measurable** if:
$$f^{-1}(B) \in \mathcal{F} \quad \text{for all } B \in \mathcal{G}$$

**Theorem 4.1** (Measurability Criteria): For $f: X \to \mathbb{R}$ and Borel σ-algebra $\mathcal{B}(\mathbb{R})$, the following are equivalent:
1. $f$ is $(\mathcal{F}, \mathcal{B}(\mathbb{R}))$-measurable
2. $\{x : f(x) \leq a\} \in \mathcal{F}$ for all $a \in \mathbb{R}$
3. $\{x : f(x) < a\} \in \mathcal{F}$ for all $a \in \mathbb{R}$
4. $\{x : f(x) \geq a\} \in \mathcal{F}$ for all $a \in \mathbb{R}$
5. $\{x : f(x) > a\} \in \mathcal{F}$ for all $a \in \mathbb{R}$

### Lebesgue Integration

**Definition 4.2** (Simple Function): A function $s: X \to \mathbb{R}$ is **simple** if it takes finitely many values:
$$s = \sum_{i=1}^n a_i \mathbb{1}_{A_i}$$
where $a_i \in \mathbb{R}$, $A_i \in \mathcal{F}$ are disjoint, and $\mathbb{1}_{A_i}$ is the indicator function.

**Definition 4.3** (Integral of Simple Function): For simple function $s = \sum_{i=1}^n a_i \mathbb{1}_{A_i}$:
$$\int s \, d\mu = \sum_{i=1}^n a_i \mu(A_i)$$

**Definition 4.4** (Lebesgue Integral): For non-negative measurable $f: X \to [0, \infty]$:
$$\int f \, d\mu = \sup \left\{\int s \, d\mu : s \text{ simple, } 0 \leq s \leq f\right\}$$

For general $f$, write $f = f^+ - f^-$ where $f^+ = \max(f, 0)$ and $f^- = \max(-f, 0)$:
$$\int f \, d\mu = \int f^+ \, d\mu - \int f^- \, d\mu$$
provided at least one integral is finite.

### Properties of Integration

**Theorem 4.2** (Linearity): For integrable functions $f, g$ and constants $a, b \in \mathbb{R}$:
$$\int (af + bg) \, d\mu = a\int f \, d\mu + b\int g \, d\mu$$

**Theorem 4.3** (Monotonicity): If $f \leq g$ almost everywhere and both are integrable:
$$\int f \, d\mu \leq \int g \, d\mu$$

**Theorem 4.4** (Dominated Convergence Theorem - DCT): Let $f_n$ be measurable with $f_n \to f$ pointwise. If there exists integrable $g$ with $|f_n| \leq g$ for all $n$, then:
$$\lim_{n \to \infty} \int f_n \, d\mu = \int f \, d\mu$$

**Theorem 4.5** (Monotone Convergence Theorem - MCT): If $0 \leq f_1 \leq f_2 \leq \cdots$ and $f_n \to f$ pointwise, then:
$$\lim_{n \to \infty} \int f_n \, d\mu = \int f \, d\mu$$

**Theorem 4.6** (Fatou's Lemma): For non-negative measurable functions $(f_n)$:
$$\int \liminf_{n \to \infty} f_n \, d\mu \leq \liminf_{n \to \infty} \int f_n \, d\mu$$

### $L^p$ Spaces

**Definition 4.5** ($L^p$ Space): For $1 \leq p < \infty$:
$$L^p(X, \mathcal{F}, \mu) = \left\{f : \int |f|^p \, d\mu < \infty\right\}$$

The **$L^p$ norm** is:
$$\|f\|_p = \left(\int |f|^p \, d\mu\right)^{1/p}$$

For $p = \infty$:
$$\|f\|_\infty = \inf\{M : |f| \leq M \text{ a.e.}\}$$

**Theorem 4.7** (Hölder's Inequality): For $1/p + 1/q = 1$ with $p, q \in [1, \infty]$:
$$\int |fg| \, d\mu \leq \|f\|_p \|g\|_q$$

**Theorem 4.8** (Minkowski's Inequality): For $1 \leq p \leq \infty$:
$$\|f + g\|_p \leq \|f\|_p + \|g\|_p$$

**Theorem 4.9** (Riesz-Fischer): $L^p$ is complete for $1 \leq p \leq \infty$ (is a Banach space).

```python
# Demonstration: Product Measures, Radon-Nikodym, and ML Applications

def demonstrate_advanced_measure_theory():
    """Advanced demonstrations: product measures, R-N theorem, and ML applications"""
    
    print("="*80)
    print("ADVANCED MEASURE THEORY AND APPLICATIONS")
    print("="*80)
    
    fig, axes = plt.subplots(2, 3, figsize=(16, 11))
    
    # 1. Product Measures and Fubini's Theorem
    print("\n1. PRODUCT MEASURES AND FUBINI'S THEOREM")
    print("-" * 60)
    
    # Double integral demonstration
    x_vals = np.linspace(0, 1, 100)
    y_vals = np.linspace(0, 1, 100)
    X, Y = np.meshgrid(x_vals, y_vals)
    
    # Function f(x,y) = xy
    Z = X * Y
    
    # Double integral
    dx = x_vals[1] - x_vals[0]
    dy = y_vals[1] - y_vals[0]
    double_integral = np.sum(Z) * dx * dy
    
    # Iterated integrals
    inner_y = np.trapz(Z, y_vals, axis=0)
    iterated_xy = np.trapz(inner_y, x_vals)
    
    inner_x = np.trapz(Z, x_vals, axis=1)
    iterated_yx = np.trapz(inner_x, y_vals)
    
    theoretical = 0.25  # ∫∫ xy dxdy = 1/4
    
    print("f(x,y) = xy on [0,1]²")
    print(f"Double integral: ∫∫ f dxdy = {double_integral:.6f}")
    print(f"Iterated (x,y): ∫(∫ f dy)dx = {iterated_xy:.6f}")
    print(f"Iterated (y,x): ∫(∫ f dx)dy = {iterated_yx:.6f}")
    print(f"Theoretical: 1/4 = {theoretical:.6f}")
    print(f"Fubini verified: all equal ✓")
    
    # Visualize
    im = axes[0, 0].contourf(X, Y, Z, levels=20, cmap='viridis')
    axes[0, 0].set_title('f(x,y) = xy (Product Measure)', fontsize=12, fontweight='bold')
    axes[0, 0].set_xlabel('x')
    axes[0, 0].set_ylabel('y')
    plt.colorbar(im, ax=axes[0, 0])
    
    # 2. Radon-Nikodym Theorem
    print("\n2. RADON-NIKODYM THEOREM AND DENSITIES")
    print("-" * 60)
    
    x_rn = np.linspace(0, 1, 500)
    
    # Reference measure: uniform
    mu_density = np.ones_like(x_rn)
    mu_density = mu_density / np.trapz(mu_density, x_rn)
    
    # Density function (Radon-Nikodym derivative)
    rn_derivative = 2 * x_rn  # dν/dμ = 2x
    
    # Absolutely continuous measure
    nu_density = rn_derivative * mu_density
    nu_density = nu_density / np.trapz(nu_density, x_rn)
    
    print("Reference measure μ: uniform on [0,1]")
    print("Radon-Nikodym derivative: dν/dμ = 2x")
    print(f"∫(dν/dμ) dμ = {np.trapz(rn_derivative * mu_density, x_rn):.6f}")
    print("Measure ν is absolutely continuous w.r.t. μ")
    
    # Verify by integration
    test_set_mask = (x_rn >= 0.3) & (x_rn <= 0.7)
    mu_measure = np.trapz(mu_density[test_set_mask], x_rn[test_set_mask])
    nu_measure = np.trapz(nu_density[test_set_mask], x_rn[test_set_mask])
    rn_computed = np.trapz(rn_derivative[test_set_mask] * mu_density[test_set_mask], 
                           x_rn[test_set_mask])
    
    print(f"\nVerification on [0.3, 0.7]:")
    print(f"  ν([0.3,0.7]) = {nu_measure:.6f}")
    print(f"  ∫_{[0.3,0.7]} (dν/dμ) dμ = {rn_computed:.6f}")
    print(f"  Match: {abs(nu_measure - rn_computed) < 1e-6} ✓")
    
    axes[0, 1].plot(x_rn, mu_density, label='μ (uniform)', linewidth=2)
    axes[0, 1].plot(x_rn, nu_density, label='ν (dν/dμ = 2x)', linewidth=2)
    axes[0, 1].fill_between(x_rn, nu_density, alpha=0.3)
    axes[0, 1].axvspan(0.3, 0.7, alpha=0.2, color='gray', label='Test set')
    axes[0, 1].set_title('Radon-Nikodym Theorem', fontsize=12, fontweight='bold')
    axes[0, 1].set_xlabel('x')
    axes[0, 1].set_ylabel('Density')
    axes[0, 1].legend(fontsize=9)
    axes[0, 1].grid(True, alpha=0.3)
    
    # 3. Empirical measures and convergence
    print("\n3. EMPIRICAL MEASURES (Law of Large Numbers)")
    print("-" * 60)
    
    # True distribution: N(0, 1)
    true_mean = 0
    true_std = 1
    
    sample_sizes = [10, 50, 100, 500, 1000, 5000]
    empirical_means = []
    empirical_stds = []
    
    np.random.seed(42)
    
    print("True distribution: N(0,1)")
    print("Empirical convergence:")
    
    for n in sample_sizes:
        samples = np.random.normal(true_mean, true_std, n)
        emp_mean = np.mean(samples)
        emp_std = np.std(samples, ddof=1)
        empirical_means.append(emp_mean)
        empirical_stds.append(emp_std)
        print(f"  n={n:5d}: mean={emp_mean:+.4f}, std={emp_std:.4f}")
    
    print(f"True values: mean={true_mean}, std={true_std}")
    
    axes[0, 2].semilogx(sample_sizes, empirical_means, 'o-', label='Empirical mean', 
                        linewidth=2, markersize=8)
    axes[0, 2].axhline(y=true_mean, color='r', linestyle='--', linewidth=2, 
                       label=f'True mean = {true_mean}')
    axes[0, 2].set_title('Empirical Measure Convergence', fontsize=12, fontweight='bold')
    axes[0, 2].set_xlabel('Sample size n')
    axes[0, 2].set_ylabel('Empirical mean')
    axes[0, 2].legend(fontsize=9)
    axes[0, 2].grid(True, alpha=0.3)
    
    # 4. Concentration of measure
    print("\n4. CONCENTRATION OF MEASURE")
    print("-" * 60)
    
    dimensions = [2, 5, 10, 20, 50, 100]
    n_samples = 1000
    
    concentration_data = []
    
    print("Random points on unit sphere:")
    print(f"{'Dim':>3} {'Mean dist':>10} {'Std dist':>10} {'CV':>8}")
    print("-" * 35)
    
    for d in dimensions:
        # Sample from unit sphere
        X = np.random.randn(n_samples, d)
        norms = np.linalg.norm(X, axis=1, keepdims=True)
        X_sphere = X / norms
        
        # Pairwise distances (sample)
        sample_idx = np.random.choice(n_samples, min(100, n_samples), replace=False)
        distances = []
        for i in range(len(sample_idx)-1):
            dist = np.linalg.norm(X_sphere[sample_idx[i]] - X_sphere[sample_idx[i+1]])
            distances.append(dist)
        
        mean_dist = np.mean(distances)
        std_dist = np.std(distances)
        cv = std_dist / mean_dist  # Coefficient of variation
        
        concentration_data.append((d, mean_dist, std_dist, cv))
        print(f"{d:3d} {mean_dist:10.4f} {std_dist:10.4f} {cv:8.4f}")
    
    print("Concentration effect: CV decreases with dimension")
    
    dims, means, stds, cvs = zip(*concentration_data)
    axes[1, 0].plot(dims, cvs, 'o-', linewidth=2, markersize=8, color='darkblue')
    axes[1, 0].set_title('Concentration of Measure', fontsize=12, fontweight='bold')
    axes[1, 0].set_xlabel('Dimension')
    axes[1, 0].set_ylabel('Coefficient of Variation')
    axes[1, 0].grid(True, alpha=0.3)
    axes[1, 0].set_xscale('log')
    
    # 5. KL Divergence and Information Theory
    print("\n5. KL DIVERGENCE (Information Theory)")
    print("-" * 60)
    
    # Two distributions
    n_bins = 20
    x_kl = np.linspace(-3, 3, n_bins)
    
    # P: N(0, 1)
    p = np.exp(-x_kl**2/2) / np.sqrt(2*np.pi)
    p = p / np.sum(p)  # Normalize
    
    # Q: N(0.5, 1.2²)
    q = np.exp(-(x_kl-0.5)**2/(2*1.2**2)) / (1.2*np.sqrt(2*np.pi))
    q = q / np.sum(q)  # Normalize
    
    # KL divergences
    epsilon = 1e-10
    kl_pq = np.sum(p * np.log((p + epsilon) / (q + epsilon)))
    kl_qp = np.sum(q * np.log((q + epsilon) / (p + epsilon)))
    
    # Entropy
    H_p = -np.sum(p * np.log(p + epsilon))
    H_q = -np.sum(q * np.log(q + epsilon))
    
    print("P ~ N(0, 1), Q ~ N(0.5, 1.2²)")
    print(f"H(P) = {H_p:.6f} bits")
    print(f"H(Q) = {H_q:.6f} bits")
    print(f"KL(P||Q) = {kl_pq:.6f}")
    print(f"KL(Q||P) = {kl_qp:.6f}")
    print(f"KL divergence is asymmetric: {kl_pq:.4f} ≠ {kl_qp:.4f}")
    
    axes[1, 1].bar(x_kl - 0.1, p, width=0.2, label='P ~ N(0,1)', alpha=0.7)
    axes[1, 1].bar(x_kl + 0.1, q, width=0.2, label='Q ~ N(0.5,1.2²)', alpha=0.7)
    axes[1, 1].set_title(f'KL Divergence: KL(P||Q)={kl_pq:.3f}', 
                         fontsize=12, fontweight='bold')
    axes[1, 1].set_xlabel('x')
    axes[1, 1].set_ylabel('Probability')
    axes[1, 1].legend(fontsize=9)
    axes[1, 1].grid(True, alpha=0.3, axis='y')
    
    # 6. PAC Learning and Generalization
    print("\n6. PAC LEARNING AND GENERALIZATION BOUNDS")
    print("-" * 60)
    
    # Simulate classification with different sample sizes
    n_features = 2
    sample_sizes_pac = [20, 50, 100, 200, 500, 1000]
    
    generalization_gaps = []
    
    print("Binary classification with linear hypothesis")
    print(f"{'n':>4} {'Emp Risk':>10} {'True Risk':>10} {'Gap':>8}")
    print("-" * 35)
    
    for n in sample_sizes_pac:
        # Generate data
        X_train = np.random.randn(n, n_features)
        y_train = (X_train[:, 0] + X_train[:, 1] > 0).astype(int)
        
        # Add noise
        noise_idx = np.random.rand(n) < 0.1
        y_train[noise_idx] = 1 - y_train[noise_idx]
        
        # Simple linear classifier
        w = np.array([1, 1]) / np.sqrt(2)
        
        # Empirical risk
        y_pred_train = (X_train @ w > 0).astype(int)
        emp_risk = np.mean(y_pred_train != y_train)
        
        # True risk (large test set)
        X_test = np.random.randn(10000, n_features)
        y_test = (X_test[:, 0] + X_test[:, 1] > 0).astype(int)
        noise_test = np.random.rand(10000) < 0.1
        y_test[noise_test] = 1 - y_test[noise_test]
        
        y_pred_test = (X_test @ w > 0).astype(int)
        true_risk = np.mean(y_pred_test != y_test)
        
        gap = abs(true_risk - emp_risk)
        generalization_gaps.append(gap)
        
        print(f"{n:4d} {emp_risk:10.6f} {true_risk:10.6f} {gap:8.6f}")
    
    # Plot generalization gap
    axes[1, 2].loglog(sample_sizes_pac, generalization_gaps, 'o-', linewidth=2, 
                      markersize=8, label='Empirical gap')
    
    # Theoretical bound: O(√(log n / n))
    theoretical_bound = np.sqrt(np.log(np.array(sample_sizes_pac)) / np.array(sample_sizes_pac))
    theoretical_bound = theoretical_bound * generalization_gaps[0] / theoretical_bound[0]
    axes[1, 2].loglog(sample_sizes_pac, theoretical_bound, '--', linewidth=2, 
                      label='O(√(log n / n))', color='red')
    
    axes[1, 2].set_title('PAC Learning: Generalization Gap', fontsize=12, fontweight='bold')
    axes[1, 2].set_xlabel('Sample size n')
    axes[1, 2].set_ylabel('|True Risk - Emp Risk|')
    axes[1, 2].legend(fontsize=9)
    axes[1, 2].grid(True, alpha=0.3, which='both')
    
    plt.tight_layout()
    plt.savefig('/Users/ruichaowang/GitHub/Mathematics_for_AI/04_Advanced_Mathematics/measure_theory_applications.png', 
                dpi=150, bbox_inches='tight')
    plt.show()
    
    print("\n" + "="*80)
    print("Advanced measure theory demonstrations completed!")
    print("="*80)

# Run demonstration
demonstrate_advanced_measure_theory()
```
