## 1. Introduction to Computational Complexity Theory {#introduction}

Computational complexity theory provides the mathematical framework for analyzing the computational resources required to solve problems. This field is fundamental to understanding the theoretical limits of algorithms and has profound implications for machine learning, optimization, and artificial intelligence.

### Fundamental Concepts

**Definition 1.1** (Decision Problem): A decision problem is a function $L: \{0,1\}^* \to \{0,1\}$ that maps strings to boolean values. The **language** $L$ is the set of strings $x$ such that $L(x) = 1$.

**Definition 1.2** (Time Complexity): For a Turing machine $M$, the time complexity function is $T_M(n) = \max_{|x|=n} \text{steps}(M, x)$ where $\text{steps}(M,x)$ is the number of computation steps on input $x$.

**Definition 1.3** (Space Complexity): The space complexity function is $S_M(n) = \max_{|x|=n} \text{space}(M, x)$ where $\text{space}(M,x)$ is the maximum tape cells used.

### Asymptotic Notation Theory

**Definition 1.4** (Landau Notation): For functions $f, g: \mathbb{N} \to \mathbb{R}^+$:
- $f(n) = O(g(n))$ if $\exists c, n_0: f(n) \leq c \cdot g(n)$ for all $n \geq n_0$
- $f(n) = \Omega(g(n))$ if $\exists c, n_0: f(n) \geq c \cdot g(n)$ for all $n \geq n_0$  
- $f(n) = \Theta(g(n))$ if $f(n) = O(g(n))$ and $f(n) = \Omega(g(n))$
- $f(n) = o(g(n))$ if $\lim_{n \to \infty} \frac{f(n)}{g(n)} = 0$

**Theorem 1.1** (Hierarchy Properties): The complexity measures satisfy:
1. **Transitivity**: If $f = O(g)$ and $g = O(h)$, then $f = O(h)$
2. **Reflexivity**: $f = O(f)$ for any function $f$
3. **Sum Rule**: If $f_1 = O(g)$ and $f_2 = O(g)$, then $f_1 + f_2 = O(g)$
4. **Product Rule**: If $f_1 = O(g_1)$ and $f_2 = O(g_2)$, then $f_1 \cdot f_2 = O(g_1 \cdot g_2)$

### Machine Learning Complexity Implications

**Sample Complexity**: Number of training examples needed for learning
**Computational Complexity**: Time/space required for training and inference
**Statistical Complexity**: Fundamental limits on learnability (VC-dimension, Rademacher complexity)
**Approximation Complexity**: Quality-efficiency tradeoffs in approximate algorithms

## 2. Advanced Time and Space Complexity Analysis {#time-space}

### Hierarchy Theorems

**Theorem 2.1** (Time Hierarchy Theorem): For any time-constructible function $t(n)$, there exists a language $L$ such that:
- $L \in \text{DTIME}(t(n) \log t(n))$
- $L \notin \text{DTIME}(t(n))$

This establishes that more time provably allows solving strictly more problems.

**Theorem 2.2** (Space Hierarchy Theorem): For space-constructible $s(n) \geq \log n$:
$$\text{DSPACE}(s(n)) \subsetneq \text{DSPACE}(s(n) \log s(n))$$

### Advanced Complexity Measures

**Definition 2.1** (Alternating Time): An alternating Turing machine uses existential and universal states. $\text{ATIME}(t(n))$ is the class of problems solvable by alternating machines in time $O(t(n))$.

**Theorem 2.3** (Chandra-Kozen-Stockmeyer): 
$$\text{ATIME}(t(n)) = \text{DSPACE}(t(n))$$
for $t(n) \geq \log n$.

### Circuit Complexity

**Definition 2.2** (Boolean Circuit): A boolean circuit is a directed acyclic graph with AND, OR, NOT gates computing a boolean function $f: \{0,1\}^n \to \{0,1\}$.

**Definition 2.3** (Circuit Classes):
- $\text{SIZE}(s(n))$: Functions computable by circuits of size $O(s(n))$
- $\text{DEPTH}(d(n))$: Functions computable by circuits of depth $O(d(n))$
- $\text{NC}^k = \text{DEPTH}(O(\log^k n)) \cap \text{SIZE}(n^{O(1)})$

**Theorem 2.4** (Circuit Lower Bounds): Almost all boolean functions require exponential circuit size, but proving super-polynomial lower bounds for specific functions in $\text{NP}$ remains open.

### Algebraic Complexity

**Definition 2.4** (Arithmetic Circuit): Directed acyclic graph with $+, \times$ gates computing polynomials over a field.

**Valiant's Conjecture**: The permanent $\text{per}(X) = \sum_{\sigma} \prod_{i} X_{i,\sigma(i)}$ requires super-polynomial arithmetic circuit size.

### Machine Learning Applications:

**Deep Learning Expressivity**: 
- Circuit complexity characterizes neural network expressivity
- Depth-width tradeoffs in neural networks
- Boolean circuit lower bounds inform neural architecture limitations

**Kernel Computation**: 
- Computing kernel matrices has complexity implications
- Fast multipole methods and low-rank approximations
- Sparse kernel methods for scalability

**Optimization Complexity**:
- First-order methods: $O(1/\varepsilon^2)$ for convex, $O(1/\varepsilon^4)$ for non-convex
- Second-order methods: Newton's method complexity analysis
- Gradient descent in non-convex landscapes

```python
class ComplexityTheory:
    def __init__(self):
        self.results = {}
    
    def time_complexity_demo(self):
        """Demonstrate different time complexities empirically"""
        print("Time Complexity Demonstration:")
        
        def constant_time(arr):
            """O(1) - Access first element"""
            return arr[0] if len(arr) > 0 else None
        
        def linear_time(arr):
            """O(n) - Sum all elements"""
            return sum(arr)
        
        def quadratic_time(arr):
            """O(n²) - All pairs sum"""
            total = 0
            for i in range(len(arr)):
                for j in range(len(arr)):
                    total += arr[i] * arr[j]
            return total
        
        def logarithmic_time(arr, target):
            """O(log n) - Binary search"""
            arr = sorted(arr)
            left, right = 0, len(arr) - 1
            while left <= right:
                mid = (left + right) // 2
                if arr[mid] == target:
                    return mid
                elif arr[mid] < target:
                    left = mid + 1
                else:
                    right = mid - 1
            return -1
        
        # Test different input sizes
        sizes = [100, 200, 400, 800, 1600]
        
        print(f"\nTiming results (seconds):")
        print(f"{'Size':<6} {'O(1)':<10} {'O(log n)':<10} {'O(n)':<10} {'O(n²)':<10}")
        
        for n in sizes:
            arr = list(range(n))
            
            # O(1)
            start = time.time()
            for _ in range(1000):
                constant_time(arr)
            time_o1 = time.time() - start
            
            # O(log n)
            start = time.time()
            for _ in range(1000):
                logarithmic_time(arr, n//2)
            time_olog = time.time() - start
            
            # O(n)
            start = time.time()
            for _ in range(100):
                linear_time(arr)
            time_on = time.time() - start
            
            # O(n²) - only for smaller sizes
            if n <= 400:
                start = time.time()
                quadratic_time(arr[:min(n//4, 100)])  # Reduced size
                time_on2 = time.time() - start
            else:
                time_on2 = float('inf')
            
            print(f"{n:<6} {time_o1:<10.6f} {time_olog:<10.6f} {time_on:<10.6f} {time_on2:<10.6f}")
        
        return sizes
    
    def np_complete_demo(self):
        """Demonstrate NP-complete problems"""
        print(f"\nNP-Complete Problems in ML:")
        
        # Feature selection problem (NP-hard)
        print(f"\n1. Feature Selection (Subset Selection):")
        print(f"   Problem: Choose k features to maximize accuracy")
        print(f"   Complexity: Requires checking C(n,k) = n!/(k!(n-k)!) subsets")
        
        # Example with small dataset
        X, y = make_classification(n_samples=100, n_features=10, n_informative=5, 
                                 n_redundant=0, random_state=42)
        
        n_features = X.shape[1]
        k = 3  # Select 3 features
        
        # Brute force approach
        best_score = 0
        best_features = None
        
        print(f"\n   Dataset: {X.shape[0]} samples, {n_features} features")
        print(f"   Selecting k={k} features")
        print(f"   Number of combinations: C({n_features},{k}) = {np.math.comb(n_features, k)}")
        
        # Try all combinations (exponential time)
        for feature_subset in combinations(range(n_features), k):
            X_subset = X[:, feature_subset]
            
            # Quick evaluation using logistic regression
            clf = LogisticRegression(random_state=42, max_iter=100)
            clf.fit(X_subset, y)
            score = clf.score(X_subset, y)
            
            if score > best_score:
                best_score = score
                best_features = feature_subset
        
        print(f"   Best features: {best_features}")
        print(f"   Best score: {best_score:.4f}")
        
        # Compare with greedy approximation
        selector = SelectKBest(k=k)
        X_selected = selector.fit_transform(X, y)
        selected_features = selector.get_support(indices=True)
        
        clf_greedy = LogisticRegression(random_state=42, max_iter=100)
        clf_greedy.fit(X_selected, y)
        greedy_score = clf_greedy.score(X_selected, y)
        
        print(f"\n   Greedy approximation features: {tuple(selected_features)}")
        print(f"   Greedy score: {greedy_score:.4f}")
        print(f"   Approximation ratio: {greedy_score/best_score:.4f}")
        
        return best_features, selected_features
    
    def complexity_growth_demo(self):
        """Demonstrate growth rates of complexity functions"""
        print(f"\nComplexity Growth Demonstration:")
        
        n_values = [1, 2, 5, 10, 15, 20]
        
        print(f"\nGrowth of common complexity functions:")
        print(f"{'n':<4} {'log n':<8} {'n':<8} {'n log n':<12} {'n²':<10} {'2^n':<12} {'n!':<15}")
        
        for n in n_values:
            log_n = np.log2(n) if n > 0 else 0
            n_log_n = n * log_n
            n_squared = n**2
            exp_n = 2**n if n <= 20 else float('inf')
            factorial_n = np.math.factorial(n) if n <= 20 else float('inf')
            
            print(f"{n:<4} {log_n:<8.2f} {n:<8} {n_log_n:<12.2f} {n_squared:<10} "
                  f"{exp_n:<12} {factorial_n:<15}")
        
        print(f"\nKey insights:")
        print(f"- Exponential and factorial growth become intractable quickly")
        print(f"- Even n²  can be problematic for large n")
        print(f"- Logarithmic and linear algorithms scale well")
        
        return n_values

# Create demo instance
demo = ComplexityTheory()
sizes = demo.time_complexity_demo()
best_feat, greedy_feat = demo.np_complete_demo()
n_vals = demo.complexity_growth_demo()
```
