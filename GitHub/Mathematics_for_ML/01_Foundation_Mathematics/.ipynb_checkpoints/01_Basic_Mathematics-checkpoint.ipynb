```python
import numpy as np
import matplotlib.pyplot as plt
import math
from fractions import Fraction

# Set up plotting style
plt.style.use('seaborn-v0_8')
plt.rcParams['figure.figsize'] = (10, 6)
```

## 1. Arithmetic Operations and Properties

### Basic Operations

The four fundamental arithmetic operations form the basis of all mathematical computations in ML:
- **Addition (+)**: Combining quantities
- **Subtraction (-)**: Finding differences
- **Multiplication (×)**: Repeated addition
- **Division (÷)**: Splitting into equal parts

### Important Properties

These properties are crucial for understanding how ML algorithms manipulate data:

```python
# Demonstrating arithmetic properties
a, b, c = 5, 3, 2

print("Commutative Property:")
print(f"Addition: {a} + {b} = {a + b}, {b} + {a} = {b + a}")
print(f"Multiplication: {a} × {b} = {a * b}, {b} × {a} = {b * a}")

print("\nAssociative Property:")
print(f"Addition: ({a} + {b}) + {c} = {(a + b) + c}, {a} + ({b} + {c}) = {a + (b + c)}")
print(f"Multiplication: ({a} × {b}) × {c} = {(a * b) * c}, {a} × ({b} × {c}) = {a * (b * c)}")

print("\nDistributive Property:")
print(f"{a} × ({b} + {c}) = {a * (b + c)}, ({a} × {b}) + ({a} × {c}) = {(a * b) + (a * c)}")
```

### ML Application: Data Preprocessing

These basic operations are used extensively in data preprocessing:

```python
# Example: Normalizing data (common ML preprocessing step)
data = np.array([10, 20, 30, 40, 50])

# Calculate mean and standard deviation
mean = np.sum(data) / len(data)  # Using basic arithmetic
variance = np.sum((data - mean) ** 2) / len(data)
std_dev = np.sqrt(variance)

# Normalize data (z-score normalization)
normalized_data = (data - mean) / std_dev

print(f"Original data: {data}")
print(f"Mean: {mean}")
print(f"Standard deviation: {std_dev:.2f}")
print(f"Normalized data: {normalized_data}")

# Visualize the transformation
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))

ax1.bar(range(len(data)), data)
ax1.set_title('Original Data')
ax1.set_xlabel('Index')
ax1.set_ylabel('Value')

ax2.bar(range(len(normalized_data)), normalized_data)
ax2.set_title('Normalized Data')
ax2.set_xlabel('Index')
ax2.set_ylabel('Normalized Value')
ax2.axhline(y=0, color='r', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```

## 2. Fractions, Decimals, and Percentages

Understanding different number representations is crucial for ML, especially when dealing with probabilities and ratios.

```python
# Working with fractions
frac1 = Fraction(3, 4)  # 3/4
frac2 = Fraction(1, 2)  # 1/2

print("Fraction Operations:")
print(f"{frac1} + {frac2} = {frac1 + frac2}")
print(f"{frac1} × {frac2} = {frac1 * frac2}")

# Converting between representations
decimal_value = 0.75
fraction_value = Fraction(decimal_value).limit_denominator()
percentage_value = decimal_value * 100

print(f"\nConversions:")
print(f"Decimal: {decimal_value}")
print(f"Fraction: {fraction_value}")
print(f"Percentage: {percentage_value}%")
```

### ML Application: Classification Accuracy

In classification tasks, we often express model performance as percentages:

```python
# Example: Classification results
total_predictions = 1000
correct_predictions = 850
incorrect_predictions = total_predictions - correct_predictions

# Calculate accuracy as fraction, decimal, and percentage
accuracy_fraction = Fraction(correct_predictions, total_predictions)
accuracy_decimal = correct_predictions / total_predictions
accuracy_percentage = accuracy_decimal * 100

print(f"Classification Results:")
print(f"Correct predictions: {correct_predictions}")
print(f"Total predictions: {total_predictions}")
print(f"Accuracy (fraction): {accuracy_fraction}")
print(f"Accuracy (decimal): {accuracy_decimal}")
print(f"Accuracy (percentage): {accuracy_percentage}%")

# Visualize results
labels = ['Correct', 'Incorrect']
sizes = [correct_predictions, incorrect_predictions]
colors = ['lightgreen', 'lightcoral']

plt.pie(sizes, labels=labels, colors=colors, autopct='%1.1f%%', startangle=90)
plt.title('Classification Accuracy')
plt.axis('equal')
plt.show()
```

## 3. Exponents and Logarithms

Exponents and logarithms are fundamental in ML, appearing in activation functions, loss functions, and probability distributions.

```python
# Exponent operations
base = 2
exponents = [1, 2, 3, 4, 5]
powers = [base ** exp for exp in exponents]

print("Powers of 2:")
for exp, power in zip(exponents, powers):
    print(f"2^{exp} = {power}")

# Logarithm operations
print("\nLogarithms (base 2):")
for power in powers:
    log_value = math.log2(power)
    print(f"log₂({power}) = {log_value}")

# Important logarithm properties
x, y = 8, 4
print(f"\nLogarithm Properties:")
print(f"log({x} × {y}) = log({x}) + log({y}) = {math.log(x*y):.3f} = {math.log(x) + math.log(y):.3f}")
print(f"log({x} / {y}) = log({x}) - log({y}) = {math.log(x/y):.3f} = {math.log(x) - math.log(y):.3f}")
print(f"log({x}^2) = 2 × log({x}) = {math.log(x**2):.3f} = {2 * math.log(x):.3f}")
```

### ML Application: Sigmoid Activation Function

The sigmoid function uses the exponential function and is crucial in neural networks:

```python
def sigmoid(x):
    """Sigmoid activation function"""
    return 1 / (1 + np.exp(-x))

# Generate data points
x = np.linspace(-10, 10, 100)
y = sigmoid(x)

# Plot sigmoid function
plt.figure(figsize=(10, 6))
plt.plot(x, y, 'b-', linewidth=2, label='Sigmoid(x) = 1/(1 + e^(-x))')
plt.axhline(y=0.5, color='r', linestyle='--', alpha=0.7, label='y = 0.5')
plt.axvline(x=0, color='r', linestyle='--', alpha=0.7, label='x = 0')
plt.grid(True, alpha=0.3)
plt.xlabel('Input (x)')
plt.ylabel('Output (sigmoid(x))')
plt.title('Sigmoid Activation Function')
plt.legend()
plt.ylim(-0.1, 1.1)

# Add annotations
plt.annotate('Approaches 1', xy=(5, 0.99), xytext=(6, 0.8),
            arrowprops=dict(arrowstyle='->', color='green'))
plt.annotate('Approaches 0', xy=(-5, 0.01), xytext=(-6, 0.2),
            arrowprops=dict(arrowstyle='->', color='green'))

plt.show()

# Show some specific values
test_values = [-2, -1, 0, 1, 2]
print("Sigmoid function values:")
for val in test_values:
    print(f"sigmoid({val}) = {sigmoid(val):.4f}")
```

## 4. Basic Set Theory

Set theory provides the foundation for understanding data collections, feature spaces, and probability theory in ML.

```python
# Basic set operations
A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7, 8}
C = {1, 3, 5, 7, 9}

print(f"Set A: {A}")
print(f"Set B: {B}")
print(f"Set C: {C}")

print(f"\nSet Operations:")
print(f"A ∪ B (Union): {A.union(B)}")
print(f"A ∩ B (Intersection): {A.intersection(B)}")
print(f"A - B (Difference): {A.difference(B)}")
print(f"A ⊕ B (Symmetric Difference): {A.symmetric_difference(B)}")

# Subset relationships
D = {2, 4}
print(f"\nSubset Relationships:")
print(f"D = {D}")
print(f"D ⊆ A (D is subset of A): {D.issubset(A)}")
print(f"A ⊇ D (A is superset of D): {A.issuperset(D)}")

# Cardinality (size)
print(f"\nCardinality:")
print(f"|A| = {len(A)}")
print(f"|B| = {len(B)}")
print(f"|A ∩ B| = {len(A.intersection(B))}")
```

### ML Application: Feature Selection

Set theory concepts are used in feature selection and data analysis:

```python
# Example: Feature selection using set operations
all_features = {'age', 'income', 'education', 'experience', 'skills', 'location', 'gender'}
important_features = {'age', 'income', 'education', 'experience'}
demographic_features = {'age', 'gender', 'location'}
economic_features = {'income', 'education', 'experience'}

print("Feature Analysis:")
print(f"All features: {all_features}")
print(f"Important features: {important_features}")
print(f"Demographic features: {demographic_features}")
print(f"Economic features: {economic_features}")

print("\nFeature Set Operations:")
overlap = demographic_features.intersection(economic_features)
combined = demographic_features.union(economic_features)
unused_features = all_features.difference(important_features)

print(f"Overlap between demographic and economic: {overlap}")
print(f"Combined demographic and economic: {combined}")
print(f"Unused features: {unused_features}")

# Visualize with Venn diagram concept
import matplotlib.patches as patches

fig, ax = plt.subplots(1, 1, figsize=(10, 8))

# Create circles for Venn diagram
circle1 = patches.Circle((0.3, 0.5), 0.25, fill=False, linewidth=2, edgecolor='blue')
circle2 = patches.Circle((0.7, 0.5), 0.25, fill=False, linewidth=2, edgecolor='red')

ax.add_patch(circle1)
ax.add_patch(circle2)

# Add labels
ax.text(0.15, 0.5, 'Demographic\nFeatures', ha='center', va='center', fontsize=10)
ax.text(0.85, 0.5, 'Economic\nFeatures', ha='center', va='center', fontsize=10)
ax.text(0.5, 0.5, 'age', ha='center', va='center', fontsize=10, weight='bold')

# Add feature labels outside circles
ax.text(0.15, 0.3, 'gender\nlocation', ha='center', va='center', fontsize=9)
ax.text(0.85, 0.3, 'income\neducation\nexperience', ha='center', va='center', fontsize=9)

ax.set_xlim(0, 1)
ax.set_ylim(0, 1)
ax.set_aspect('equal')
ax.axis('off')
ax.set_title('Feature Set Relationships', fontsize=14)

plt.show()
```

## 5. Functions and Their Properties

Functions are fundamental to ML - they represent mappings from inputs to outputs, which is exactly what ML models do.

```python
# Define various types of functions
def linear_function(x):
    """Linear function: f(x) = 2x + 1"""
    return 2 * x + 1

def quadratic_function(x):
    """Quadratic function: f(x) = x² - 4x + 3"""
    return x**2 - 4*x + 3

def exponential_function(x):
    """Exponential function: f(x) = 2^x"""
    return 2**x

def logarithmic_function(x):
    """Logarithmic function: f(x) = log₂(x)"""
    return np.log2(x) if x > 0 else float('nan')

# Generate data points
x_linear = np.linspace(-5, 5, 100)
x_exp = np.linspace(-3, 3, 100)
x_log = np.linspace(0.1, 10, 100)

# Calculate function values
y_linear = [linear_function(x) for x in x_linear]
y_quadratic = [quadratic_function(x) for x in x_linear]
y_exp = [exponential_function(x) for x in x_exp]
y_log = [logarithmic_function(x) for x in x_log]

# Plot all functions
fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(2, 2, figsize=(12, 10))

ax1.plot(x_linear, y_linear, 'b-', linewidth=2)
ax1.set_title('Linear Function: f(x) = 2x + 1')
ax1.grid(True, alpha=0.3)
ax1.set_xlabel('x')
ax1.set_ylabel('f(x)')

ax2.plot(x_linear, y_quadratic, 'r-', linewidth=2)
ax2.set_title('Quadratic Function: f(x) = x² - 4x + 3')
ax2.grid(True, alpha=0.3)
ax2.set_xlabel('x')
ax2.set_ylabel('f(x)')

ax3.plot(x_exp, y_exp, 'g-', linewidth=2)
ax3.set_title('Exponential Function: f(x) = 2^x')
ax3.grid(True, alpha=0.3)
ax3.set_xlabel('x')
ax3.set_ylabel('f(x)')

ax4.plot(x_log, y_log, 'm-', linewidth=2)
ax4.set_title('Logarithmic Function: f(x) = log₂(x)')
ax4.grid(True, alpha=0.3)
ax4.set_xlabel('x')
ax4.set_ylabel('f(x)')

plt.tight_layout()
plt.show()
```

### Function Properties Important in ML

```python
# Function properties demonstration
def check_function_properties(func, x_values, func_name):
    """Check various properties of a function"""
    y_values = [func(x) for x in x_values]
    
    print(f"\n{func_name} Properties:")
    
    # Domain and Range (for given values)
    valid_pairs = [(x, y) for x, y in zip(x_values, y_values) if not np.isnan(y)]
    if valid_pairs:
        x_domain = [pair[0] for pair in valid_pairs]
        y_range = [pair[1] for pair in valid_pairs]
        print(f"Domain (sample): [{min(x_domain):.2f}, {max(x_domain):.2f}]")
        print(f"Range (sample): [{min(y_range):.2f}, {max(y_range):.2f}]")
    
    # Monotonicity check
    differences = [y_values[i+1] - y_values[i] for i in range(len(y_values)-1) if not np.isnan(y_values[i]) and not np.isnan(y_values[i+1])]
    if differences:
        if all(d >= 0 for d in differences):
            print("Monotonicity: Non-decreasing")
        elif all(d <= 0 for d in differences):
            print("Monotonicity: Non-increasing")
        else:
            print("Monotonicity: Neither monotonic")

# Test function properties
test_x = np.linspace(1, 10, 20)

check_function_properties(linear_function, test_x, "Linear Function")
check_function_properties(exponential_function, test_x, "Exponential Function")
check_function_properties(logarithmic_function, test_x, "Logarithmic Function")
```

### ML Application: Loss Functions

Functions are central to ML as loss functions that measure prediction errors:

```python
# Common loss functions in ML
def mean_squared_error(y_true, y_pred):
    """Mean Squared Error loss function"""
    return np.mean((y_true - y_pred)**2)

def mean_absolute_error(y_true, y_pred):
    """Mean Absolute Error loss function"""
    return np.mean(np.abs(y_true - y_pred))

def huber_loss(y_true, y_pred, delta=1.0):
    """Huber loss function (robust to outliers)"""
    error = y_true - y_pred
    condition = np.abs(error) <= delta
    squared_loss = 0.5 * error**2
    linear_loss = delta * np.abs(error) - 0.5 * delta**2
    return np.where(condition, squared_loss, linear_loss)

# Generate example data
np.random.seed(42)
y_true = np.array([1, 2, 3, 4, 5])
y_pred = y_true + np.random.normal(0, 0.5, 5)  # Add some noise

print("Loss Function Comparison:")
print(f"True values: {y_true}")
print(f"Predicted values: {y_pred}")
print(f"MSE Loss: {mean_squared_error(y_true, y_pred):.4f}")
print(f"MAE Loss: {mean_absolute_error(y_true, y_pred):.4f}")
print(f"Huber Loss: {np.mean(huber_loss(y_true, y_pred)):.4f}")

# Visualize loss functions behavior
errors = np.linspace(-3, 3, 100)
mse_values = errors**2
mae_values = np.abs(errors)
huber_values = [huber_loss(0, -err)[0] for err in errors]

plt.figure(figsize=(10, 6))
plt.plot(errors, mse_values, 'b-', label='MSE: (y_true - y_pred)²', linewidth=2)
plt.plot(errors, mae_values, 'r-', label='MAE: |y_true - y_pred|', linewidth=2)
plt.plot(errors, huber_values, 'g-', label='Huber Loss', linewidth=2)
plt.xlabel('Prediction Error')
plt.ylabel('Loss Value')
plt.title('Comparison of Loss Functions')
plt.legend()
plt.grid(True, alpha=0.3)
plt.xlim(-3, 3)
plt.ylim(0, 9)
plt.show()
```
