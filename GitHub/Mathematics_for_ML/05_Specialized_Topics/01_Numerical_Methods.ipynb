## 1. Introduction and Theoretical Foundations {#introduction}

Numerical analysis provides the mathematical framework for developing, analyzing, and implementing algorithms that solve mathematical problems computationally. As the backbone of scientific computing and machine learning, it bridges the gap between pure mathematics and computational reality, addressing the fundamental challenges of finite precision arithmetic, algorithmic complexity, and numerical stability.

### Historical Context and Development

The development of numerical analysis evolved through several key phases:
- **Classical Methods**: Newton-Raphson, Euler methods, Gaussian elimination (17th-18th centuries)
- **Matrix Theory**: Development of systematic linear algebra algorithms (19th-20th centuries)
- **Computer Era**: Algorithms designed for digital computers, floating-point arithmetic (mid-20th century)
- **Modern Computational Science**: High-performance computing, parallel algorithms, machine learning integration (late 20th-21st centuries)

### Fundamental Mathematical Framework

**Definition 1.1** (Numerical Algorithm): A **numerical algorithm** is a finite sequence of computational steps that transforms input data into approximate solutions of mathematical problems.

**Key Components**:
1. **Input Data**: Problem parameters and initial conditions
2. **Computational Steps**: Arithmetic operations and logical decisions
3. **Output**: Approximate solution with quantified accuracy
4. **Error Analysis**: Bounds on approximation error

**Definition 1.2** (Well-Posed Problem): A mathematical problem is **well-posed** in the sense of Hadamard if:
1. **Existence**: A solution exists
2. **Uniqueness**: The solution is unique  
3. **Continuous Dependence**: Small changes in input data produce small changes in solution

**Definition 1.3** (Condition Number): For problem $f: \mathbb{R}^n \to \mathbb{R}^m$, the **absolute condition number** is:
$$\kappa_{abs} = \lim_{\delta \to 0} \sup_{\|\Delta x\| \leq \delta} \frac{\|f(x + \Delta x) - f(x)\|}{\|\Delta x\|}$$

The **relative condition number** is:
$$\kappa_{rel} = \frac{\kappa_{abs} \|x\|}{\|f(x)\|}$$

### Error Analysis Framework

**Definition 1.4** (Types of Error):
1. **Modeling Error**: Difference between real problem and mathematical model
2. **Data Error**: Uncertainty in input data  
3. **Discretization Error**: Error from approximating continuous problem
4. **Roundoff Error**: Error from finite precision arithmetic

**Total Error Bound**:
$$E_{total} = E_{modeling} + E_{data} + E_{discretization} + E_{roundoff}$$

**Definition 1.5** (Forward and Backward Error Analysis):
- **Forward Error**: $\|\tilde{f} - f\|$ where $\tilde{f}$ is computed solution
- **Backward Error**: Smallest $\delta$ such that $\tilde{f} = f(x + \delta)$ for some $\delta$

**Fundamental Theorem**: If backward error is small and problem is well-conditioned, then forward error is small.

### Floating-Point Arithmetic

**IEEE 754 Standard**: Represents real number $x$ as:
$$x = \pm (1 + \text{mantissa}) \times 2^{\text{exponent}}$$

**Machine Precision** ($\epsilon_{mach}$): Smallest positive number such that:
$$fl(1 + \epsilon_{mach}) > 1$$

For double precision: $\epsilon_{mach} \approx 2.22 \times 10^{-16}$

**Fundamental Operations**: For basic arithmetic operations:
$$fl(a \circ b) = (a \circ b)(1 + \delta), \quad |\delta| \leq \epsilon_{mach}$$

### Stability and Convergence Theory

**Definition 1.6** (Numerical Stability):
- **Forward Stable**: Small perturbations in input produce small changes in output
- **Backward Stable**: Computed solution is exact solution to nearby problem

**Definition 1.7** (Convergence): Sequence of approximations $\{x_n\}$ converges to solution $x^*$ if:
$$\lim_{n \to \infty} \|x_n - x^*\| = 0$$

**Order of Convergence**: If $\|x_{n+1} - x^*\| \leq C\|x_n - x^*\|^p$ for constant $C$, then convergence is of order $p$.

### Computational Complexity

**Definition 1.8** (Algorithm Complexity):
- **Time Complexity**: Number of operations as function of problem size
- **Space Complexity**: Memory requirements
- **Communication Complexity**: Data movement in parallel algorithms

**Common Complexity Classes**:
- $O(n)$: Linear algorithms (vector operations)
- $O(n^2)$: Quadratic algorithms (matrix-vector products)
- $O(n^3)$: Cubic algorithms (matrix factorizations)
- $O(n \log n)$: Fast algorithms (FFT, fast multipole)

### Applications in Machine Learning

**Optimization Algorithms**:

1. **Gradient Descent**: First-order optimization method
   $$x_{k+1} = x_k - \alpha_k \nabla f(x_k)$$
   - **Convergence Analysis**: For convex functions with Lipschitz gradients
   - **Learning Rate Selection**: Armijo, Wolfe conditions
   - **Variants**: Momentum, adaptive methods (Adam, RMSprop)

2. **Newton's Method**: Second-order optimization
   $$x_{k+1} = x_k - H_f(x_k)^{-1} \nabla f(x_k)$$
   - **Quadratic Convergence**: Near solution
   - **Modified Newton**: Handle indefinite Hessians
   - **Quasi-Newton**: BFGS, L-BFGS approximations

**Linear Algebra in ML**:

1. **Singular Value Decomposition**: $A = U\Sigma V^T$
   - **Principal Component Analysis**: Dimensionality reduction
   - **Matrix Completion**: Collaborative filtering

2. **Eigenvalue Problems**: $Ax = \lambda x$

**Numerical Integration and Sampling**:

1. **Monte Carlo Methods**:

   - **MCMC**: Markov Chain Monte Carlo sampling

2. **Gaussian Quadrature**: High-order integration rules

```python
# Root Finding Algorithms: Newton-Raphson, Bisection, Secant Methods

def newton_raphson(f, df, x0, tol=1e-8, max_iter=50):
    """Newton-Raphson method for finding roots"""
    x = x0
    history = [x]
    
    for i in range(max_iter):
        fx = f(x)
        dfx = df(x)
        
        if abs(dfx) < 1e-12:
            print(f"Derivative too small at iteration {i}")
            break
        
        x_new = x - fx / dfx
        history.append(x_new)
        
        if abs(x_new - x) < tol:
            return x_new, np.array(history), i+1
        
        x = x_new
    
    return x, np.array(history), max_iter

def bisection(f, a, b, tol=1e-8, max_iter=100):
    """Bisection method for finding roots"""
    if f(a) * f(b) > 0:
        raise ValueError("f(a) and f(b) must have opposite signs")
    
    history = []
    for i in range(max_iter):
        c = (a + b) / 2
        history.append(c)
        
        if abs(f(c)) < tol or (b - a) / 2 < tol:
            return c, np.array(history), i+1
        
        if f(c) * f(a) < 0:
            b = c
        else:
            a = c
    
    return (a + b) / 2, np.array(history), max_iter

def secant(f, x0, x1, tol=1e-8, max_iter=50):
    """Secant method for finding roots"""
    history = [x0, x1]
    
    for i in range(max_iter):
        fx0 = f(x0)
        fx1 = f(x1)
        
        if abs(fx1 - fx0) < 1e-12:
            print(f"Division by zero risk at iteration {i}")
            break
        
        x_new = x1 - fx1 * (x1 - x0) / (fx1 - fx0)
        history.append(x_new)
        
        if abs(x_new - x1) < tol:
            return x_new, np.array(history), i+1
        
        x0, x1 = x1, x_new
    
    return x1, np.array(history), max_iter

# Test function: f(x) = x^3 - 2x - 5
f = lambda x: x**3 - 2*x - 5
df = lambda x: 3*x**2 - 2

# Find roots using different methods
x0_newton = 2.0
root_newton, hist_newton, iter_newton = newton_raphson(f, df, x0_newton)

a, b = 1.0, 3.0
root_bisect, hist_bisect, iter_bisect = bisection(f, a, b)

x0_secant, x1_secant = 1.0, 3.0
root_secant, hist_secant, iter_secant = secant(f, x0_secant, x1_secant)

# Visualization
fig, axes = plt.subplots(1, 2, figsize=(15, 5))

# Plot 1: Function and roots
x = np.linspace(0, 3, 1000)
axes[0].plot(x, f(x), 'b-', linewidth=2, label='f(x) = x³ - 2x - 5')
axes[0].axhline(y=0, color='k', linestyle='--', alpha=0.3)
axes[0].plot(root_newton, f(root_newton), 'ro', markersize=10, label=f'Newton: {root_newton:.6f}')
axes[0].plot(root_bisect, f(root_bisect), 'gs', markersize=10, label=f'Bisection: {root_bisect:.6f}')
axes[0].plot(root_secant, f(root_secant), 'm^', markersize=10, label=f'Secant: {root_secant:.6f}')
axes[0].set_xlabel('x', fontsize=12)
axes[0].set_ylabel('f(x)', fontsize=12)
axes[0].set_title('Root Finding: f(x) = x³ - 2x - 5', fontsize=14, fontweight='bold')
axes[0].legend(fontsize=10)
axes[0].grid(True, alpha=0.3)

# Plot 2: Convergence comparison
iterations_newton = np.arange(len(hist_newton))
iterations_bisect = np.arange(len(hist_bisect))
iterations_secant = np.arange(len(hist_secant))

errors_newton = np.abs(hist_newton - root_newton)
errors_bisect = np.abs(hist_bisect - root_bisect)
errors_secant = np.abs(hist_secant - root_secant)

axes[1].semilogy(iterations_newton, errors_newton, 'ro-', label=f'Newton ({iter_newton} iter)', linewidth=2)
axes[1].semilogy(iterations_bisect, errors_bisect, 'gs-', label=f'Bisection ({iter_bisect} iter)', linewidth=2)
axes[1].semilogy(iterations_secant, errors_secant, 'm^-', label=f'Secant ({iter_secant} iter)', linewidth=2)
axes[1].set_xlabel('Iteration', fontsize=12)
axes[1].set_ylabel('|Error| (log scale)', fontsize=12)
axes[1].set_title('Convergence Rate Comparison', fontsize=14, fontweight='bold')
axes[1].legend(fontsize=10)
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print(f"\nRoot Finding Results:")
print(f"Newton-Raphson: root = {root_newton:.10f}, iterations = {iter_newton}, f(root) = {f(root_newton):.2e}")
print(f"Bisection:      root = {root_bisect:.10f}, iterations = {iter_bisect}, f(root) = {f(root_bisect):.2e}")
print(f"Secant:         root = {root_secant:.10f}, iterations = {iter_secant}, f(root) = {f(root_secant):.2e}")
```

```python
# Numerical Integration: Trapezoid, Simpson's Rule, Gaussian Quadrature

def trapezoid_rule(f, a, b, n):
    """Trapezoid rule for numerical integration"""
    h = (b - a) / n
    x = np.linspace(a, b, n+1)
    y = f(x)
    return h * (0.5 * y[0] + np.sum(y[1:-1]) + 0.5 * y[-1])

def simpson_rule(f, a, b, n):
    """Simpson's rule for numerical integration (n must be even)"""
    if n % 2 != 0:
        n += 1
    h = (b - a) / n
    x = np.linspace(a, b, n+1)
    y = f(x)
    return h/3 * (y[0] + 4*np.sum(y[1:-1:2]) + 2*np.sum(y[2:-1:2]) + y[-1])

# Test function: f(x) = sin(x) * exp(-x/5)
f_int = lambda x: np.sin(x) * np.exp(-x/5)
a, b = 0, 10

# Analytical solution using scipy
true_value, _ = integrate.quad(f_int, a, b)

# Compute with different numbers of intervals
n_values = np.array([4, 8, 16, 32, 64, 128, 256])
trap_values = []
simp_values = []

for n in n_values:
    trap_values.append(trapezoid_rule(f_int, a, b, n))
    simp_values.append(simpson_rule(f_int, a, b, n))

trap_errors = np.abs(np.array(trap_values) - true_value)
simp_errors = np.abs(np.array(simp_values) - true_value)

# Gaussian quadrature
gauss_values = []
gauss_orders = [2, 3, 4, 5, 8, 12, 16]
for order in gauss_orders:
    val, _ = integrate.fixed_quad(f_int, a, b, n=order)
    gauss_values.append(val)
gauss_errors = np.abs(np.array(gauss_values) - true_value)

# Visualization
fig, axes = plt.subplots(1, 2, figsize=(15, 5))

# Plot 1: Function and integration regions
x = np.linspace(a, b, 1000)
y = f_int(x)
axes[0].plot(x, y, 'b-', linewidth=2, label='f(x) = sin(x)·exp(-x/5)')
axes[0].fill_between(x, 0, y, alpha=0.3, label=f'Area = {true_value:.6f}')

# Show trapezoid approximation with n=8
n_demo = 8
x_trap = np.linspace(a, b, n_demo+1)
y_trap = f_int(x_trap)
for i in range(n_demo):
    axes[0].plot([x_trap[i], x_trap[i+1]], [y_trap[i], y_trap[i+1]], 'r-', linewidth=1.5, alpha=0.7)
axes[0].plot(x_trap, y_trap, 'ro', markersize=6, label=f'Trapezoid (n={n_demo})')

axes[0].set_xlabel('x', fontsize=12)
axes[0].set_ylabel('f(x)', fontsize=12)
axes[0].set_title('Numerical Integration of sin(x)·exp(-x/5)', fontsize=14, fontweight='bold')
axes[0].legend(fontsize=10)
axes[0].grid(True, alpha=0.3)

# Plot 2: Error comparison
axes[1].loglog(n_values, trap_errors, 'ro-', linewidth=2, markersize=8, label='Trapezoid (O(h²))')
axes[1].loglog(n_values, simp_errors, 'bs-', linewidth=2, markersize=8, label="Simpson's (O(h⁴))")
axes[1].loglog(gauss_orders, gauss_errors, 'g^-', linewidth=2, markersize=8, label='Gaussian Quadrature')

# Add reference lines for convergence rates
axes[1].loglog(n_values, trap_errors[0] * (n_values[0]/n_values)**2, 'r--', alpha=0.5, label='O(n⁻²) reference')
axes[1].loglog(n_values, simp_errors[0] * (n_values[0]/n_values)**4, 'b--', alpha=0.5, label='O(n⁻⁴) reference')

axes[1].set_xlabel('Number of intervals (n)', fontsize=12)
axes[1].set_ylabel('Absolute Error', fontsize=12)
axes[1].set_title('Convergence Rates of Integration Methods', fontsize=14, fontweight='bold')
axes[1].legend(fontsize=9)
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print(f"\nNumerical Integration Results (True value = {true_value:.10f}):")
print(f"\nTrapezoid rule (n=256): {trap_values[-1]:.10f}, error = {trap_errors[-1]:.2e}")
print(f"Simpson's rule (n=256): {simp_values[-1]:.10f}, error = {simp_errors[-1]:.2e}")
print(f"Gaussian quad (n=16):   {gauss_values[-1]:.10f}, error = {gauss_errors[-1]:.2e}")
```

```python
# ODE Solvers: Euler, Runge-Kutta, and Adaptive Methods

def euler_method(f, y0, t):
    """Euler's method for solving ODEs"""
    y = np.zeros((len(t), len(y0) if hasattr(y0, '__len__') else 1))
    y[0] = y0
    
    for i in range(len(t) - 1):
        h = t[i+1] - t[i]
        y[i+1] = y[i] + h * np.array(f(t[i], y[i]))
    
    return y

def rk4_method(f, y0, t):
    """Fourth-order Runge-Kutta method"""
    y = np.zeros((len(t), len(y0) if hasattr(y0, '__len__') else 1))
    y[0] = y0
    
    for i in range(len(t) - 1):
        h = t[i+1] - t[i]
        k1 = np.array(f(t[i], y[i]))
        k2 = np.array(f(t[i] + h/2, y[i] + h*k1/2))
        k3 = np.array(f(t[i] + h/2, y[i] + h*k2/2))
        k4 = np.array(f(t[i] + h, y[i] + h*k3))
        
        y[i+1] = y[i] + (h/6) * (k1 + 2*k2 + 2*k3 + k4)
    
    return y

# Example 1: Simple exponential decay dy/dt = -y, y(0) = 1
# Analytical solution: y(t) = exp(-t)
def exponential_decay(t, y):
    return -y

t_span = (0, 5)
t_euler = np.linspace(t_span[0], t_span[1], 20)
t_rk4 = np.linspace(t_span[0], t_span[1], 10)
t_fine = np.linspace(t_span[0], t_span[1], 1000)
y0 = 1.0

# Solve with different methods
y_euler = euler_method(exponential_decay, y0, t_euler)
y_rk4 = rk4_method(exponential_decay, y0, t_rk4)
y_analytical = np.exp(-t_fine)

# Scipy's solve_ivp for comparison
sol_scipy = solve_ivp(exponential_decay, t_span, [y0], t_eval=t_fine, method='RK45')

# Example 2: Lotka-Volterra (predator-prey) equations
def lotka_volterra(t, y):
    """Lotka-Volterra predator-prey model"""
    prey, predator = y
    alpha, beta, gamma, delta = 1.0, 0.5, 0.5, 0.2  # Parameters
    
    dprey_dt = alpha * prey - beta * prey * predator
    dpredator_dt = delta * prey * predator - gamma * predator
    
    return [dprey_dt, dpredator_dt]

t_lv = np.linspace(0, 30, 1000)
y0_lv = [10, 5]  # Initial: 10 prey, 5 predators
y_lv = rk4_method(lotka_volterra, y0_lv, t_lv)

# Visualization
fig = plt.figure(figsize=(16, 10))
gs = fig.add_gridspec(2, 2, hspace=0.3, wspace=0.3)

# Plot 1: Exponential decay - comparison of methods
ax1 = fig.add_subplot(gs[0, 0])
ax1.plot(t_fine, y_analytical, 'k-', linewidth=2, label='Analytical: y = e⁻ᵗ')
ax1.plot(t_euler, y_euler, 'ro--', linewidth=1.5, markersize=6, label=f'Euler (n={len(t_euler)})')
ax1.plot(t_rk4, y_rk4, 'bs-', linewidth=1.5, markersize=7, label=f'RK4 (n={len(t_rk4)})')
ax1.set_xlabel('t', fontsize=11)
ax1.set_ylabel('y(t)', fontsize=11)
ax1.set_title('ODE Solution: dy/dt = -y', fontsize=12, fontweight='bold')
ax1.legend(fontsize=9)
ax1.grid(True, alpha=0.3)

# Plot 2: Error comparison
ax2 = fig.add_subplot(gs[0, 1])
y_euler_fine = euler_method(exponential_decay, y0, t_fine)
y_rk4_fine = rk4_method(exponential_decay, y0, t_fine)
error_euler = np.abs(y_euler_fine.flatten() - y_analytical)
error_rk4 = np.abs(y_rk4_fine.flatten() - y_analytical)
ax2.semilogy(t_fine, error_euler, 'r-', linewidth=2, label='Euler Method')
ax2.semilogy(t_fine, error_rk4, 'b-', linewidth=2, label='RK4 Method')
ax2.set_xlabel('t', fontsize=11)
ax2.set_ylabel('Absolute Error (log scale)', fontsize=11)
ax2.set_title('Error Accumulation Over Time', fontsize=12, fontweight='bold')
ax2.legend(fontsize=9)
ax2.grid(True, alpha=0.3)

# Plot 3: Lotka-Volterra time series
ax3 = fig.add_subplot(gs[1, 0])
ax3.plot(t_lv, y_lv[:, 0], 'b-', linewidth=2, label='Prey')
ax3.plot(t_lv, y_lv[:, 1], 'r-', linewidth=2, label='Predator')
ax3.set_xlabel('Time', fontsize=11)
ax3.set_ylabel('Population', fontsize=11)
ax3.set_title('Lotka-Volterra: Predator-Prey Dynamics', fontsize=12, fontweight='bold')
ax3.legend(fontsize=9)
ax3.grid(True, alpha=0.3)

# Plot 4: Phase portrait
ax4 = fig.add_subplot(gs[1, 1])
ax4.plot(y_lv[:, 0], y_lv[:, 1], 'g-', linewidth=2, label='Trajectory')
ax4.plot(y_lv[0, 0], y_lv[0, 1], 'go', markersize=10, label='Start')
ax4.plot(y_lv[-1, 0], y_lv[-1, 1], 'rs', markersize=10, label='End')
ax4.set_xlabel('Prey Population', fontsize=11)
ax4.set_ylabel('Predator Population', fontsize=11)
ax4.set_title('Phase Portrait: Predator vs Prey', fontsize=12, fontweight='bold')
ax4.legend(fontsize=9)
ax4.grid(True, alpha=0.3)

plt.show()

print(f"\nODE Solver Results:")
print(f"Exponential decay at t=5:")
print(f"  Analytical: {np.exp(-5):.10f}")
print(f"  Euler:      {y_euler_fine[-1,0]:.10f} (error: {error_euler[-1]:.2e})")
print(f"  RK4:        {y_rk4_fine[-1,0]:.10f} (error: {error_rk4[-1]:.2e})")
print(f"\nLotka-Volterra final populations: Prey={y_lv[-1,0]:.2f}, Predator={y_lv[-1,1]:.2f}")
```

```python
# Numerical Linear Algebra: QR, SVD, and Iterative Methods

# Generate a test matrix
np.random.seed(42)
n = 100
A = np.random.randn(n, n)
A = A + A.T  # Make symmetric for better conditioning
b = np.random.randn(n)

# 1. QR Decomposition
Q, R = linalg.qr(A)

# Verify QR properties
is_orthogonal = np.allclose(Q.T @ Q, np.eye(n))
is_upper_triangular = np.allclose(R, np.triu(R))
reconstruction_error = np.linalg.norm(A - Q @ R, 'fro')

print("QR Decomposition:")
print(f"  Q is orthogonal: {is_orthogonal}")
print(f"  R is upper triangular: {is_upper_triangular}")
print(f"  ||A - QR||_F = {reconstruction_error:.2e}\n")

# 2. SVD (Singular Value Decomposition)
# Create a rectangular matrix for better visualization
m, n_rect = 50, 30
A_rect = np.random.randn(m, n_rect)

U, s, Vt = linalg.svd(A_rect, full_matrices=False)

# Verify SVD properties
reconstruction_svd = U @ np.diag(s) @ Vt
svd_error = np.linalg.norm(A_rect - reconstruction_svd, 'fro')

print("SVD (Singular Value Decomposition):")
print(f"  Matrix shape: {A_rect.shape}")
print(f"  U shape: {U.shape}, S shape: {s.shape}, Vt shape: {Vt.shape}")
print(f"  ||A - USVᵀ||_F = {svd_error:.2e}")
print(f"  Condition number: {s[0]/s[-1]:.2e}")
print(f"  Rank: {np.sum(s > 1e-10)}\n")

# 3. Iterative Methods for Linear Systems
# Make A positive definite for CG method
A_pd = A.T @ A + 0.1 * np.eye(n)
b_pd = np.random.randn(n)
x_true = linalg.solve(A_pd, b_pd)

def jacobi_iteration(A, b, x0, max_iter=100, tol=1e-6):
    """Jacobi iterative method"""
    D = np.diag(np.diag(A))
    R = A - D
    x = x0.copy()
    errors = []
    
    for i in range(max_iter):
        x = linalg.solve(D, b - R @ x)
        error = np.linalg.norm(x - x_true)
        errors.append(error)
        if error < tol:
            break
    
    return x, errors

def gauss_seidel(A, b, x0, max_iter=100, tol=1e-6):
    """Gauss-Seidel iterative method"""
    L = np.tril(A)
    U = A - L
    x = x0.copy()
    errors = []
    
    for i in range(max_iter):
        x = linalg.solve(L, b - U @ x)
        error = np.linalg.norm(x - x_true)
        errors.append(error)
        if error < tol:
            break
    
    return x, errors

def conjugate_gradient(A, b, x0, max_iter=100, tol=1e-6):
    """Conjugate Gradient method for positive definite systems"""
    x = x0.copy()
    r = b - A @ x
    p = r.copy()
    errors = []
    
    for i in range(max_iter):
        Ap = A @ p
        alpha = (r.T @ r) / (p.T @ Ap)
        x = x + alpha * p
        r_new = r - alpha * Ap
        
        error = np.linalg.norm(x - x_true)
        errors.append(error)
        
        if error < tol:
            break
        
        beta = (r_new.T @ r_new) / (r.T @ r)
        p = r_new + beta * p
        r = r_new
    
    return x, errors

# Test iterative methods
x0 = np.zeros(n)
x_jacobi, err_jacobi = jacobi_iteration(A_pd, b_pd, x0, max_iter=200)
x_gs, err_gs = gauss_seidel(A_pd, b_pd, x0, max_iter=200)
x_cg, err_cg = conjugate_gradient(A_pd, b_pd, x0, max_iter=200)

print(f"Iterative Methods (system size n={n}):")
print(f"  Jacobi:        {len(err_jacobi)} iterations, final error: {err_jacobi[-1]:.2e}")
print(f"  Gauss-Seidel:  {len(err_gs)} iterations, final error: {err_gs[-1]:.2e}")
print(f"  Conj Gradient: {len(err_cg)} iterations, final error: {err_cg[-1]:.2e}")

# Visualization
fig, axes = plt.subplots(2, 2, figsize=(15, 12))

# Plot 1: Singular values
axes[0, 0].semilogy(s, 'bo-', linewidth=2, markersize=6)
axes[0, 0].axhline(y=1e-10, color='r', linestyle='--', label='Threshold')
axes[0, 0].set_xlabel('Index', fontsize=11)
axes[0, 0].set_ylabel('Singular Value (log scale)', fontsize=11)
axes[0, 0].set_title('SVD: Singular Value Spectrum', fontsize=12, fontweight='bold')
axes[0, 0].legend(fontsize=9)
axes[0, 0].grid(True, alpha=0.3)

# Plot 2: Low-rank approximation
ranks = [1, 3, 5, 10, 20, 30]
reconstruction_errors = []
for r in ranks:
    A_r = U[:, :r] @ np.diag(s[:r]) @ Vt[:r, :]
    reconstruction_errors.append(np.linalg.norm(A_rect - A_r, 'fro'))

axes[0, 1].plot(ranks, reconstruction_errors, 'rs-', linewidth=2, markersize=8)
axes[0, 1].set_xlabel('Rank k', fontsize=11)
axes[0, 1].set_ylabel('||A - Aₖ||_F', fontsize=11)
axes[0, 1].set_title('SVD: Low-Rank Approximation Error', fontsize=12, fontweight='bold')
axes[0, 1].grid(True, alpha=0.3)

# Plot 3: Convergence of iterative methods
axes[1, 0].semilogy(err_jacobi, 'r-', linewidth=2, label=f'Jacobi ({len(err_jacobi)} iter)')
axes[1, 0].semilogy(err_gs, 'b-', linewidth=2, label=f'Gauss-Seidel ({len(err_gs)} iter)')
axes[1, 0].semilogy(err_cg, 'g-', linewidth=2, label=f'Conjugate Gradient ({len(err_cg)} iter)')
axes[1, 0].set_xlabel('Iteration', fontsize=11)
axes[1, 0].set_ylabel('Error ||x - x*|| (log scale)', fontsize=11)
axes[1, 0].set_title('Iterative Methods: Convergence Comparison', fontsize=12, fontweight='bold')
axes[1, 0].legend(fontsize=9)
axes[1, 0].grid(True, alpha=0.3)

# Plot 4: QR visualization (structure)
# Show sparsity pattern of R
R_small = linalg.qr(np.random.randn(20, 20))[1]
axes[1, 1].imshow(np.abs(R_small) > 1e-10, cmap='binary', interpolation='nearest')
axes[1, 1].set_xlabel('Column', fontsize=11)
axes[1, 1].set_ylabel('Row', fontsize=11)
axes[1, 1].set_title('QR: Upper Triangular Structure of R', fontsize=12, fontweight='bold')

plt.tight_layout()
plt.show()
```

```python
# Optimization Methods: Gradient Descent Variants, Newton's Method, Conjugate Gradient

# Test function: Rosenbrock function (banana function)
def rosenbrock(x):
    """Rosenbrock function: f(x,y) = (1-x)² + 100(y-x²)²"""
    return (1 - x[0])**2 + 100 * (x[1] - x[0]**2)**2

def rosenbrock_grad(x):
    """Gradient of Rosenbrock function"""
    df_dx = -2*(1 - x[0]) - 400*x[0]*(x[1] - x[0]**2)
    df_dy = 200*(x[1] - x[0]**2)
    return np.array([df_dx, df_dy])

def rosenbrock_hess(x):
    """Hessian of Rosenbrock function"""
    h11 = 2 - 400*x[1] + 1200*x[0]**2
    h12 = -400*x[0]
    h21 = -400*x[0]
    h22 = 200
    return np.array([[h11, h12], [h21, h22]])

# 1. Gradient Descent with different variants
def gradient_descent(f, grad, x0, lr=0.001, max_iter=1000, tol=1e-6):
    """Vanilla gradient descent"""
    x = x0.copy()
    history = [x.copy()]
    
    for i in range(max_iter):
        g = grad(x)
        x = x - lr * g
        history.append(x.copy())
        
        if np.linalg.norm(g) < tol:
            break
    
    return x, np.array(history)

def momentum_gd(f, grad, x0, lr=0.001, momentum=0.9, max_iter=1000, tol=1e-6):
    """Gradient descent with momentum"""
    x = x0.copy()
    v = np.zeros_like(x)
    history = [x.copy()]
    
    for i in range(max_iter):
        g = grad(x)
        v = momentum * v - lr * g
        x = x + v
        history.append(x.copy())
        
        if np.linalg.norm(g) < tol:
            break
    
    return x, np.array(history)

def adam_optimizer(f, grad, x0, lr=0.01, beta1=0.9, beta2=0.999, eps=1e-8, max_iter=1000, tol=1e-6):
    """Adam optimizer"""
    x = x0.copy()
    m = np.zeros_like(x)
    v = np.zeros_like(x)
    history = [x.copy()]
    
    for i in range(1, max_iter + 1):
        g = grad(x)
        m = beta1 * m + (1 - beta1) * g
        v = beta2 * v + (1 - beta2) * g**2
        
        m_hat = m / (1 - beta1**i)
        v_hat = v / (1 - beta2**i)
        
        x = x - lr * m_hat / (np.sqrt(v_hat) + eps)
        history.append(x.copy())
        
        if np.linalg.norm(g) < tol:
            break
    
    return x, np.array(history)

def newton_method(f, grad, hess, x0, max_iter=100, tol=1e-6):
    """Newton's method for optimization"""
    x = x0.copy()
    history = [x.copy()]
    
    for i in range(max_iter):
        g = grad(x)
        H = hess(x)
        
        try:
            delta = linalg.solve(H, -g)
        except:
            break
        
        x = x + delta
        history.append(x.copy())
        
        if np.linalg.norm(g) < tol:
            break
    
    return x, np.array(history)

# Test all methods
x0 = np.array([-1.0, 1.0])

x_gd, hist_gd = gradient_descent(rosenbrock, rosenbrock_grad, x0, lr=0.001, max_iter=5000)
x_momentum, hist_momentum = momentum_gd(rosenbrock, rosenbrock_grad, x0, lr=0.001, max_iter=5000)
x_adam, hist_adam = adam_optimizer(rosenbrock, rosenbrock_grad, x0, lr=0.01, max_iter=5000)
x_newton, hist_newton = newton_method(rosenbrock, rosenbrock_grad, rosenbrock_hess, x0, max_iter=100)

# Scipy's optimization for comparison
result_bfgs = optimize.minimize(rosenbrock, x0, method='BFGS', jac=rosenbrock_grad)

print("Optimization Results (Rosenbrock function, minimum at [1, 1]):")
print(f"  Gradient Descent:  x = [{x_gd[0]:.6f}, {x_gd[1]:.6f}], f(x) = {rosenbrock(x_gd):.2e}, iters = {len(hist_gd)}")
print(f"  Momentum GD:       x = [{x_momentum[0]:.6f}, {x_momentum[1]:.6f}], f(x) = {rosenbrock(x_momentum):.2e}, iters = {len(hist_momentum)}")
print(f"  Adam:              x = [{x_adam[0]:.6f}, {x_adam[1]:.6f}], f(x) = {rosenbrock(x_adam):.2e}, iters = {len(hist_adam)}")
print(f"  Newton's Method:   x = [{x_newton[0]:.6f}, {x_newton[1]:.6f}], f(x) = {rosenbrock(x_newton):.2e}, iters = {len(hist_newton)}")
print(f"  BFGS (scipy):      x = [{result_bfgs.x[0]:.6f}, {result_bfgs.x[1]:.6f}], f(x) = {result_bfgs.fun:.2e}")

# Visualization
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Create contour plot
x_range = np.linspace(-2, 2, 400)
y_range = np.linspace(-1, 3, 400)
X, Y = np.meshgrid(x_range, y_range)
Z = np.zeros_like(X)
for i in range(X.shape[0]):
    for j in range(X.shape[1]):
        Z[i, j] = rosenbrock([X[i, j], Y[i, j]])

# Plot 1: Optimization paths
levels = np.logspace(-1, 3.5, 20)
axes[0].contour(X, Y, Z, levels=levels, cmap='viridis', alpha=0.6)
axes[0].contourf(X, Y, Z, levels=levels, cmap='viridis', alpha=0.3)

# Plot paths (subsample for clarity)
step = max(1, len(hist_gd) // 50)
axes[0].plot(hist_gd[::step, 0], hist_gd[::step, 1], 'r.-', linewidth=1.5, markersize=4, label='GD', alpha=0.7)
axes[0].plot(hist_momentum[::step, 0], hist_momentum[::step, 1], 'b.-', linewidth=1.5, markersize=4, label='Momentum', alpha=0.7)
axes[0].plot(hist_adam[::step, 0], hist_adam[::step, 1], 'g.-', linewidth=1.5, markersize=4, label='Adam', alpha=0.7)
axes[0].plot(hist_newton[:, 0], hist_newton[:, 1], 'm.-', linewidth=2, markersize=6, label='Newton', alpha=0.9)

axes[0].plot(1, 1, 'r*', markersize=20, label='Minimum')
axes[0].plot(x0[0], x0[1], 'ko', markersize=10, label='Start')
axes[0].set_xlabel('x', fontsize=12)
axes[0].set_ylabel('y', fontsize=12)
axes[0].set_title('Optimization Paths on Rosenbrock Function', fontsize=13, fontweight='bold')
axes[0].legend(fontsize=9, loc='upper left')
axes[0].grid(True, alpha=0.3)

# Plot 2: Convergence rates
f_vals_gd = [rosenbrock(x) for x in hist_gd]
f_vals_momentum = [rosenbrock(x) for x in hist_momentum]
f_vals_adam = [rosenbrock(x) for x in hist_adam]
f_vals_newton = [rosenbrock(x) for x in hist_newton]

axes[1].semilogy(f_vals_gd, 'r-', linewidth=2, label=f'GD ({len(hist_gd)} iter)', alpha=0.8)
axes[1].semilogy(f_vals_momentum, 'b-', linewidth=2, label=f'Momentum ({len(hist_momentum)} iter)', alpha=0.8)
axes[1].semilogy(f_vals_adam, 'g-', linewidth=2, label=f'Adam ({len(hist_adam)} iter)', alpha=0.8)
axes[1].semilogy(f_vals_newton, 'm-', linewidth=2, label=f'Newton ({len(hist_newton)} iter)', alpha=0.8)
axes[1].axhline(y=1e-6, color='k', linestyle='--', alpha=0.5, label='Target')

axes[1].set_xlabel('Iteration', fontsize=12)
axes[1].set_ylabel('f(x) - log scale', fontsize=12)
axes[1].set_title('Convergence Rate Comparison', fontsize=13, fontweight='bold')
axes[1].legend(fontsize=9)
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# ML Application 1: Gradient-Based Learning and Backpropagation

# Generate synthetic regression dataset
np.random.seed(42)
n_samples = 200
X = np.random.randn(n_samples, 1)
y_true = 3 * X.squeeze() + 2 + np.random.randn(n_samples) * 0.5

# Add polynomial features
X_poly = np.c_[np.ones(n_samples), X, X**2, X**3]

# 1. Linear Regression using Normal Equations (closed form)
theta_normal = linalg.solve(X_poly.T @ X_poly, X_poly.T @ y_true)
y_pred_normal = X_poly @ theta_normal
mse_normal = np.mean((y_true - y_pred_normal)**2)

print("Linear Regression Results:")
print(f"Normal Equations: MSE = {mse_normal:.6f}")
print(f"  Coefficients: {theta_normal}\n")

# 2. Gradient Descent for Linear Regression
def mse_loss(theta, X, y):
    """Mean squared error loss"""
    predictions = X @ theta
    return np.mean((y - predictions)**2)

def mse_gradient(theta, X, y):
    """Gradient of MSE loss"""
    predictions = X @ theta
    return -2 * X.T @ (y - predictions) / len(y)

# Train with batch gradient descent
theta_gd = np.zeros(X_poly.shape[1])
lr = 0.01
n_epochs = 500
loss_history = []

for epoch in range(n_epochs):
    grad = mse_gradient(theta_gd, X_poly, y_true)
    theta_gd = theta_gd - lr * grad
    loss = mse_loss(theta_gd, X_poly, y_true)
    loss_history.append(loss)

y_pred_gd = X_poly @ theta_gd
print(f"Gradient Descent: MSE = {loss_history[-1]:.6f}")
print(f"  Coefficients: {theta_gd}")
print(f"  Converged in {n_epochs} epochs\n")

# 3. Stochastic Gradient Descent (SGD)
theta_sgd = np.zeros(X_poly.shape[1])
lr_sgd = 0.01
n_epochs_sgd = 100
batch_size = 32
loss_history_sgd = []

for epoch in range(n_epochs_sgd):
    # Shuffle data
    indices = np.random.permutation(n_samples)
    X_shuffled = X_poly[indices]
    y_shuffled = y_true[indices]
    
    # Mini-batch updates
    for i in range(0, n_samples, batch_size):
        X_batch = X_shuffled[i:i+batch_size]
        y_batch = y_shuffled[i:i+batch_size]
        
        grad = mse_gradient(theta_sgd, X_batch, y_batch)
        theta_sgd = theta_sgd - lr_sgd * grad
    
    loss = mse_loss(theta_sgd, X_poly, y_true)
    loss_history_sgd.append(loss)

y_pred_sgd = X_poly @ theta_sgd
print(f"Stochastic GD (batch_size={batch_size}): MSE = {loss_history_sgd[-1]:.6f}")
print(f"  Coefficients: {theta_sgd}\n")

# 4. Simple Neural Network with Backpropagation
class SimpleNN:
    def __init__(self, input_dim, hidden_dim, output_dim):
        # Initialize weights
        self.W1 = np.random.randn(input_dim, hidden_dim) * 0.1
        self.b1 = np.zeros(hidden_dim)
        self.W2 = np.random.randn(hidden_dim, output_dim) * 0.1
        self.b2 = np.zeros(output_dim)
    
    def relu(self, x):
        return np.maximum(0, x)
    
    def relu_derivative(self, x):
        return (x > 0).astype(float)
    
    def forward(self, X):
        self.z1 = X @ self.W1 + self.b1
        self.a1 = self.relu(self.z1)
        self.z2 = self.a1 @ self.W2 + self.b2
        return self.z2
    
    def backward(self, X, y, lr=0.01):
        m = X.shape[0]
        
        # Forward pass
        y_pred = self.forward(X)
        
        # Backward pass
        dz2 = (y_pred - y.reshape(-1, 1)) / m
        dW2 = self.a1.T @ dz2
        db2 = np.sum(dz2, axis=0)
        
        da1 = dz2 @ self.W2.T
        dz1 = da1 * self.relu_derivative(self.z1)
        dW1 = X.T @ dz1
        db1 = np.sum(dz1, axis=0)
        
        # Update weights
        self.W1 -= lr * dW1
        self.b1 -= lr * db1
        self.W2 -= lr * dW2
        self.b2 -= lr * db2
        
        # Return loss
        return np.mean((y_pred.squeeze() - y)**2)

# Train neural network
nn = SimpleNN(input_dim=1, hidden_dim=10, output_dim=1)
nn_loss_history = []
n_epochs_nn = 1000

for epoch in range(n_epochs_nn):
    loss = nn.backward(X, y_true, lr=0.01)
    nn_loss_history.append(loss)

y_pred_nn = nn.forward(X).squeeze()
print(f"Neural Network: MSE = {nn_loss_history[-1]:.6f}")
print(f"  Hidden units: {nn.W1.shape[1]}, trained for {n_epochs_nn} epochs")

# Visualization
fig, axes = plt.subplots(2, 2, figsize=(16, 12))

# Plot 1: Data and fits
X_plot = np.linspace(X.min(), X.max(), 300).reshape(-1, 1)
X_plot_poly = np.c_[np.ones(len(X_plot)), X_plot, X_plot**2, X_plot**3]

axes[0, 0].scatter(X, y_true, alpha=0.5, s=30, label='Data')
axes[0, 0].plot(X_plot, X_plot_poly @ theta_normal, 'r-', linewidth=2, label='Normal Equations')
axes[0, 0].plot(X_plot, X_plot_poly @ theta_gd, 'b--', linewidth=2, label='Batch GD')
axes[0, 0].plot(X_plot, X_plot_poly @ theta_sgd, 'g:', linewidth=2, label='SGD')
y_plot_nn = nn.forward(X_plot).squeeze()
axes[0, 0].plot(X_plot, y_plot_nn, 'm-.', linewidth=2, label='Neural Network')
axes[0, 0].set_xlabel('x', fontsize=11)
axes[0, 0].set_ylabel('y', fontsize=11)
axes[0, 0].set_title('Model Predictions Comparison', fontsize=12, fontweight='bold')
axes[0, 0].legend(fontsize=9)
axes[0, 0].grid(True, alpha=0.3)

# Plot 2: Training loss curves
axes[0, 1].semilogy(loss_history, 'b-', linewidth=2, label='Batch GD', alpha=0.8)
axes[0, 1].semilogy(loss_history_sgd, 'g-', linewidth=2, label='SGD', alpha=0.8)
axes[0, 1].semilogy(nn_loss_history, 'm-', linewidth=2, label='Neural Network', alpha=0.8)
axes[0, 1].axhline(y=mse_normal, color='r', linestyle='--', label='Normal Eq. MSE')
axes[0, 1].set_xlabel('Epoch', fontsize=11)
axes[0, 1].set_ylabel('MSE Loss (log scale)', fontsize=11)
axes[0, 1].set_title('Training Loss Convergence', fontsize=12, fontweight='bold')
axes[0, 1].legend(fontsize=9)
axes[0, 1].grid(True, alpha=0.3)

# Plot 3: Residuals comparison
residuals_normal = y_true - y_pred_normal
residuals_gd = y_true - y_pred_gd
residuals_nn = y_true - y_pred_nn

axes[1, 0].scatter(y_pred_normal, residuals_normal, alpha=0.5, s=30, label='Normal Eq.')
axes[1, 0].scatter(y_pred_gd, residuals_gd, alpha=0.5, s=30, label='Batch GD')
axes[1, 0].scatter(y_pred_nn, residuals_nn, alpha=0.5, s=30, label='Neural Net')
axes[1, 0].axhline(y=0, color='k', linestyle='--', alpha=0.5)
axes[1, 0].set_xlabel('Predicted Value', fontsize=11)
axes[1, 0].set_ylabel('Residual', fontsize=11)
axes[1, 0].set_title('Residual Analysis', fontsize=12, fontweight='bold')
axes[1, 0].legend(fontsize=9)
axes[1, 0].grid(True, alpha=0.3)

# Plot 4: Neural network weights visualization
axes[1, 1].bar(range(len(nn.W1.squeeze())), nn.W1.squeeze(), alpha=0.7, label='Input→Hidden')
axes[1, 1].bar(range(len(nn.W2.squeeze())), nn.W2.squeeze(), alpha=0.7, label='Hidden→Output')
axes[1, 1].set_xlabel('Weight Index', fontsize=11)
axes[1, 1].set_ylabel('Weight Value', fontsize=11)
axes[1, 1].set_title('Neural Network Learned Weights', fontsize=12, fontweight='bold')
axes[1, 1].legend(fontsize=9)
axes[1, 1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# ML Application 2: Eigenvalue Methods (PCA) and Sparse Systems

# Generate synthetic high-dimensional data
np.random.seed(42)
n_samples = 300
n_features = 50

# Create data with underlying low-dimensional structure
latent_dim = 3
latent = np.random.randn(n_samples, latent_dim)
W = np.random.randn(n_features, latent_dim)
X_original = latent @ W.T + 0.5 * np.random.randn(n_samples, n_features)

# Standardize data
X_mean = X_original.mean(axis=0)
X_std = X_original.std(axis=0) + 1e-8
X = (X_original - X_mean) / X_std

print("Dataset Information:")
print(f"  Samples: {n_samples}, Features: {n_features}")
print(f"  True latent dimension: {latent_dim}\n")

# 1. Principal Component Analysis (PCA) using SVD
def pca_svd(X, n_components):
    """PCA using SVD"""
    # X should be centered
    U, s, Vt = linalg.svd(X, full_matrices=False)
    
    # Components are rows of Vt
    components = Vt[:n_components]
    
    # Explained variance
    explained_variance = (s ** 2) / (X.shape[0] - 1)
    
    # Transform data
    X_transformed = X @ components.T
    
    return components, explained_variance, X_transformed, s

# Apply PCA
n_components = 10
components, explained_var, X_pca, singular_values = pca_svd(X, n_components)

# Calculate cumulative explained variance
total_var = np.sum(explained_var)
explained_var_ratio = explained_var / total_var
cumulative_var_ratio = np.cumsum(explained_var_ratio)

print("PCA Results (SVD method):")
print(f"  Top 3 components explain: {cumulative_var_ratio[2]:.2%} of variance")
print(f"  Top 5 components explain: {cumulative_var_ratio[4]:.2%} of variance")
print(f"  Top 10 components explain: {cumulative_var_ratio[9]:.2%} of variance\n")

# 2. PCA using Eigenvalue Decomposition
def pca_eigen(X, n_components):
    """PCA using eigenvalue decomposition of covariance matrix"""
    # Compute covariance matrix
    cov = (X.T @ X) / (X.shape[0] - 1)
    
    # Eigenvalue decomposition
    eigenvalues, eigenvectors = linalg.eigh(cov)
    
    # Sort by eigenvalues (descending)
    idx = eigenvalues.argsort()[::-1]
    eigenvalues = eigenvalues[idx]
    eigenvectors = eigenvectors[:, idx]
    
    # Select top components
    components = eigenvectors[:, :n_components].T
    
    # Transform data
    X_transformed = X @ components.T
    
    return components, eigenvalues, X_transformed

components_eigen, eigenvalues, X_pca_eigen = pca_eigen(X, n_components)

print("PCA Results (Eigenvalue decomposition):")
print(f"  Top 3 eigenvalues: {eigenvalues[:3]}")
print(f"  Reconstruction error: {np.linalg.norm(X - X_pca_eigen @ components_eigen):.4f}\n")

# 3. Sparse Linear System (relevant for L1-regularized learning)
# Create a sparse system
n_sparse = 1000
density = 0.01  # 1% non-zero entries

# Create sparse matrix
from scipy.sparse import random as sparse_random
A_sparse = sparse_random(n_sparse, n_sparse, density=density, format='csr')
A_sparse = A_sparse + A_sparse.T + 10 * sparse.eye(n_sparse)  # Make symmetric positive definite

x_true_sparse = np.random.randn(n_sparse)
b_sparse = A_sparse @ x_true_sparse

# Solve sparse system
import time

# Dense solution (for small comparison)
if n_sparse <= 500:  # Only for small systems
    start = time.time()
    x_dense = linalg.solve(A_sparse.toarray(), b_sparse)
    time_dense = time.time() - start
else:
    time_dense = None
    x_dense = None

# Sparse solution
start = time.time()
x_sparse_solve = spsolve(A_sparse, b_sparse)
time_sparse = time.time() - start

print(f"Sparse Linear System (size {n_sparse}×{n_sparse}, density {density:.1%}):")
print(f"  Non-zero entries: {A_sparse.nnz:,} out of {n_sparse**2:,}")
print(f"  Sparse solver time: {time_sparse:.4f} seconds")
if time_dense is not None:
    print(f"  Dense solver time: {time_dense:.4f} seconds")
    print(f"  Speedup: {time_dense/time_sparse:.1f}x")
print(f"  Solution error: {np.linalg.norm(x_sparse_solve - x_true_sparse):.2e}\n")

# 4. Power Method for finding largest eigenvalue
def power_method(A, max_iter=100, tol=1e-6):
    """Power method for finding largest eigenvalue and eigenvector"""
    n = A.shape[0]
    v = np.random.randn(n)
    v = v / np.linalg.norm(v)
    
    eigenvalues_history = []
    
    for i in range(max_iter):
        v_new = A @ v
        v_new = v_new / np.linalg.norm(v_new)
        
        # Rayleigh quotient
        eigenvalue = v_new.T @ A @ v_new
        eigenvalues_history.append(eigenvalue)
        
        if np.linalg.norm(v_new - v) < tol:
            return eigenvalue, v_new, eigenvalues_history
        
        v = v_new
    
    return eigenvalue, v, eigenvalues_history

# Test power method on a small matrix
A_test = np.random.randn(20, 20)
A_test = A_test @ A_test.T  # Make symmetric positive definite

eigenval_power, eigenvec_power, eig_history = power_method(A_test)
eigenvals_true = linalg.eigvalsh(A_test)

print(f"Power Method Results:")
print(f"  Computed largest eigenvalue: {eigenval_power:.6f}")
print(f"  True largest eigenvalue: {eigenvals_true[-1]:.6f}")
print(f"  Error: {abs(eigenval_power - eigenvals_true[-1]):.2e}")
print(f"  Converged in {len(eig_history)} iterations")

# Visualization
fig = plt.figure(figsize=(16, 12))
gs = fig.add_gridspec(3, 2, hspace=0.35, wspace=0.3)

# Plot 1: Explained variance
ax1 = fig.add_subplot(gs[0, 0])
ax1.bar(range(1, n_components+1), explained_var_ratio[:n_components], alpha=0.7, label='Individual')
ax1.plot(range(1, n_components+1), cumulative_var_ratio[:n_components], 'ro-', linewidth=2, markersize=6, label='Cumulative')
ax1.axhline(y=0.95, color='g', linestyle='--', alpha=0.7, label='95% threshold')
ax1.set_xlabel('Principal Component', fontsize=11)
ax1.set_ylabel('Explained Variance Ratio', fontsize=11)
ax1.set_title('PCA: Explained Variance by Component', fontsize=12, fontweight='bold')
ax1.legend(fontsize=9)
ax1.grid(True, alpha=0.3)

# Plot 2: 2D projection
ax2 = fig.add_subplot(gs[0, 1])
scatter = ax2.scatter(X_pca[:, 0], X_pca[:, 1], c=latent[:, 0], cmap='viridis', alpha=0.6, s=30)
plt.colorbar(scatter, ax=ax2, label='True latent variable')
ax2.set_xlabel('PC1', fontsize=11)
ax2.set_ylabel('PC2', fontsize=11)
ax2.set_title('PCA: 2D Projection (colored by true latent)', fontsize=12, fontweight='bold')
ax2.grid(True, alpha=0.3)

# Plot 3: 3D projection
ax3 = fig.add_subplot(gs[1, :], projection='3d')
ax3.scatter(X_pca[:, 0], X_pca[:, 1], X_pca[:, 2], c=latent[:, 0], cmap='viridis', alpha=0.6, s=20)
ax3.set_xlabel('PC1', fontsize=10)
ax3.set_ylabel('PC2', fontsize=10)
ax3.set_zlabel('PC3', fontsize=10)
ax3.set_title('PCA: 3D Projection', fontsize=12, fontweight='bold')

# Plot 4: Sparse matrix structure
ax4 = fig.add_subplot(gs[2, 0])
# Show a subset for visualization
subset_size = min(200, n_sparse)
A_subset = A_sparse[:subset_size, :subset_size].toarray()
im = ax4.imshow(A_subset != 0, cmap='binary', interpolation='nearest', aspect='auto')
ax4.set_xlabel('Column', fontsize=11)
ax4.set_ylabel('Row', fontsize=11)
ax4.set_title(f'Sparse Matrix Structure ({subset_size}×{subset_size} subset)', fontsize=12, fontweight='bold')

# Plot 5: Power method convergence
ax5 = fig.add_subplot(gs[2, 1])
ax5.plot(eig_history, 'b-', linewidth=2)
ax5.axhline(y=eigenvals_true[-1], color='r', linestyle='--', linewidth=2, label='True eigenvalue')
ax5.set_xlabel('Iteration', fontsize=11)
ax5.set_ylabel('Estimated Eigenvalue', fontsize=11)
ax5.set_title('Power Method: Convergence to Largest Eigenvalue', fontsize=12, fontweight='bold')
ax5.legend(fontsize=9)
ax5.grid(True, alpha=0.3)

plt.show()

print(f"\nVisualization complete!")
```

```python
# Advanced Topic: Partial Differential Equations (PDEs)
# Solving the 1D Heat Equation using Finite Difference Method

# Heat equation: ∂u/∂t = α ∂²u/∂x²
# Boundary conditions: u(0,t) = u(L,t) = 0
# Initial condition: u(x,0) = sin(πx/L)

# Parameters
L = 1.0  # Length of domain
T = 0.5  # Total time
alpha = 0.01  # Thermal diffusivity
nx = 50  # Number of spatial points
nt = 1000  # Number of time steps

# Grid
dx = L / (nx - 1)
dt = T / nt
x = np.linspace(0, L, nx)
t = np.linspace(0, T, nt)

# Stability condition (CFL condition)
r = alpha * dt / dx**2
print("1D Heat Equation Solver")
print(f"  Domain: [0, {L}], Time: [0, {T}]")
print(f"  Grid: {nx} spatial points, {nt} time steps")
print(f"  Stability parameter r = α·Δt/Δx² = {r:.4f}")
if r > 0.5:
    print(f"  WARNING: Unstable! r should be ≤ 0.5 for FTCS scheme")
else:
    print(f"  Stable: r ≤ 0.5 ✓\n")

# Initialize solution array
u = np.zeros((nt, nx))

# Initial condition: u(x,0) = sin(πx/L)
u[0, :] = np.sin(np.pi * x / L)

# Analytical solution for comparison
def analytical_solution(x, t, L, alpha):
    return np.exp(-alpha * (np.pi/L)**2 * t) * np.sin(np.pi * x / L)

# Finite Difference Method (FTCS - Forward Time, Central Space)
for n in range(nt - 1):
    for i in range(1, nx - 1):
        u[n+1, i] = u[n, i] + r * (u[n, i+1] - 2*u[n, i] + u[n, i-1])
    
    # Boundary conditions: u(0,t) = u(L,t) = 0
    u[n+1, 0] = 0
    u[n+1, -1] = 0

# Compute analytical solution at final time
u_analytical = analytical_solution(x, T, L, alpha)
error = np.linalg.norm(u[-1, :] - u_analytical) / np.linalg.norm(u_analytical)

print(f"Numerical Solution:")
print(f"  Final time T = {T}")
print(f"  Relative error vs analytical: {error:.2e}")
print(f"  Max value at t=0: {u[0, :].max():.6f}")
print(f"  Max value at t=T: {u[-1, :].max():.6f}")
print(f"  Analytical max at t=T: {u_analytical.max():.6f}\n")

# 2D Heat Equation (bonus)
# ∂u/∂t = α(∂²u/∂x² + ∂²u/∂y²)
Lx, Ly = 1.0, 1.0
nx2d, ny2d = 40, 40
nt2d = 500
T2d = 0.1

dx2d = Lx / (nx2d - 1)
dy2d = Ly / (ny2d - 1)
dt2d = T2d / nt2d

x2d = np.linspace(0, Lx, nx2d)
y2d = np.linspace(0, Ly, ny2d)
X2d, Y2d = np.meshgrid(x2d, y2d)

# Stability check
r2d = alpha * dt2d * (1/dx2d**2 + 1/dy2d**2)
print(f"2D Heat Equation:")
print(f"  Grid: {nx2d}×{ny2d}, {nt2d} time steps")
print(f"  Stability parameter: {r2d:.4f} (should be ≤ 0.25)")

# Initial condition: Gaussian peak
u2d = np.exp(-50 * ((X2d - 0.5)**2 + (Y2d - 0.5)**2))
u2d_initial = u2d.copy()

# Store snapshots
snapshots = [u2d.copy()]
snapshot_times = [0]

# Solve using finite differences
rx = alpha * dt2d / dx2d**2
ry = alpha * dt2d / dy2d**2

for n in range(nt2d):
    u2d_new = u2d.copy()
    
    for i in range(1, nx2d - 1):
        for j in range(1, ny2d - 1):
            u2d_new[i, j] = u2d[i, j] + rx * (u2d[i+1, j] - 2*u2d[i, j] + u2d[i-1, j]) + \
                                          ry * (u2d[i, j+1] - 2*u2d[i, j] + u2d[i, j-1])
    
    # Boundary conditions (Dirichlet: u = 0 on boundary)
    u2d_new[0, :] = 0
    u2d_new[-1, :] = 0
    u2d_new[:, 0] = 0
    u2d_new[:, -1] = 0
    
    u2d = u2d_new
    
    # Save snapshots
    if n in [nt2d//4, nt2d//2, 3*nt2d//4, nt2d-1]:
        snapshots.append(u2d.copy())
        snapshot_times.append((n+1) * dt2d)

print(f"  2D simulation complete!\n")

# Visualization
fig = plt.figure(figsize=(18, 12))
gs = fig.add_gridspec(3, 3, hspace=0.35, wspace=0.35)

# Plot 1: 1D Heat equation evolution
ax1 = fig.add_subplot(gs[0, :])
time_indices = [0, nt//4, nt//2, 3*nt//4, nt-1]
colors = plt.cm.coolwarm(np.linspace(0, 1, len(time_indices)))

for idx, c in zip(time_indices, colors):
    ax1.plot(x, u[idx, :], color=c, linewidth=2, label=f't = {t[idx]:.3f}')

ax1.plot(x, u_analytical, 'k--', linewidth=2, alpha=0.7, label='Analytical (final)')
ax1.set_xlabel('x', fontsize=11)
ax1.set_ylabel('u(x,t)', fontsize=11)
ax1.set_title('1D Heat Equation: Temperature Evolution', fontsize=13, fontweight='bold')
ax1.legend(fontsize=9, ncol=3)
ax1.grid(True, alpha=0.3)

# Plot 2-6: 2D Heat equation snapshots
snapshot_positions = [(1, 0), (1, 1), (1, 2), (2, 0), (2, 1)]
vmin, vmax = 0, u2d_initial.max()

for idx, (i, j) in enumerate(snapshot_positions):
    ax = fig.add_subplot(gs[i, j])
    im = ax.contourf(X2d, Y2d, snapshots[idx], levels=20, cmap='hot', vmin=vmin, vmax=vmax)
    ax.contour(X2d, Y2d, snapshots[idx], levels=10, colors='black', alpha=0.3, linewidths=0.5)
    ax.set_xlabel('x', fontsize=10)
    ax.set_ylabel('y', fontsize=10)
    ax.set_title(f't = {snapshot_times[idx]:.4f}', fontsize=11, fontweight='bold')
    ax.set_aspect('equal')
    plt.colorbar(im, ax=ax, fraction=0.046, pad=0.04)

# Plot 7: 3D surface of final state
ax7 = fig.add_subplot(gs[2, 2], projection='3d')
surf = ax7.plot_surface(X2d, Y2d, snapshots[-1], cmap='hot', alpha=0.9, 
                        linewidth=0, antialiased=True, vmin=vmin, vmax=vmax)
ax7.set_xlabel('x', fontsize=9)
ax7.set_ylabel('y', fontsize=9)
ax7.set_zlabel('u', fontsize=9)
ax7.set_title(f'3D View at t = {snapshot_times[-1]:.4f}', fontsize=11, fontweight='bold')
ax7.view_init(elev=25, azim=45)

plt.show()

print("PDE Visualization complete!")
print("\nKey Insights:")
print("  • Heat diffusion smooths out temperature gradients over time")
print("  • FTCS scheme requires stability condition: r ≤ 0.5 (1D) or 0.25 (2D)")
print("  • Numerical and analytical solutions match well for stable parameters")
print("  • 2D diffusion spreads heat radially from the initial Gaussian peak")
```

### Critical Algorithms and Their ML Applications

**Matrix Decompositions in Machine Learning**:

1. **Singular Value Decomposition**: $A = U\Sigma V^T$
   - **Computational Complexity**: $O(mn^2)$ for $m \times n$ matrix
   - **Numerical Stability**: Backward stable algorithms via Householder reflections
   - **Applications**: PCA, matrix completion, recommender systems
   - **Randomized Variants**: $O(mnk)$ for rank-$k$ approximation

2. **QR Decomposition**: $A = QR$ where $Q$ orthogonal, $R$ upper triangular
   - **Householder Method**: Numerically stable, $O(mn^2 - n^3/3)$ flops
   - **Gram-Schmidt**: Unstable, but provides insight
   - **Applications**: Linear regression, least squares, eigenvalue computation

3. **Cholesky Decomposition**: $A = LL^T$ for positive definite $A$
   - **Computational Cost**: $O(n^3/3)$, half that of LU decomposition
   - **Numerical Benefits**: Enhanced stability for well-conditioned problems
   - **ML Applications**: Gaussian processes, multivariate normal sampling

**Optimization Algorithms for Machine Learning**:

1. **Gradient Descent Variants**:
   - **Vanilla GD**: $\theta_{t+1} = \theta_t - \alpha \nabla L(\theta_t)$
   - **Momentum**: $v_{t+1} = \beta v_t + \nabla L(\theta_t)$, $\theta_{t+1} = \theta_t - \alpha v_{t+1}$
   - **Adam**: Adaptive moments with bias correction

2. **Second-Order Methods**:
   - **Newton's Method**: $\theta_{t+1} = \theta_t - H^{-1}\nabla L(\theta_t)$
   - **L-BFGS**: Limited memory approximation for large-scale problems
   - **Natural Gradients**: $\theta_{t+1} = \theta_t - \alpha F^{-1}\nabla L(\theta_t)$

**Convergence Analysis**:

**Theorem 2.1** (Gradient Descent Convergence): For $L$-Lipschitz, $\mu$-strongly convex function:
$$L(\theta_k) - L(\theta^*) \leq \left(1 - \frac{\mu}{L}\right)^k (L(\theta_0) - L(\theta^*))$$

**Condition Number Impact**: Convergence rate depends on $\kappa = L/\mu$.

### Numerical Challenges in Deep Learning

**Vanishing and Exploding Gradients**:

**Mathematical Analysis**: For deep network with $L$ layers:
$$\frac{\partial L}{\partial W^{(1)}} = \frac{\partial L}{\partial a^{(L)}} \prod_{\ell=2}^L \frac{\partial a^{(\ell)}}{\partial a^{(\ell-1)}}$$

**Gradient Magnitude**: $\|\nabla_{W^{(1)}} L\| \approx \gamma^{L-1}$ where $\gamma$ depends on weights and activations.

**Solutions**:
1. **Proper Initialization**: Xavier, He initialization to control $\gamma \approx 1$
2. **Residual Connections**: $x_{l+1} = x_l + F(x_l)$ provides gradient shortcuts
3. **Batch Normalization**: Normalizes layer inputs, stabilizes training

**Numerical Stability in Activations**:

1. **Softmax Numerical Stability**:
   $$\text{softmax}(x_i) = \frac{e^{x_i - \max_j x_j}}{\sum_j e^{x_j - \max_j x_j}}$$
   
2. **Log-Sum-Exp Trick**: For computing $\log\sum_i e^{x_i}$:
   $$\log\sum_i e^{x_i} = a + \log\sum_i e^{x_i - a}$$
   where $a = \max_i x_i$.

### Monte Carlo Methods in Machine Learning

**Basic Monte Carlo Integration**:
$$\mathbb{E}[f(X)] = \int f(x) p(x) dx \approx \frac{1}{n}\sum_{i=1}^n f(X_i)$$

**Error Analysis**: Standard error $\approx \frac{\sigma}{\sqrt{n}}$ where $\sigma^2 = \text{Var}[f(X)]$.

**Advanced Sampling Methods**:

1. **Markov Chain Monte Carlo (MCMC)**:
   - **Metropolis-Hastings**: Accept/reject based on ratio $\frac{\pi(x')}{\pi(x)}$
   - **Gibbs Sampling**: Sample from conditional distributions
   - **Hamiltonian Monte Carlo**: Use gradient information for efficient sampling

2. **Variational Inference**:
   - **Mean Field**: $q(z) = \prod_i q_i(z_i)$ factorization
   - **Evidence Lower Bound**: $\log p(x) \geq \mathcal{L}(q) = \mathbb{E}_q[\log p(x,z)] - \mathbb{E}_q[\log q(z)]$
   - **Stochastic Optimization**: Optimize ELBO via gradient descent
