# Advanced Statistical Learning Theory: Mathematical Foundations

Statistical learning theory provides the rigorous mathematical foundation for understanding generalization in machine learning systems. This comprehensive framework encompasses classical results from PAC learning and VC theory through modern developments in deep learning theory, online learning, and information-theoretic bounds.

## Theoretical Architecture

### Level 1: Classical Foundations
- **PAC Learning Framework**: Probably Approximately Correct paradigm for learnability
- **VC Theory**: Vapnik-Chervonenkis dimension and combinatorial complexity measures
- **Uniform Convergence**: Fundamental convergence properties of empirical processes
- **Sample Complexity**: Quantifying data requirements for reliable learning

### Level 2: Advanced Complexity Measures  
- **Rademacher Complexity**: Data-dependent and adaptive complexity measures
- **Algorithmic Stability**: Generalization through algorithm properties rather than hypothesis class
- **Concentration Inequalities**: Hoeffding, McDiarmid, and Bernstein bounds
- **Growth Functions**: Combinatorial characterizations of hypothesis class richness

### Level 3: Modern Developments
- **Deep Learning Theory**: Understanding overparameterized models and implicit regularization

- **Information Theory**: Mutual information bounds and compression perspectives
- **Neural Tangent Kernels**: Infinite-width limits and kernel perspectives on neural networks

## Mathematical Foundations

### Definition 1.1 (Learning Problem Setup)
A learning problem consists of:
- **Input Space**: $\mathcal{X}$ (e.g., $\mathbb{R}^d$ for $d$-dimensional features)
- **Output Space**: $\mathcal{Y}$ (e.g., $\{-1,+1\}$ for binary classification)
- **Hypothesis Space**: $\mathcal{H} = \{h : \mathcal{X} \to \mathcal{Y}\}$
- **Loss Function**: $\ell: \mathcal{Y} \times \mathcal{Y} \to [0, \infty)$
- **Data Distribution**: $\mathcal{D}$ over $\mathcal{X} \times \mathcal{Y}$

### Definition 1.2 (Risk Functions)
For hypothesis $h \in \mathcal{H}$:

**Population Risk (True Risk)**:
$$R(h) = \mathbb{E}_{(x,y) \sim \mathcal{D}}[\ell(h(x), y)]$$

**Empirical Risk (Training Error)**:
$$\hat{R}_n(h) = \frac{1}{n}\sum_{i=1}^n \ell(h(x_i), y_i)$$

**Generalization Gap**:
$$\text{Gap}(h) = R(h) - \hat{R}_n(h)$$

## PAC Learning Theory

### Definition 1.3 (PAC Learnable)
A concept class $\mathcal{C}$ is **PAC learnable** if there exists an algorithm $A$ and polynomial $p(\cdot, \cdot, \cdot, \cdot)$ such that:

For all $\epsilon, \delta \in (0, 1)$, all distributions $\mathcal{D}$ over $\mathcal{X}$, and all target concepts $c \in \mathcal{C}$:

If $A$ receives $m \geq p(1/\epsilon, 1/\delta, n, \text{size}(c))$ training examples, then with probability $\geq 1 - \delta$:
$$\mathbb{P}_{x \sim \mathcal{D}}[A(S)(x) \neq c(x)] \leq \epsilon$$

### Theorem 1.1 (Sample Complexity for Finite Classes)
For finite concept class $|\mathcal{C}| < \infty$:
$$m_{\mathcal{C}}(\epsilon, \delta) \leq \frac{1}{\epsilon}\left(\ln|\mathcal{C}| + \ln\frac{1}{\delta}\right)$$

## VC Theory: Combinatorial Complexity

### Definition 1.4 (VC Dimension)
The **Vapnik-Chervonenkis dimension** of hypothesis class $\mathcal{H}$ is:
$$\text{VC}(\mathcal{H}) = \max\{n : \exists S = \{x_1, \ldots, x_n\} \text{ such that } |\mathcal{H}_S| = 2^n\}$$

where $\mathcal{H}_S = \{(h(x_1), \ldots, h(x_n)) : h \in \mathcal{H}\}$.

### Definition 1.5 (Growth Function)
$$\Pi_{\mathcal{H}}(n) = \max_{|S|=n} |\mathcal{H}_S|$$

### Theorem 1.2 (Sauer's Lemma)
If $\text{VC}(\mathcal{H}) = d < \infty$, then:
$$\Pi_{\mathcal{H}}(n) \leq \sum_{i=0}^d \binom{n}{i} \leq \left(\frac{en}{d}\right)^d \text{ for } n \geq d$$

### Theorem 1.3 (VC Generalization Bound)
For hypothesis class with $\text{VC}(\mathcal{H}) = d$, with probability $\geq 1-\delta$:
$$\sup_{h \in \mathcal{H}} |R(h) - \hat{R}_n(h)| \leq \sqrt{\frac{8d \log\frac{2en}{d} + 8\log\frac{4}{\delta}}{n}}$$

## Rademacher Complexity

### Definition 1.6 (Empirical Rademacher Complexity)
For function class $\mathcal{F}$ and sample $S = \{x_1, \ldots, x_n\}$:
$$\hat{\mathcal{R}}_n(\mathcal{F}) = \mathbb{E}_{\sigma}\left[\sup_{f \in \mathcal{F}} \frac{1}{n}\sum_{i=1}^n \sigma_i f(x_i)\right]$$

where $\sigma_i$ are independent Rademacher variables ($\pm 1$ with equal probability).

### Theorem 1.4 (Rademacher Generalization Bound)
With probability $\geq 1-\delta$:
$$R(\hat{h}_n) \leq \hat{R}_n(\hat{h}_n) + 2\hat{\mathcal{R}}_n(\mathcal{H}) + \sqrt{\frac{\ln(2/\delta)}{2n}}$$

## Deep Learning Theory

### Neural Tangent Kernel Framework
For infinite-width neural networks, training dynamics become:
$$\frac{df(x; \theta_t)}{dt} = -\eta \sum_{i=1}^n K(x, x_i)(f(x_i; \theta_t) - y_i)$$

where $K(x, x')$ is the Neural Tangent Kernel.

### Double Descent Phenomenon
Modern overparameterized models exhibit non-monotonic test error:
1. **Classical Regime**: Error decreases with model complexity
2. **Interpolation Threshold**: Peak in test error
3. **Modern Regime**: Error decreases again in overparameterized models

## Online Learning and Regret Bounds

### Definition 1.7 (Regret)
For online algorithm against competitor class $\mathcal{F}$:
$$\text{Regret}_T = \max_{f \in \mathcal{F}} \sum_{t=1}^T \ell_t(f(x_t), y_t) - \sum_{t=1}^T \ell_t(\hat{f}_t(x_t), y_t)$$

### Theorem 1.5 (Hedge Algorithm Bound)
Exponential weights algorithm achieves:
$$\text{Regret}_T \leq \sqrt{2T \ln|\mathcal{A}|}$$

for finite action set $\mathcal{A}$.

## Information-Theoretic Bounds

### Theorem 1.6 (Mutual Information Bound)
For any learning algorithm $A$:
$$\mathbb{E}[R(A(S)) - \hat{R}(A(S))] \leq \sqrt{\frac{I(S; A(S))}{2n}}$$

where $I(S; A(S))$ is mutual information between data and hypothesis.

### PAC-Bayes Framework
For prior $P$ and posterior $Q$ on $\mathcal{H}$:
$$\mathbb{E}_{h \sim Q}[R(h)] \leq \mathbb{E}_{h \sim Q}[\hat{R}_n(h)] + \sqrt{\frac{\text{KL}(Q \| P) + \ln(2n/\delta)}{2n}}$$

## Advanced Applications

### Meta-Learning Theory
Sample complexity for meta-learning with $T$ tasks and $k$ shots per task:
$$\mathcal{O}\left(\frac{\text{complexity}_{\text{source}}}{\sqrt{T}} + \frac{\text{complexity}_{\text{target}}}{\sqrt{k}}\right)$$

### Federated Learning
Convergence bounds accounting for:
- **Data Heterogeneity**: Non-IID distributions across clients
- **Communication Constraints**: Limited rounds of communication
- **Local Updates**: Multiple SGD steps between communications

### Adversarial Robustness
Robust generalization bounds:
$$R_{\text{adv}}(h) \leq \hat{R}_{\text{adv}}(h) + \text{standard bound} + L \cdot \epsilon$$

where $L$ is Lipschitz constant and $\epsilon$ is perturbation budget.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import learning_curve, validation_curve
from sklearn.linear_model import Ridge, LogisticRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.pipeline import Pipeline
from sklearn.datasets import make_regression, make_classification
from sklearn.metrics import mean_squared_error, log_loss
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from scipy import stats
from scipy.special import comb
from scipy.optimize import minimize
import warnings
warnings.filterwarnings('ignore')

# Set random seed for reproducibility
np.random.seed(42)
plt.style.use('default')
plt.rcParams['figure.figsize'] = (14, 10)
plt.rcParams['font.size'] = 12

class AdvancedStatisticalLearningFramework:
    """
    Comprehensive framework for advanced statistical learning theory
    covering PAC learning, VC theory, generalization bounds, and modern developments
    """
    
    def __init__(self):
        self.results = {}
        self.theoretical_constants = {
            'euler': np.e,
            'pi': np.pi,
            'sqrt_2pi': np.sqrt(2 * np.pi)
        }
    
    def demonstrate_pac_learning_fundamentals(self):
        """
        Comprehensive demonstration of PAC learning theory fundamentals
        """
        print("=" * 80)
        print("PAC LEARNING THEORY: MATHEMATICAL FOUNDATIONS")
        print("=" * 80)
        
        print("\n1. PAC LEARNING DEFINITIONS AND FRAMEWORK")
        print("-" * 50)
        
        print("Definition: A concept class C is PAC learnable if ∃ algorithm A and polynomial p such that:")
        print("∀ε,δ ∈ (0,1), ∀ distribution D, ∀ target concept c ∈ C:")
        print("If m ≥ p(1/ε, 1/δ, n, size(c)) training examples, then with probability ≥ 1-δ:")
        print("P[A(S)(x) ≠ c(x)] ≤ ε")
        
        # Sample complexity demonstration
        def finite_class_sample_complexity(num_hypotheses, epsilon, delta):
            """Sample complexity for finite hypothesis class"""
            return int(np.ceil((np.log(num_hypotheses) + np.log(1/delta)) / epsilon))
        
        print("\n2. SAMPLE COMPLEXITY FOR FINITE HYPOTHESIS CLASSES")
        print("-" * 50)
        
        # Various scenarios
        scenarios = [
            (10, 0.1, 0.05),      # Small class
            (100, 0.1, 0.05),     # Medium class
            (1000, 0.1, 0.05),    # Large class
            (100, 0.05, 0.05),    # Higher accuracy
            (100, 0.1, 0.01)      # Higher confidence
        ]
        
        print("Sample Complexity: m ≥ (ln|H| + ln(1/δ)) / ε")
        print(f"{'H_size':<8} {'ε':<8} {'δ':<8} {'Sample Complexity':<18}")
        print("-" * 50)
        
        for H_size, eps, delta in scenarios:
            m = finite_class_sample_complexity(H_size, eps, delta)
            print(f"{H_size:<8} {eps:<8} {delta:<8} {m:<18}")
        
        # Demonstrate PAC learning with polynomial regression
        print("\n3. PRACTICAL PAC LEARNING DEMONSTRATION")
        print("-" * 50)
        
        def true_concept(x):
            """True underlying concept: polynomial of degree 3"""
            return 0.5 * x + 0.3 * x**2 - 0.1 * x**3
        
        # Test PAC learning performance
        sample_sizes = [50, 100, 200, 500, 1000]
        polynomial_degrees = [1, 2, 3, 5, 8]
        
        # Fixed test set for evaluation
        x_test = np.linspace(-2, 2, 200)
        y_test_true = true_concept(x_test)
        
        pac_results = np.zeros((len(polynomial_degrees), len(sample_sizes)))
        
        for i, degree in enumerate(polynomial_degrees):
            for j, n_samples in enumerate(sample_sizes):
                # Multiple trials for statistical robustness
                errors = []
                for trial in range(10):
                    # Generate training data
                    x_train = np.random.uniform(-2, 2, n_samples)
                    y_train = true_concept(x_train) + 0.2 * np.random.randn(n_samples)
                    
                    # Fit polynomial with regularization
                    poly_features = PolynomialFeatures(degree=degree)
                    X_train = poly_features.fit_transform(x_train.reshape(-1, 1))
                    X_test = poly_features.transform(x_test.reshape(-1, 1))
                    
                    ridge = Ridge(alpha=0.01)
                    ridge.fit(X_train, y_train)
                    
                    # Compute test error
                    y_pred = ridge.predict(X_test)
                    test_error = np.mean((y_test_true - y_pred)**2)
                    errors.append(test_error)
                
                pac_results[i, j] = np.mean(errors)
        
        # Display results
        print(f"Mean Squared Error (averaged over 10 trials):")
        print(f"{'Degree':<8}", end="")
        for n in sample_sizes:
            print(f"{n:<10}", end="")
        print()
        print("-" * 58)
        
        for i, degree in enumerate(polynomial_degrees):
            print(f"{degree:<8}", end="")
            for j in range(len(sample_sizes)):
                print(f"{pac_results[i,j]:<10.4f}", end="")
            print()
        
        print(f"\nObservations:")
        print(f"- Degree 3 polynomial matches true concept (minimum bias)")
        print(f"- Higher degrees show overfitting with small samples")
        print(f"- Error decreases with sample size for appropriate complexity")
        print(f"- Demonstrates bias-variance trade-off in PAC learning")
        
        return pac_results, sample_sizes, polynomial_degrees
    
    def demonstrate_vc_theory_and_dimension(self):
        """
        Comprehensive demonstration of VC theory and Vapnik-Chervonenkis dimension
        """
        print("\n" + "=" * 80)
        print("VC THEORY: COMBINATORIAL COMPLEXITY AND GENERALIZATION")
        print("=" * 80)
        
        print("\n1. VC DIMENSION DEFINITIONS AND CONCEPTS")
        print("-" * 50)
        
        print("Definition: VC dimension of hypothesis class H is the size of the largest set")
        print("that can be shattered by H.")
        print("Set S is shattered by H if for every subset T ⊆ S, ∃h ∈ H such that")
        print("h⁻¹(1) ∩ S = T")
        
        # Growth function and Sauer's lemma
        def sauer_lemma_bound(n, d):
            """Sauer's lemma: upper bound on growth function"""
            if n <= d:
                return 2**n
            else:
                result = 0
                for i in range(d + 1):
                    result += comb(n, i, exact=True)
                return result
        
        def asymptotic_sauer_bound(n, d):
            """Asymptotic form of Sauer's lemma"""
            if n <= d or d == 0:
                return 2**n
            return (np.e * n / d)**d
        
        print("\n2. SAUER'S LEMMA AND GROWTH FUNCTION")
        print("-" * 50)
        
        print("Sauer's Lemma: If VC(H) = d, then Π_H(n) ≤ Σ(i=0 to d) C(n,i)")
        print("Asymptotic bound: Π_H(n) ≤ (en/d)^d for n ≥ d")
        
        # Demonstrate growth function bounds
        vc_dimensions = [2, 5, 10, 20]
        sample_sizes = [5, 10, 20, 50, 100]
        
        print(f"\nGrowth Function Bounds:")
        print(f"{'n\\VC(H)':<8}", end="")
        for d in vc_dimensions:
            print(f"{d:<12}", end="")
        print()
        print("-" * 60)
        
        for n in sample_sizes:
            print(f"{n:<8}", end="")
            for d in vc_dimensions:
                exact_bound = sauer_lemma_bound(n, d)
                asymp_bound = asymptotic_sauer_bound(n, d)
                if exact_bound == asymp_bound:
                    print(f"{exact_bound:<12.0f}", end="")
                else:
                    print(f"{min(exact_bound, asymp_bound):<12.0f}", end="")
            print()
        
        # VC dimension examples
        print("\n3. VC DIMENSION EXAMPLES")
        print("-" * 50)
        
        def linear_classifier_vc_demo():
            """Demonstrate VC dimension of linear classifiers"""
            print("Linear Classifiers in R^d:")
            print("VC dimension = d + 1")
            
            # 2D case: VC dimension = 3
            print("\nExample: Linear classifiers in R^2 (VC dimension = 3)")
            
            # Three points that can be shattered
            points_3 = np.array([[0, 0], [1, 0], [0, 1]])
            
            def can_realize_labeling(points, labels):
                """Check if linear classifier can realize given labeling"""
                n = len(points)
                X = np.column_stack([points, np.ones(n)])  # Add bias
                y = np.array(labels)
                
                try:
                    # Solve for weights
                    w = np.linalg.lstsq(X, y, rcond=None)[0]
                    predictions = np.sign(X @ w)
                    return np.array_equal(predictions, y)
                except:
                    return False
            
            # Test all 2^3 = 8 labelings
            realizable_count = 0
            total_labelings = 2**3
            
            print(f"Testing {total_labelings} labelings on 3 points:")
            for i in range(total_labelings):
                labels = [(i >> j) & 1 for j in range(3)]
                labels = [1 if l else -1 for l in labels]
                
                realizable = can_realize_labeling(points_3, labels)
                if realizable:
                    realizable_count += 1
                
                print(f"Labeling {labels}: {'✓' if realizable else '✗'}")
            
            print(f"\nRealizable labelings: {realizable_count}/{total_labelings}")
            print(f"Can shatter 3 points: {realizable_count == total_labelings}")
            
            # Four points that cannot be shattered
            points_4 = np.array([[0, 0], [1, 0], [0, 1], [1, 1]])
            print(f"\nTesting 4 points (should not be fully shatterable):")
            
            realizable_count_4 = 0
            total_labelings_4 = 2**4
            non_realizable_examples = []
            
            for i in range(total_labelings_4):
                labels = [(i >> j) & 1 for j in range(4)]
                labels = [1 if l else -1 for l in labels]
                
                realizable = can_realize_labeling(points_4, labels)
                if realizable:
                    realizable_count_4 += 1
                else:
                    non_realizable_examples.append(labels)
            
            print(f"Realizable labelings: {realizable_count_4}/{total_labelings_4}")
            print(f"Can shatter 4 points: {realizable_count_4 == total_labelings_4}")
            
            if non_realizable_examples:
                print(f"Examples of non-realizable labelings:")
                for example in non_realizable_examples[:3]:
                    print(f"  {example}")
            
            return points_3, points_4, realizable_count, realizable_count_4
        
        points_3, points_4, real_3, real_4 = linear_classifier_vc_demo()
        
        # Additional VC dimension examples
        print("\n4. OTHER VC DIMENSION EXAMPLES")
        print("-" * 50)
        
        vc_examples = [
            ("Threshold functions on R", 1),
            ("Intervals on R", 2),
            ("Linear classifiers in R^d", "d + 1"),
            ("Axis-aligned rectangles in R^2", 4),
            ("Neural networks with W weights", "O(W log W)"),
            ("Decision trees of depth d", "O(2^d)"),
            ("Polynomial classifiers of degree k in R^d", "O((dk)^d)")
        ]
        
        print(f"{'Hypothesis Class':<35} {'VC Dimension':<15}")
        print("-" * 50)
        for class_name, vc_dim in vc_examples:
            print(f"{class_name:<35} {vc_dim:<15}")
        
        return points_3, points_4, real_3, real_4
    
    def demonstrate_generalization_bounds(self):
        """
        Comprehensive demonstration of generalization bounds and complexity measures
        """
        print("\n" + "=" * 80)
        print("GENERALIZATION BOUNDS: FROM THEORY TO PRACTICE")
        print("=" * 80)
        
        print("\n1. FUNDAMENTAL GENERALIZATION BOUND")
        print("-" * 50)
        
        print("Theorem (VC Generalization Bound):")
        print("For hypothesis class H with VC(H) = d, with probability ≥ 1-δ:")
        print("sup_{h∈H} |R(h) - R̂_n(h)| ≤ √[(8d log(2en/d) + 8 log(4/δ)) / n]")
        
        def vc_generalization_bound(n, d, delta):
            """Compute VC generalization bound"""
            if n <= d:
                return float('inf')
            log_term = np.log(2 * np.e * n / d) if d > 0 else np.log(2 * np.e * n)
            numerator = 8 * d * log_term + 8 * np.log(4 / delta)
            return np.sqrt(numerator / n)
        
        def rademacher_bound_approximation(n, d, delta):
            """Approximate Rademacher complexity bound"""
            if n <= d:
                return float('inf')
            # Simplified approximation for linear classifiers
            return np.sqrt(2 * d / n) + np.sqrt(2 * np.log(2 / delta) / n)
        
        print("\n2. COMPARISON OF GENERALIZATION BOUNDS")
        print("-" * 50)
        
        # Parameter settings
        sample_sizes = [100, 500, 1000, 5000, 10000]
        vc_dimensions = [3, 10, 50, 100]
        confidence = 0.95
        delta = 1 - confidence
        
        print(f"Generalization Bounds (δ = {delta}, confidence = {confidence})")
        print(f"{'n':<8} {'d':<6} {'VC Bound':<12} {'Rademacher':<12} {'Ratio':<8}")
        print("-" * 50)
        
        for n in sample_sizes:
            for d in vc_dimensions:
                if n > 10 * d:  # Only show meaningful comparisons
                    vc_bound = vc_generalization_bound(n, d, delta)
                    rad_bound = rademacher_bound_approximation(n, d, delta)
                    ratio = vc_bound / rad_bound if rad_bound > 0 else float('inf')
                    
                    print(f"{n:<8} {d:<6} {vc_bound:<12.4f} {rad_bound:<12.4f} {ratio:<8.2f}")
        
        print("\n3. SAMPLE COMPLEXITY ANALYSIS")
        print("-" * 50)
        
        def sample_complexity_vc(epsilon, delta, d):
            """Sample complexity from VC theory"""
            c = 8  # Constant from VC bound
            log_term = np.log(4 / delta)
            # Solve n such that bound ≤ epsilon
            # Approximate solution for large n
            return int(np.ceil(c * (d + log_term) / (epsilon**2)))
        
        epsilons = [0.1, 0.05, 0.01]
        deltas = [0.1, 0.05, 0.01]
        vc_dims = [3, 10, 50]
        
        print(f"Sample Complexity: m(ε,δ,d)")
        print(f"{'ε':<8} {'δ':<8} {'d=3':<10} {'d=10':<10} {'d=50':<10}")
        print("-" * 50)
        
        for eps in epsilons:
            for delta in deltas:
                row = f"{eps:<8} {delta:<8}"
                for d in vc_dims:
                    m = sample_complexity_vc(eps, delta, d)
                    row += f"{m:<10}"
                print(row)
        
        print("\n4. EMPIRICAL VALIDATION OF BOUNDS")
        print("-" * 50)
        
        # Generate synthetic data to validate bounds
        def empirical_generalization_gap(n_samples=1000, n_trials=100, d=10):
            """Measure empirical generalization gap"""
            gaps = []
            
            for trial in range(n_trials):
                # Generate data
                X = np.random.randn(n_samples, d)
                true_w = np.random.randn(d)
                y = np.sign(X @ true_w) + 0.1 * np.random.randn(n_samples)
                y = np.sign(y)
                
                # Split data
                split = n_samples // 2
                X_train, X_test = X[:split], X[split:]
                y_train, y_test = y[:split], y[split:]
                
                # Train linear classifier
                clf = LogisticRegression(penalty='l2', C=1.0, max_iter=1000)
                clf.fit(X_train, y_train)
                
                # Compute empirical risk (training error)
                train_pred = clf.predict(X_train)
                train_error = np.mean(train_pred != y_train)
                
                # Compute true risk (test error)
                test_pred = clf.predict(X_test)
                test_error = np.mean(test_pred != y_test)
                
                gap = test_error - train_error
                gaps.append(gap)
            
            return np.array(gaps)
        
        # Test different sample sizes
        test_sizes = [100, 200, 500, 1000]
        dimension = 5
        
        print(f"Empirical Generalization Gap Analysis (d = {dimension})")
        print(f"{'Sample Size':<12} {'Mean Gap':<12} {'Std Gap':<12} {'VC Bound':<12}")
        print("-" * 50)
        
        for n in test_sizes:
            gaps = empirical_generalization_gap(n, n_trials=50, d=dimension)
            mean_gap = np.mean(gaps)
            std_gap = np.std(gaps)
            theoretical_bound = vc_generalization_bound(n//2, dimension + 1, 0.05)
            
            print(f"{n:<12} {mean_gap:<12.4f} {std_gap:<12.4f} {theoretical_bound:<12.4f}")
        
        print(f"\nObservations:")
        print(f"- Empirical gaps are much smaller than theoretical bounds")
        print(f"- Bounds are often loose but provide worst-case guarantees")
        print(f"- Gap decreases approximately as O(1/√n)")
        print(f"- Theory provides valuable insights despite looseness")
        
        return gaps
    
    def demonstrate_rademacher_complexity(self):
        """
        Advanced demonstration of Rademacher complexity theory
        """
        print("\n" + "=" * 80)
        print("RADEMACHER COMPLEXITY: ADVANCED COMPLEXITY MEASURES")
        print("=" * 80)
        
        print("\n1. RADEMACHER COMPLEXITY DEFINITION")
        print("-" * 50)
        
        print("Definition: Empirical Rademacher complexity of function class F:")
        print("R̂_n(F) = E_σ[sup_{f∈F} (1/n) Σᵢ σᵢ f(xᵢ)]")
        print("where σᵢ are independent Rademacher variables (±1 with equal probability)")
        
        print("\nRademacher Generalization Bound:")
        print("With probability ≥ 1-δ: R(ĥ) ≤ R̂(ĥ) + 2R̂_n(H) + √[ln(2/δ)/(2n)]")
        
        def empirical_rademacher_complexity(X, function_class, n_trials=1000):
            """Estimate empirical Rademacher complexity"""
            n, d = X.shape
            complexities = []
            
            for trial in range(n_trials):
                # Generate Rademacher variables
                sigma = np.random.choice([-1, 1], size=n)
                
                # Compute supremum over function class
                sup_value = -float('inf')
                for f_params in function_class:
                    # Linear functions: f(x) = w^T x
                    f_values = X @ f_params
                    empirical_process = np.mean(sigma * f_values)
                    sup_value = max(sup_value, empirical_process)
                
                complexities.append(sup_value)
            
            return np.mean(complexities), np.std(complexities)
        
        print("\n2. RADEMACHER COMPLEXITY EXAMPLES")
        print("-" * 50)
        
        # Example: Linear functions with bounded norm
        n_samples = 200
        dimensions = [2, 5, 10, 20]
        
        for d in dimensions:
            # Generate random data
            X = np.random.randn(n_samples, d)
            X = X / np.linalg.norm(X, axis=1, keepdims=True)  # Normalize
            
            # Function class: linear functions with ||w||₂ ≤ 1
            n_functions = 100
            function_class = []
            for _ in range(n_functions):
                w = np.random.randn(d)
                w = w / np.linalg.norm(w)  # Normalize to unit sphere
                function_class.append(w)
            
            rad_mean, rad_std = empirical_rademacher_complexity(X, function_class)
            
            # Theoretical bound for linear functions
            theoretical_bound = 1 / np.sqrt(n_samples)  # For ||w||₂ ≤ 1, ||x||₂ ≤ 1
            
            print(f"d={d:2d}: Empirical={rad_mean:.4f}±{rad_std:.4f}, "
                  f"Theoretical≤{theoretical_bound:.4f}")
        
        print("\n3. COMPARISON WITH VC BOUNDS")
        print("-" * 50)
        
        def compare_bounds(n, d, delta=0.05):
            """Compare different generalization bounds"""
            # VC bound
            vc_bound = vc_generalization_bound(n, d, delta)
            
            # Rademacher bound (approximate)
            rad_bound = np.sqrt(2 * d / n) + np.sqrt(2 * np.log(2 / delta) / n)
            
            # Stability bound (approximate for SGD)
            learning_rate = 0.01
            stability_bound = 2 * learning_rate + np.sqrt(2 * np.log(1 / delta) / n)
            
            return vc_bound, rad_bound, stability_bound
        
        sample_sizes = [100, 500, 1000, 5000]
        vc_dim = 10
        
        print(f"Bound Comparison (d = {vc_dim}, δ = 0.05)")
        print(f"{'n':<8} {'VC Bound':<12} {'Rademacher':<12} {'Stability':<12}")
        print("-" * 48)
        
        for n in sample_sizes:
            vc_b, rad_b, stab_b = compare_bounds(n, vc_dim)
            print(f"{n:<8} {vc_b:<12.4f} {rad_b:<12.4f} {stab_b:<12.4f}")
        
        print("\n4. CONCENTRATION INEQUALITIES")
        print("-" * 50)
        
        print("Key Concentration Inequalities in Learning Theory:")
        print("\n• Hoeffding's Inequality:")
        print("  P(|X̄ - E[X̄]| ≥ t) ≤ 2 exp(-2nt²/(b-a)²)")
        print("  For bounded random variables X ∈ [a,b]")
        
        print("\n• McDiarmid's Inequality:")
        print("  P(|f(X₁,...,Xₙ) - E[f]| ≥ t) ≤ 2 exp(-2t²/Σcᵢ²)")
        print("  For functions with bounded differences |f(...,xᵢ,...) - f(...,x'ᵢ,...)| ≤ cᵢ")
        
        print("\n• Bernstein's Inequality:")
        print("  P(|X̄ - E[X̄]| ≥ t) ≤ 2 exp(-nt²/(2σ² + 2bt/3))")
        print("  For random variables with variance σ² and bound b")
        
        # Demonstrate concentration
        def demonstrate_concentration(n=1000, n_trials=1000):
            """Demonstrate concentration of empirical mean"""
            true_mean = 0.5
            
            # Hoeffding bound
            t_values = np.linspace(0.01, 0.5, 20)
            empirical_probs = []
            hoeffding_bounds = []
            
            for t in t_values:
                # Count violations
                violations = 0
                for trial in range(n_trials):
                    X = np.random.uniform(0, 1, n)  # Bounded in [0,1]
                    empirical_mean = np.mean(X)
                    if abs(empirical_mean - true_mean) >= t:
                        violations += 1
                
                empirical_prob = violations / n_trials
                hoeffding_bound = 2 * np.exp(-2 * n * t**2)
                
                empirical_probs.append(empirical_prob)
                hoeffding_bounds.append(hoeffding_bound)
            
            return t_values, empirical_probs, hoeffding_bounds
        
        print(f"\n5. CONCENTRATION INEQUALITY VALIDATION")
        print("-" * 50)
        
        t_vals, emp_probs, hoff_bounds = demonstrate_concentration(n=500, n_trials=500)
        
        print(f"Hoeffding Inequality Validation (n=500, 500 trials)")
        print(f"{'t':<8} {'Empirical':<12} {'Hoeffding':<12} {'Ratio':<8}")
        print("-" * 40)
        
        for i in range(0, len(t_vals), 3):  # Show every 3rd value
            t = t_vals[i]
            emp = emp_probs[i]
            hoff = hoff_bounds[i]
            ratio = emp / hoff if hoff > 0 else 0
            print(f"{t:<8.3f} {emp:<12.4f} {hoff:<12.4f} {ratio:<8.3f}")
        
        print(f"\nObservations:")
        print(f"- Empirical probabilities well below Hoeffding bounds")
        print(f"- Concentration inequalities provide conservative guarantees")
        print(f"- Tighter bounds possible with additional assumptions")
        
        return t_vals, emp_probs, hoff_bounds
    
    def demonstrate_deep_learning_theory(self):
        """
        Advanced demonstration of modern deep learning theory
        """
        print("\n" + "=" * 80)
        print("DEEP LEARNING THEORY: MODERN DEVELOPMENTS")
        print("=" * 80)
        
        print("\n1. NEURAL TANGENT KERNEL THEORY")
        print("-" * 50)
        
        print("Neural Tangent Kernel (NTK) Framework:")
        print("For infinite-width neural networks, training dynamics become:")
        print("df/dt = -η Σᵢ K(x,xᵢ)(f(xᵢ;θₜ) - yᵢ)")
        print("where K(x,x') is the Neural Tangent Kernel")
        
        def neural_tangent_kernel_2layer(X1, X2, sigma=1.0):
            """
            Compute NTK for 2-layer ReLU network
            Simplified implementation for demonstration
            """
            n1, d = X1.shape
            n2, _ = X2.shape
            
            # Compute dot products
            dot_products = X1 @ X2.T / d
            
            # NTK for ReLU activation (simplified)
            # K(x,x') = σ²[(x·x')θ + (1-θ)π] where θ = arccos(x·x')
            kernel_matrix = np.zeros((n1, n2))
            
            for i in range(n1):
                for j in range(n2):
                    dot_prod = np.clip(dot_products[i, j], -1, 1)
                    theta = np.arccos(np.abs(dot_prod))
                    kernel_value = sigma**2 * (dot_prod * (np.pi - theta) + np.sqrt(1 - dot_prod**2)) / np.pi
                    kernel_matrix[i, j] = kernel_value
            
            return kernel_matrix
        
        # Demonstrate NTK behavior
        n_samples = 50
        d = 10
        X = np.random.randn(n_samples, d)
        X = X / np.linalg.norm(X, axis=1, keepdims=True)  # Normalize
        
        K = neural_tangent_kernel_2layer(X, X)
        
        print(f"\nNTK Matrix Properties (n={n_samples}, d={d}):")
        print(f"Eigenvalue decomposition:")
        eigenvals = np.linalg.eigvals(K)
        eigenvals = np.sort(eigenvals)[::-1]
        
        print(f"Top 5 eigenvalues: {eigenvals[:5]}")
        print(f"Condition number: {eigenvals[0] / eigenvals[-1]:.2f}")
        print(f"Trace: {np.trace(K):.2f}")
        print(f"Frobenius norm: {np.linalg.norm(K, 'fro'):.2f}")
        
        print("\n2. GENERALIZATION IN OVERPARAMETERIZED MODELS")
        print("-" * 50)
        
        print("Double Descent Phenomenon:")
        print("Test error exhibits non-monotonic behavior:")
        print("1. Classical regime: error decreases with model complexity")
        print("2. Interpolation threshold: peak in test error")
        print("3. Modern regime: error decreases again in overparameterized models")
        
        def simulate_double_descent():
            """Simulate double descent with polynomial regression"""
            # Generate data
            n_samples = 100
            d_true = 5
            X = np.random.randn(n_samples, d_true)
            true_w = np.random.randn(d_true)
            y = X @ true_w + 0.3 * np.random.randn(n_samples)
            
            # Test different polynomial degrees (model complexities)
            degrees = range(1, 50)
            train_errors = []
            test_errors = []
            
            # Fixed test set
            X_test = np.random.randn(200, d_true)
            y_test = X_test @ true_w + 0.3 * np.random.randn(200)
            
            for degree in degrees:
                # Create polynomial features
                poly = PolynomialFeatures(degree=degree, include_bias=False)
                X_poly = poly.fit_transform(X)
                X_test_poly = poly.transform(X_test)
                
                # Regularization decreases with increasing overparameterization
                alpha = 1e-6 if X_poly.shape[1] > n_samples else 0.01
                
                # Fit model
                ridge = Ridge(alpha=alpha)
                ridge.fit(X_poly, y)
                
                # Compute errors
                train_pred = ridge.predict(X_poly)
                test_pred = ridge.predict(X_test_poly)
                
                train_error = np.mean((y - train_pred)**2)
                test_error = np.mean((y_test - test_pred)**2)
                
                train_errors.append(train_error)
                test_errors.append(test_error)
            
            return degrees, train_errors, test_errors
        
        degrees, train_errs, test_errs = simulate_double_descent()
        
        # Find interpolation threshold
        interpolation_threshold = None
        for i, (deg, train_err) in enumerate(zip(degrees, train_errs)):
            if train_err < 1e-4:  # Approximately zero training error
                interpolation_threshold = deg
                break
        
        print(f"\nDouble Descent Simulation Results:")
        print(f"Interpolation threshold: degree ≈ {interpolation_threshold}")
        
        # Show key points
        min_test_idx = np.argmin(test_errs[:20])  # Classical minimum
        print(f"Classical minimum: degree {degrees[min_test_idx]}, "
              f"test error {test_errs[min_test_idx]:.4f}")
        
        if interpolation_threshold and interpolation_threshold < len(test_errs) - 5:
            overparameterized_errors = test_errs[interpolation_threshold+5:]
            if overparameterized_errors:
                min_over_idx = np.argmin(overparameterized_errors) + interpolation_threshold + 5
                print(f"Overparameterized minimum: degree {degrees[min_over_idx]}, "
                      f"test error {test_errs[min_over_idx]:.4f}")
        
        print("\n3. IMPLICIT REGULARIZATION IN SGD")
        print("-" * 50)
        
        print("Implicit Bias of Gradient Descent:")
        print("• SGD finds solutions with specific geometric properties")
        print("• For linear models: converges to minimum norm solution")
        print("• For neural networks: edge of stability, feature learning")
        print("• Learning rate and batch size affect implicit regularization")
        
        def demonstrate_implicit_regularization():
            """Show implicit regularization in linear regression"""
            # Overparameterized linear regression
            n_samples = 50
            n_features = 100  # More features than samples
            
            X = np.random.randn(n_samples, n_features)
            true_w = np.random.randn(n_features)
            true_w[50:] = 0  # Sparse true weights
            y = X @ true_w + 0.1 * np.random.randn(n_samples)
            
            # Different optimization methods
            methods = {
                'Ridge (λ=1.0)': Ridge(alpha=1.0),
                'Ridge (λ=0.01)': Ridge(alpha=0.01),
                'Ridge (λ=1e-6)': Ridge(alpha=1e-6)
            }
            
            print(f"Implicit Regularization Comparison:")
            print(f"Problem: {n_samples} samples, {n_features} features")
            print(f"True solution has {np.sum(true_w != 0)} non-zero coefficients")
            print()
            
            for name, model in methods.items():
                model.fit(X, y)
                w_estimated = model.coef_
                
                # Metrics
                l2_norm = np.linalg.norm(w_estimated)
                l1_norm = np.linalg.norm(w_estimated, ord=1)
                train_error = np.mean((y - model.predict(X))**2)
                reconstruction_error = np.linalg.norm(w_estimated - true_w)
                
                print(f"{name}:")
                print(f"  ‖w‖₂ = {l2_norm:.3f}")
                print(f"  ‖w‖₁ = {l1_norm:.3f}")
                print(f"  Train error = {train_error:.4f}")
                print(f"  ‖w - w*‖₂ = {reconstruction_error:.3f}")
                print()
            
            return methods
        
        reg_methods = demonstrate_implicit_regularization()
        
        print("4. MODERN GENERALIZATION BOUNDS")
        print("-" * 50)
        
        print("Advances Beyond Classical VC Theory:")
        print("• Norm-based bounds for neural networks")
        print("• PAC-Bayes bounds with data-dependent priors")
        print("• Stability-based generalization bounds")
        print("• Compression-based bounds")
        print("• Information-theoretic bounds")
        
        def norm_based_bound(network_weights, data_norm, n_samples, delta=0.05):
            """
            Simplified norm-based generalization bound for neural networks
            Based on spectral complexity
            """
            # Product of spectral norms of weight matrices
            spectral_complexity = 1.0
            for W in network_weights:
                spectral_norm = np.linalg.norm(W, ord=2)
                spectral_complexity *= spectral_norm
            
            # Simplified bound (actual bounds are more complex)
            bound = spectral_complexity * data_norm / np.sqrt(n_samples)
            bound += np.sqrt(np.log(1/delta) / (2 * n_samples))
            
            return bound
        
        # Example network
        layer_sizes = [10, 50, 20, 1]
        network_weights = []
        for i in range(len(layer_sizes) - 1):
            W = np.random.randn(layer_sizes[i+1], layer_sizes[i]) / np.sqrt(layer_sizes[i])
            network_weights.append(W)
        
        data_norms = [1.0, 2.0, 5.0]
        sample_sizes = [100, 500, 1000, 5000]
        
        print(f"\nNorm-based Generalization Bounds:")
        print(f"Network: {' → '.join(map(str, layer_sizes))}")
        print(f"{'n':<8} {'‖x‖=1':<10} {'‖x‖=2':<10} {'‖x‖=5':<10}")
        print("-" * 40)
        
        for n in sample_sizes:
            row = f"{n:<8}"
            for data_norm in data_norms:
                bound = norm_based_bound(network_weights, data_norm, n)
                row += f"{bound:<10.3f}"
            print(row)
        
        print(f"\nKey Insights from Modern Theory:")
        print(f"• Generalization depends on network norms, not just parameter count")
        print(f"• Implicit regularization explains generalization in practice")
        print(f"• Multiple regimes: underparameterized vs overparameterized")
        print(f"• Data-dependent bounds often tighter than worst-case bounds")
        
        return degrees, train_errs, test_errs
    
    def demonstrate_online_learning_theory(self):
        """
        Comprehensive demonstration of online learning and regret bounds
        """
        print("\n" + "=" * 80)
        print("ONLINE LEARNING THEORY: SEQUENTIAL DECISION MAKING")
        print("=" * 80)
        
        print("\n1. ONLINE LEARNING FRAMEWORK")
        print("-" * 50)
        
        print("Online Learning Protocol:")
        print("For t = 1, 2, ..., T:")
        print("1. Algorithm chooses action aₜ")
        print("2. Environment reveals loss ℓₜ(aₜ)")
        print("3. Algorithm suffers loss ℓₜ(aₜ)")
        print()
        print("Regret: R_T = Σₜ ℓₜ(aₜ) - min_{a∈A} Σₜ ℓₜ(a)")
        print("Goal: Achieve sublinear regret O(√T)")
        
        class OnlineLearningAlgorithms:
            """Collection of online learning algorithms"""
            
            @staticmethod
            def follow_the_leader(losses_history):
                """Follow the Leader algorithm"""
                if not losses_history:
                    return 0  # Default action
                
                # Find action with minimum cumulative loss so far
                cumulative_losses = np.sum(losses_history, axis=0)
                return np.argmin(cumulative_losses)
            
            @staticmethod
            def follow_the_regularized_leader(losses_history, regularization=0.1):
                """Follow the Regularized Leader with entropy regularization"""
                if not losses_history:
                    return np.random.choice(len(losses_history[0]) if losses_history else 2)
                
                n_actions = len(losses_history[0])
                cumulative_losses = np.sum(losses_history, axis=0)
                
                # Add entropy regularization
                regularized_losses = cumulative_losses + regularization * np.log(np.arange(1, n_actions + 1))
                return np.argmin(regularized_losses)
            
            @staticmethod
            def exponential_weights(losses_history, learning_rate=0.1):
                """Exponential Weights (Hedge) algorithm"""
                if not losses_history:
                    n_actions = len(losses_history[0]) if losses_history else 2
                    return np.random.choice(n_actions)
                
                n_actions = len(losses_history[0])
                cumulative_losses = np.sum(losses_history, axis=0)
                
                # Compute weights
                weights = np.exp(-learning_rate * cumulative_losses)
                probabilities = weights / np.sum(weights)
                
                # Sample action according to probabilities
                return np.random.choice(n_actions, p=probabilities)
            
            @staticmethod
            def online_gradient_descent(gradients_history, learning_rate=0.1, projection_radius=1.0):
                """Online Gradient Descent for continuous actions"""
                if not gradients_history:
                    return np.zeros(2)  # Default action in R²
                
                # Initialize at origin
                action = np.zeros_like(gradients_history[0])
                
                # Gradient descent updates with projection
                for gradient in gradients_history:
                    action -= learning_rate * gradient
                    # Project onto ball of radius projection_radius
                    norm = np.linalg.norm(action)
                    if norm > projection_radius:
                        action = projection_radius * action / norm
                
                return action
        
        print("\n2. REGRET ANALYSIS FOR SPECIFIC ALGORITHMS")
        print("-" * 50)
        
        def simulate_online_learning(T=1000, n_actions=3, noise_level=0.1):
            """Simulate online learning scenario"""
            # Generate adversarial loss sequence
            # Each action has time-varying expected loss
            time_steps = np.arange(T)
            expected_losses = np.zeros((T, n_actions))
            
            # Action 0: sine wave
            expected_losses[:, 0] = 0.5 + 0.3 * np.sin(2 * np.pi * time_steps / 100)
            # Action 1: cosine wave
            expected_losses[:, 1] = 0.5 + 0.3 * np.cos(2 * np.pi * time_steps / 150)
            # Action 2: linear trend
            expected_losses[:, 2] = 0.3 + 0.4 * time_steps / T
            
            # Add noise
            actual_losses = expected_losses + noise_level * np.random.randn(T, n_actions)
            actual_losses = np.clip(actual_losses, 0, 1)  # Ensure losses in [0,1]
            
            # Run algorithms
            algorithms = {
                'Follow the Leader': OnlineLearningAlgorithms.follow_the_leader,
                'Regularized Leader': OnlineLearningAlgorithms.follow_the_regularized_leader,
                'Exponential Weights': OnlineLearningAlgorithms.exponential_weights
            }
            
            results = {}
            
            for alg_name, algorithm in algorithms.items():
                chosen_actions = []
                cumulative_loss = 0
                losses_history = []
                
                for t in range(T):
                    # Choose action
                    action = algorithm(losses_history)
                    chosen_actions.append(action)
                    
                    # Observe loss
                    loss = actual_losses[t, action]
                    cumulative_loss += loss
                    
                    # Update history
                    losses_history.append(actual_losses[t])
                
                # Compute regret
                best_fixed_action = np.argmin(np.sum(actual_losses, axis=0))
                best_cumulative_loss = np.sum(actual_losses[:, best_fixed_action])
                regret = cumulative_loss - best_cumulative_loss
                
                results[alg_name] = {
                    'actions': chosen_actions,
                    'cumulative_loss': cumulative_loss,
                    'regret': regret,
                    'regret_per_round': regret / T
                }
            
            return results, actual_losses, expected_losses
        
        # Run simulation
        results, actual_losses, expected_losses = simulate_online_learning(T=1000)
        
        print(f"Online Learning Simulation Results (T=1000):")
        print(f"{'Algorithm':<20} {'Cumulative Loss':<15} {'Regret':<10} {'Regret/T':<10}")
        print("-" * 55)
        
        for alg_name, result in results.items():
            print(f"{alg_name:<20} {result['cumulative_loss']:<15.3f} "
                  f"{result['regret']:<10.3f} {result['regret_per_round']:<10.4f}")
        
        # Theoretical regret bounds
        print(f"\n3. THEORETICAL REGRET BOUNDS")
        print("-" * 50)
        
        T_values = [100, 500, 1000, 5000, 10000]
        n_actions = 3
        
        print(f"Theoretical Regret Bounds (K = {n_actions} actions):")
        print(f"{'T':<8} {'FTL':<12} {'Hedge':<12} {'OGD':<12}")
        print("-" * 44)
        
        for T in T_values:
            # Follow the Leader: O(T^(2/3)) for general losses
            ftl_bound = T**(2/3)
            
            # Hedge (Exponential Weights): O(√(T log K))
            hedge_bound = np.sqrt(T * np.log(n_actions))
            
            # Online Gradient Descent: O(√T) for convex losses
            ogd_bound = np.sqrt(T)
            
            print(f"{T:<8} {ftl_bound:<12.1f} {hedge_bound:<12.1f} {ogd_bound:<12.1f}")
        
        print(f"\n4. ADAPTIVE ALGORITHMS AND LOWER BOUNDS")
        print("-" * 50)
        
        print("Advanced Online Learning Concepts:")
        print()
        print("• Adaptive Regret: Competing with different comparators over time")
        print("• Second-order bounds: O(√(VT)) where V is loss variance")
        print("• Parameter-free algorithms: No need to know T in advance")
        print("• Strongly convex losses: O(log T) regret possible")
        print("• Lower bounds: Ω(√T) regret unavoidable in worst case")
        
        def demonstrate_adaptive_regret():
            """Demonstrate adaptive regret concept"""
            T = 1000
            window_size = 100
            
            # Generate data with regime changes
            losses = np.zeros((T, 2))
            
            # Regime 1: Action 0 better
            losses[:300, 0] = 0.3 + 0.1 * np.random.randn(300)
            losses[:300, 1] = 0.7 + 0.1 * np.random.randn(300)
            
            # Regime 2: Action 1 better
            losses[300:700, 0] = 0.8 + 0.1 * np.random.randn(400)
            losses[300:700, 1] = 0.2 + 0.1 * np.random.randn(400)
            
            # Regime 3: Action 0 better again
            losses[700:, 0] = 0.1 + 0.1 * np.random.randn(300)
            losses[700:, 1] = 0.9 + 0.1 * np.random.randn(300)
            
            losses = np.clip(losses, 0, 1)
            
            # Compute sliding window regret
            sliding_regrets = []
            for t in range(window_size, T):
                window_losses = losses[t-window_size:t]
                
                # Best action in this window
                best_action = np.argmin(np.sum(window_losses, axis=0))
                
                # Regret if we had chosen best action throughout window
                # This is a simplified calculation
                sliding_regret = window_size * 0.1  # Placeholder
                sliding_regrets.append(sliding_regret)
            
            return sliding_regrets
        
        sliding_regrets = demonstrate_adaptive_regret()
        
        print(f"\nAdaptive Regret Demonstration:")
        print(f"Sliding window regret analysis for non-stationary environment")
        print(f"Average sliding regret: {np.mean(sliding_regrets):.3f}")
        print(f"Max sliding regret: {np.max(sliding_regrets):.3f}")
        print(f"Min sliding regret: {np.min(sliding_regrets):.3f}")
        
        return results, actual_losses, sliding_regrets
    
    def demonstrate_information_theoretic_bounds(self):
        """
        Advanced demonstration of information-theoretic learning bounds
        """
        print("\n" + "=" * 80)
        print("INFORMATION-THEORETIC LEARNING BOUNDS")
        print("=" * 80)
        
        print("\n1. MUTUAL INFORMATION BOUNDS")
        print("-" * 50)
        
        print("Theorem (Xu & Raginsky, 2017):")
        print("For any learning algorithm A:")
        print("E[R(A(S)) - R̂(A(S))] ≤ √[I(S; A(S)) / (2n)]")
        print("where I(S; A(S)) is mutual information between data and hypothesis")
        
        print("\nKey Insights:")
        print("• Generalization depends on information learned about training data")
        print("• Algorithms that memorize training data have poor generalization")
        print("• Compression leads to generalization")
        
        def estimate_mutual_information(X, Y, bins=10):
            """
            Estimate mutual information using histogram method
            Simplified implementation for demonstration
            """
            # Discretize continuous variables
            X_discrete = np.digitize(X, bins=np.histogram(X, bins=bins)[1]) - 1
            Y_discrete = np.digitize(Y, bins=np.histogram(Y, bins=bins)[1]) - 1
            
            # Compute joint and marginal distributions
            joint_hist = np.histogram2d(X_discrete, Y_discrete, bins=bins)[0]
            joint_prob = joint_hist / np.sum(joint_hist)
            
            X_prob = np.sum(joint_prob, axis=1)
            Y_prob = np.sum(joint_prob, axis=0)
            
            # Compute mutual information
            mi = 0
            for i in range(bins):
                for j in range(bins):
                    if joint_prob[i, j] > 0:
                        mi += joint_prob[i, j] * np.log(
                            joint_prob[i, j] / (X_prob[i] * Y_prob[j])
                        )
            
            return mi
        
        print("\n2. PAC-BAYES THEORY")
        print("-" * 50)
        
        print("PAC-Bayes Bound:")
        print("For any prior P and posterior Q:")
        print("E_{h~Q}[R(h)] ≤ E_{h~Q}[R̂(h)] + √[(KL(Q||P) + ln(2n/δ)) / (2n)]")
        print("where KL(Q||P) is KL divergence between posterior and prior")
        
        def pac_bayes_bound(empirical_risk, kl_divergence, n_samples, delta=0.05):
            """Compute PAC-Bayes generalization bound"""
            complexity_term = (kl_divergence + np.log(2 * n_samples / delta)) / (2 * n_samples)
            return empirical_risk + np.sqrt(complexity_term)
        
        # Demonstrate PAC-Bayes with different priors
        n_samples = 1000
        empirical_risks = [0.1, 0.2, 0.3]
        kl_divergences = [0.5, 1.0, 2.0, 5.0, 10.0]
        
        print(f"\nPAC-Bayes Bounds (n = {n_samples}, δ = 0.05):")
        print(f"{'KL(Q||P)':<10} {'R̂=0.1':<10} {'R̂=0.2':<10} {'R̂=0.3':<10}")
        print("-" * 40)
        
        for kl in kl_divergences:
            row = f"{kl:<10.1f}"
            for emp_risk in empirical_risks:
                bound = pac_bayes_bound(emp_risk, kl, n_samples)
                row += f"{bound:<10.3f}"
            print(row)
        
        print("\n3. COMPRESSION BOUNDS")
        print("-" * 50)
        
        print("Compression-based Generalization:")
        print("If hypothesis can be compressed to C bits, then:")
        print("R(h) ≤ R̂(h) + √[(C + ln(2/δ)) / (2n)]")
        
        def compression_bound(empirical_risk, compression_bits, n_samples, delta=0.05):
            """Compute compression-based generalization bound"""
            complexity_term = (compression_bits + np.log(2 / delta)) / (2 * n_samples)
            return empirical_risk + np.sqrt(complexity_term)
        
        # Model compression scenarios
        model_types = [
            ("Linear (d=10)", 10 * 32),      # 10 parameters × 32 bits
            ("Linear (d=100)", 100 * 32),   # 100 parameters × 32 bits
            ("Neural Net (1000 params)", 1000 * 32),
            ("Compressed NN (100 bits)", 100),
            ("Quantized NN (8-bit)", 1000 * 8)
        ]
        
        empirical_risk = 0.15
        sample_sizes = [100, 500, 1000, 5000]
        
        print(f"\nCompression Bounds (R̂ = {empirical_risk}):")
        print(f"{'Model Type':<25} {'Bits':<8}", end="")
        for n in sample_sizes:
            print(f"n={n:<6}", end="")
        print()
        print("-" * 70)
        
        for model_name, bits in model_types:
            row = f"{model_name:<25} {bits:<8}"
            for n in sample_sizes:
                bound = compression_bound(empirical_risk, bits, n)
                row += f"{bound:<8.3f}"
            print(row)
        
        print("\n4. STABILITY AND INFORMATION")
        print("-" * 50)
        
        print("Connection between Stability and Information:")
        print("• Stable algorithms have low mutual information with training data")
        print("• Differential privacy provides stability and information bounds")
        print("• SGD with noise injection reduces mutual information")
        
        class DifferentiallyPrivateSGD:
            """Simplified DP-SGD implementation"""
            
            def __init__(self, noise_scale=1.0, clip_norm=1.0):
                self.noise_scale = noise_scale
                self.clip_norm = clip_norm
                self.privacy_spent = 0
            
            def private_gradient(self, gradient):
                """Add noise to gradient for differential privacy"""
                # Clip gradient
                norm = np.linalg.norm(gradient)
                if norm > self.clip_norm:
                    gradient = self.clip_norm * gradient / norm
                
                # Add Gaussian noise
                noise = np.random.normal(0, self.noise_scale, gradient.shape)
                private_grad = gradient + noise
                
                # Update privacy budget (simplified)
                self.privacy_spent += 1 / (self.noise_scale**2)
                
                return private_grad
            
            def get_privacy_bound(self, delta=1e-5):
                """Simplified privacy bound computation"""
                # This is a very simplified bound
                epsilon = self.privacy_spent * delta
                return epsilon
        
        # Demonstrate privacy-generalization trade-off
        noise_scales = [0.1, 0.5, 1.0, 2.0, 5.0]
        
        print(f"\nPrivacy-Generalization Trade-off:")
        print(f"{'Noise Scale':<12} {'Privacy ε':<12} {'Stability':<12} {'Info Bound':<12}")
        print("-" * 48)
        
        for noise_scale in noise_scales:
            dp_sgd = DifferentiallyPrivateSGD(noise_scale=noise_scale)
            
            # Simulate some gradient updates
            for _ in range(100):
                grad = np.random.randn(10)  # Random gradient
                dp_sgd.private_gradient(grad)
            
            epsilon = dp_sgd.get_privacy_bound()
            stability = 1 / noise_scale  # Inverse relationship
            info_bound = np.sqrt(epsilon)  # Simplified information bound
            
            print(f"{noise_scale:<12.1f} {epsilon:<12.3f} {stability:<12.3f} {info_bound:<12.3f}")
        
        print("\n5. MODERN INFORMATION-THEORETIC RESULTS")
        print("-" * 50)
        
        print("Recent Developments:")
        print("• Conditional Mutual Information bounds")
        print("• Information complexity of learning algorithms")
        print("• Connections to minimum description length (MDL)")
        print("• Information-theoretic analysis of deep learning")
        
        print("\nOpen Questions:")
        print("• Tight information-theoretic characterization of generalization")
        print("• Role of algorithmic information theory in learning")
        print("• Information bottleneck principle in deep networks")
        print("• Mutual information in federated learning settings")
        
        return model_types, noise_scales
    
    def demonstrate_advanced_applications(self):
        """
        Demonstrate advanced applications and research frontiers
        """
        print("\n" + "=" * 80)
        print("ADVANCED APPLICATIONS AND RESEARCH FRONTIERS")
        print("=" * 80)
        
        print("\n1. META-LEARNING AND FEW-SHOT LEARNING")
        print("-" * 50)
        
        print("Meta-Learning Theory:")
        print("• Learning to learn from limited data")
        print("• Theory of representation learning")
        print("• Sample complexity of meta-learning")
        print("• Transfer learning guarantees")
        
        def meta_learning_sample_complexity(n_tasks, k_shots, complexity_source, complexity_target):
            """
            Simplified meta-learning sample complexity bound
            Based on Baxter (2000) and recent extensions
            """
            # Source domain sample complexity
            source_complexity = complexity_source / np.sqrt(n_tasks)
            
            # Target domain sample complexity  
            target_complexity = complexity_target / np.sqrt(k_shots)
            
            # Combined bound
            total_complexity = source_complexity + target_complexity
            
            return total_complexity
        
        # Meta-learning scenarios
        scenarios = [
            (10, 5, 100, 20),    # Few tasks, few shots
            (100, 5, 100, 20),   # Many tasks, few shots
            (100, 20, 100, 20),  # Many tasks, many shots
            (1000, 1, 100, 20),  # Many tasks, one shot
        ]
        
        print(f"\nMeta-Learning Sample Complexity:")
        print(f"{'Tasks':<8} {'Shots':<8} {'Source Complexity':<18} {'Target Complexity':<18} {'Total':<10}")
        print("-" * 70)
        
        for n_tasks, k_shots, comp_src, comp_tgt in scenarios:
            complexity = meta_learning_sample_complexity(n_tasks, k_shots, comp_src, comp_tgt)
            src_term = comp_src / np.sqrt(n_tasks)
            tgt_term = comp_tgt / np.sqrt(k_shots)
            print(f"{n_tasks:<8} {k_shots:<8} {src_term:<18.3f} {tgt_term:<18.3f} {complexity:<10.3f}")
        
        print("\n2. FEDERATED LEARNING THEORY")
        print("-" * 50)
        
        print("Federated Learning Challenges:")
        print("• Heterogeneous data distributions")
        print("• Communication constraints")
        print("• Privacy preservation")
        print("• Non-IID data analysis")
        
        def federated_learning_bound(n_clients, local_steps, heterogeneity, communication_rounds):
            """
            Simplified federated learning convergence bound
            Based on recent federated optimization theory
            """
            # Heterogeneity penalty
            heterogeneity_term = heterogeneity / communication_rounds
            
            # Local update error
            local_error = 1 / (local_steps * n_clients)
            
            # Communication efficiency
            comm_efficiency = 1 / np.sqrt(communication_rounds)
            
            total_error = heterogeneity_term + local_error + comm_efficiency
            
            return total_error
        
        # Federated scenarios
        fed_scenarios = [
            (10, 10, 0.1, 100),   # Few clients, low heterogeneity
            (100, 10, 0.1, 100),  # Many clients, low heterogeneity  
            (100, 10, 1.0, 100),  # Many clients, high heterogeneity
            (100, 1, 0.1, 100),   # Many clients, single local step
        ]
        
        print(f"\nFederated Learning Convergence:")
        print(f"{'Clients':<8} {'Local Steps':<12} {'Heterogeneity':<14} {'Comm Rounds':<12} {'Error':<10}")
        print("-" * 58)
        
        for n_clients, local_steps, heterogeneity, comm_rounds in fed_scenarios:
            error = federated_learning_bound(n_clients, local_steps, heterogeneity, comm_rounds)
            print(f"{n_clients:<8} {local_steps:<12} {heterogeneity:<14.1f} {comm_rounds:<12} {error:<10.4f}")
        
        print("\n3. ADVERSARIAL ROBUSTNESS THEORY")
        print("-" * 50)
        
        print("Adversarial Learning Framework:")
        print("• Robust risk: R_adv(h) = E[max_{||δ||≤ε} ℓ(h(x+δ), y)]")
        print("• Trade-off between accuracy and robustness")
        print("• Certified defenses and verification")
        
        def adversarial_generalization_bound(standard_bound, perturbation_size, lipschitz_constant):
            """
            Adversarial generalization bound
            Accounts for perturbation size and function smoothness
            """
            # Standard generalization term
            standard_term = standard_bound
            
            # Adversarial perturbation term
            adversarial_term = lipschitz_constant * perturbation_size
            
            return standard_term + adversarial_term
        
        # Adversarial scenarios
        perturbation_sizes = [0.01, 0.05, 0.1, 0.3]
        lipschitz_constants = [1.0, 5.0, 10.0, 50.0]
        base_bound = 0.1
        
        print(f"\nAdversarial Generalization Bounds:")
        print(f"Standard bound = {base_bound}")
        print(f"{'ε':<8} {'L=1':<10} {'L=5':<10} {'L=10':<10} {'L=50':<10}")
        print("-" * 48)
        
        for eps in perturbation_sizes:
            row = f"{eps:<8}"
            for L in lipschitz_constants:
                bound = adversarial_generalization_bound(base_bound, eps, L)
                row += f"{bound:<10.3f}"
            print(row)
        
        print("\n4. QUANTUM MACHINE LEARNING")
        print("-" * 50)
        
        print("Quantum Learning Theory:")
        print("• Quantum sample complexity advantages")
        print("• Quantum PAC learning framework")
        print("• Variational quantum algorithms")
        print("• Quantum advantage for specific problems")
        
        def quantum_sample_complexity(classical_complexity, quantum_advantage_factor):
            """
            Theoretical quantum sample complexity
            Some problems show quadratic speedup
            """
            return classical_complexity / quantum_advantage_factor
        
        # Quantum learning scenarios
        problems = [
            ("Dense linear systems", 1000, 4),      # Quadratic speedup
            ("Sparse problems", 1000, 2),           # Moderate speedup
            ("General learning", 1000, 1.1),       # Minimal speedup
            ("Quantum-specific", 1000, 10),        # Large speedup
        ]
        
        print(f"\nQuantum vs Classical Sample Complexity:")
        print(f"{'Problem Type':<20} {'Classical':<12} {'Quantum':<12} {'Speedup':<10}")
        print("-" * 54)
        
        for problem, classical, advantage in problems:
            quantum = quantum_sample_complexity(classical, advantage)
            print(f"{problem:<20} {classical:<12} {quantum:<12.1f} {advantage:<10.1f}x")
        
        print("\n5. CAUSAL LEARNING THEORY")
        print("-" * 50)
        
        print("Causal Machine Learning:")
        print("• Learning causal relationships from observational data")
        print("• Sample complexity of causal discovery")
        print("• Confounding and identification")
        print("• Intervention and counterfactual learning")
        
        def causal_sample_complexity(n_variables, graph_density, intervention_strength):
            """
            Simplified causal discovery sample complexity
            Depends on graph structure and intervention quality
            """
            # Combinatorial factor for graph structure
            structure_complexity = n_variables**2 * graph_density
            
            # Intervention quality factor
            intervention_factor = 1 / intervention_strength
            
            return structure_complexity * intervention_factor
        
        # Causal learning scenarios
        causal_scenarios = [
            (5, 0.2, 1.0),    # Small dense graph, strong interventions
            (10, 0.2, 1.0),   # Medium dense graph, strong interventions
            (10, 0.2, 0.1),   # Medium dense graph, weak interventions
            (20, 0.1, 0.5),   # Large sparse graph, medium interventions
        ]
        
        print(f"\nCausal Discovery Sample Complexity:")
        print(f"{'Variables':<10} {'Density':<10} {'Intervention':<12} {'Complexity':<12}")
        print("-" * 44)
        
        for n_vars, density, intervention in causal_scenarios:
            complexity = causal_sample_complexity(n_vars, density, intervention)
            print(f"{n_vars:<10} {density:<10.1f} {intervention:<12.1f} {complexity:<12.1f}")
        
        print("\n6. FUTURE RESEARCH DIRECTIONS")
        print("-" * 50)
        
        print("Open Problems and Future Directions:")
        print()
        print("• Foundation Models Theory:")
        print("  - Scaling laws for large language models")
        print("  - In-context learning theoretical foundations")
        print("  - Emergence and phase transitions")
        print()
        print("• Multimodal Learning:")
        print("  - Cross-modal generalization bounds")
        print("  - Representation alignment theory")
        print("  - Multimodal sample complexity")
        print()
        print("• Continual Learning:")
        print("  - Catastrophic forgetting analysis")
        print("  - Lifelong learning bounds")
        print("  - Plasticity-stability trade-offs")
        print()
        print("• AI Safety and Alignment:")
        print("  - Theoretical frameworks for AI safety")
        print("  - Robustness guarantees for deployed systems")
        print("  - Value alignment and reward learning")
        print()
        print("• Computational Complexity:")
        print("  - Statistical-computational trade-offs")
        print("  - Lower bounds for learning algorithms")
        print("  - Average-case complexity analysis")
        
        return scenarios, fed_scenarios, problems
    
    def comprehensive_demonstration(self):
        """
        Run comprehensive demonstration of all advanced statistical learning theory concepts
        """
        print("ADVANCED STATISTICAL LEARNING THEORY")
        print("Comprehensive Mathematical Framework and Applications")
        print("=" * 80)
        
        # Run all demonstrations
        pac_results = self.demonstrate_pac_learning_fundamentals()
        vc_results = self.demonstrate_vc_theory_and_dimension()
        bounds_results = self.demonstrate_generalization_bounds()
        rademacher_results = self.demonstrate_rademacher_complexity()
        deep_results = self.demonstrate_deep_learning_theory()
        online_results = self.demonstrate_online_learning_theory()
        info_results = self.demonstrate_information_theoretic_bounds()
        advanced_results = self.demonstrate_advanced_applications()
        
        print("\n" + "=" * 80)
        print("COMPREHENSIVE SUMMARY: THEORETICAL FOUNDATIONS")
        print("=" * 80)
        
        print("\nKey Theoretical Results Demonstrated:")
        print("1. PAC Learning Framework: Sample complexity bounds for learnable concepts")
        print("2. VC Theory: Combinatorial complexity and uniform convergence")
        print("3. Generalization Bounds: From training to test performance guarantees")
        print("4. Rademacher Complexity: Data-dependent complexity measures")
        print("5. Deep Learning Theory: Modern overparameterization phenomena")
        print("6. Online Learning: Sequential decision making and regret bounds")
        print("7. Information Theory: Mutual information and compression bounds")
        print("8. Advanced Applications: Meta-learning, federated learning, quantum ML")
        
        print(f"\nMathematical Sophistication:")
        print(f"• Graduate-level mathematical rigor and proof techniques")
        print(f"• Integration of probability theory, analysis, and optimization")
        print(f"• Concentration inequalities and uniform convergence theory")
        print(f"• Information-theoretic and algorithmic perspectives")
        print(f"• Modern developments in deep learning and AI")
        
        print(f"\nPractical Impact:")
        print(f"• Algorithm design principles based on theoretical insights")
        print(f"• Model selection and hyperparameter tuning guidance")
        print(f"• Understanding of fundamental limits and trade-offs")
        print(f"• Bridge between theory and practice in modern ML")
        
        return {
            'pac_results': pac_results,
            'vc_results': vc_results,
            'bounds_results': bounds_results,
            'rademacher_results': rademacher_results,
            'deep_results': deep_results,
            'online_results': online_results,
            'info_results': info_results,
            'advanced_results': advanced_results
        }

# Create and run the comprehensive framework
framework = AdvancedStatisticalLearningFramework()
comprehensive_results = framework.comprehensive_demonstration()
```

## 2. PAC Learning and Agnostic Framework {#pac-learning}

### PAC Learning Theory

**Definition 2.1** (PAC Learnable): A concept class $\mathcal{C}$ over input space $\mathcal{X}$ is **PAC learnable** if there exists an algorithm $A$ and a polynomial $p(\cdot, \cdot, \cdot, \cdot)$ such that:

For all $\epsilon, \delta \in (0, 1)$, all distributions $\mathcal{D}$ over $\mathcal{X}$, and all target concepts $c \in \mathcal{C}$:

If $A$ is given $m \geq p(1/\epsilon, 1/\delta, n, \text{size}(c))$ training examples drawn i.i.d. from $\mathcal{D}$ according to $c$, then with probability at least $1 - \delta$:

$$R_{0-1}(A(S)) \leq \epsilon$$

where $R_{0-1}(h) = \mathbb{P}_{x \sim \mathcal{D}}[h(x) \neq c(x)]$ is the 0-1 risk.

**Definition 2.2** (Sample Complexity): The **sample complexity** $m_{\mathcal{C}}(\epsilon, \delta)$ is the minimum number of samples needed to PAC learn $\mathcal{C}$ with accuracy $\epsilon$ and confidence $1-\delta$.

**Theorem 2.1** (Fundamental Sample Complexity Bound): For finite concept class $|\mathcal{C}| < \infty$:
$$m_{\mathcal{C}}(\epsilon, \delta) \leq \frac{1}{\epsilon}\left(\log|\mathcal{C}| + \log\frac{1}{\delta}\right)$$

**Proof Sketch**: Use union bound over all hypotheses inconsistent with true concept.

### Agnostic PAC Learning

**Definition 2.3** (Agnostic PAC Learning): Hypothesis class $\mathcal{H}$ is **agnostic PAC learnable** if there exists algorithm $A$ and polynomial $p$ such that:

For all $\epsilon, \delta \in (0, 1)$ and all distributions $\mathcal{D}$ over $\mathcal{X} \times \mathcal{Y}$:

With $m \geq p(1/\epsilon, 1/\delta, n)$ samples, with probability $\geq 1-\delta$:
$$R(A(S)) \leq \min_{h \in \mathcal{H}} R(h) + \epsilon$$

**Key Difference**: No assumption that optimal hypothesis achieves zero error.

### Uniform Convergence and PAC Learning

**Theorem 2.2** (Uniform Convergence ⟺ Agnostic PAC Learning): For hypothesis class $\mathcal{H}$:
$$\mathcal{H} \text{ is agnostic PAC learnable} \iff \mathcal{H} \text{ has uniform convergence property}$$

**Definition 2.4** (Uniform Convergence Property): $\mathcal{H}$ has uniform convergence if:
$$\lim_{n \to \infty} \mathbb{E}_{S \sim \mathcal{D}^n}\left[\sup_{h \in \mathcal{H}} |R(h) - \hat{R}_n(h)|\right] = 0$$

### The Growth Function and Sauer's Lemma

**Definition 2.5** (Growth Function): 
$$\Pi_{\mathcal{H}}(n) = \max_{S: |S|=n} |\{(h(x_1), \ldots, h(x_n)) : h \in \mathcal{H}\}|$$

**Theorem 2.3** (Sauer's Lemma): If $\text{VC}(\mathcal{H}) = d < \infty$, then:
$$\Pi_{\mathcal{H}}(n) \leq \sum_{i=0}^d \binom{n}{i}$$

When $n \geq d$:
$$\Pi_{\mathcal{H}}(n) \leq \left(\frac{en}{d}\right)^d$$

**Proof Outline**: Combinatorial argument using dual shatter coefficient and induction.

### VC Dimension Examples

**Example 2.1** (Linear Classifiers): In $\mathbb{R}^d$, linear classifiers $\{x \mapsto \text{sign}(\langle w, x \rangle + b)\}$ have:
$$\text{VC}(\text{linear}) = d + 1$$

**Example 2.2** (Axis-Aligned Rectangles): In $\mathbb{R}^2$, axis-aligned rectangles have:
$$\text{VC}(\text{rectangles}) = 4$$

**Example 2.3** (Neural Networks): For neural networks with $W$ weights and piecewise polynomial activation:
$$\text{VC}(\text{neural}) = O(W \log W)$$

### Sample Complexity via VC Theory

**Theorem 2.4** (VC Generalization Bound): For hypothesis class with $\text{VC}(\mathcal{H}) = d$, with probability $\geq 1-\delta$:
$$\sup_{h \in \mathcal{H}} |R(h) - \hat{R}_n(h)| \leq \sqrt{\frac{8d \log\frac{2en}{d} + 8\log\frac{4}{\delta}}{n}}$$

**Corollary 2.1** (Sample Complexity): For agnostic PAC learning with accuracy $\epsilon$ and confidence $1-\delta$:
$$m \geq \frac{8d \log\frac{2em}{d} + 8\log\frac{4}{\delta}}{\epsilon^2}$$

### Lower Bounds

**Theorem 2.5** (VC Lower Bound): Any algorithm that agnostic PAC learns $\mathcal{H}$ requires:
$$m \geq \Omega\left(\frac{d + \log(1/\delta)}{\epsilon^2}\right)$$

This shows VC dimension characterizes sample complexity up to logarithmic factors.

### Beyond Binary Classification

**Multi-Class Learning**: 
- **Natarajan Dimension**: Generalization of VC dimension
- **Sample Complexity**: $O(k \cdot d \cdot \log(k/\epsilon)/\epsilon^2)$ for $k$ classes, dimension $d$

**Regression**:
- **Pseudo-Dimension**: Complexity measure for real-valued functions
- **Fat-Shattering Dimension**: Scale-sensitive complexity measure

### Computational Considerations

**Definition 2.6** (Efficient PAC Learning): Class $\mathcal{C}$ is **efficiently PAC learnable** if:
1. $\mathcal{C}$ is PAC learnable
2. Learning algorithm runs in polynomial time
3. Learned hypothesis can be evaluated in polynomial time

**Hardness Results**: Many natural concept classes (e.g., Boolean formulas, neural networks) are believed to be computationally hard to learn.

### Structural Risk Minimization

**Principle**: Balance empirical risk and model complexity.

For nested sequence $\mathcal{H}_1 \subset \mathcal{H}_2 \subset \cdots \subset \mathcal{H}$:

$$\hat{h}_n = \arg\min_{h \in \bigcup_k \mathcal{H}_k} \left[\hat{R}_n(h) + \text{complexity}(\mathcal{H}_k)\right]$$

where $h \in \mathcal{H}_k$.

**Theorem 2.6** (SRM Bound): With appropriate complexity penalties:
$$R(\hat{h}_n) \leq \min_k \left[\inf_{h \in \mathcal{H}_k} R(h) + O\left(\sqrt{\frac{\text{VC}(\mathcal{H}_k) + \log k + \log(1/\delta)}{n}}\right)\right]$$

```python
class StatisticalLearningTheory:
    def __init__(self):
        self.results = {}
    
    def pac_learning_demo(self):
        """Demonstrate PAC learning concepts with polynomial regression"""
        print("PAC Learning Demo - Polynomial Regression:")
        
        # Generate synthetic data
        def true_function(x):
            return 0.5 * x + 0.3 * x**2 - 0.1 * x**3
        
        # Different sample sizes
        sample_sizes = [20, 50, 100, 200, 500, 1000]
        degrees = [1, 2, 3, 5, 10]
        
        # Test data (fixed)
        x_test = np.linspace(-1, 1, 100)
        y_test_true = true_function(x_test)
        
        results = {}
        
        for degree in degrees:
            test_errors = []
            
            for m in sample_sizes:
                # Generate training data
                x_train = np.random.uniform(-1, 1, m)
                y_train = true_function(x_train) + 0.1 * np.random.randn(m)
                
                # Fit polynomial
                poly_features = PolynomialFeatures(degree=degree)
                X_train_poly = poly_features.fit_transform(x_train.reshape(-1, 1))
                X_test_poly = poly_features.transform(x_test.reshape(-1, 1))
                
                # Ridge regression for regularization
                ridge = Ridge(alpha=0.01)
                ridge.fit(X_train_poly, y_train)
                
                # Test error
                y_pred = ridge.predict(X_test_poly)
                test_error = np.mean((y_test_true - y_pred)**2)
                test_errors.append(test_error)
            
            results[degree] = test_errors
        
        # Display results
        print(f"\nTest Error vs Sample Size:")
        print(f"{'Degree':<8}", end="")
        for m in sample_sizes:
            print(f"{m:<8}", end="")
        print()
        
        for degree in degrees:
            print(f"{degree:<8}", end="")
            for error in results[degree]:
                print(f"{error:<8.4f}", end="")
            print()
        
        print(f"\nObservations:")
        print(f"- Higher degree polynomials need more data to generalize well")
        print(f"- Test error generally decreases with more training data")
        print(f"- Degree 3 matches true function, performs best")
        
        return results, true_function
    
    def vc_dimension_demo(self):
        """Demonstrate VC dimension concept"""
        print(f"\nVC Dimension Demo:")
        
        def can_shatter_points(points, labels):
            """Check if linear classifier can realize given labeling"""
            n = len(points)
            if n == 0:
                return True
            
            # Add bias term
            X = np.column_stack([points, np.ones(n)])
            y = np.array(labels)
            
            # Try to solve Xw = y (approximately)
            try:
                w = np.linalg.lstsq(X, y, rcond=None)[0]
                predictions = np.sign(X @ w)
                return np.allclose(predictions, y)
            except:
                return False
        
        # Test VC dimension for linear classifiers in 2D
        print(f"Testing VC dimension of linear classifiers in 2D:")
        
        # 3 points (should be shatterable)
        points_3 = np.array([[0, 0], [1, 0], [0, 1]])
        
        print(f"\n3 points: {points_3.tolist()}")
        shatterable_count = 0
        total_labelings = 2**3
        
        for i in range(total_labelings):
            labels = [(i >> j) & 1 for j in range(3)]
            labels = [1 if l else -1 for l in labels]  # Convert to ±1
            if can_shatter_points(points_3, labels):
                shatterable_count += 1
        
        print(f"Realizable labelings: {shatterable_count}/{total_labelings}")
        print(f"Can shatter 3 points: {shatterable_count == total_labelings}")
        
        # 4 points (should not be fully shatterable)
        points_4 = np.array([[0, 0], [1, 0], [0, 1], [1, 1]])
        
        print(f"\n4 points: {points_4.tolist()}")
        shatterable_count = 0
        total_labelings = 2**4
        
        for i in range(total_labelings):
            labels = [(i >> j) & 1 for j in range(4)]
            labels = [1 if l else -1 for l in labels]
            if can_shatter_points(points_4, labels):
                shatterable_count += 1
        
        print(f"Realizable labelings: {shatterable_count}/{total_labelings}")
        print(f"Can shatter 4 points: {shatterable_count == total_labelings}")
        
        print(f"\nConclusion: VC dimension of linear classifiers in 2D is 3")
        print(f"General result: VC dimension in d dimensions is d+1")
        
        return points_3, points_4
    
    def generalization_bound_demo(self):
        """Demonstrate generalization bounds"""
        print(f"\nGeneralization Bounds Demo:")
        
        # VC generalization bound
        def vc_bound(m, d, delta):
            """VC generalization bound"""
            if m <= d:
                return float('inf')
            return np.sqrt((d * np.log(2*m/d) + np.log(4/delta)) / (2*m))
        
        # Parameters
        sample_sizes = np.logspace(1, 4, 50).astype(int)
        vc_dimensions = [3, 10, 50, 100]
        delta = 0.05  # 95% confidence
        
        print(f"VC Generalization Bound (δ = {delta}):")
        print(f"Bound: ε ≤ √[(d log(2m/d) + log(4/δ)) / (2m)]")
        print()
        
        # Calculate bounds for different VC dimensions
        for d in vc_dimensions:
            print(f"VC dimension d = {d}:")
            
            selected_sizes = [100, 500, 1000, 5000, 10000]
            for m in selected_sizes:
                if m > d:
                    bound = vc_bound(m, d, delta)
                    print(f"  m = {m:5d}: ε ≤ {bound:.4f}")
            print()
        
        print(f"Key insights:")
        print(f"- Generalization bound decreases as O(1/√m)")
        print(f"- Higher VC dimension requires more data for same guarantee")
        print(f"- Bounds are often loose but provide important theoretical insights")
        
        return vc_bound

    def rademacher_complexity_demo(self):
        """Demonstrate Rademacher complexity"""
        print(f"\nRademacher Complexity Demo:")
        print(f"Rademacher complexity measures the richness of a function class")
        print(f"by how well it can fit random noise\n")

        # Generate data
        np.random.seed(42)
        n_samples = 100
        X = np.random.randn(n_samples, 2)

        # Different hypothesis classes with varying complexity
        def compute_empirical_rademacher(X, predictions_list):
            """Compute empirical Rademacher complexity"""
            n = len(X)
            n_trials = 1000
            rad_values = []

            for _ in range(n_trials):
                # Random Rademacher variables (±1)
                sigma = 2 * np.random.randint(0, 2, n) - 1

                # Maximum correlation with random labels
                max_corr = max([np.abs(np.dot(sigma, pred)) / n
                               for pred in predictions_list])
                rad_values.append(max_corr)

            return np.mean(rad_values)

        # 1. Linear classifiers
        linear_preds = []
        for _ in range(50):
            w = np.random.randn(2)
            linear_preds.append(np.sign(X @ w))

        rad_linear = compute_empirical_rademacher(X, linear_preds)

        # 2. Polynomial classifiers (degree 2)
        poly_preds = []
        X_poly = np.column_stack([X, X[:, 0]**2, X[:, 1]**2, X[:, 0]*X[:, 1]])
        for _ in range(50):
            w = np.random.randn(5)
            poly_preds.append(np.sign(X_poly @ w))

        rad_poly = compute_empirical_rademacher(X, poly_preds)

        # 3. High-capacity classifiers (degree 5)
        X_poly5 = PolynomialFeatures(degree=5).fit_transform(X)
        high_preds = []
        for _ in range(50):
            w = np.random.randn(X_poly5.shape[1])
            high_preds.append(np.sign(X_poly5 @ w))

        rad_high = compute_empirical_rademacher(X, high_preds)

        print(f"Empirical Rademacher Complexity (n={n_samples}):")
        print(f"  Linear classifiers (d=2):     {rad_linear:.4f}")
        print(f"  Polynomial classifiers (d=2): {rad_poly:.4f}")
        print(f"  High-degree poly (d=5):       {rad_high:.4f}")

        print(f"\nTheoretical bound: R̂(F) ≤ √(d/n)")
        print(f"  Linear: √(2/{n_samples}) = {np.sqrt(2/n_samples):.4f}")
        print(f"  Poly-2: √(5/{n_samples}) = {np.sqrt(5/n_samples):.4f}")
        print(f"  Poly-5: √(21/{n_samples}) = {np.sqrt(21/n_samples):.4f}")

        print(f"\nKey insights:")
        print(f"- More complex models have higher Rademacher complexity")
        print(f"- Rademacher complexity decreases as O(1/√n)")
        print(f"- It provides data-dependent generalization bounds")

        return rad_linear, rad_poly, rad_high

    def bias_variance_tradeoff_demo(self):
        """Demonstrate bias-variance decomposition"""
        print(f"\nBias-Variance Tradeoff Demo:")

        # True function
        def f_true(x):
            return np.sin(2 * np.pi * x)

        # Generate test points
        x_test = np.linspace(0, 1, 100)
        y_true = f_true(x_test)

        # Different model complexities
        degrees = [1, 3, 9, 15]
        n_trials = 100
        n_train = 25

        results = {}

        for degree in degrees:
            predictions = []

            # Run multiple trials with different training sets
            for trial in range(n_trials):
                # Generate training data with noise
                x_train = np.random.uniform(0, 1, n_train)
                y_train = f_true(x_train) + np.random.randn(n_train) * 0.3

                # Fit polynomial
                poly = PolynomialFeatures(degree=degree)
                X_train = poly.fit_transform(x_train.reshape(-1, 1))
                X_test = poly.transform(x_test.reshape(-1, 1))

                ridge = Ridge(alpha=0.001)
                ridge.fit(X_train, y_train)
                y_pred = ridge.predict(X_test)

                predictions.append(y_pred)

            predictions = np.array(predictions)

            # Compute bias and variance
            mean_pred = predictions.mean(axis=0)
            bias_squared = np.mean((mean_pred - y_true)**2)
            variance = np.mean(predictions.var(axis=0))
            noise = 0.3**2  # Known noise variance

            # Expected test error = bias² + variance + noise
            expected_error = bias_squared + variance + noise

            results[degree] = {
                'bias_squared': bias_squared,
                'variance': variance,
                'expected_error': expected_error,
                'predictions': predictions,
                'mean_pred': mean_pred
            }

        # Display results
        print(f"\n{'Degree':<8} {'Bias²':<12} {'Variance':<12} {'Total Error':<12}")
        print(f"{'-'*50}")
        for degree in degrees:
            r = results[degree]
            print(f"{degree:<8} {r['bias_squared']:<12.4f} {r['variance']:<12.4f} {r['expected_error']:<12.4f}")

        print(f"\nObservations:")
        print(f"- Low complexity: High bias, low variance (underfitting)")
        print(f"- High complexity: Low bias, high variance (overfitting)")
        print(f"- Optimal model balances bias and variance")

        return results, x_test, y_true

    def empirical_risk_minimization_demo(self):
        """Demonstrate empirical risk minimization"""
        print(f"\nEmpirical Risk Minimization (ERM) Demo:")

        # Generate data
        np.random.seed(42)
        n_train = 200
        n_test = 1000

        # True decision boundary: XOR-like pattern
        X_train = np.random.randn(n_train, 2)
        y_train = ((X_train[:, 0] > 0) != (X_train[:, 1] > 0)).astype(int) * 2 - 1
        y_train = y_train * (1 - 2 * (np.random.rand(n_train) < 0.1))  # 10% noise

        X_test = np.random.randn(n_test, 2)
        y_test = ((X_test[:, 0] > 0) != (X_test[:, 1] > 0)).astype(int) * 2 - 1

        # Different hypothesis classes
        models = {
            'Linear': LogisticRegression(penalty=None, max_iter=1000),
            'Poly-2': Pipeline([
                ('poly', PolynomialFeatures(degree=2)),
                ('clf', LogisticRegression(penalty=None, max_iter=1000))
            ]),
            'RBF SVM': SVC(kernel='rbf', gamma='scale', C=1.0),
            'Random Forest': RandomForestClassifier(n_estimators=100, max_depth=10, random_state=42)
        }

        results = {}

        print(f"\n{'Model':<15} {'Train Error':<15} {'Test Error':<15} {'Generalization Gap':<20}")
        print(f"{'-'*70}")

        for name, model in models.items():
            # Fit model (ERM: minimize empirical risk)
            model.fit(X_train, y_train)

            # Compute errors
            train_error = 1 - model.score(X_train, y_train)
            test_error = 1 - model.score(X_test, y_test)
            gen_gap = test_error - train_error

            results[name] = {
                'train_error': train_error,
                'test_error': test_error,
                'gen_gap': gen_gap,
                'model': model
            }

            print(f"{name:<15} {train_error:<15.4f} {test_error:<15.4f} {gen_gap:<20.4f}")

        print(f"\nKey insights:")
        print(f"- ERM minimizes training error within hypothesis class")
        print(f"- Generalization gap depends on model complexity and sample size")
        print(f"- More complex models can fit training data better but may overfit")

        return results, X_train, y_train, X_test, y_test

    def sample_complexity_demo(self):
        """Demonstrate sample complexity analysis"""
        print(f"\nSample Complexity Demo:")
        print(f"How much data is needed to learn a good model?\n")

        # True model: logistic regression with 5 features
        np.random.seed(42)
        d = 5  # dimensions
        w_true = np.random.randn(d)
        w_true = w_true / np.linalg.norm(w_true)

        sample_sizes = [20, 50, 100, 200, 500, 1000, 2000, 5000]
        n_trials = 50

        results = {'train_errors': [], 'test_errors': [], 'param_errors': []}

        # Fixed large test set
        X_test = np.random.randn(10000, d)
        y_test_prob = 1 / (1 + np.exp(-X_test @ w_true))
        y_test = (y_test_prob > 0.5).astype(int) * 2 - 1

        for m in sample_sizes:
            train_errs = []
            test_errs = []
            param_errs = []

            for _ in range(n_trials):
                # Generate training data
                X_train = np.random.randn(m, d)
                y_train_prob = 1 / (1 + np.exp(-X_train @ w_true))
                y_train = (y_train_prob > 0.5).astype(int) * 2 - 1

                # Fit model
                clf = LogisticRegression(penalty='l2', C=1.0, max_iter=1000)
                clf.fit(X_train, y_train)

                # Errors
                train_err = 1 - clf.score(X_train, y_train)
                test_err = 1 - clf.score(X_test, y_test)

                # Parameter error (normalized)
                w_learned = clf.coef_[0]
                w_learned = w_learned / np.linalg.norm(w_learned)
                param_err = np.linalg.norm(w_learned - w_true)

                train_errs.append(train_err)
                test_errs.append(test_err)
                param_errs.append(param_err)

            results['train_errors'].append(np.mean(train_errs))
            results['test_errors'].append(np.mean(test_errs))
            results['param_errors'].append(np.mean(param_errs))

        print(f"{'Sample Size':<12} {'Train Error':<15} {'Test Error':<15} {'Param Error':<15}")
        print(f"{'-'*60}")
        for i, m in enumerate(sample_sizes):
            print(f"{m:<12} {results['train_errors'][i]:<15.4f} "
                  f"{results['test_errors'][i]:<15.4f} {results['param_errors'][i]:<15.4f}")

        print(f"\nTheoretical sample complexity:")
        print(f"- To learn d-dimensional linear classifier: m = O(d/ε²)")
        print(f"- For ε=0.1 and d=5: m ≈ {int(5 / 0.1**2)} samples")
        print(f"- Observed: ~200-500 samples gives good performance")

        return results, sample_sizes

    def structural_risk_minimization_demo(self):
        """Demonstrate structural risk minimization (SRM)"""
        print(f"\nStructural Risk Minimization Demo:")
        print(f"SRM chooses model complexity based on data-dependent bounds\n")

        # Generate data
        np.random.seed(42)
        n = 100
        X = np.random.uniform(-3, 3, n)
        y_true = np.sin(X) + 0.1 * X**2
        y = y_true + 0.5 * np.random.randn(n)

        # Validation set
        X_val = np.random.uniform(-3, 3, 200)
        y_val = np.sin(X_val) + 0.1 * X_val**2 + 0.5 * np.random.randn(200)

        # Try different polynomial degrees (model structure)
        degrees = range(1, 16)

        train_errors = []
        val_errors = []
        srm_scores = []

        for d in degrees:
            # Fit model
            poly = PolynomialFeatures(degree=d)
            X_poly = poly.fit_transform(X.reshape(-1, 1))
            X_val_poly = poly.transform(X_val.reshape(-1, 1))

            ridge = Ridge(alpha=0.1)
            ridge.fit(X_poly, y)

            # Empirical risk
            y_pred_train = ridge.predict(X_poly)
            train_err = np.mean((y - y_pred_train)**2)

            # Validation error
            y_pred_val = ridge.predict(X_val_poly)
            val_err = np.mean((y_val - y_pred_val)**2)

            # SRM score: empirical risk + complexity penalty
            # Using VC-like penalty: sqrt((d * log(n) + log(1/δ)) / n)
            delta = 0.05
            complexity_penalty = np.sqrt((d * np.log(n) + np.log(1/delta)) / n)
            srm_score = train_err + complexity_penalty

            train_errors.append(train_err)
            val_errors.append(val_err)
            srm_scores.append(srm_score)

        # Find best models
        best_srm = np.argmin(srm_scores) + 1
        best_val = np.argmin(val_errors) + 1

        print(f"{'Degree':<8} {'Train Error':<15} {'Val Error':<15} {'SRM Score':<15}")
        print(f"{'-'*60}")
        for i, d in enumerate(degrees):
            marker = ""
            if d == best_srm:
                marker += " <- SRM"
            if d == best_val:
                marker += " <- Best Val"
            print(f"{d:<8} {train_errors[i]:<15.4f} {val_errors[i]:<15.4f} "
                  f"{srm_scores[i]:<15.4f}{marker}")

        print(f"\nSRM selected degree: {best_srm}")
        print(f"Best validation degree: {best_val}")
        print(f"\nKey insights:")
        print(f"- SRM balances empirical risk and model complexity")
        print(f"- Provides principled way to choose model structure")
        print(f"- Often aligns with cross-validation results")

        return train_errors, val_errors, srm_scores, list(degrees)

# Create demo instance
demo = StatisticalLearningTheory()
pac_results, true_func = demo.pac_learning_demo()
points_3, points_4 = demo.vc_dimension_demo()
vc_bound_func = demo.generalization_bound_demo()
rad_results = demo.rademacher_complexity_demo()
bv_results, x_test_bv, y_true_bv = demo.bias_variance_tradeoff_demo()
erm_results, X_train_erm, y_train_erm, X_test_erm, y_test_erm = demo.empirical_risk_minimization_demo()
sc_results, sample_sizes = demo.sample_complexity_demo()
srm_train, srm_val, srm_scores, degrees = demo.structural_risk_minimization_demo()
```

```python
# Visualization 8: Real-World Application - Learning Curves
from sklearn.datasets import load_digits, load_breast_cancer
from sklearn.model_selection import learning_curve

fig = plt.figure(figsize=(15, 5))

# Dataset 1: Digits (multiclass)
digits = load_digits()
X_digits, y_digits = digits.data, digits.target

# Dataset 2: Breast Cancer (binary)
cancer = load_breast_cancer()
X_cancer, y_cancer = cancer.data, cancer.target

datasets = [
    (X_digits, y_digits, 'Digits Classification', LogisticRegression(max_iter=1000)),
    (X_cancer, y_cancer, 'Breast Cancer', LogisticRegression(max_iter=1000)),
]

for idx, (X, y, title, model) in enumerate(datasets):
    ax = plt.subplot(1, 2, idx + 1)
    
    # Compute learning curves
    train_sizes, train_scores, val_scores = learning_curve(
        model, X, y, cv=5, n_jobs=-1,
        train_sizes=np.linspace(0.1, 1.0, 10),
        scoring='accuracy'
    )
    
    train_mean = train_scores.mean(axis=1)
    train_std = train_scores.std(axis=1)
    val_mean = val_scores.mean(axis=1)
    val_std = val_scores.std(axis=1)
    
    # Plot
    ax.plot(train_sizes, train_mean, 'o-', linewidth=2, markersize=6, label='Training Score')
    ax.fill_between(train_sizes, train_mean - train_std, train_mean + train_std, alpha=0.2)
    
    ax.plot(train_sizes, val_mean, 's-', linewidth=2, markersize=6, label='Validation Score')
    ax.fill_between(train_sizes, val_mean - val_std, val_mean + val_std, alpha=0.2)
    
    # Generalization gap
    gen_gap = train_mean - val_mean
    ax2 = ax.twinx()
    ax2.plot(train_sizes, gen_gap, '^-', linewidth=2, markersize=5, 
            color='red', alpha=0.6, label='Generalization Gap')
    ax2.set_ylabel('Generalization Gap', fontsize=10, color='red')
    ax2.tick_params(axis='y', labelcolor='red')
    
    ax.set_xlabel('Training Set Size', fontsize=11)
    ax.set_ylabel('Accuracy', fontsize=11)
    ax.set_title(f'{title}\n(d={X.shape[1]} features)', fontsize=12, fontweight='bold')
    ax.legend(loc='lower right', fontsize=9)
    ax.grid(True, alpha=0.3)

plt.suptitle('Learning Curves on Real Datasets', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()

print("\nKey Observations:")
print("- Training accuracy starts high and may decrease slightly (more data = harder to fit perfectly)")
print("- Validation accuracy increases with more training data")
print("- Generalization gap decreases as sample size increases")
print("- Plateaus indicate we've reached the model's capacity or Bayes error")
```
