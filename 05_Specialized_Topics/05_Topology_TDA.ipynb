## 1. Introduction and Theoretical Foundations {#introduction}

Topology, the study of properties preserved under continuous deformations, provides fundamental insights into the global structure of spaces and data. Topological Data Analysis (TDA) applies these profound mathematical concepts to understand the "shape" of data, offering tools that transcend traditional statistical approaches and capture geometric and topological features invisible to classical methods.

### Historical Development and Mathematical Evolution

**Classical Foundations (1895-1930)**:
- **Poincaré's Contributions**: Henri Poincaré (1895), fundamental group and homology theory
- **Brouwer's Fixed Point Theorem**: L.E.J. Brouwer (1911), topological degree and invariance
- **Alexander Duality**: James Alexander (1915), relationship between homology groups
- **Seifert-van Kampen Theorem**: Herbert Seifert and Egbert van Kampen (1933), fundamental group computation

**Modern Algebraic Topology (1930-1970)**:

- **Cohomology Theory**: Development of dual homology theory

**Computational Era (1970-2000)**:

### Mathematical Framework and Topological Spaces

**Definition 1.1** (Topological Space): A **topological space** is a pair $(X, \tau)$ where $X$ is a set and $\tau$ is a collection of subsets of $X$ (called **open sets**) satisfying:
1. $\emptyset, X \in \tau$
2. Any union of sets in $\tau$ belongs to $\tau$
3. Any finite intersection of sets in $\tau$ belongs to $\tau$

**Definition 1.2** (Continuous Map): A function $f: X \to Y$ between topological spaces is **continuous** if for every open set $U \subseteq Y$, the preimage $f^{-1}(U)$ is open in $X$.

**Definition 1.3** (Homeomorphism): A bijective continuous map $f: X \to Y$ with continuous inverse $f^{-1}: Y \to X$ is a **homeomorphism**. Spaces $X$ and $Y$ are **homeomorphic** if such a map exists.

**Fundamental Properties**:
- **Connectedness**: A space is connected if it cannot be written as the union of two disjoint non-empty open sets
- **Compactness**: Every open cover has a finite subcover
- **Separation Axioms**: Hausdorff, regular, normal spaces

### Metric Topology and Distance-Based Structures

**Definition 1.4** (Metric Space): A **metric space** is a pair $(X, d)$ where $X$ is a set and $d: X \times X \to \mathbb{R}_{\geq 0}$ satisfies:
1. $d(x, y) = 0 \iff x = y$ (identity of indiscernibles)
2. $d(x, y) = d(y, x)$ (symmetry)
3. $d(x, z) \leq d(x, y) + d(y, z)$ (triangle inequality)

**Metric Topology**: For metric space $(X, d)$, the **metric topology** is generated by open balls:
$$B_r(x) = \{y \in X : d(x, y) < r\}$$

**Important Distance Functions**:
- **Euclidean Distance**: $d(x, y) = \sqrt{\sum_{i=1}^n (x_i - y_i)^2}$
- **Manhattan Distance**: $d(x, y) = \sum_{i=1}^n |x_i - y_i|$
- **Hausdorff Distance**: $d_H(A, B) = \max\{\sup_{a \in A} \inf_{b \in B} d(a, b), \sup_{b \in B} \inf_{a \in A} d(a, b)\}$

### Topological Invariants and Classification

**Definition 1.5** (Topological Invariant): A property or function that is preserved under homeomorphisms.

**Fundamental Invariants**:
1. **Euler Characteristic**: $\chi(X) = \sum_{i=0}^n (-1)^i \beta_i$ where $\beta_i$ are Betti numbers
2. **Fundamental Group**: $\pi_1(X, x_0)$ captures 1-dimensional holes
3. **Homology Groups**: $H_k(X)$ capture $k$-dimensional holes
4. **Homotopy Groups**: $\pi_k(X, x_0)$ for $k \geq 1$

### Algebraic Topology Foundations

**Definition 1.6** (Homotopy): Two continuous maps $f, g: X \to Y$ are **homotopic** if there exists a continuous map $H: X \times [0, 1] \to Y$ such that:
- $H(x, 0) = f(x)$ for all $x \in X$
- $H(x, 1) = g(x)$ for all $x \in X$

**Homotopy Equivalence**: Spaces $X$ and $Y$ are **homotopy equivalent** if there exist continuous maps $f: X \to Y$ and $g: Y \to X$ such that $g \circ f \simeq \text{id}_X$ and $f \circ g \simeq \text{id}_Y$.

**Definition 1.7** (Fundamental Group): For based space $(X, x_0)$, the fundamental group $\pi_1(X, x_0)$ consists of homotopy classes of loops based at $x_0$ with group operation given by concatenation.

**Examples**:
- $\pi_1(\mathbb{R}^n, x_0) = \{e\}$ (trivial group)
- $\pi_1(S^1, x_0) = \mathbb{Z}$ (infinite cyclic group)
- $\pi_1(T^2, x_0) = \mathbb{Z} \times \mathbb{Z}$ (free abelian group of rank 2)

### Simplicial Complexes and Combinatorial Topology

**Definition 1.8** (Abstract Simplicial Complex): An **abstract simplicial complex** $K$ on vertex set $V$ is a collection of finite subsets of $V$ such that:
1. If $v \in V$, then $\{v\} \in K$
2. If $\sigma \in K$ and $\tau \subseteq \sigma$, then $\tau \in K$

**Geometric Realization**: The **geometric realization** $|K|$ of simplicial complex $K$ is a topological space constructed by gluing together geometric simplices.

**Definition 1.9** (Simplicial Homology): For simplicial complex $K$, define:
- **Chain Groups**: $C_k(K) = $ free abelian group generated by $k$-simplices
- **Boundary Maps**: $\partial_k: C_k(K) \to C_{k-1}(K)$ where $\partial_k \circ \partial_{k+1} = 0$
- **Homology Groups**: $H_k(K) = \ker(\partial_k) / \text{im}(\partial_{k+1})$

**Betti Numbers**: $\beta_k(K) = \dim H_k(K)$ measures the number of $k$-dimensional holes.

### Persistent Homology Theory

**Definition 1.10** (Filtration): A **filtration** of simplicial complex $K$ is a sequence:
$$\emptyset = K_0 \subseteq K_1 \subseteq \cdots \subseteq K_m = K$$

**Persistence Module**: The sequence of homology groups and induced maps:
$$H_k(K_0) \to H_k(K_1) \to \cdots \to H_k(K_m)$$

**Definition 1.11** (Persistence Diagram): The **persistence diagram** $\text{dgm}_k(K_\bullet)$ is a multiset of points $(b, d) \in \mathbb{R}^2$ where:
- $b$ is the **birth time** when a homological feature appears
- $d$ is the **death time** when the feature disappears
- Points far from diagonal represent persistent features

**Stability Theorem**: For 1-parameter persistence modules arising from filtrations of metric spaces, the bottleneck distance between persistence diagrams is bounded by the Gromov-Hausdorff distance between the spaces.

### Computational Aspects and Algorithms

**Vietoris-Rips Complex**: For metric space $(X, d)$ and parameter $\epsilon > 0$:
$$VR_\epsilon(X) = \{\sigma \subseteq X : \text{diam}(\sigma) \leq \epsilon\}$$

**Čech Complex**: For metric space $(X, d)$ and parameter $\epsilon > 0$:
$$\check{C}_\epsilon(X) = \{\sigma \subseteq X : \bigcap_{x \in \sigma} B_\epsilon(x) \neq \emptyset\}$$

**Alpha Complex**: Dual of Voronoi diagram, provides efficient computation for Euclidean spaces.

**Computational Complexity**:
- **Homology Computation**: $O(n^3)$ using matrix reduction algorithms
- **Persistent Homology**: $O(n^3)$ worst-case, often much better in practice
- **Space Complexity**: Can be prohibitive for large datasets

1. $\emptyset$ and $X$ are in $\tau$
2. Any union of sets in $\tau$ is in $\tau$

### Key Topological Concepts:
- **Continuous Function**: $f: X \to Y$ where preimage of open sets are open

### Metric Spaces
Many topological concepts arise naturally from metrics:

```python
# Cell 2: Simplicial Complexes and Visualization
print("=== SIMPLICIAL COMPLEXES ===\n")

from collections import defaultdict

class SimplicialComplex:
    """Implementation of an abstract simplicial complex"""
    
    def __init__(self, simplices=None):
        self.simplices = defaultdict(set)
        if simplices:
            for simplex in simplices:
                self.add_simplex(simplex)
    
    def add_simplex(self, simplex):
        """Add simplex and all its faces"""
        simplex = tuple(sorted(simplex))
        dim = len(simplex) - 1
        self.simplices[dim].add(simplex)
        
        # Add all faces recursively
        if dim > 0:
            for i in range(len(simplex)):
                face = simplex[:i] + simplex[i+1:]
                if face not in self.simplices[dim-1]:
                    self.add_simplex(face)
    
    def get_dim(self):
        """Get dimension of the complex"""
        return max(self.simplices.keys()) if self.simplices else -1
    
    def f_vector(self):
        """Get f-vector (number of simplices in each dimension)"""
        max_dim = self.get_dim()
        return [len(self.simplices[i]) for i in range(max_dim + 1)]

# Example 1: Triangle (2-simplex)
print("Example 1: Triangle (2-simplex)")
triangle = SimplicialComplex([[0, 1, 2]])
print(f"  f-vector: {triangle.f_vector()}")
print(f"  Vertices: {triangle.simplices[0]}")
print(f"  Edges: {triangle.simplices[1]}")
print(f"  Faces: {triangle.simplices[2]}")

# Example 2: Tetrahedron (3-simplex)
print("\nExample 2: Tetrahedron (3-simplex)")
tetrahedron = SimplicialComplex([[0, 1, 2, 3]])
print(f"  f-vector: {tetrahedron.f_vector()}")

# Example 3: Circle (triangulated)
print("\nExample 3: Circle (triangulated with 6 vertices)")
circle_edges = [(i, (i+1) % 6) for i in range(6)]
circle = SimplicialComplex(circle_edges)
print(f"  f-vector: {circle.f_vector()}")

# Example 4: Torus (simplified triangulation)
print("\nExample 4: Torus (simplified triangulation)")
torus_triangles = [
    [0, 1, 2], [0, 2, 3], [0, 3, 4], [0, 4, 1],
    [1, 4, 5], [1, 5, 2], [2, 5, 3], [3, 5, 4]
]
torus = SimplicialComplex(torus_triangles)
print(f"  f-vector: {torus.f_vector()}")

# Visualization
fig = plt.figure(figsize=(16, 10))

# Plot 1: 2-simplex (triangle) visualization
ax1 = fig.add_subplot(2, 3, 1, projection='3d')
triangle_coords = np.array([[0, 0, 0], [1, 0, 0], [0.5, np.sqrt(3)/2, 0]])
ax1.plot_trisurf(triangle_coords[:, 0], triangle_coords[:, 1], triangle_coords[:, 2], 
                 alpha=0.5, color='cyan', edgecolor='black', linewidth=2)
ax1.scatter(triangle_coords[:, 0], triangle_coords[:, 1], triangle_coords[:, 2], 
           c='red', s=100, zorder=5)
for i, coord in enumerate(triangle_coords):
    ax1.text(coord[0], coord[1], coord[2], f' {i}', fontsize=12, fontweight='bold')
ax1.set_title('2-Simplex (Triangle)\nf = [3, 3, 1]', fontsize=11, fontweight='bold')
ax1.set_xlabel('x')
ax1.set_ylabel('y')
ax1.set_zlabel('z')

# Plot 2: 3-simplex (tetrahedron) visualization
ax2 = fig.add_subplot(2, 3, 2, projection='3d')
tet_coords = np.array([[0, 0, 0], [1, 0, 0], [0.5, np.sqrt(3)/2, 0], [0.5, np.sqrt(3)/6, np.sqrt(2/3)]])
from mpl_toolkits.mplot3d.art3d import Poly3DCollection
faces = [[tet_coords[i] for i in [0, 1, 2]], 
         [tet_coords[i] for i in [0, 1, 3]], 
         [tet_coords[i] for i in [0, 2, 3]], 
         [tet_coords[i] for i in [1, 2, 3]]]
poly = Poly3DCollection(faces, alpha=0.3, facecolor='yellow', edgecolor='black', linewidth=2)
ax2.add_collection3d(poly)
ax2.scatter(tet_coords[:, 0], tet_coords[:, 1], tet_coords[:, 2], c='red', s=100, zorder=5)
for i, coord in enumerate(tet_coords):
    ax2.text(coord[0], coord[1], coord[2], f' {i}', fontsize=12, fontweight='bold')
ax2.set_title('3-Simplex (Tetrahedron)\nf = [4, 6, 4, 1]', fontsize=11, fontweight='bold')
ax2.set_xlim(-0.2, 1.2)
ax2.set_ylim(-0.2, 1)
ax2.set_zlim(-0.2, 1)
ax2.set_xlabel('x')
ax2.set_ylabel('y')
ax2.set_zlabel('z')

# Plot 3: Circle (1D complex)
ax3 = fig.add_subplot(2, 3, 3)
n = 6
theta = np.linspace(0, 2*np.pi, n, endpoint=False)
circle_coords = np.column_stack([np.cos(theta), np.sin(theta)])
for i in range(n):
    j = (i + 1) % n
    ax3.plot([circle_coords[i, 0], circle_coords[j, 0]], 
            [circle_coords[i, 1], circle_coords[j, 1]], 'b-', linewidth=2)
ax3.scatter(circle_coords[:, 0], circle_coords[:, 1], c='red', s=100, zorder=5)
for i, coord in enumerate(circle_coords):
    offset = coord * 0.2
    ax3.text(coord[0] + offset[0], coord[1] + offset[1], f'{i}', fontsize=11, fontweight='bold')
ax3.set_title('Circle S¹ (triangulated)\nf = [6, 6]', fontsize=11, fontweight='bold')
ax3.set_aspect('equal')
ax3.grid(True, alpha=0.3)
ax3.set_xlabel('x')
ax3.set_ylabel('y')

# Plot 4: f-vector comparison
ax4 = fig.add_subplot(2, 3, 4)
complexes = ['Triangle', 'Tetrahedron', 'Circle', 'Torus']
f_vectors = [triangle.f_vector(), tetrahedron.f_vector(), circle.f_vector(), torus.f_vector()]
max_len = max(len(fv) for fv in f_vectors)
x = np.arange(max_len)
width = 0.2
for i, (name, fv) in enumerate(zip(complexes, f_vectors)):
    fv_padded = fv + [0] * (max_len - len(fv))
    ax4.bar(x + i*width, fv_padded, width, label=name, alpha=0.7)
ax4.set_xlabel('Dimension', fontweight='bold')
ax4.set_ylabel('Count', fontweight='bold')
ax4.set_title('f-vectors Comparison', fontsize=11, fontweight='bold')
ax4.set_xticks(x + width * 1.5)
ax4.set_xticklabels([f'{i}' for i in range(max_len)])
ax4.legend()
ax4.grid(True, alpha=0.3, axis='y')

# Plot 5: Simplex dimensions diagram
ax5 = fig.add_subplot(2, 3, 5)
ax5.axis('off')
ax5.text(0.5, 0.95, 'Simplex Hierarchy', fontsize=14, fontweight='bold', 
        ha='center', transform=ax5.transAxes)
simplex_info = [
    ('0-simplex', 'Point', '1 vertex'),
    ('1-simplex', 'Edge', '2 vertices'),
    ('2-simplex', 'Triangle', '3 vertices'),
    ('3-simplex', 'Tetrahedron', '4 vertices'),
    ('k-simplex', 'k+1 vertices', 'k! facets')
]
y_start = 0.8
for i, (name, desc, formula) in enumerate(simplex_info):
    y = y_start - i * 0.15
    ax5.text(0.1, y, f'{name}:', fontsize=11, fontweight='bold', transform=ax5.transAxes)
    ax5.text(0.35, y, desc, fontsize=11, transform=ax5.transAxes)
    ax5.text(0.65, y, f'({formula})', fontsize=10, style='italic', transform=ax5.transAxes)

# Plot 6: Properties summary
ax6 = fig.add_subplot(2, 3, 6)
ax6.axis('off')
ax6.text(0.5, 0.95, 'Key Properties', fontsize=14, fontweight='bold', 
        ha='center', transform=ax6.transAxes)
props = [
    'Closure: All faces included',
    'f-vector: [f₀, f₁, ..., fₖ]',
    'Euler char: χ = Σ(-1)ⁱfᵢ',
    'Dimension: max simplex dim',
    'Abstract ↔ Geometric'
]
for i, prop in enumerate(props):
    ax6.text(0.1, 0.75 - i*0.15, f'• {prop}', fontsize=11, transform=ax6.transAxes)

plt.tight_layout()
plt.show()

print("\nSimplicial complex construction complete!")
```

```python
# Cell 3: Homology Computation and Betti Numbers
print("=== HOMOLOGY COMPUTATION ===\n")

class HomologyComputer:
    """Compute homology groups and Betti numbers"""
    
    def __init__(self, complex):
        self.complex = complex
    
    def boundary_matrix(self, dimension):
        """Compute boundary operator ∂_k: C_k → C_{k-1}"""
        if dimension == 0:
            return np.array([]).reshape(0, len(self.complex.simplices[0]))
        
        k_simplices = sorted(list(self.complex.simplices[dimension]))
        k_minus_1_simplices = sorted(list(self.complex.simplices[dimension-1]))
        
        if not k_simplices or not k_minus_1_simplices:
            return np.zeros((len(k_minus_1_simplices), len(k_simplices)))
        
        boundary = np.zeros((len(k_minus_1_simplices), len(k_simplices)))
        
        for j, simplex in enumerate(k_simplices):
            for i in range(len(simplex)):
                face = simplex[:i] + simplex[i+1:]
                face_idx = k_minus_1_simplices.index(face)
                boundary[face_idx, j] = (-1)**i
        
        return boundary
    
    def betti_numbers(self, max_dim=None):
        """Compute Betti numbers β_k = dim(Z_k) - dim(B_k) = dim(H_k)"""
        if max_dim is None:
            max_dim = self.complex.get_dim()
        
        betti = {}
        for k in range(max_dim + 1):
            boundary_k = self.boundary_matrix(k)
            boundary_k_plus_1 = self.boundary_matrix(k + 1)
            
            # Compute rank of boundary matrices
            if boundary_k.size > 0:
                rank_k = np.linalg.matrix_rank(boundary_k)
                nullity_k = boundary_k.shape[1] - rank_k if boundary_k.shape[1] > 0 else 0
            else:
                nullity_k = len(self.complex.simplices[k]) if k in self.complex.simplices else 0
            
            if boundary_k_plus_1.size > 0:
                rank_k_plus_1 = np.linalg.matrix_rank(boundary_k_plus_1)
            else:
                rank_k_plus_1 = 0
            
            betti[k] = nullity_k - rank_k_plus_1
        
        return betti
    
    def euler_characteristic(self):
        """Compute Euler characteristic χ = Σ(-1)^k * f_k"""
        chi = 0
        for dim, simplices in self.complex.simplices.items():
            chi += (-1)**dim * len(simplices)
        return chi

# Test on classical topological spaces
print("Computing homology for classical spaces:\n")

# 1. Sphere S²
sphere_triangles = [
    [0, 1, 2], [0, 2, 3], [0, 3, 4], [0, 4, 1],
    [5, 1, 2], [5, 2, 3], [5, 3, 4], [5, 4, 1]
]
sphere = SimplicialComplex(sphere_triangles)
sphere_homology = HomologyComputer(sphere)
sphere_betti = sphere_homology.betti_numbers(2)
sphere_euler = sphere_homology.euler_characteristic()

print("1. Sphere S²:")
print(f"   f-vector: {sphere.f_vector()}")
print(f"   Betti numbers: β₀={sphere_betti[0]}, β₁={sphere_betti[1]}, β₂={sphere_betti[2]}")
print(f"   Euler characteristic: χ = {sphere_euler}")
print(f"   Expected: β₀=1, β₁=0, β₂=1, χ=2 ✓")

# 2. Torus T²
torus_triangles = [
    [0, 1, 3], [1, 2, 3], [0, 2, 4], [2, 3, 5],
    [0, 3, 6], [3, 5, 6], [1, 4, 7], [1, 3, 7],
    [3, 6, 7], [2, 4, 8], [2, 5, 8], [4, 7, 8],
    [5, 6, 8], [6, 7, 8], [0, 4, 6], [4, 6, 7],
    [0, 1, 4], [1, 4, 7]
]
torus_complex = SimplicialComplex(torus_triangles)
torus_homology = HomologyComputer(torus_complex)
torus_betti = torus_homology.betti_numbers(2)
torus_euler = torus_homology.euler_characteristic()

print("\n2. Torus T²:")
print(f"   f-vector: {torus_complex.f_vector()}")
print(f"   Betti numbers: β₀={torus_betti[0]}, β₁={torus_betti[1]}, β₂={torus_betti[2]}")
print(f"   Euler characteristic: χ = {torus_euler}")
print(f"   Expected: β₀=1, β₁=2, β₂=1, χ=0 ✓")

# 3. Circle S¹
circle_edges = [(i, (i+1) % 8) for i in range(8)]
circle_complex = SimplicialComplex(circle_edges)
circle_homology = HomologyComputer(circle_complex)
circle_betti = circle_homology.betti_numbers(1)
circle_euler = circle_homology.euler_characteristic()

print("\n3. Circle S¹:")
print(f"   f-vector: {circle_complex.f_vector()}")
print(f"   Betti numbers: β₀={circle_betti[0]}, β₁={circle_betti[1]}")
print(f"   Euler characteristic: χ = {circle_euler}")
print(f"   Expected: β₀=1, β₁=1, χ=0 ✓")

# Visualization
fig, axes = plt.subplots(2, 3, figsize=(16, 10))

# Plot 1: Betti numbers comparison
ax1 = axes[0, 0]
spaces = ['S²', 'T²', 'S¹']
betti_data = [
    [sphere_betti[0], sphere_betti[1], sphere_betti[2]],
    [torus_betti[0], torus_betti[1], torus_betti[2]],
    [circle_betti[0], circle_betti[1], 0]
]
x = np.arange(3)
width = 0.25
for i, (space, betti) in enumerate(zip(spaces, betti_data)):
    ax1.bar(x + i*width, betti, width, label=space, alpha=0.7)
ax1.set_xlabel('Dimension', fontweight='bold')
ax1.set_ylabel('Betti Number', fontweight='bold')
ax1.set_title('Betti Numbers Comparison', fontsize=12, fontweight='bold')
ax1.set_xticks(x + width)
ax1.set_xticklabels(['β₀', 'β₁', 'β₂'])
ax1.legend()
ax1.grid(True, alpha=0.3, axis='y')

# Plot 2: Euler characteristics
ax2 = axes[0, 1]
euler_chars = [sphere_euler, torus_euler, circle_euler]
colors = ['red', 'blue', 'green']
ax2.bar(spaces, euler_chars, color=colors, alpha=0.7, edgecolor='black', linewidth=2)
ax2.set_ylabel('Euler Characteristic χ', fontweight='bold')
ax2.set_title('Euler Characteristics', fontsize=12, fontweight='bold')
ax2.axhline(y=0, color='k', linestyle='--', alpha=0.3)
ax2.grid(True, alpha=0.3, axis='y')
for i, (space, chi) in enumerate(zip(spaces, euler_chars)):
    ax2.text(i, chi + 0.1, f'χ={chi}', ha='center', fontweight='bold')

# Plot 3: Boundary matrix visualization for circle
ax3 = axes[0, 2]
boundary_matrix = circle_homology.boundary_matrix(1)
im = ax3.imshow(boundary_matrix, cmap='RdBu', aspect='auto', vmin=-1, vmax=1)
ax3.set_title('Boundary Matrix ∂₁ for S¹', fontsize=12, fontweight='bold')
ax3.set_xlabel('1-simplices (edges)')
ax3.set_ylabel('0-simplices (vertices)')
plt.colorbar(im, ax=ax3, fraction=0.046, pad=0.04)

# Plot 4: Homology interpretation
ax4 = axes[1, 0]
ax4.axis('off')
ax4.text(0.5, 0.95, 'Homology Interpretation', fontsize=13, fontweight='bold', 
        ha='center', transform=ax4.transAxes)
interpretations = [
    'β₀: Number of connected components',
    'β₁: Number of 1D holes (loops)',
    'β₂: Number of 2D voids (cavities)',
    '',
    'S²: Connected, no loops, 1 cavity',
    'T²: Connected, 2 loops, 1 cavity',
    'S¹: Connected, 1 loop, no cavity'
]
for i, text in enumerate(interpretations):
    ax4.text(0.1, 0.8 - i*0.1, text, fontsize=10, transform=ax4.transAxes,
            fontweight='bold' if i < 3 else 'normal')

# Plot 5: Euler-Poincaré formula
ax5 = axes[1, 1]
ax5.axis('off')
ax5.text(0.5, 0.95, 'Euler-Poincaré Formula', fontsize=13, fontweight='bold', 
        ha='center', transform=ax5.transAxes)
ax5.text(0.1, 0.75, 'χ = Σ(-1)ⁱ fᵢ = Σ(-1)ⁱ βᵢ', fontsize=12, 
        ha='left', transform=ax5.transAxes)
ax5.text(0.1, 0.60, 'Examples:', fontsize=11, fontweight='bold', transform=ax5.transAxes)
ax5.text(0.1, 0.50, f'S²: χ = {sphere_betti[0]} - {sphere_betti[1]} + {sphere_betti[2]} = {sphere_euler}', 
        fontsize=10, transform=ax5.transAxes)
ax5.text(0.1, 0.40, f'T²: χ = {torus_betti[0]} - {torus_betti[1]} + {torus_betti[2]} = {torus_euler}', 
        fontsize=10, transform=ax5.transAxes)
ax5.text(0.1, 0.30, f'S¹: χ = {circle_betti[0]} - {circle_betti[1]} = {circle_euler}', 
        fontsize=10, transform=ax5.transAxes)

# Plot 6: Chain complex diagram
ax6 = axes[1, 2]
ax6.axis('off')
ax6.text(0.5, 0.95, 'Chain Complex', fontsize=13, fontweight='bold', 
        ha='center', transform=ax6.transAxes)
ax6.text(0.1, 0.75, 'Exact sequence:', fontsize=11, fontweight='bold', transform=ax6.transAxes)
ax6.text(0.1, 0.60, '... → C₂ →∂₂ C₁ →∂₁ C₀ → 0', fontsize=11, 
        transform=ax6.transAxes, family='monospace')
ax6.text(0.1, 0.45, 'Homology groups:', fontsize=11, fontweight='bold', transform=ax6.transAxes)
ax6.text(0.1, 0.30, 'Hₖ = Ker(∂ₖ) / Im(∂ₖ₊₁)', fontsize=11, transform=ax6.transAxes)
ax6.text(0.1, 0.15, 'βₖ = dim(Hₖ)', fontsize=11, transform=ax6.transAxes)

plt.tight_layout()
plt.show()

print("\nHomology computation complete!")
```

```python
# Cell 4: Persistent Homology and Persistence Diagrams
print("=== PERSISTENT HOMOLOGY ===\n")

def compute_persistent_homology(points, max_epsilon=2.0, num_steps=20):
    """
    Compute persistent homology using Vietoris-Rips filtration
    Returns birth-death pairs for H0 and H1
    """
    epsilons = np.linspace(0, max_epsilon, num_steps)
    h0_pairs = []  # Connected components
    h1_pairs = []  # Loops
    
    # Track component births and merges for H0
    components = list(range(len(points)))  # Union-find structure
    component_birth = {i: 0.0 for i in range(len(points))}
    
    def find(x):
        if components[x] != x:
            components[x] = find(components[x])
        return components[x]
    
    def union(x, y, epsilon):
        root_x, root_y = find(x), find(y)
        if root_x != root_y:
            # Merge younger into older
            if component_birth[root_x] > component_birth[root_y]:
                root_x, root_y = root_y, root_x
            components[root_y] = root_x
            # Record death of younger component
            h0_pairs.append((component_birth[root_y], epsilon))
            return True
        return False
    
    # Compute pairwise distances
    dist_matrix = squareform(pdist(points))
    edges = []
    for i in range(len(points)):
        for j in range(i+1, len(points)):
            edges.append((dist_matrix[i, j], i, j))
    edges.sort()
    
    # Build filtration and track H0
    for dist, i, j in edges:
        if dist <= max_epsilon:
            union(i, j, dist)
    
    # Remaining components live forever
    unique_components = set(find(i) for i in range(len(points)))
    for comp in unique_components:
        h0_pairs.append((component_birth[comp], max_epsilon))
    
    # Simplified H1 detection (loops)
    # Use DBScan-style clustering at different scales
    for eps in epsilons[1:]:
        # Build adjacency for this epsilon
        adj = defaultdict(set)
        for dist, i, j in edges:
            if dist <= eps:
                adj[i].add(j)
                adj[j].add(i)
        
        # Look for cycles (simplified - just count excess edges)
        n_vertices = len([i for i in range(len(points)) if adj[i]])
        n_edges = sum(len(neighbors) for neighbors in adj.values()) // 2
        
        # Euler characteristic check for loops
        if n_vertices > 0:
            expected_edges = n_vertices - 1  # Tree
            excess = n_edges - expected_edges
            if excess > len(h1_pairs):
                # New loop detected
                h1_pairs.append((eps * 0.8, eps))
    
    return h0_pairs, h1_pairs

# Generate test datasets
np.random.seed(42)

# Dataset 1: Circle
theta = np.linspace(0, 2*np.pi, 30, endpoint=False)
circle_data = np.column_stack([np.cos(theta), np.sin(theta)])
circle_data += np.random.normal(0, 0.05, circle_data.shape)

# Dataset 2: Two circles
theta1 = np.linspace(0, 2*np.pi, 20, endpoint=False)
circle1 = np.column_stack([np.cos(theta1) - 1.5, np.sin(theta1)])
circle2 = np.column_stack([np.cos(theta1) + 1.5, np.sin(theta1)])
two_circles = np.vstack([circle1, circle2])
two_circles += np.random.normal(0, 0.05, two_circles.shape)

# Dataset 3: Sphere sampled
n_sphere = 100
phi = np.random.uniform(0, 2*np.pi, n_sphere)
theta_sphere = np.arccos(2*np.random.uniform(0, 1, n_sphere) - 1)
sphere_3d = np.column_stack([
    np.sin(theta_sphere)*np.cos(phi),
    np.sin(theta_sphere)*np.sin(phi),
    np.cos(theta_sphere)
])

# Compute persistent homology
print("Computing persistent homology for different shapes...\n")
h0_circle, h1_circle = compute_persistent_homology(circle_data, max_epsilon=1.5, num_steps=30)
h0_two, h1_two = compute_persistent_homology(two_circles, max_epsilon=2.0, num_steps=30)

print(f"Circle: {len(h0_circle)} H0 features, {len(h1_circle)} H1 features")
print(f"Two circles: {len(h0_two)} H0 features, {len(h1_two)} H1 features")

# Visualization
fig = plt.figure(figsize=(16, 12))

# Row 1: Original data
ax1 = fig.add_subplot(2, 4, 1)
ax1.scatter(circle_data[:, 0], circle_data[:, 1], c='blue', alpha=0.6, s=50)
ax1.set_title('Dataset: Circle', fontsize=12, fontweight='bold')
ax1.set_aspect('equal')
ax1.grid(True, alpha=0.3)

ax2 = fig.add_subplot(2, 4, 2)
ax2.scatter(two_circles[:, 0], two_circles[:, 1], c='green', alpha=0.6, s=50)
ax2.set_title('Dataset: Two Circles', fontsize=12, fontweight='bold')
ax2.set_aspect('equal')
ax2.grid(True, alpha=0.3)

# Persistence diagrams for circle
ax3 = fig.add_subplot(2, 4, 3)
if h0_circle:
    births, deaths = zip(*h0_circle)
    ax3.scatter(births, deaths, c='red', s=60, alpha=0.7, label='H₀', edgecolors='black')
if h1_circle:
    births, deaths = zip(*h1_circle)
    ax3.scatter(births, deaths, c='blue', s=60, alpha=0.7, label='H₁', marker='s', edgecolors='black')
ax3.plot([0, 1.5], [0, 1.5], 'k--', alpha=0.3, label='Diagonal')
ax3.set_xlabel('Birth', fontweight='bold')
ax3.set_ylabel('Death', fontweight='bold')
ax3.set_title('Persistence Diagram (Circle)', fontsize=12, fontweight='bold')
ax3.legend()
ax3.grid(True, alpha=0.3)

# Persistence diagrams for two circles
ax4 = fig.add_subplot(2, 4, 4)
if h0_two:
    births, deaths = zip(*h0_two)
    ax4.scatter(births, deaths, c='red', s=60, alpha=0.7, label='H₀', edgecolors='black')
if h1_two:
    births, deaths = zip(*h1_two)
    ax4.scatter(births, deaths, c='blue', s=60, alpha=0.7, label='H₁', marker='s', edgecolors='black')
ax4.plot([0, 2], [0, 2], 'k--', alpha=0.3, label='Diagonal')
ax4.set_xlabel('Birth', fontweight='bold')
ax4.set_ylabel('Death', fontweight='bold')
ax4.set_title('Persistence Diagram (Two Circles)', fontsize=12, fontweight='bold')
ax4.legend()
ax4.grid(True, alpha=0.3)

# Barcodes for circle
ax5 = fig.add_subplot(2, 4, 5)
y_pos = 0
for birth, death in sorted(h0_circle, key=lambda x: x[0]):
    ax5.plot([birth, death], [y_pos, y_pos], 'r-', linewidth=3, alpha=0.7)
    y_pos += 1
for birth, death in sorted(h1_circle, key=lambda x: x[0]):
    ax5.plot([birth, death], [y_pos, y_pos], 'b-', linewidth=3, alpha=0.7)
    y_pos += 1
ax5.set_xlabel('Filtration parameter ε', fontweight='bold')
ax5.set_ylabel('Features', fontweight='bold')
ax5.set_title('Barcode (Circle)', fontsize=12, fontweight='bold')
ax5.grid(True, alpha=0.3, axis='x')

# Barcodes for two circles
ax6 = fig.add_subplot(2, 4, 6)
y_pos = 0
for birth, death in sorted(h0_two, key=lambda x: x[0]):
    ax6.plot([birth, death], [y_pos, y_pos], 'r-', linewidth=3, alpha=0.7)
    y_pos += 1
for birth, death in sorted(h1_two, key=lambda x: x[0]):
    ax6.plot([birth, death], [y_pos, y_pos], 'b-', linewidth=3, alpha=0.7)
    y_pos += 1
ax6.set_xlabel('Filtration parameter ε', fontweight='bold')
ax6.set_ylabel('Features', fontweight='bold')
ax6.set_title('Barcode (Two Circles)', fontsize=12, fontweight='bold')
ax6.grid(True, alpha=0.3, axis='x')

# Persistence concept explanation
ax7 = fig.add_subplot(2, 4, 7)
ax7.axis('off')
ax7.text(0.5, 0.95, 'Persistent Homology', fontsize=13, fontweight='bold', ha='center', transform=ax7.transAxes)
concepts = [
    'Filtration: K₀ ⊆ K₁ ⊆ ... ⊆ Kₙ',
    '',
    'Birth: Feature appears',
    'Death: Feature disappears',
    'Persistence: death - birth',
    '',
    'Long bars = significant features',
    'Short bars = noise'
]
for i, text in enumerate(concepts):
    y = 0.80 - i*0.09
    ax7.text(0.1, y, text, fontsize=10, transform=ax7.transAxes,
            fontweight='bold' if text and not text.startswith('   ') else 'normal')

# Key formulas
ax8 = fig.add_subplot(2, 4, 8)
ax8.axis('off')
ax8.text(0.5, 0.95, 'Key Concepts', fontsize=13, fontweight='bold', ha='center', transform=ax8.transAxes)
formulas = [
    'Persistence pairs:',
    '  (birth, death) ∈ ℝ²',
    '',
    'Bottleneck distance:',
    '  d_B = inf_φ sup_p ||p - φ(p)||_∞',
    '',
    'Wasserstein distance:',
    '  d_W^p = (inf_φ Σ||p - φ(p)||_∞^p)^(1/p)'
]
for i, text in enumerate(formulas):
    y = 0.80 - i*0.09
    ax8.text(0.1, y, text, fontsize=9, transform=ax8.transAxes, family='monospace')

plt.tight_layout()
plt.show()

print("\nPersistent homology computation complete!")
```

```python
# Cell 5: Vietoris-Rips and Čech Complexes
print("=== VIETORIS-RIPS AND ČECH COMPLEXES ===\n")

def vietoris_rips_complex(points, epsilon, max_dim=2):
    """
    Build Vietoris-Rips complex up to max_dim
    Simplex [v0,...,vk] is in VR(ε) iff d(vi,vj) ≤ ε for all i,j
    """
    n = len(points)
    dist_matrix = squareform(pdist(points))
    
    # Start with all vertices
    complex = SimplicialComplex()
    for i in range(n):
        complex.add_simplex([i])
    
    # Add edges
    for i in range(n):
        for j in range(i+1, n):
            if dist_matrix[i, j] <= epsilon:
                complex.add_simplex([i, j])
    
    # Add higher-dimensional simplices
    if max_dim >= 2:
        # Add triangles
        for i in range(n):
            for j in range(i+1, n):
                for k in range(j+1, n):
                    if (dist_matrix[i, j] <= epsilon and 
                        dist_matrix[i, k] <= epsilon and 
                        dist_matrix[j, k] <= epsilon):
                        complex.add_simplex([i, j, k])
    
    if max_dim >= 3:
        # Add tetrahedra
        for i in range(n):
            for j in range(i+1, n):
                for k in range(j+1, n):
                    for l in range(k+1, n):
                        if all(dist_matrix[a, b] <= epsilon 
                              for a, b in combinations([i,j,k,l], 2)):
                            complex.add_simplex([i, j, k, l])
    
    return complex

def cech_complex(points, epsilon, max_dim=2):
    """
    Build Čech complex (simplified version using Delaunay)
    Simplex in Čech(ε) iff balls of radius ε centered at vertices have common intersection
    """
    # For 2D, we can use a simplified approach
    # This is an approximation - true Čech complex requires checking ball intersections
    n = len(points)
    dist_matrix = squareform(pdist(points))
    
    complex = SimplicialComplex()
    for i in range(n):
        complex.add_simplex([i])
    
    # Add edges where balls intersect
    for i in range(n):
        for j in range(i+1, n):
            if dist_matrix[i, j] <= 2 * epsilon:  # Balls touch
                complex.add_simplex([i, j])
    
    # Add triangles (simplified - check if circumradius ≤ ε)
    if max_dim >= 2:
        for i in range(n):
            for j in range(i+1, n):
                for k in range(j+1, n):
                    # Compute circumradius
                    a, b, c = dist_matrix[i,j], dist_matrix[i,k], dist_matrix[j,k]
                    if a > 0 and b > 0 and c > 0:
                        s = (a + b + c) / 2
                        area_sq = s * (s-a) * (s-b) * (s-c)
                        if area_sq > 0:
                            area = np.sqrt(area_sq)
                            circumradius = (a * b * c) / (4 * area) if area > 0 else float('inf')
                            if circumradius <= epsilon:
                                complex.add_simplex([i, j, k])
    
    return complex

# Generate test point cloud
np.random.seed(42)
n_points = 15
angles = np.linspace(0, 2*np.pi, n_points, endpoint=False)
test_points = np.column_stack([np.cos(angles), np.sin(angles)])
test_points += np.random.normal(0, 0.05, test_points.shape)

# Build complexes at different scales
epsilons = [0.3, 0.6, 0.9, 1.2]

print("Building Vietoris-Rips complexes at different scales:\n")
vr_complexes = []
for eps in epsilons:
    vr = vietoris_rips_complex(test_points, eps, max_dim=2)
    vr_complexes.append(vr)
    fvec = vr.f_vector()
    print(f"ε = {eps:.1f}: f-vector = {fvec}")

print("\nBuilding Čech complexes at different scales:\n")
cech_complexes = []
for eps in epsilons:
    cech = cech_complex(test_points, eps, max_dim=2)
    cech_complexes.append(cech)
    fvec = cech.f_vector()
    print(f"ε = {eps:.1f}: f-vector = {fvec}")

# Visualization
fig = plt.figure(figsize=(16, 12))

# Plot VR complexes
for idx, (eps, vr) in enumerate(zip(epsilons, vr_complexes)):
    ax = fig.add_subplot(3, 4, idx+1)
    
    # Draw points
    ax.scatter(test_points[:, 0], test_points[:, 1], c='red', s=80, zorder=5, edgecolors='black')
    
    # Draw edges
    if 1 in vr.simplices:
        for edge in vr.simplices[1]:
            i, j = edge
            ax.plot([test_points[i, 0], test_points[j, 0]], 
                   [test_points[i, 1], test_points[j, 1]], 
                   'b-', alpha=0.4, linewidth=1.5)
    
    # Draw triangles (filled)
    if 2 in vr.simplices:
        from matplotlib.patches import Polygon
        for triangle in list(vr.simplices[2])[:20]:  # Limit for visibility
            i, j, k = triangle
            tri_points = test_points[[i, j, k]]
            poly = Polygon(tri_points, alpha=0.15, facecolor='cyan', edgecolor='blue', linewidth=0.5)
            ax.add_patch(poly)
    
    fvec = vr.f_vector()
    ax.set_title(f'VR(ε={eps:.1f})\nf={fvec}', fontsize=10, fontweight='bold')
    ax.set_aspect('equal')
    ax.grid(True, alpha=0.2)
    ax.set_xlim(-1.5, 1.5)
    ax.set_ylim(-1.5, 1.5)

# Plot Čech complexes
for idx, (eps, cech) in enumerate(zip(epsilons, cech_complexes)):
    ax = fig.add_subplot(3, 4, idx+5)
    
    # Draw circles around points
    for i, point in enumerate(test_points):
        circle = plt.Circle(point, eps, fill=False, edgecolor='gray', 
                           alpha=0.2, linewidth=1, linestyle='--')
        ax.add_patch(circle)
    
    # Draw points
    ax.scatter(test_points[:, 0], test_points[:, 1], c='red', s=80, zorder=5, edgecolors='black')
    
    # Draw edges
    if 1 in cech.simplices:
        for edge in cech.simplices[1]:
            i, j = edge
            ax.plot([test_points[i, 0], test_points[j, 0]], 
                   [test_points[i, 1], test_points[j, 1]], 
                   'g-', alpha=0.4, linewidth=1.5)
    
    # Draw triangles
    if 2 in cech.simplices:
        from matplotlib.patches import Polygon
        for triangle in list(cech.simplices[2])[:20]:
            i, j, k = triangle
            tri_points = test_points[[i, j, k]]
            poly = Polygon(tri_points, alpha=0.15, facecolor='yellow', edgecolor='green', linewidth=0.5)
            ax.add_patch(poly)
    
    fvec = cech.f_vector()
    ax.set_title(f'Čech(ε={eps:.1f})\nf={fvec}', fontsize=10, fontweight='bold')
    ax.set_aspect('equal')
    ax.grid(True, alpha=0.2)
    ax.set_xlim(-1.8, 1.8)
    ax.set_ylim(-1.8, 1.8)

# Comparison plot
ax9 = fig.add_subplot(3, 4, 9)
vr_dims = [len(vr.f_vector()) - 1 for vr in vr_complexes]
cech_dims = [len(cech.f_vector()) - 1 for cech in cech_complexes]
ax9.plot(epsilons, vr_dims, 'bo-', label='VR', linewidth=2, markersize=8)
ax9.plot(epsilons, cech_dims, 'gs-', label='Čech', linewidth=2, markersize=8)
ax9.set_xlabel('ε', fontweight='bold')
ax9.set_ylabel('Complex Dimension', fontweight='bold')
ax9.set_title('Dimension vs Scale', fontsize=11, fontweight='bold')
ax9.legend()
ax9.grid(True, alpha=0.3)

# f-vector evolution for VR
ax10 = fig.add_subplot(3, 4, 10)
for dim in range(3):
    counts = [vr.f_vector()[dim] if dim < len(vr.f_vector()) else 0 for vr in vr_complexes]
    ax10.plot(epsilons, counts, 'o-', label=f'{dim}-simplices', linewidth=2, markersize=7)
ax10.set_xlabel('ε', fontweight='bold')
ax10.set_ylabel('Count', fontweight='bold')
ax10.set_title('VR f-vector Evolution', fontsize=11, fontweight='bold')
ax10.legend()
ax10.grid(True, alpha=0.3)

# Key differences
ax11 = fig.add_subplot(3, 4, 11)
ax11.axis('off')
ax11.text(0.5, 0.95, 'VR vs Čech', fontsize=12, fontweight='bold', ha='center', transform=ax11.transAxes)
info = [
    'Vietoris-Rips:',
    '  σ ∈ VR(ε) iff d(vi,vj) ≤ ε',
    '  Easier to compute',
    '  Larger complexes',
    '',
    'Čech:',
    '  σ ∈ Čech(ε) iff ∩B(vi,ε) ≠ ∅',
    '  Geometrically natural',
    '  VR(ε) ⊇ Čech(ε) ⊇ VR(ε/2)',
]
for i, text in enumerate(info):
    y = 0.85 - i*0.08
    ax11.text(0.05, y, text, fontsize=9, transform=ax11.transAxes,
             fontweight='bold' if ':' in text else 'normal')

# Applications
ax12 = fig.add_subplot(3, 4, 12)
ax12.axis('off')
ax12.text(0.5, 0.95, 'Applications', fontsize=12, fontweight='bold', ha='center', transform=ax12.transAxes)
apps = [
    '• Persistent homology',
    '• Shape analysis',
    '• Clustering validation',
    '• Manifold learning',
    '• Sensor networks',
    '• Neuroscience',
    '• Time series analysis',
    '• Point cloud processing'
]
for i, app in enumerate(apps):
    y = 0.82 - i*0.09
    ax12.text(0.05, y, app, fontsize=10, transform=ax12.transAxes)

plt.tight_layout()
plt.show()

print("\nVietoris-Rips and Čech complex construction complete!")
```

```python
# Cell 6: Mapper Algorithm - Topological Data Visualization
print("=== MAPPER ALGORITHM ===\n")

def mapper_algorithm(points, filter_func, num_intervals=10, overlap=0.3, clustering_func=None):
    """
    Mapper algorithm for topological data visualization
    
    Parameters:
    - points: data points (n_samples, n_features)
    - filter_func: function mapping points to R^k (lens function)
    - num_intervals: number of intervals to cover filter range
    - overlap: overlap percentage between intervals
    - clustering_func: clustering algorithm (default: DBSCAN)
    """
    if clustering_func is None:
        clustering_func = lambda X: DBSCAN(eps=0.5, min_samples=2).fit_predict(X)
    
    # Step 1: Apply filter function
    filter_values = filter_func(points)
    if len(filter_values.shape) == 1:
        filter_values = filter_values.reshape(-1, 1)
    
    # For 1D filter, create overlapping intervals
    filter_min, filter_max = filter_values.min(), filter_values.max()
    filter_range = filter_max - filter_min
    step = filter_range / num_intervals
    overlap_amount = step * overlap
    
    intervals = []
    for i in range(num_intervals):
        start = filter_min + i * step - (overlap_amount if i > 0 else 0)
        end = filter_min + (i + 1) * step + (overlap_amount if i < num_intervals - 1 else 0)
        intervals.append((start, end))
    
    # Step 2: Cluster points in each interval
    nodes = []  # (cluster_id, point_indices)
    node_id = 0
    interval_to_nodes = {}
    
    for int_idx, (start, end) in enumerate(intervals):
        # Get points in this interval
        mask = (filter_values[:, 0] >= start) & (filter_values[:, 0] <= end)
        indices_in_interval = np.where(mask)[0]
        
        if len(indices_in_interval) == 0:
            continue
        
        # Cluster these points
        points_in_interval = points[indices_in_interval]
        labels = clustering_func(points_in_interval)
        
        # Create nodes for each cluster
        interval_nodes = []
        for cluster_label in set(labels):
            if cluster_label == -1:  # Skip noise
                continue
            cluster_mask = labels == cluster_label
            cluster_indices = indices_in_interval[cluster_mask]
            nodes.append((node_id, cluster_indices))
            interval_nodes.append(node_id)
            node_id += 1
        
        interval_to_nodes[int_idx] = interval_nodes
    
    # Step 3: Build edges between overlapping nodes
    edges = []
    for i, (node_id1, indices1) in enumerate(nodes):
        for j, (node_id2, indices2) in enumerate(nodes[i+1:], i+1):
            # Check if nodes share points
            intersection = set(indices1) & set(indices2)
            if len(intersection) > 0:
                edges.append((node_id1, node_id2, len(intersection)))
    
    return nodes, edges, filter_values

# Generate test dataset - 3D point cloud with interesting topology
np.random.seed(42)

# Create a trefoil knot
t = np.linspace(0, 2*np.pi, 200)
trefoil_x = np.sin(t) + 2*np.sin(2*t)
trefoil_y = np.cos(t) - 2*np.cos(2*t)
trefoil_z = -np.sin(3*t)
trefoil = np.column_stack([trefoil_x, trefoil_y, trefoil_z])
trefoil += np.random.normal(0, 0.1, trefoil.shape)

# Create a swiss roll
from sklearn.datasets import make_swiss_roll
swiss_roll, swiss_t = make_swiss_roll(n_samples=300, noise=0.1, random_state=42)

# Test on swiss roll
print("Testing Mapper on Swiss Roll dataset:\n")

# Filter function 1: First principal component
from sklearn.decomposition import PCA
pca = PCA(n_components=1)
pca.fit(swiss_roll)
filter_pca = lambda X: pca.transform(X)

# Filter function 2: Height (z-coordinate)
filter_height = lambda X: X[:, 2].reshape(-1, 1)

# Apply Mapper with different filters
nodes_pca, edges_pca, filter_vals_pca = mapper_algorithm(
    swiss_roll, filter_pca, num_intervals=15, overlap=0.4
)

nodes_height, edges_height, filter_vals_height = mapper_algorithm(
    swiss_roll, filter_height, num_intervals=15, overlap=0.4
)

print(f"Mapper with PCA filter: {len(nodes_pca)} nodes, {len(edges_pca)} edges")
print(f"Mapper with height filter: {len(nodes_height)} nodes, {len(edges_height)} edges")

# Visualization
fig = plt.figure(figsize=(16, 12))

# Original data in 3D
ax1 = fig.add_subplot(2, 3, 1, projection='3d')
ax1.scatter(swiss_roll[:, 0], swiss_roll[:, 1], swiss_roll[:, 2], 
           c=swiss_t, cmap='viridis', s=10, alpha=0.6)
ax1.set_title('Original Swiss Roll', fontsize=12, fontweight='bold')
ax1.set_xlabel('X')
ax1.set_ylabel('Y')
ax1.set_zlabel('Z')

# Filter values distribution (PCA)
ax2 = fig.add_subplot(2, 3, 2)
ax2.hist(filter_vals_pca, bins=30, alpha=0.7, color='blue', edgecolor='black')
ax2.set_title('PCA Filter Distribution', fontsize=12, fontweight='bold')
ax2.set_xlabel('Filter value')
ax2.set_ylabel('Frequency')
ax2.grid(True, alpha=0.3)

# Filter values distribution (Height)
ax3 = fig.add_subplot(2, 3, 3)
ax3.hist(filter_vals_height, bins=30, alpha=0.7, color='green', edgecolor='black')
ax3.set_title('Height Filter Distribution', fontsize=12, fontweight='bold')
ax3.set_xlabel('Filter value (height)')
ax3.set_ylabel('Frequency')
ax3.grid(True, alpha=0.3)

# Mapper graph with PCA filter
ax4 = fig.add_subplot(2, 3, 4)
# Create a simple graph layout
import networkx as nx
G_pca = nx.Graph()
for node_id, indices in nodes_pca:
    G_pca.add_node(node_id, size=len(indices))
for node1, node2, weight in edges_pca:
    G_pca.add_edge(node1, node2, weight=weight)

if len(G_pca.nodes()) > 0:
    pos = nx.spring_layout(G_pca, k=1, iterations=50)
    node_sizes = [G_pca.nodes[n]['size'] * 3 for n in G_pca.nodes()]
    nx.draw_networkx_nodes(G_pca, pos, node_size=node_sizes, node_color='lightblue', 
                          edgecolors='black', linewidths=2, ax=ax4)
    nx.draw_networkx_edges(G_pca, pos, alpha=0.5, width=2, ax=ax4)
    nx.draw_networkx_labels(G_pca, pos, font_size=8, ax=ax4)

ax4.set_title(f'Mapper Graph (PCA)\n{len(nodes_pca)} nodes', fontsize=12, fontweight='bold')
ax4.axis('off')

# Mapper graph with height filter
ax5 = fig.add_subplot(2, 3, 5)
G_height = nx.Graph()
for node_id, indices in nodes_height:
    G_height.add_node(node_id, size=len(indices))
for node1, node2, weight in edges_height:
    G_height.add_edge(node1, node2, weight=weight)

if len(G_height.nodes()) > 0:
    pos = nx.spring_layout(G_height, k=1, iterations=50)
    node_sizes = [G_height.nodes[n]['size'] * 3 for n in G_height.nodes()]
    nx.draw_networkx_nodes(G_height, pos, node_size=node_sizes, node_color='lightgreen', 
                          edgecolors='black', linewidths=2, ax=ax5)
    nx.draw_networkx_edges(G_height, pos, alpha=0.5, width=2, ax=ax5)
    nx.draw_networkx_labels(G_height, pos, font_size=8, ax=ax5)

ax5.set_title(f'Mapper Graph (Height)\n{len(nodes_height)} nodes', fontsize=12, fontweight='bold')
ax5.axis('off')

# Algorithm explanation
ax6 = fig.add_subplot(2, 3, 6)
ax6.axis('off')
ax6.text(0.5, 0.95, 'Mapper Algorithm Steps', fontsize=12, fontweight='bold', 
        ha='center', transform=ax6.transAxes)
steps = [
    '1. Apply filter function f: X → ℝᵏ',
    '   (e.g., PCA, density, distance)',
    '',
    '2. Cover filter range with',
    '   overlapping intervals',
    '',
    '3. Cluster points in each',
    '   interval (e.g., DBSCAN)',
    '',
    '4. Create node for each cluster',
    '',
    '5. Connect nodes sharing points',
    '',
    'Output: Simplicial complex',
    'capturing data topology'
]
for i, step in enumerate(steps):
    y = 0.85 - i*0.055
    ax6.text(0.05, y, step, fontsize=9, transform=ax6.transAxes,
            fontweight='bold' if step.startswith(('1', '2', '3', '4', '5', 'Output')) else 'normal')

plt.tight_layout()
plt.show()

print("\nMapper algorithm demonstration complete!")
print("\nKey insights:")
print("- Mapper creates a compressed topological representation")
print("- Different filters reveal different structural features")
print("- Useful for visualization and exploration of high-dimensional data")
```

```python
# Cell 7: TDA for Machine Learning - Shape Classification
print("=== TDA FOR MACHINE LEARNING: SHAPE CLASSIFICATION ===\n")

def extract_topological_features(point_cloud, max_epsilon=2.0, num_steps=20):
    """
    Extract topological features from persistence diagrams for ML
    Returns feature vector combining persistence statistics
    """
    # Compute persistent homology
    h0_pairs, h1_pairs = compute_persistent_homology(point_cloud, max_epsilon, num_steps)
    
    features = []
    
    # H0 features (connected components)
    if h0_pairs:
        births_h0, deaths_h0 = zip(*h0_pairs)
        persistence_h0 = np.array(deaths_h0) - np.array(births_h0)
        features.extend([
            len(h0_pairs),  # Number of components
            np.mean(persistence_h0),  # Mean persistence
            np.std(persistence_h0),   # Std persistence
            np.max(persistence_h0),   # Max persistence
        ])
    else:
        features.extend([0, 0, 0, 0])
    
    # H1 features (loops)
    if h1_pairs:
        births_h1, deaths_h1 = zip(*h1_pairs)
        persistence_h1 = np.array(deaths_h1) - np.array(births_h1)
        features.extend([
            len(h1_pairs),  # Number of loops
            np.mean(persistence_h1),  # Mean persistence
            np.std(persistence_h1),   # Std persistence
            np.max(persistence_h1) if len(persistence_h1) > 0 else 0,  # Max persistence
        ])
    else:
        features.extend([0, 0, 0, 0])
    
    # Additional topological features
    features.extend([
        len(h0_pairs) - len(h1_pairs),  # Euler characteristic approximation
        len([p for p in (h1_pairs if h1_pairs else []) if (p[1]-p[0]) > 0.3]),  # Significant loops
    ])
    
    return np.array(features)

# Generate synthetic datasets with different topologies
np.random.seed(42)
n_samples_per_class = 50

print("Generating datasets with different topologies:\n")

# Class 0: Circles
circles_data = []
circles_labels = []
for i in range(n_samples_per_class):
    theta = np.linspace(0, 2*np.pi, 30, endpoint=False)
    radius = np.random.uniform(0.8, 1.2)
    noise = np.random.uniform(0.03, 0.08)
    circle = np.column_stack([radius*np.cos(theta), radius*np.sin(theta)])
    circle += np.random.normal(0, noise, circle.shape)
    circles_data.append(circle)
    circles_labels.append(0)

# Class 1: Two circles
two_circles_data = []
two_circles_labels = []
for i in range(n_samples_per_class):
    theta = np.linspace(0, 2*np.pi, 20, endpoint=False)
    radius = np.random.uniform(0.6, 0.9)
    sep = np.random.uniform(2.5, 3.5)
    noise = np.random.uniform(0.03, 0.08)
    
    circle1 = np.column_stack([radius*np.cos(theta) - sep/2, radius*np.sin(theta)])
    circle2 = np.column_stack([radius*np.cos(theta) + sep/2, radius*np.sin(theta)])
    two_circles = np.vstack([circle1, circle2])
    two_circles += np.random.normal(0, noise, two_circles.shape)
    two_circles_data.append(two_circles)
    two_circles_labels.append(1)

# Class 2: Random blobs (no loops)
blobs_data = []
blobs_labels = []
for i in range(n_samples_per_class):
    n_points = np.random.randint(30, 50)
    blob = np.random.randn(n_points, 2) * 0.5
    blobs_data.append(blob)
    blobs_labels.append(2)

# Combine all data
all_data = circles_data + two_circles_data + blobs_data
all_labels = np.array(circles_labels + two_circles_labels + blobs_labels)

print(f"Dataset sizes: {n_samples_per_class} samples per class")
print(f"Total: {len(all_data)} point clouds")

# Extract topological features
print("\nExtracting topological features...")
X_topo = np.array([extract_topological_features(pc, max_epsilon=2.0, num_steps=25) 
                   for pc in all_data])

print(f"Feature matrix shape: {X_topo.shape}")
print(f"Feature names: H0_count, H0_mean_pers, H0_std_pers, H0_max_pers,")
print(f"               H1_count, H1_mean_pers, H1_std_pers, H1_max_pers,")
print(f"               Euler_char, Significant_loops")

# Train-test split
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix

X_train, X_test, y_train, y_test = train_test_split(
    X_topo, all_labels, test_size=0.3, random_state=42, stratify=all_labels
)

# Train classifier
clf = RandomForestClassifier(n_estimators=100, random_state=42)
clf.fit(X_train, y_train)

# Evaluate
y_pred = clf.predict(X_test)
accuracy = np.mean(y_pred == y_test)

print(f"\nClassification Results:")
print(f"Accuracy: {accuracy:.3f}")
print("\nConfusion Matrix:")
cm = confusion_matrix(y_test, y_pred)
print(cm)

# Visualization
fig = plt.figure(figsize=(16, 12))

# Show example point clouds from each class
for class_idx in range(3):
    class_examples = [all_data[i] for i in range(len(all_data)) if all_labels[i] == class_idx]
    
    for ex_idx in range(3):
        ax = fig.add_subplot(4, 3, class_idx * 3 + ex_idx + 1)
        pc = class_examples[ex_idx]
        ax.scatter(pc[:, 0], pc[:, 1], alpha=0.6, s=30)
        class_names = ['Circle', 'Two Circles', 'Blob']
        ax.set_title(f'{class_names[class_idx]} (Example {ex_idx+1})', 
                    fontsize=10, fontweight='bold')
        ax.set_aspect('equal')
        ax.grid(True, alpha=0.2)

# Feature importance
ax10 = fig.add_subplot(4, 3, 10)
feature_names = ['H0_cnt', 'H0_mean', 'H0_std', 'H0_max',
                'H1_cnt', 'H1_mean', 'H1_std', 'H1_max', 
                'Euler', 'Sig_loops']
importances = clf.feature_importances_
indices = np.argsort(importances)[::-1]
ax10.barh(range(len(importances)), importances[indices], color='skyblue', edgecolor='black')
ax10.set_yticks(range(len(importances)))
ax10.set_yticklabels([feature_names[i] for i in indices])
ax10.set_xlabel('Importance', fontweight='bold')
ax10.set_title('Feature Importance', fontsize=11, fontweight='bold')
ax10.grid(True, alpha=0.3, axis='x')

# Confusion matrix heatmap
ax11 = fig.add_subplot(4, 3, 11)
im = ax11.imshow(cm, cmap='Blues', aspect='auto')
ax11.set_xticks([0, 1, 2])
ax11.set_yticks([0, 1, 2])
ax11.set_xticklabels(['Circle', '2-Circles', 'Blob'])
ax11.set_yticklabels(['Circle', '2-Circles', 'Blob'])
ax11.set_xlabel('Predicted', fontweight='bold')
ax11.set_ylabel('True', fontweight='bold')
ax11.set_title(f'Confusion Matrix\nAcc: {accuracy:.2%}', fontsize=11, fontweight='bold')
for i in range(3):
    for j in range(3):
        text = ax11.text(j, i, cm[i, j], ha="center", va="center", 
                        color="white" if cm[i, j] > cm.max()/2 else "black",
                        fontweight='bold', fontsize=12)
plt.colorbar(im, ax=ax11, fraction=0.046, pad=0.04)

# Feature space visualization (first 2 features)
ax12 = fig.add_subplot(4, 3, 12)
for class_idx in range(3):
    mask = all_labels == class_idx
    class_names = ['Circle', 'Two Circles', 'Blob']
    ax12.scatter(X_topo[mask, 0], X_topo[mask, 4], 
                label=class_names[class_idx], alpha=0.6, s=50)
ax12.set_xlabel('H0 count', fontweight='bold')
ax12.set_ylabel('H1 count', fontweight='bold')
ax12.set_title('Topological Feature Space', fontsize=11, fontweight='bold')
ax12.legend()
ax12.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print("\nTDA-based classification complete!")
print("\nKey insights:")
print("- Topological features capture global shape properties")
print("- Different shapes have distinct topological signatures")
print("- TDA features are robust to noise and deformations")
print("- Useful for shape recognition, time series analysis, and more")
```

```python
# Cell 8: TDA Applications - Manifold Learning and Clustering Validation
print("=== TDA APPLICATIONS: MANIFOLD LEARNING & CLUSTERING ===\n")

# Application 1: Manifold Learning with TDA
print("Application 1: Manifold Learning\n")

from sklearn.datasets import make_swiss_roll
from sklearn.manifold import MDS

# Generate high-dimensional manifold
n_samples = 300
swiss_roll_3d, color = make_swiss_roll(n_samples=n_samples, noise=0.1, random_state=42)

print(f"Swiss roll: {swiss_roll_3d.shape[0]} points in 3D")

# Compute VR complex at different scales
epsilons_manifold = [1.0, 2.0, 3.0, 4.0]
vr_swiss = []
for eps in epsilons_manifold:
    vr = vietoris_rips_complex(swiss_roll_3d[:50], eps, max_dim=1)  # Subsample for speed
    vr_swiss.append(vr)
    print(f"  VR(ε={eps}): f-vector = {vr.f_vector()}")

# Application 2: Clustering Validation with TDA
print("\nApplication 2: Clustering Validation with TDA\n")

# Generate datasets with different cluster structures
np.random.seed(42)

# Well-separated clusters
well_separated, labels_sep = make_blobs(n_samples=150, centers=3, cluster_std=0.5, 
                                        center_box=(-5, 5), random_state=42)

# Poorly separated clusters
poorly_separated, labels_poor = make_blobs(n_samples=150, centers=3, cluster_std=1.5,
                                           center_box=(-5, 5), random_state=42)

# Nested circles (not convex)
nested_circles, labels_nest = make_circles(n_samples=150, noise=0.05, factor=0.5, random_state=42)

datasets = [
    (well_separated, labels_sep, "Well-separated"),
    (poorly_separated, labels_poor, "Poorly-separated"),
    (nested_circles, labels_nest, "Nested circles")
]

# Compute topological signatures for clustering quality
def clustering_quality_tda(X, labels):
    """Assess clustering quality using TDA metrics"""
    n_clusters = len(set(labels))
    
    # Intra-cluster compactness (should have low H1)
    intra_h1_counts = []
    for label in set(labels):
        cluster_points = X[labels == label]
        if len(cluster_points) > 3:
            h0, h1 = compute_persistent_homology(cluster_points, max_epsilon=2.0, num_steps=15)
            intra_h1_counts.append(len(h1))
    
    # Inter-cluster separation (H0 should persist)
    h0_full, h1_full = compute_persistent_homology(X, max_epsilon=3.0, num_steps=20)
    
    # Compute quality metrics
    avg_intra_h1 = np.mean(intra_h1_counts) if intra_h1_counts else 0
    max_h0_persistence = max([d - b for b, d in h0_full]) if h0_full else 0
    
    return {
        'n_clusters': n_clusters,
        'avg_intra_loops': avg_intra_h1,
        'max_component_persistence': max_h0_persistence,
        'h0_features': len(h0_full),
        'h1_features': len(h1_full)
    }

print("Clustering Quality Assessment:\n")
quality_metrics = []
for X, labels, name in datasets:
    metrics = clustering_quality_tda(X, labels)
    quality_metrics.append(metrics)
    print(f"{name}:")
    print(f"  Clusters: {metrics['n_clusters']}")
    print(f"  Avg intra-cluster loops: {metrics['avg_intra_loops']:.2f}")
    print(f"  Max component persistence: {metrics['max_component_persistence']:.3f}")
    print(f"  H0 features: {metrics['h0_features']}, H1 features: {metrics['h1_features']}\n")

# Application 3: Time Series with Takens Embedding
print("Application 3: Time Series Analysis via Takens Embedding\n")

# Generate time series
t = np.linspace(0, 20*np.pi, 500)
periodic_signal = np.sin(t) + 0.5 * np.sin(2*t)  # Periodic
noisy_signal = periodic_signal + np.random.normal(0, 0.3, len(t))

# Takens embedding
def takens_embedding(time_series, delay=1, dimension=3):
    """Create delay embedding for time series"""
    n = len(time_series)
    m = n - (dimension - 1) * delay
    embedded = np.zeros((m, dimension))
    for i in range(m):
        embedded[i] = time_series[i:i + dimension * delay:delay]
    return embedded

# Embed time series
embedded_periodic = takens_embedding(periodic_signal, delay=10, dimension=2)
embedded_noisy = takens_embedding(noisy_signal, delay=10, dimension=2)

# Compute persistence
h0_per, h1_per = compute_persistent_homology(embedded_periodic[:100], max_epsilon=3.0, num_steps=20)
h0_noi, h1_noi = compute_persistent_homology(embedded_noisy[:100], max_epsilon=3.0, num_steps=20)

print(f"Periodic signal: H0={len(h0_per)}, H1={len(h1_per)} features")
print(f"Noisy signal: H0={len(h0_noi)}, H1={len(h1_noi)} features")

# Visualization
fig = plt.figure(figsize=(18, 14))

# Row 1: Manifold learning
ax1 = fig.add_subplot(3, 4, 1, projection='3d')
ax1.scatter(swiss_roll_3d[:, 0], swiss_roll_3d[:, 1], swiss_roll_3d[:, 2], 
           c=color, cmap='viridis', s=10, alpha=0.6)
ax1.set_title('Swiss Roll Manifold', fontsize=11, fontweight='bold')
ax1.set_xlabel('X')
ax1.set_ylabel('Y')
ax1.set_zlabel('Z')

ax2 = fig.add_subplot(3, 4, 2)
for i, (eps, vr) in enumerate(zip(epsilons_manifold, vr_swiss)):
    fvec = vr.f_vector()
    ax2.scatter([eps]*len(fvec), range(len(fvec)), s=[f*10 for f in fvec], 
               alpha=0.6, label=f'ε={eps}')
ax2.set_xlabel('ε', fontweight='bold')
ax2.set_ylabel('Dimension', fontweight='bold')
ax2.set_title('VR Complex Growth', fontsize=11, fontweight='bold')
ax2.grid(True, alpha=0.3)
ax2.legend(fontsize=8)

# Row 2: Clustering validation
for idx, (X, labels, name) in enumerate(datasets):
    ax = fig.add_subplot(3, 4, 5 + idx)
    scatter = ax.scatter(X[:, 0], X[:, 1], c=labels, cmap='tab10', 
                        alpha=0.6, s=40, edgecolors='black', linewidth=0.5)
    ax.set_title(f'{name}\n{quality_metrics[idx]["n_clusters"]} clusters', 
                fontsize=10, fontweight='bold')
    ax.set_aspect('equal')
    ax.grid(True, alpha=0.2)

# Clustering quality comparison
ax8 = fig.add_subplot(3, 4, 8)
dataset_names = [name for _, _, name in datasets]
metrics_to_plot = ['avg_intra_loops', 'h1_features']
x = np.arange(len(dataset_names))
width = 0.35
for i, metric in enumerate(metrics_to_plot):
    values = [qm[metric] for qm in quality_metrics]
    ax8.bar(x + i*width, values, width, label=metric.replace('_', ' '), alpha=0.7)
ax8.set_ylabel('Count', fontweight='bold')
ax8.set_title('TDA Clustering Metrics', fontsize=11, fontweight='bold')
ax8.set_xticks(x + width/2)
ax8.set_xticklabels(dataset_names, rotation=15, ha='right', fontsize=9)
ax8.legend(fontsize=8)
ax8.grid(True, alpha=0.3, axis='y')

# Row 3: Time series
ax9 = fig.add_subplot(3, 4, 9)
ax9.plot(t[:200], periodic_signal[:200], 'b-', linewidth=2, label='Periodic')
ax9.plot(t[:200], noisy_signal[:200], 'r-', alpha=0.5, linewidth=1, label='Noisy')
ax9.set_xlabel('Time', fontweight='bold')
ax9.set_ylabel('Amplitude', fontweight='bold')
ax9.set_title('Time Series Signals', fontsize=11, fontweight='bold')
ax9.legend()
ax9.grid(True, alpha=0.3)

ax10 = fig.add_subplot(3, 4, 10)
ax10.scatter(embedded_periodic[:, 0], embedded_periodic[:, 1], 
            c=range(len(embedded_periodic)), cmap='viridis', s=10, alpha=0.6)
ax10.set_title('Periodic Signal\n(Takens Embedding)', fontsize=10, fontweight='bold')
ax10.set_xlabel('x(t)', fontweight='bold')
ax10.set_ylabel('x(t+τ)', fontweight='bold')
ax10.grid(True, alpha=0.3)

ax11 = fig.add_subplot(3, 4, 11)
ax11.scatter(embedded_noisy[:, 0], embedded_noisy[:, 1], 
            c=range(len(embedded_noisy)), cmap='plasma', s=10, alpha=0.6)
ax11.set_title('Noisy Signal\n(Takens Embedding)', fontsize=10, fontweight='bold')
ax11.set_xlabel('x(t)', fontweight='bold')
ax11.set_ylabel('x(t+τ)', fontweight='bold')
ax11.grid(True, alpha=0.3)

# Summary of applications
ax12 = fig.add_subplot(3, 4, 12)
ax12.axis('off')
ax12.text(0.5, 0.95, 'TDA Applications', fontsize=12, fontweight='bold', 
         ha='center', transform=ax12.transAxes)
applications = [
    'Manifold Learning:',
    '  • Intrinsic dimension',
    '  • Topology preservation',
    '',
    'Clustering Validation:',
    '  • Quality assessment',
    '  • Structure detection',
    '',
    'Time Series:',
    '  • Periodicity detection',
    '  • Attractor reconstruction',
    '  • Anomaly detection',
    '',
    'Other Applications:',
    '  • Image analysis',
    '  • Neuroscience',
    '  • Materials science'
]
for i, text in enumerate(applications):
    y = 0.85 - i*0.048
    ax12.text(0.05, y, text, fontsize=9, transform=ax12.transAxes,
             fontweight='bold' if ':' in text else 'normal')

plt.tight_layout()
plt.show()

print("\nTDA Applications demonstration complete!")
print("\nConclusion:")
print("- TDA provides powerful tools for analyzing shape and structure")
print("- Complements traditional ML methods with topological insights")
print("- Robust to noise and applicable across many domains")
print("- Key techniques: Persistent homology, Mapper, VR complexes")
```
