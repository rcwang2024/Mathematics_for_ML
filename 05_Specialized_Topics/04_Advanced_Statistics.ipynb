```python
## 7. ML Applications: A/B Testing, Bayesian Optimization, and Uncertainty Quantification

from sklearn.gaussian_process import GaussianProcessRegressor
from sklearn.gaussian_process.kernels import RBF, ConstantKernel as C
from scipy.optimize import minimize

print("="*60)
print("MACHINE LEARNING APPLICATIONS")
print("="*60)

# 7.1 A/B Testing with Sequential Analysis
np.random.seed(42)

# Simulate A/B test data
n_per_group = 1000
conversion_rate_A = 0.10  # Control: 10% conversion
conversion_rate_B = 0.12  # Treatment: 12% conversion (20% relative lift)

conversions_A = np.random.binomial(1, conversion_rate_A, n_per_group)
conversions_B = np.random.binomial(1, conversion_rate_B, n_per_group)

# Cumulative conversion rates over time
cumulative_A = np.cumsum(conversions_A) / (np.arange(n_per_group) + 1)
cumulative_B = np.cumsum(conversions_B) / (np.arange(n_per_group) + 1)

# Statistical test
total_conversions_A = conversions_A.sum()
total_conversions_B = conversions_B.sum()

# Z-test for proportions
p_pooled = (total_conversions_A + total_conversions_B) / (2 * n_per_group)
se = np.sqrt(p_pooled * (1 - p_pooled) * (2 / n_per_group))
z_stat = (conversion_rate_B - conversion_rate_A) / se
p_value_ab = 2 * (1 - stats.norm.cdf(abs(z_stat)))

# Bayesian A/B Test
alpha_prior, beta_prior = 1, 1  # Uniform prior
alpha_A = alpha_prior + total_conversions_A
beta_A = beta_prior + (n_per_group - total_conversions_A)
alpha_B = alpha_prior + total_conversions_B
beta_B = beta_prior + (n_per_group - total_conversions_B)

# Probability that B > A
samples_A = np.random.beta(alpha_A, beta_A, 100000)
samples_B = np.random.beta(alpha_B, beta_B, 100000)
prob_B_better = np.mean(samples_B > samples_A)

print(f"\n1. A/B Testing Results:")
print(f"   Control (A): {total_conversions_A}/{n_per_group} = {conversion_rate_A:.3f}")
print(f"   Treatment (B): {total_conversions_B}/{n_per_group} = {conversion_rate_B:.3f}")
print(f"   Relative lift: {(conversion_rate_B/conversion_rate_A - 1)*100:.1f}%")
print(f"   Frequentist p-value: {p_value_ab:.4f}")
print(f"   Bayesian P(B > A): {prob_B_better:.4f}")

# 7.2 Bayesian Optimization
def objective_function(x):
    """Black-box function to optimize"""
    return -(x**2 * np.sin(5 * np.pi * x))**0.5

# Initial samples
X_init = np.array([[0.1], [0.3], [0.6]])
y_init = np.array([objective_function(x[0]) for x in X_init])

# Gaussian Process model
kernel = C(1.0, (1e-3, 1e3)) * RBF(0.1, (1e-2, 1e2))
gp = GaussianProcessRegressor(kernel=kernel, alpha=1e-6, n_restarts_optimizer=10)

# Bayesian Optimization iterations
n_iterations = 7
X_sample = list(X_init)
y_sample = list(y_init)

for i in range(n_iterations):
    gp.fit(np.array(X_sample), np.array(y_sample))
    
    # Acquisition function: Upper Confidence Bound (UCB)
    def acquisition(x):
        x = np.atleast_2d(x)
        mu, sigma = gp.predict(x, return_std=True)
        return -(mu + 2 * sigma)  # Negative for minimization
    
    # Optimize acquisition function
    x_tries = np.random.uniform(0, 1, size=(1000, 1))
    acq_values = [acquisition(x) for x in x_tries]
    x_next = x_tries[np.argmin(acq_values)]
    
    # Evaluate objective at new point
    y_next = objective_function(x_next[0])
    X_sample.append(x_next)
    y_sample.append(y_next)

best_idx = np.argmax(y_sample)
best_x = X_sample[best_idx]
best_y = y_sample[best_idx]

print(f"\n2. Bayesian Optimization:")
print(f"   Initial samples: {len(X_init)}")
print(f"   Total iterations: {n_iterations}")
print(f"   Best x found: {best_x[0]:.4f}")
print(f"   Best y found: {best_y:.4f}")

# 7.3 Uncertainty Quantification in Predictions
from sklearn.ensemble import RandomForestRegressor

# Generate regression data with heteroscedastic noise
X_uq = np.random.uniform(0, 10, 200).reshape(-1, 1)
y_true_uq = np.sin(X_uq).ravel()
noise_std = 0.1 + 0.1 * X_uq.ravel()  # Increasing noise
y_uq = y_true_uq + np.random.normal(0, noise_std)

# Train model
rf = RandomForestRegressor(n_estimators=100, random_state=42)
rf.fit(X_uq, y_uq)

# Prediction with uncertainty (using tree variance)
X_test = np.linspace(0, 10, 100).reshape(-1, 1)
predictions = []
for tree in rf.estimators_:
    predictions.append(tree.predict(X_test))
predictions = np.array(predictions)

y_pred = predictions.mean(axis=0)
y_std = predictions.std(axis=0)

print(f"\n3. Uncertainty Quantification:")
print(f"   Training samples: {len(X_uq)}")
print(f"   Model: Random Forest with {rf.n_estimators} trees")
print(f"   Mean prediction uncertainty: {y_std.mean():.4f}")
print(f"   Max prediction uncertainty: {y_std.max():.4f}")

# 7.4 Statistical Power Analysis for Experiment Design
from statsmodels.stats.power import tt_ind_solve_power

# Calculate required sample size for desired power
effect_size = 0.3  # Cohen's d
alpha = 0.05
power = 0.80

n_required = tt_ind_solve_power(effect_size=effect_size, alpha=alpha, power=power, 
                                  ratio=1.0, alternative='two-sided')

print(f"\n4. Power Analysis for Experiment Design:")
print(f"   Effect size (Cohen's d): {effect_size}")
print(f"   Significance level (α): {alpha}")
print(f"   Desired power (1-β): {power}")
print(f"   Required sample size per group: {int(np.ceil(n_required))}")

# Visualizations
fig, axes = plt.subplots(2, 3, figsize=(16, 10))

# Plot 1: A/B Test Sequential Analysis
axes[0, 0].plot(cumulative_A, label='Control (A)', linewidth=2, color='blue')
axes[0, 0].plot(cumulative_B, label='Treatment (B)', linewidth=2, color='red')
axes[0, 0].axhline(y=conversion_rate_A, color='blue', linestyle='--', alpha=0.5, label='True rate A')
axes[0, 0].axhline(y=conversion_rate_B, color='red', linestyle='--', alpha=0.5, label='True rate B')
axes[0, 0].set_xlabel('Sample Size')
axes[0, 0].set_ylabel('Conversion Rate')
axes[0, 0].set_title(f'A/B Test Sequential Analysis (p={p_value_ab:.4f})')
axes[0, 0].legend()
axes[0, 0].grid(alpha=0.3)

# Plot 2: Bayesian A/B Test Posteriors
x_beta = np.linspace(0, 0.25, 500)
axes[0, 1].plot(x_beta, stats.beta.pdf(x_beta, alpha_A, beta_A), 
                'b-', linewidth=2, label='Control (A)')
axes[0, 1].plot(x_beta, stats.beta.pdf(x_beta, alpha_B, beta_B), 
                'r-', linewidth=2, label='Treatment (B)')
axes[0, 1].fill_between(x_beta, 0, stats.beta.pdf(x_beta, alpha_A, beta_A), alpha=0.3, color='blue')
axes[0, 1].fill_between(x_beta, 0, stats.beta.pdf(x_beta, alpha_B, beta_B), alpha=0.3, color='red')
axes[0, 1].set_xlabel('Conversion Rate')
axes[0, 1].set_ylabel('Posterior Density')
axes[0, 1].set_title(f'Bayesian A/B Test (P(B>A)={prob_B_better:.3f})')
axes[0, 1].legend()
axes[0, 1].grid(alpha=0.3)

# Plot 3: Bayesian Optimization Progress
X_plot = np.linspace(0, 1, 500).reshape(-1, 1)
y_plot = np.array([objective_function(x[0]) for x in X_plot])
mu, sigma = gp.predict(X_plot, return_std=True)

axes[0, 2].plot(X_plot, y_plot, 'k-', linewidth=2, label='True function')
axes[0, 2].plot(X_plot, mu, 'b-', linewidth=2, label='GP mean')
axes[0, 2].fill_between(X_plot.ravel(), mu - 2*sigma, mu + 2*sigma, 
                        alpha=0.3, color='blue', label='95% CI')
axes[0, 2].scatter(X_sample, y_sample, c=range(len(X_sample)), 
                  cmap='viridis', s=100, zorder=10, edgecolors='black', linewidth=1)
axes[0, 2].scatter(best_x, best_y, c='red', s=200, marker='*', 
                  zorder=11, edgecolors='black', linewidth=1, label='Best')
axes[0, 2].set_xlabel('x')
axes[0, 2].set_ylabel('f(x)')
axes[0, 2].set_title('Bayesian Optimization')
axes[0, 2].legend()
axes[0, 2].grid(alpha=0.3)

# Plot 4: Uncertainty Quantification
axes[1, 0].scatter(X_uq, y_uq, alpha=0.5, s=20, color='gray', label='Training data')
axes[1, 0].plot(X_test, y_pred, 'b-', linewidth=2, label='Prediction')
axes[1, 0].fill_between(X_test.ravel(), y_pred - 2*y_std, y_pred + 2*y_std, 
                        alpha=0.3, color='blue', label='95% CI')
axes[1, 0].plot(X_test, np.sin(X_test), 'r--', linewidth=2, alpha=0.7, label='True function')
axes[1, 0].set_xlabel('X')
axes[1, 0].set_ylabel('y')
axes[1, 0].set_title('Uncertainty Quantification (Random Forest)')
axes[1, 0].legend()
axes[1, 0].grid(alpha=0.3)

# Plot 5: Power Analysis Curve
effect_sizes = np.linspace(0.1, 1.0, 50)
power_curve = [tt_ind_solve_power(effect_size=es, alpha=0.05, nobs1=50, 
                                   ratio=1.0, alternative='two-sided') for es in effect_sizes]
axes[1, 1].plot(effect_sizes, power_curve, 'b-', linewidth=2)
axes[1, 1].axhline(y=0.8, color='green', linestyle='--', linewidth=2, label='80% power')
axes[1, 1].axvline(x=effect_size, color='red', linestyle='--', linewidth=2, 
                   label=f'Effect size={effect_size}')
axes[1, 1].set_xlabel('Effect Size (Cohen\'s d)')
axes[1, 1].set_ylabel('Statistical Power')
axes[1, 1].set_title('Power Analysis (n=50 per group)')
axes[1, 1].legend()
axes[1, 1].grid(alpha=0.3)

# Plot 6: Sample Size vs Power
sample_sizes = np.arange(10, 500, 10)
power_by_n = [tt_ind_solve_power(effect_size=0.3, alpha=0.05, nobs1=n, 
                                  ratio=1.0, alternative='two-sided') for n in sample_sizes]
axes[1, 2].plot(sample_sizes, power_by_n, 'b-', linewidth=2)
axes[1, 2].axhline(y=0.8, color='green', linestyle='--', linewidth=2, label='80% power')
axes[1, 2].axvline(x=n_required, color='red', linestyle='--', linewidth=2, 
                   label=f'Required n={int(np.ceil(n_required))}')
axes[1, 2].set_xlabel('Sample Size per Group')
axes[1, 2].set_ylabel('Statistical Power')
axes[1, 2].set_title('Sample Size Determination (effect size=0.3)')
axes[1, 2].legend()
axes[1, 2].grid(alpha=0.3)

plt.tight_layout()
plt.show()

print("\n✓ ML applications demonstrations complete!")
```

```python
## 6. Nonparametric Statistics: Bootstrap, Permutation Tests, and KDE

print("="*60)
print("NONPARAMETRIC STATISTICS DEMONSTRATIONS")
print("="*60)

# 6.1 Bootstrap for Confidence Intervals
np.random.seed(42)
sample_size = 100
original_sample = np.random.exponential(scale=2, size=sample_size)

# Calculate statistic of interest (median)
observed_median = np.median(original_sample)

# Bootstrap resampling
n_bootstrap = 10000
bootstrap_medians = np.zeros(n_bootstrap)

for i in range(n_bootstrap):
    bootstrap_sample = np.random.choice(original_sample, size=sample_size, replace=True)
    bootstrap_medians[i] = np.median(bootstrap_sample)

# Confidence intervals
ci_lower = np.percentile(bootstrap_medians, 2.5)
ci_upper = np.percentile(bootstrap_medians, 97.5)

print(f"\n1. Bootstrap Confidence Interval:")
print(f"   Sample size: {sample_size}")
print(f"   Observed median: {observed_median:.4f}")
print(f"   Bootstrap 95% CI: [{ci_lower:.4f}, {ci_upper:.4f}]")
print(f"   CI width: {ci_upper - ci_lower:.4f}")

# 6.2 Permutation Test for Two Samples
group_a = np.random.normal(5, 2, 50)
group_b = np.random.normal(5.5, 2, 50)

# Observed difference
observed_diff = group_a.mean() - group_b.mean()

# Permutation test
n_permutations = 10000
combined = np.concatenate([group_a, group_b])
perm_diffs = np.zeros(n_permutations)

for i in range(n_permutations):
    permuted = np.random.permutation(combined)
    perm_diffs[i] = permuted[:50].mean() - permuted[50:].mean()

# P-value (two-tailed)
p_value_perm = np.mean(np.abs(perm_diffs) >= np.abs(observed_diff))

print(f"\n2. Permutation Test:")
print(f"   Group A mean: {group_a.mean():.4f}")
print(f"   Group B mean: {group_b.mean():.4f}")
print(f"   Observed difference: {observed_diff:.4f}")
print(f"   p-value: {p_value_perm:.4f}")
print(f"   Result: {'Significant' if p_value_perm < 0.05 else 'Not significant'} at α=0.05")

# 6.3 Kernel Density Estimation
from scipy.stats import gaussian_kde

# Generate bimodal data
data_kde = np.concatenate([np.random.normal(-2, 0.5, 200), 
                           np.random.normal(2, 0.8, 200)])

# KDE with different bandwidths
kde_scott = gaussian_kde(data_kde, bw_method='scott')
kde_silverman = gaussian_kde(data_kde, bw_method='silverman')
kde_custom = gaussian_kde(data_kde, bw_method=0.2)

x_range = np.linspace(-5, 5, 500)

print(f"\n3. Kernel Density Estimation:")
print(f"   Sample size: {len(data_kde)}")
print(f"   Scott bandwidth: {kde_scott.factor:.4f}")
print(f"   Silverman bandwidth: {kde_silverman.factor:.4f}")
print(f"   Custom bandwidth: 0.2000")

# 6.4 Rank-based Tests (Mann-Whitney U)
from scipy.stats import mannwhitneyu

# Non-normal data
non_normal_a = np.random.exponential(2, 40)
non_normal_b = np.random.exponential(2.5, 40)

u_stat, p_value_mw = mannwhitneyu(non_normal_a, non_normal_b, alternative='two-sided')

print(f"\n4. Mann-Whitney U Test (Rank-based):")
print(f"   U-statistic: {u_stat:.2f}")
print(f"   p-value: {p_value_mw:.4f}")
print(f"   Result: {'Significant' if p_value_mw < 0.05 else 'Not significant'} difference")

# 6.5 Bootstrap Regression Confidence Bands
X_reg = np.random.uniform(0, 10, 100)
y_reg = 2 * X_reg + 5 + np.random.normal(0, 3, 100)

n_boot_reg = 1000
boot_slopes = np.zeros(n_boot_reg)
boot_intercepts = np.zeros(n_boot_reg)

for i in range(n_boot_reg):
    indices = np.random.choice(len(X_reg), len(X_reg), replace=True)
    X_boot = X_reg[indices]
    y_boot = y_reg[indices]
    
    # Fit linear regression
    slope, intercept = np.polyfit(X_boot, y_boot, 1)
    boot_slopes[i] = slope
    boot_intercepts[i] = intercept

print(f"\n5. Bootstrap Regression:")
print(f"   Slope 95% CI: [{np.percentile(boot_slopes, 2.5):.4f}, {np.percentile(boot_slopes, 97.5):.4f}]")
print(f"   Intercept 95% CI: [{np.percentile(boot_intercepts, 2.5):.4f}, {np.percentile(boot_intercepts, 97.5):.4f}]")

# Visualizations
fig, axes = plt.subplots(2, 3, figsize=(16, 10))

# Plot 1: Bootstrap distribution
axes[0, 0].hist(bootstrap_medians, bins=50, density=True, alpha=0.7, color='skyblue', edgecolor='black')
axes[0, 0].axvline(observed_median, color='red', linewidth=2, linestyle='--', label=f'Observed: {observed_median:.3f}')
axes[0, 0].axvline(ci_lower, color='green', linewidth=2, linestyle=':', label=f'95% CI')
axes[0, 0].axvline(ci_upper, color='green', linewidth=2, linestyle=':')
axes[0, 0].set_xlabel('Median')
axes[0, 0].set_ylabel('Density')
axes[0, 0].set_title(f'Bootstrap Distribution (n={n_bootstrap})')
axes[0, 0].legend()
axes[0, 0].grid(alpha=0.3)

# Plot 2: Permutation distribution
axes[0, 1].hist(perm_diffs, bins=50, density=True, alpha=0.7, color='lightcoral', edgecolor='black')
axes[0, 1].axvline(observed_diff, color='blue', linewidth=2, linestyle='--', label=f'Observed: {observed_diff:.3f}')
axes[0, 1].axvline(-observed_diff, color='blue', linewidth=2, linestyle='--')
axes[0, 1].set_xlabel('Mean Difference')
axes[0, 1].set_ylabel('Density')
axes[0, 1].set_title(f'Permutation Test (p={p_value_perm:.4f})')
axes[0, 1].legend()
axes[0, 1].grid(alpha=0.3)

# Plot 3: Kernel Density Estimation
axes[0, 2].hist(data_kde, bins=30, density=True, alpha=0.4, color='gray', label='Histogram')
axes[0, 2].plot(x_range, kde_scott(x_range), 'b-', linewidth=2, label="Scott's rule")
axes[0, 2].plot(x_range, kde_silverman(x_range), 'r-', linewidth=2, label="Silverman's rule")
axes[0, 2].plot(x_range, kde_custom(x_range), 'g-', linewidth=2, label='Custom (0.2)')
axes[0, 2].set_xlabel('Value')
axes[0, 2].set_ylabel('Density')
axes[0, 2].set_title('Kernel Density Estimation')
axes[0, 2].legend()
axes[0, 2].grid(alpha=0.3)

# Plot 4: Group comparison (permutation test)
axes[1, 0].violinplot([group_a, group_b], positions=[1, 2], showmeans=True, showmedians=True)
axes[1, 0].set_xticks([1, 2])
axes[1, 0].set_xticklabels(['Group A', 'Group B'])
axes[1, 0].set_ylabel('Value')
axes[1, 0].set_title(f'Permutation Test Groups (p={p_value_perm:.4f})')
axes[1, 0].grid(alpha=0.3, axis='y')

# Plot 5: Rank comparison (Mann-Whitney)
axes[1, 1].boxplot([non_normal_a, non_normal_b], labels=['Sample A', 'Sample B'], 
                   patch_artist=True, showmeans=True)
axes[1, 1].set_ylabel('Value')
axes[1, 1].set_title(f'Mann-Whitney U Test (p={p_value_mw:.4f})')
axes[1, 1].grid(alpha=0.3, axis='y')

# Plot 6: Bootstrap regression bands
axes[1, 2].scatter(X_reg, y_reg, alpha=0.5, s=30, color='blue', label='Data')
# Plot bootstrap regression lines
X_plot = np.linspace(0, 10, 100)
for i in range(100):  # Plot subset of bootstrap lines
    y_boot_line = boot_intercepts[i] + boot_slopes[i] * X_plot
    axes[1, 2].plot(X_plot, y_boot_line, 'r-', alpha=0.05)
# Plot mean regression
mean_slope = np.mean(boot_slopes)
mean_intercept = np.mean(boot_intercepts)
axes[1, 2].plot(X_plot, mean_intercept + mean_slope * X_plot, 'g-', 
                linewidth=3, label='Bootstrap mean')
axes[1, 2].set_xlabel('X')
axes[1, 2].set_ylabel('y')
axes[1, 2].set_title('Bootstrap Regression Confidence Bands')
axes[1, 2].legend()
axes[1, 2].grid(alpha=0.3)

plt.tight_layout()
plt.show()

print("\n✓ Nonparametric statistics demonstrations complete!")
```

```python
## 5. Causal Inference: Potential Outcomes and Instrumental Variables

print("="*60)
print("CAUSAL INFERENCE DEMONSTRATIONS")
print("="*60)

# 5.1 Simulate Causal Data with Confounding
np.random.seed(42)
n = 500

# Confounder (e.g., age)
Z = np.random.normal(50, 10, n)

# Treatment assignment (influenced by confounder)
treatment_prob = 1 / (1 + np.exp(-(Z - 50) / 5))  # Logistic relationship
T = np.random.binomial(1, treatment_prob, n)

# Outcome (influenced by both treatment and confounder)
# True causal effect of treatment: +10
Y0 = 100 + 2 * Z + np.random.normal(0, 10, n)  # Potential outcome under control
Y1 = Y0 + 10  # Potential outcome under treatment (constant treatment effect)
Y = T * Y1 + (1 - T) * Y0  # Observed outcome

print(f"\n1. Simulated Causal Data:")
print(f"   Sample size: {n}")
print(f"   Treatment assignment: {T.sum()} treated, {n - T.sum()} control")
print(f"   True Average Treatment Effect (ATE): 10.0")

# 5.2 Naive Comparison (Biased due to confounding)
naive_ate = Y[T == 1].mean() - Y[T == 0].mean()
print(f"\n2. Naive Treatment Effect (Biased):")
print(f"   E[Y|T=1] - E[Y|T=0] = {naive_ate:.2f}")
print(f"   Bias: {naive_ate - 10:.2f}")

# 5.3 Propensity Score Matching (to reduce confounding)
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import NearestNeighbors

# Estimate propensity scores
ps_model = LogisticRegression()
ps_model.fit(Z.reshape(-1, 1), T)
propensity_scores = ps_model.predict_proba(Z.reshape(-1, 1))[:, 1]

# Matching: for each treated unit, find nearest control unit by propensity score
treated_idx = np.where(T == 1)[0]
control_idx = np.where(T == 0)[0]

nn = NearestNeighbors(n_neighbors=1)
nn.fit(propensity_scores[control_idx].reshape(-1, 1))
distances, matches = nn.kneighbors(propensity_scores[treated_idx].reshape(-1, 1))

# Calculate ATE on matched sample
matched_control_outcomes = Y[control_idx[matches.flatten()]]
matched_treated_outcomes = Y[treated_idx]
psm_ate = matched_treated_outcomes.mean() - matched_control_outcomes.mean()

print(f"\n3. Propensity Score Matching:")
print(f"   Estimated ATE: {psm_ate:.2f}")
print(f"   Bias: {psm_ate - 10:.2f}")

# 5.4 Regression Adjustment
from sklearn.linear_model import LinearRegression

# Fit outcome model including confounder
X_reg = np.c_[T, Z]
reg_model = LinearRegression()
reg_model.fit(X_reg, Y)
reg_ate = reg_model.coef_[0]

print(f"\n4. Regression Adjustment:")
print(f"   Treatment coefficient: {reg_ate:.2f}")
print(f"   Confounder coefficient: {reg_model.coef_[1]:.2f}")
print(f"   Bias: {reg_ate - 10:.2f}")

# 5.5 Instrumental Variable (IV) Analysis
# Create an instrument: randomized encouragement
IV = np.random.binomial(1, 0.5, n)

# Treatment is influenced by IV and confounder
T_iv = np.random.binomial(1, 0.3 + 0.4 * IV + 0.001 * Z, n)

# Outcome (same causal structure)
Y0_iv = 100 + 2 * Z + np.random.normal(0, 10, n)
Y1_iv = Y0_iv + 10
Y_iv = T_iv * Y1_iv + (1 - T_iv) * Y0_iv

# Two-Stage Least Squares (2SLS)
# First stage: regress T on IV
first_stage = LinearRegression()
first_stage.fit(IV.reshape(-1, 1), T_iv)
T_hat = first_stage.predict(IV.reshape(-1, 1))

# Second stage: regress Y on predicted T
second_stage = LinearRegression()
second_stage.fit(T_hat.reshape(-1, 1), Y_iv)
iv_ate = second_stage.coef_[0]

print(f"\n5. Instrumental Variables (2SLS):")
print(f"   First stage F-statistic: {np.var(T_hat) / np.var(T_iv - T_hat):.2f}")
print(f"   IV estimated ATE: {iv_ate:.2f}")
print(f"   Bias: {iv_ate - 10:.2f}")

# Visualizations
fig, axes = plt.subplots(2, 3, figsize=(16, 10))

# Plot 1: Confounding structure
axes[0, 0].scatter(Z[T == 0], Y[T == 0], alpha=0.5, s=30, label='Control', color='blue')
axes[0, 0].scatter(Z[T == 1], Y[T == 1], alpha=0.5, s=30, label='Treated', color='red')
axes[0, 0].set_xlabel('Confounder (Z)')
axes[0, 0].set_ylabel('Outcome (Y)')
axes[0, 0].set_title('Confounding: Z affects both T and Y')
axes[0, 0].legend()
axes[0, 0].grid(alpha=0.3)

# Plot 2: Treatment assignment by confounder
axes[0, 1].scatter(Z, treatment_prob, alpha=0.3, s=20, label='True propensity')
axes[0, 1].scatter(Z, T, alpha=0.3, s=20, label='Observed treatment')
z_range = np.linspace(Z.min(), Z.max(), 100)
axes[0, 1].plot(z_range, 1 / (1 + np.exp(-(z_range - 50) / 5)), 
                'r-', linewidth=2, label='Treatment probability')
axes[0, 1].set_xlabel('Confounder (Z)')
axes[0, 1].set_ylabel('Treatment Probability')
axes[0, 1].set_title('Selection Bias: Treatment Assignment')
axes[0, 1].legend()
axes[0, 1].grid(alpha=0.3)

# Plot 3: Treatment effects comparison
methods = ['True ATE', 'Naive', 'PSM', 'Regression', 'IV']
estimates = [10, naive_ate, psm_ate, reg_ate, iv_ate]
colors = ['green', 'red', 'orange', 'blue', 'purple']
axes[0, 2].barh(methods, estimates, color=colors, alpha=0.7)
axes[0, 2].axvline(x=10, color='green', linestyle='--', linewidth=2, label='True ATE')
axes[0, 2].set_xlabel('Estimated Treatment Effect')
axes[0, 2].set_title('Causal Effect Estimates')
axes[0, 2].legend()
axes[0, 2].grid(alpha=0.3, axis='x')

# Plot 4: Propensity score distribution
axes[1, 0].hist(propensity_scores[T == 0], bins=30, alpha=0.6, 
                label='Control', color='blue', density=True)
axes[1, 0].hist(propensity_scores[T == 1], bins=30, alpha=0.6, 
                label='Treated', color='red', density=True)
axes[1, 0].set_xlabel('Propensity Score')
axes[1, 0].set_ylabel('Density')
axes[1, 0].set_title('Propensity Score Distribution')
axes[1, 0].legend()
axes[1, 0].grid(alpha=0.3)

# Plot 5: Covariate balance before/after matching
balance_before = Z[T == 1].mean() - Z[T == 0].mean()
balance_after = Z[treated_idx].mean() - Z[control_idx[matches.flatten()]].mean()

axes[1, 1].bar(['Before Matching', 'After Matching'], 
               [balance_before, balance_after], 
               color=['red', 'green'], alpha=0.7)
axes[1, 1].axhline(y=0, color='black', linestyle='--', linewidth=1)
axes[1, 1].set_ylabel('Mean Difference in Z')
axes[1, 1].set_title('Covariate Balance (PSM)')
axes[1, 1].grid(alpha=0.3, axis='y')

# Plot 6: IV first stage
axes[1, 2].scatter(IV, T_iv, alpha=0.5, s=30)
iv_means = [T_iv[IV == 0].mean(), T_iv[IV == 1].mean()]
axes[1, 2].plot([0, 1], iv_means, 'ro-', linewidth=2, markersize=10, 
                label=f'Treatment difference: {iv_means[1] - iv_means[0]:.3f}')
axes[1, 2].set_xlabel('Instrument (IV)')
axes[1, 2].set_ylabel('Treatment (T)')
axes[1, 2].set_title('Instrumental Variable: First Stage')
axes[1, 2].set_xticks([0, 1])
axes[1, 2].set_xticklabels(['No IV', 'IV'])
axes[1, 2].legend()
axes[1, 2].grid(alpha=0.3)

plt.tight_layout()
plt.show()

print(f"\n6. Summary - Bias Reduction:")
print(f"   Naive bias: {abs(naive_ate - 10):.2f}")
print(f"   PSM bias: {abs(psm_ate - 10):.2f}")
print(f"   Regression bias: {abs(reg_ate - 10):.2f}")
print(f"   IV bias: {abs(iv_ate - 10):.2f}")
print(f"   Best method: {'PSM' if abs(psm_ate - 10) < min(abs(naive_ate - 10), abs(reg_ate - 10), abs(iv_ate - 10)) else 'Regression' if abs(reg_ate - 10) < min(abs(psm_ate - 10), abs(iv_ate - 10)) else 'IV'}")

print("\n✓ Causal inference demonstrations complete!")
```

```python
## 4. Time Series Analysis: Stationarity, ARIMA, and Forecasting

from statsmodels.graphics.tsaplots import plot_acf, plot_pacf
from statsmodels.tsa.statespace.sarimax import SARIMAX

print("="*60)
print("TIME SERIES ANALYSIS DEMONSTRATIONS")
print("="*60)

# 4.1 Generate time series data
np.random.seed(42)
n_points = 200

# Non-stationary series (with trend and seasonality)
t = np.arange(n_points)
trend = 0.5 * t
seasonality = 10 * np.sin(2 * np.pi * t / 12)
noise = np.random.normal(0, 5, n_points)
ts_nonstationary = trend + seasonality + noise

# Stationary series (ARMA process)
# AR(1) process: y_t = 0.7*y_{t-1} + ε_t
ar_coef = 0.7
ts_stationary = np.zeros(n_points)
for i in range(1, n_points):
    ts_stationary[i] = ar_coef * ts_stationary[i-1] + np.random.normal(0, 1)

print(f"\n1. Generated Time Series:")
print(f"   Length: {n_points} points")
print(f"   Non-stationary: trend + seasonality + noise")
print(f"   Stationary: AR(1) process with φ=0.7")

# 4.2 Stationarity Tests (Augmented Dickey-Fuller)
adf_nonstat = adfuller(ts_nonstationary, autolag='AIC')
adf_stat = adfuller(ts_stationary, autolag='AIC')

print(f"\n2. Augmented Dickey-Fuller Test:")
print(f"   Non-stationary series:")
print(f"     ADF statistic: {adf_nonstat[0]:.4f}")
print(f"     p-value: {adf_nonstat[1]:.4f}")
print(f"     Result: {'Stationary' if adf_nonstat[1] < 0.05 else 'Non-stationary'}")
print(f"   Stationary series:")
print(f"     ADF statistic: {adf_stat[0]:.4f}")
print(f"     p-value: {adf_stat[1]:.4f}")
print(f"     Result: {'Stationary' if adf_stat[1] < 0.05 else 'Non-stationary'}")

# 4.3 Differencing to achieve stationarity
ts_diff = np.diff(ts_nonstationary)
adf_diff = adfuller(ts_diff, autolag='AIC')

print(f"\n3. First Differencing:")
print(f"   ADF statistic: {adf_diff[0]:.4f}")
print(f"   p-value: {adf_diff[1]:.4f}")
print(f"   Result: {'Stationary' if adf_diff[1] < 0.05 else 'Non-stationary'}")

# 4.4 ARIMA Model Fitting
# Fit ARIMA(1,1,1) to non-stationary data
arima_model = SARIMAX(ts_nonstationary, order=(1, 1, 1), trend='c')
arima_results = arima_model.fit(disp=False)

print(f"\n4. ARIMA(1,1,1) Model:")
print(f"   AR coefficient: {arima_results.params[1]:.4f}")
print(f"   MA coefficient: {arima_results.params[2]:.4f}")
print(f"   AIC: {arima_results.aic:.2f}")
print(f"   BIC: {arima_results.bic:.2f}")

# 4.5 Forecasting
n_forecast = 20
forecast = arima_results.forecast(steps=n_forecast)
forecast_conf = arima_results.get_forecast(steps=n_forecast).conf_int()

print(f"\n5. Forecasting:")
print(f"   Forecast horizon: {n_forecast} steps")
print(f"   First 5 forecasts: {forecast[:5].values}")

# Visualizations
fig = plt.figure(figsize=(16, 12))
gs = fig.add_gridspec(3, 3, hspace=0.3, wspace=0.3)

# Plot 1: Non-stationary time series
ax1 = fig.add_subplot(gs[0, :2])
ax1.plot(ts_nonstationary, linewidth=1.5, label='Non-stationary series')
ax1.plot(trend, 'r--', linewidth=2, alpha=0.7, label='Trend')
ax1.axhline(y=ts_nonstationary.mean(), color='green', linestyle=':', 
            linewidth=2, label=f'Mean: {ts_nonstationary.mean():.2f}')
ax1.set_xlabel('Time')
ax1.set_ylabel('Value')
ax1.set_title(f'Non-stationary Series (ADF p-value: {adf_nonstat[1]:.4f})')
ax1.legend()
ax1.grid(alpha=0.3)

# Plot 2: Stationary time series
ax2 = fig.add_subplot(gs[0, 2])
ax2.plot(ts_stationary, linewidth=1.5, label='Stationary series')
ax2.axhline(y=ts_stationary.mean(), color='red', linestyle='--', 
            linewidth=2, label=f'Mean: {ts_stationary.mean():.2f}')
ax2.set_xlabel('Time')
ax2.set_ylabel('Value')
ax2.set_title(f'Stationary AR(1) Series\n(ADF p-value: {adf_stat[1]:.4f})')
ax2.legend()
ax2.grid(alpha=0.3)

# Plot 3: ACF of non-stationary
ax3 = fig.add_subplot(gs[1, 0])
plot_acf(ts_nonstationary, lags=40, ax=ax3, alpha=0.05)
ax3.set_title('ACF: Non-stationary')
ax3.grid(alpha=0.3)

# Plot 4: PACF of non-stationary
ax4 = fig.add_subplot(gs[1, 1])
plot_pacf(ts_nonstationary, lags=40, ax=ax4, alpha=0.05)
ax4.set_title('PACF: Non-stationary')
ax4.grid(alpha=0.3)

# Plot 5: First difference
ax5 = fig.add_subplot(gs[1, 2])
ax5.plot(ts_diff, linewidth=1.5, color='purple')
ax5.axhline(y=ts_diff.mean(), color='red', linestyle='--', linewidth=2)
ax5.set_xlabel('Time')
ax5.set_ylabel('Δy')
ax5.set_title(f'First Difference (ADF p-value: {adf_diff[1]:.4f})')
ax5.grid(alpha=0.3)

# Plot 6: ARIMA Forecast
ax6 = fig.add_subplot(gs[2, :])
ax6.plot(ts_nonstationary, linewidth=1.5, label='Observed', color='blue')
ax6.plot(arima_results.fittedvalues, linewidth=1.5, label='Fitted', 
         color='red', alpha=0.7)
forecast_index = np.arange(n_points, n_points + n_forecast)
ax6.plot(forecast_index, forecast, linewidth=2, label='Forecast', 
         color='green', linestyle='--')
ax6.fill_between(forecast_index, 
                  forecast_conf.iloc[:, 0], 
                  forecast_conf.iloc[:, 1],
                  alpha=0.3, color='green', label='95% CI')
ax6.axvline(x=n_points, color='black', linestyle=':', linewidth=2, 
            label='Forecast start')
ax6.set_xlabel('Time')
ax6.set_ylabel('Value')
ax6.set_title(f'ARIMA(1,1,1) Forecast (AIC: {arima_results.aic:.2f})')
ax6.legend(loc='upper left')
ax6.grid(alpha=0.3)

plt.show()

# Additional: Residual diagnostics
print(f"\n6. Residual Diagnostics:")
residuals = arima_results.resid
print(f"   Mean of residuals: {residuals.mean():.6f}")
print(f"   Std of residuals: {residuals.std():.4f}")

# Ljung-Box test for residual autocorrelation
from statsmodels.stats.diagnostic import acorr_ljungbox
lb_test = acorr_ljungbox(residuals, lags=[10], return_df=True)
print(f"   Ljung-Box test (lag 10):")
print(f"     Statistic: {lb_test['lb_stat'].values[0]:.4f}")
print(f"     p-value: {lb_test['lb_pvalue'].values[0]:.4f}")
print(f"     Result: {'No significant autocorrelation' if lb_test['lb_pvalue'].values[0] > 0.05 else 'Autocorrelation present'}")

print("\n✓ Time series analysis demonstrations complete!")
```

```python
## 3. Multivariate Analysis: PCA, MANOVA, and Factor Analysis

from sklearn.decomposition import PCA, FactorAnalysis
from sklearn.preprocessing import StandardScaler

print("="*60)
print("MULTIVARIATE ANALYSIS DEMONSTRATIONS")
print("="*60)

# 3.1 Generate multivariate data
np.random.seed(42)
n_samples = 200
n_features = 4

# Create correlated features
mean = [0, 0, 0, 0]
cov = [[1.0, 0.8, 0.6, 0.3],
       [0.8, 1.0, 0.7, 0.4],
       [0.6, 0.7, 1.0, 0.5],
       [0.3, 0.4, 0.5, 1.0]]
X_mv = np.random.multivariate_normal(mean, cov, n_samples)

# Add group labels for MANOVA
group1 = X_mv[:100] + np.array([0.5, 0.3, 0.2, 0.1])
group2 = X_mv[100:] + np.array([-0.3, -0.2, -0.1, 0.0])
X_manova = np.vstack([group1, group2])
groups = np.array([0]*100 + [1]*100)

print(f"\n1. Data Shape: {X_mv.shape}")
print(f"   Features: {n_features}, Samples: {n_samples}")

# 3.2 Principal Component Analysis
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_mv)

pca = PCA()
X_pca = pca.fit_transform(X_scaled)

print(f"\n2. Principal Component Analysis:")
print(f"   Explained variance ratio: {pca.explained_variance_ratio_}")
print(f"   Cumulative variance: {np.cumsum(pca.explained_variance_ratio_)}")
print(f"   First 2 PCs explain: {np.sum(pca.explained_variance_ratio_[:2]):.2%} of variance")

# 3.3 MANOVA (Multivariate Analysis of Variance)
# Using Wilks' Lambda test statistic
def manova_wilks_lambda(X, groups):
    """Simplified MANOVA using Wilks' Lambda"""
    unique_groups = np.unique(groups)
    n = len(X)
    p = X.shape[1]
    
    # Total sum of squares and cross products (T)
    grand_mean = X.mean(axis=0)
    T = (X - grand_mean).T @ (X - grand_mean)
    
    # Within-group sum of squares and cross products (W)
    W = np.zeros((p, p))
    for g in unique_groups:
        X_g = X[groups == g]
        group_mean = X_g.mean(axis=0)
        W += (X_g - group_mean).T @ (X_g - group_mean)
    
    # Between-group sum of squares (B = T - W)
    B = T - W
    
    # Wilks' Lambda
    wilks = np.linalg.det(W) / np.linalg.det(T)
    
    # Approximate F-test
    k = len(unique_groups)
    df1 = p
    df2 = n - k - p + 1
    f_stat = ((1 - wilks) / wilks) * (df2 / df1)
    p_value = 1 - stats.f.cdf(f_stat, df1, df2)
    
    return wilks, f_stat, p_value

wilks, f_stat, p_value = manova_wilks_lambda(X_manova, groups)

print(f"\n3. MANOVA (Wilks' Lambda):")
print(f"   Wilks' Λ: {wilks:.4f}")
print(f"   F-statistic: {f_stat:.4f}")
print(f"   p-value: {p_value:.4e}")
print(f"   Result: {'Significant difference' if p_value < 0.05 else 'No significant difference'} between groups")

# 3.4 Factor Analysis
fa = FactorAnalysis(n_components=2, random_state=42)
X_fa = fa.fit_transform(X_scaled)

print(f"\n4. Factor Analysis:")
print(f"   Number of factors: 2")
print(f"   Noise variance: {fa.noise_variance_}")

# Visualizations
fig, axes = plt.subplots(2, 3, figsize=(16, 10))

# Plot 1: Correlation matrix
im = axes[0, 0].imshow(np.corrcoef(X_mv.T), cmap='coolwarm', vmin=-1, vmax=1, aspect='auto')
axes[0, 0].set_xticks(range(n_features))
axes[0, 0].set_yticks(range(n_features))
axes[0, 0].set_xticklabels([f'Var{i+1}' for i in range(n_features)])
axes[0, 0].set_yticklabels([f'Var{i+1}' for i in range(n_features)])
axes[0, 0].set_title('Correlation Matrix')
plt.colorbar(im, ax=axes[0, 0])
for i in range(n_features):
    for j in range(n_features):
        text = axes[0, 0].text(j, i, f'{np.corrcoef(X_mv.T)[i, j]:.2f}',
                              ha="center", va="center", color="black", fontsize=9)

# Plot 2: Scree plot
axes[0, 1].bar(range(1, len(pca.explained_variance_ratio_)+1), 
               pca.explained_variance_ratio_, alpha=0.6, color='blue', label='Individual')
axes[0, 1].plot(range(1, len(pca.explained_variance_ratio_)+1), 
                np.cumsum(pca.explained_variance_ratio_), 'ro-', 
                linewidth=2, markersize=8, label='Cumulative')
axes[0, 1].axhline(y=0.8, color='green', linestyle='--', label='80% threshold')
axes[0, 1].set_xlabel('Principal Component')
axes[0, 1].set_ylabel('Explained Variance Ratio')
axes[0, 1].set_title('PCA Scree Plot')
axes[0, 1].legend()
axes[0, 1].grid(alpha=0.3, axis='y')

# Plot 3: PCA biplot (first 2 components)
axes[0, 2].scatter(X_pca[:, 0], X_pca[:, 1], alpha=0.6, s=40)
# Add loading vectors
loadings = pca.components_.T * np.sqrt(pca.explained_variance_)
for i, (x, y) in enumerate(loadings[:2].T):
    axes[0, 2].arrow(0, 0, x*3, y*3, head_width=0.1, head_length=0.1, 
                     fc='red', ec='red', linewidth=2, alpha=0.7)
    axes[0, 2].text(x*3.3, y*3.3, f'Var{i+1}', fontsize=10, color='red', 
                    fontweight='bold')
axes[0, 2].set_xlabel(f'PC1 ({pca.explained_variance_ratio_[0]:.1%})')
axes[0, 2].set_ylabel(f'PC2 ({pca.explained_variance_ratio_[1]:.1%})')
axes[0, 2].set_title('PCA Biplot')
axes[0, 2].grid(alpha=0.3)
axes[0, 2].axhline(y=0, color='k', linestyle='-', linewidth=0.5, alpha=0.3)
axes[0, 2].axvline(x=0, color='k', linestyle='-', linewidth=0.5, alpha=0.3)

# Plot 4: MANOVA group comparison
for feature_idx in range(min(2, n_features)):
    axes[1, 0].violinplot([X_manova[groups==0, feature_idx], 
                           X_manova[groups==1, feature_idx]], 
                          positions=[feature_idx*2, feature_idx*2+0.8],
                          widths=0.6, showmeans=True)
axes[1, 0].set_xticks([0.4, 2.4])
axes[1, 0].set_xticklabels(['Var1', 'Var2'])
axes[1, 0].set_ylabel('Value')
axes[1, 0].set_title(f'MANOVA: Group Comparison (Wilks Λ={wilks:.3f}, p={p_value:.4f})')
axes[1, 0].grid(alpha=0.3, axis='y')

# Plot 5: PCA vs original space
pca_2d = PCA(n_components=2)
X_pca_2d = pca_2d.fit_transform(X_scaled)
axes[1, 1].scatter(X_scaled[:, 0], X_scaled[:, 1], alpha=0.5, s=40, label='Original space (Var1 vs Var2)')
axes[1, 1].set_xlabel('Var1 (standardized)')
axes[1, 1].set_ylabel('Var2 (standardized)')
axes[1, 1].set_title('Original Feature Space')
axes[1, 1].legend()
axes[1, 1].grid(alpha=0.3)

# Plot 6: Factor Analysis loading heatmap
loadings_fa = fa.components_.T
im2 = axes[1, 2].imshow(loadings_fa, cmap='RdBu_r', vmin=-1, vmax=1, aspect='auto')
axes[1, 2].set_xticks([0, 1])
axes[1, 2].set_yticks(range(n_features))
axes[1, 2].set_xticklabels(['Factor 1', 'Factor 2'])
axes[1, 2].set_yticklabels([f'Var{i+1}' for i in range(n_features)])
axes[1, 2].set_title('Factor Analysis Loadings')
plt.colorbar(im2, ax=axes[1, 2])
for i in range(n_features):
    for j in range(2):
        text = axes[1, 2].text(j, i, f'{loadings_fa[i, j]:.2f}',
                              ha="center", va="center", color="black", fontsize=9)

plt.tight_layout()
plt.show()

print("\n✓ Multivariate analysis demonstrations complete!")
```

```python
## 2. Bayesian Inference: Priors, Posteriors, and MCMC

print("="*60)
print("BAYESIAN INFERENCE DEMONSTRATIONS")
print("="*60)

# 2.1 Bayesian Parameter Estimation (Coin Flip Example)
# Prior: Beta(2, 2) - slightly biased toward fairness
# Likelihood: Binomial
# Posterior: Beta(α + successes, β + failures)

n_flips = 100
n_heads = 65  # Observed heads

# Prior parameters
alpha_prior, beta_prior = 2, 2

# Posterior parameters (conjugate prior)
alpha_post = alpha_prior + n_heads
beta_post = beta_prior + (n_flips - n_heads)

# Create probability grid
theta = np.linspace(0, 1, 1000)
prior = stats.beta.pdf(theta, alpha_prior, beta_prior)
likelihood = stats.binom.pmf(n_heads, n_flips, theta)
posterior = stats.beta.pdf(theta, alpha_post, beta_post)

print(f"\n1. Bayesian Coin Flip Analysis:")
print(f"   Observed: {n_heads} heads in {n_flips} flips")
print(f"   Prior: Beta({alpha_prior}, {beta_prior})")
print(f"   Posterior: Beta({alpha_post}, {beta_post})")
print(f"   Posterior mean: {alpha_post/(alpha_post + beta_post):.4f}")
print(f"   95% Credible Interval: [{stats.beta.ppf(0.025, alpha_post, beta_post):.4f}, "
      f"{stats.beta.ppf(0.975, alpha_post, beta_post):.4f}]")

# 2.2 Bayesian Linear Regression
np.random.seed(42)
X_true = np.linspace(0, 10, 50)
y_true = 2.5 * X_true + 1.5 + np.random.normal(0, 2, 50)

# Bayesian regression with conjugate priors
X_design = np.c_[np.ones(len(X_true)), X_true]

# Prior: β ~ N(0, λI)
lambda_prior = 1.0
prior_precision = lambda_prior * np.eye(2)

# Likelihood precision (assuming known noise variance)
noise_precision = 1.0 / 4.0

# Posterior (conjugate normal-normal)
posterior_precision = prior_precision + noise_precision * X_design.T @ X_design
posterior_cov = np.linalg.inv(posterior_precision)
posterior_mean = noise_precision * posterior_cov @ X_design.T @ y_true

print(f"\n2. Bayesian Linear Regression:")
print(f"   Posterior β₀ (intercept): {posterior_mean[0]:.4f} ± {np.sqrt(posterior_cov[0,0]):.4f}")
print(f"   Posterior β₁ (slope): {posterior_mean[1]:.4f} ± {np.sqrt(posterior_cov[1,1]):.4f}")
print(f"   True parameters: β₀=1.5, β₁=2.5")

# 2.3 Simple MCMC: Metropolis-Hastings for Normal mean
def log_posterior(mu, data, prior_mean=0, prior_std=10):
    """Log posterior for normal mean with normal prior"""
    log_prior = stats.norm.logpdf(mu, prior_mean, prior_std)
    log_likelihood = np.sum(stats.norm.logpdf(data, mu, 1))
    return log_prior + log_likelihood

# Generate data
true_mu = 5.0
data = np.random.normal(true_mu, 1, 100)

# Metropolis-Hastings
n_samples = 5000
samples = np.zeros(n_samples)
current = 0.0  # Initial value

for i in range(n_samples):
    # Propose new value
    proposal = current + np.random.normal(0, 0.5)
    
    # Acceptance ratio
    log_accept_ratio = log_posterior(proposal, data) - log_posterior(current, data)
    
    if np.log(np.random.rand()) < log_accept_ratio:
        current = proposal  # Accept
    
    samples[i] = current

# Burn-in
burn_in = 1000
samples_burned = samples[burn_in:]

print(f"\n3. MCMC (Metropolis-Hastings) for Normal Mean:")
print(f"   True mean: {true_mu:.4f}")
print(f"   Sample mean: {data.mean():.4f}")
print(f"   MCMC posterior mean: {samples_burned.mean():.4f}")
print(f"   MCMC posterior std: {samples_burned.std():.4f}")
print(f"   Acceptance rate: {len(np.unique(samples)) / len(samples):.2%}")

# Visualizations
fig, axes = plt.subplots(2, 3, figsize=(16, 10))

# Plot 1: Prior, Likelihood, Posterior for coin flip
axes[0, 0].plot(theta, prior, 'b-', label='Prior', linewidth=2)
axes[0, 0].plot(theta, likelihood / likelihood.max() * prior.max(), 'g-', 
                label='Likelihood (scaled)', linewidth=2)
axes[0, 0].plot(theta, posterior, 'r-', label='Posterior', linewidth=2)
axes[0, 0].axvline(n_heads/n_flips, color='black', linestyle='--', 
                   label=f'MLE: {n_heads/n_flips:.3f}')
axes[0, 0].fill_between(theta, 0, posterior, alpha=0.3, color='red')
axes[0, 0].set_xlabel('θ (probability of heads)')
axes[0, 0].set_ylabel('Density')
axes[0, 0].set_title('Bayesian Inference: Coin Flip')
axes[0, 0].legend()
axes[0, 0].grid(alpha=0.3)

# Plot 2: Posterior predictive for coin flip
axes[0, 1].hist(stats.beta.rvs(alpha_post, beta_post, size=10000), bins=50, 
                density=True, alpha=0.6, color='red', label='Posterior samples')
axes[0, 1].axvline(alpha_post/(alpha_post + beta_post), color='red', 
                   linestyle='--', linewidth=2, label='Posterior mean')
axes[0, 1].axvline(n_heads/n_flips, color='black', linestyle=':', 
                   linewidth=2, label='Observed proportion')
axes[0, 1].set_xlabel('θ')
axes[0, 1].set_ylabel('Density')
axes[0, 1].set_title('Posterior Distribution Samples')
axes[0, 1].legend()
axes[0, 1].grid(alpha=0.3)

# Plot 3: Bayesian Linear Regression
axes[0, 2].scatter(X_true, y_true, alpha=0.5, s=30, label='Data')
X_pred = np.linspace(0, 10, 100)
# Sample from posterior predictive
n_posterior_samples = 100
for _ in range(n_posterior_samples):
    beta_sample = np.random.multivariate_normal(posterior_mean, posterior_cov)
    y_pred = beta_sample[0] + beta_sample[1] * X_pred
    axes[0, 2].plot(X_pred, y_pred, 'r-', alpha=0.02)
# Posterior mean prediction
y_mean = posterior_mean[0] + posterior_mean[1] * X_pred
axes[0, 2].plot(X_pred, y_mean, 'b-', linewidth=2, label='Posterior mean')
axes[0, 2].set_xlabel('X')
axes[0, 2].set_ylabel('y')
axes[0, 2].set_title('Bayesian Linear Regression')
axes[0, 2].legend()
axes[0, 2].grid(alpha=0.3)

# Plot 4: MCMC Trace Plot
axes[1, 0].plot(samples, alpha=0.6, linewidth=0.5)
axes[1, 0].axhline(true_mu, color='red', linestyle='--', linewidth=2, label='True mean')
axes[1, 0].axvline(burn_in, color='green', linestyle='--', linewidth=2, label='Burn-in')
axes[1, 0].set_xlabel('Iteration')
axes[1, 0].set_ylabel('μ')
axes[1, 0].set_title('MCMC Trace Plot')
axes[1, 0].legend()
axes[1, 0].grid(alpha=0.3)

# Plot 5: MCMC Histogram (after burn-in)
axes[1, 1].hist(samples_burned, bins=50, density=True, alpha=0.6, color='blue', label='MCMC samples')
axes[1, 1].axvline(samples_burned.mean(), color='blue', linestyle='--', 
                   linewidth=2, label=f'MCMC mean: {samples_burned.mean():.3f}')
axes[1, 1].axvline(true_mu, color='red', linestyle='--', 
                   linewidth=2, label=f'True mean: {true_mu:.3f}')
# Overlay theoretical posterior
theoretical_post_mean = data.mean()
theoretical_post_std = 1 / np.sqrt(len(data))
x_range = np.linspace(samples_burned.min(), samples_burned.max(), 100)
axes[1, 1].plot(x_range, stats.norm.pdf(x_range, theoretical_post_mean, theoretical_post_std),
                'g-', linewidth=2, label='Theoretical posterior')
axes[1, 1].set_xlabel('μ')
axes[1, 1].set_ylabel('Density')
axes[1, 1].set_title('MCMC Posterior Distribution')
axes[1, 1].legend()
axes[1, 1].grid(alpha=0.3)

# Plot 6: Autocorrelation of MCMC samples
from statsmodels.graphics.tsaplots import plot_acf
plot_acf(samples_burned, lags=50, ax=axes[1, 2], alpha=0.05)
axes[1, 2].set_xlabel('Lag')
axes[1, 2].set_ylabel('Autocorrelation')
axes[1, 2].set_title('MCMC Autocorrelation')
axes[1, 2].grid(alpha=0.3)

plt.tight_layout()
plt.show()

print("\n✓ Bayesian inference demonstrations complete!")
```

```python
## 1. Hypothesis Testing: Classical Statistical Inference

# Generate sample data for various tests
np.random.seed(42)
group1 = np.random.normal(100, 15, 50)  # Control group
group2 = np.random.normal(108, 15, 50)  # Treatment group
group3 = np.random.normal(105, 15, 50)  # Another treatment

# 1.1 Independent t-test
t_stat, p_value_t = stats.ttest_ind(group1, group2)
print("="*60)
print("HYPOTHESIS TESTING DEMONSTRATIONS")
print("="*60)
print(f"\n1. Independent t-test:")
print(f"   Null hypothesis: μ1 = μ2")
print(f"   t-statistic: {t_stat:.4f}")
print(f"   p-value: {p_value_t:.4f}")
print(f"   Result: {'Reject H0' if p_value_t < 0.05 else 'Fail to reject H0'} (α=0.05)")

# 1.2 One-way ANOVA
f_stat, p_value_anova = stats.f_oneway(group1, group2, group3)
print(f"\n2. One-way ANOVA (3 groups):")
print(f"   Null hypothesis: μ1 = μ2 = μ3")
print(f"   F-statistic: {f_stat:.4f}")
print(f"   p-value: {p_value_anova:.4f}")
print(f"   Result: {'Reject H0' if p_value_anova < 0.05 else 'Fail to reject H0'}")

# 1.3 Chi-square test for independence
observed = np.array([[30, 20], [15, 35]])
chi2_stat, p_value_chi2, dof, expected = stats.chi2_contingency(observed)
print(f"\n3. Chi-square test of independence:")
print(f"   Contingency table:\n{observed}")
print(f"   χ² statistic: {chi2_stat:.4f}")
print(f"   p-value: {p_value_chi2:.4f}")
print(f"   Degrees of freedom: {dof}")

# 1.4 Multiple testing correction
n_tests = 20
p_values = np.random.uniform(0, 0.1, n_tests)  # Simulated p-values
reject_bonf, pvals_bonf, _, _ = multipletests(p_values, alpha=0.05, method='bonferroni')
reject_bh, pvals_bh, _, _ = multipletests(p_values, alpha=0.05, method='fdr_bh')

print(f"\n4. Multiple testing correction ({n_tests} tests):")
print(f"   Uncorrected rejections: {np.sum(p_values < 0.05)}")
print(f"   Bonferroni rejections: {np.sum(reject_bonf)}")
print(f"   Benjamini-Hochberg (FDR) rejections: {np.sum(reject_bh)}")

# Visualizations
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Plot 1: Distribution comparison (t-test)
axes[0, 0].hist(group1, bins=20, alpha=0.6, label='Group 1 (Control)', color='blue')
axes[0, 0].hist(group2, bins=20, alpha=0.6, label='Group 2 (Treatment)', color='red')
axes[0, 0].axvline(group1.mean(), color='blue', linestyle='--', linewidth=2, label=f'Mean 1: {group1.mean():.2f}')
axes[0, 0].axvline(group2.mean(), color='red', linestyle='--', linewidth=2, label=f'Mean 2: {group2.mean():.2f}')
axes[0, 0].set_xlabel('Value')
axes[0, 0].set_ylabel('Frequency')
axes[0, 0].set_title(f'Independent t-test (p={p_value_t:.4f})')
axes[0, 0].legend()
axes[0, 0].grid(alpha=0.3)

# Plot 2: ANOVA visualization
bp = axes[0, 1].boxplot([group1, group2, group3], labels=['Group 1', 'Group 2', 'Group 3'],
                         patch_artist=True, showmeans=True)
for patch, color in zip(bp['boxes'], ['lightblue', 'lightcoral', 'lightgreen']):
    patch.set_facecolor(color)
axes[0, 1].set_ylabel('Value')
axes[0, 1].set_title(f'One-way ANOVA (F={f_stat:.2f}, p={p_value_anova:.4f})')
axes[0, 1].grid(alpha=0.3, axis='y')

# Plot 3: Chi-square expected vs observed
x = np.arange(len(observed.flatten()))
width = 0.35
axes[1, 0].bar(x - width/2, observed.flatten(), width, label='Observed', alpha=0.8)
axes[1, 0].bar(x + width/2, expected.flatten(), width, label='Expected', alpha=0.8)
axes[1, 0].set_xlabel('Cell')
axes[1, 0].set_ylabel('Frequency')
axes[1, 0].set_title(f'Chi-square Test (χ²={chi2_stat:.2f}, p={p_value_chi2:.4f})')
axes[1, 0].legend()
axes[1, 0].grid(alpha=0.3, axis='y')

# Plot 4: Multiple testing correction
sorted_idx = np.argsort(p_values)
axes[1, 1].scatter(range(n_tests), p_values[sorted_idx], c='gray', s=50, alpha=0.6, label='Original p-values')
axes[1, 1].scatter(np.where(reject_bonf[sorted_idx])[0], p_values[sorted_idx][reject_bonf[sorted_idx]], 
                   c='red', s=100, marker='x', label='Bonferroni reject')
axes[1, 1].scatter(np.where(reject_bh[sorted_idx])[0], p_values[sorted_idx][reject_bh[sorted_idx]], 
                   c='green', s=80, marker='^', alpha=0.7, label='BH-FDR reject')
axes[1, 1].axhline(y=0.05, color='black', linestyle='--', label='α=0.05')
axes[1, 1].set_xlabel('Test (sorted by p-value)')
axes[1, 1].set_ylabel('p-value')
axes[1, 1].set_title('Multiple Testing Correction')
axes[1, 1].legend()
axes[1, 1].grid(alpha=0.3)

plt.tight_layout()
plt.show()

print("\n✓ Hypothesis testing demonstrations complete!")
```

### Mathematical Framework and Statistical Spaces

**Definition 1.1** (Statistical Model): A **statistical model** is a family of probability distributions $\mathcal{P} = \{P_\theta : \theta \in \Theta\}$ on a measurable space $(\mathcal{X}, \mathcal{F})$, where:
- $\mathcal{X}$ is the sample space
- $\mathcal{F}$ is a σ-algebra on $\mathcal{X}$
- $\Theta$ is the parameter space
- Each $P_\theta$ is a probability measure

**Definition 1.2** (Sufficient Statistic): A statistic $T(X)$ is **sufficient** for parameter $\theta$ if the conditional distribution of $X$ given $T(X)$ does not depend on $\theta$.

**Factorization Theorem**: $T(X)$ is sufficient for $\theta$ if and only if the likelihood can be factored as:
$$L(\theta; x) = g(T(x), \theta) \cdot h(x)$$
where $g$ depends on $x$ only through $T(x)$ and $h$ does not depend on $\theta$.

**Definition 1.3** (Exponential Family): A family of distributions has the form:
$$f(x|\theta) = h(x) \exp\left(\sum_{i=1}^k \eta_i(\theta) T_i(x) - A(\theta)\right)$$
where:
- $T(x) = (T_1(x), \ldots, T_k(x))$ is the sufficient statistic
- $\eta(\theta) = (\eta_1(\theta), \ldots, \eta_k(\theta))$ is the natural parameter
- $A(\theta)$ is the log-partition function

**Properties of Exponential Families**:
1. **Sufficient Statistics**: $T(x)$ is sufficient for $\theta$
2. **Conjugate Priors**: Natural priors for Bayesian inference
3. **Maximum Likelihood**: Efficient estimation procedures
4. **Asymptotic Properties**: Regular asymptotic behavior

### Information Theory and Statistical Information

**Definition 1.4** (Fisher Information): For parameter $\theta$, the Fisher information is:
$$I(\theta) = \mathbb{E}\left[\left(\frac{\partial \log f(X|\theta)}{\partial \theta}\right)^2\right] = -\mathbb{E}\left[\frac{\partial^2 \log f(X|\theta)}{\partial \theta^2}\right]$$

**Cramér-Rao Lower Bound**: For unbiased estimator $\hat{\theta}$:
$$\text{Var}(\hat{\theta}) \geq \frac{1}{nI(\theta)}$$

**Definition 1.5** (Kullback-Leibler Divergence): For distributions $P$ and $Q$:
$$D_{KL}(P || Q) = \int p(x) \log \frac{p(x)}{q(x)} dx$$

**Properties**:
- $D_{KL}(P || Q) \geq 0$ with equality iff $P = Q$
- Generally not symmetric: $D_{KL}(P || Q) \neq D_{KL}(Q || P)$
- Not a metric (triangle inequality fails)

### Asymptotic Theory and Large Sample Properties

**Definition 1.6** (Consistency): An estimator sequence $\{\hat{\theta}_n\}$ is **consistent** if:
$$\hat{\theta}_n \xrightarrow{P} \theta_0 \text{ as } n \to \infty$$

**Definition 1.7** (Asymptotic Normality): An estimator $\hat{\theta}_n$ is **asymptotically normal** if:
$$\sqrt{n}(\hat{\theta}_n - \theta_0) \xrightarrow{d} N(0, \sigma^2)$$

**Central Limit Theorem for Statistics**: Under regularity conditions:
$$\sqrt{n}(\hat{\theta}_{MLE} - \theta_0) \xrightarrow{d} N(0, I(\theta_0)^{-1})$$

**Delta Method**: If $\sqrt{n}(\hat{\theta}_n - \theta_0) \xrightarrow{d} N(0, \sigma^2)$ and $g$ is differentiable at $\theta_0$:
$$\sqrt{n}(g(\hat{\theta}_n) - g(\theta_0)) \xrightarrow{d} N(0, \sigma^2 [g'(\theta_0)]^2)$$

### Decision Theory and Risk Functions

**Definition 1.8** (Decision Problem): A statistical decision problem consists of:
- **Sample Space**: $\mathcal{X}$ with probability measures $\{P_\theta : \theta \in \Theta\}$
- **Action Space**: $\mathcal{A}$ of possible actions
- **Loss Function**: $L: \Theta \times \mathcal{A} \to \mathbb{R}$
- **Decision Rule**: $\delta: \mathcal{X} \to \mathcal{A}$

**Risk Function**: $R(\theta, \delta) = \mathbb{E}_\theta[L(\theta, \delta(X))]$

**Bayes Risk**: For prior $\pi$ on $\Theta$:
$$r(\pi, \delta) = \int R(\theta, \delta) \pi(\theta) d\theta$$

**Minimax Risk**: $R^* = \inf_\delta \sup_\theta R(\theta, \delta)$

**Admissibility**: Decision rule $\delta_1$ **dominates** $\delta_2$ if $R(\theta, \delta_1) \leq R(\theta, \delta_2)$ for all $\theta$ with strict inequality for some $\theta$.

### High-Dimensional Statistics

**Curse of Dimensionality**: In $d$ dimensions, the volume of unit sphere is concentrated near the surface for large $d$:
$$\frac{\text{Volume of shell }(1-\epsilon, 1)}{\text{Volume of unit ball}} = 1 - (1-\epsilon)^d \to 1 \text{ as } d \to \infty$$

**Concentration of Measure**: High-dimensional phenomena:
1. **Distances become uniform**: All pairwise distances approximately equal
2. **Volume concentration**: Most volume near boundary
3. **Orthogonality**: Random vectors tend to be orthogonal

**High-Dimensional Asymptotics**: When $p/n \to c \in (0, \infty)$ as $n, p \to \infty$:

```python
## 2. Bayesian Statistics and Inference {#bayesian}

Bayesian statistics provides a coherent framework for uncertainty quantification and inference, treating parameters as random variables and updating beliefs through Bayes' theorem. This approach is fundamental to modern machine learning and AI systems that require principled uncertainty quantification.

### Foundational Theory

**Bayes' Theorem**: For parameters $\theta$ and data $x$:
$$p(\theta|x) = \frac{p(x|\theta)p(\theta)}{p(x)} = \frac{p(x|\theta)p(\theta)}{\int p(x|\theta)p(\theta) d\theta}$$

**Components**:
- **Prior**: $p(\theta)$ represents prior beliefs about parameters
- **Likelihood**: $p(x|\theta)$ describes data generation process
- **Posterior**: $p(\theta|x)$ combines prior beliefs with observed data
- **Evidence**: $p(x)$ normalizes the posterior distribution

### Conjugate Priors and Exponential Families

**Definition 2.1** (Conjugate Prior): Prior $p(\theta)$ is **conjugate** to likelihood $p(x|\theta)$ if the posterior $p(\theta|x)$ belongs to the same family as the prior.

**Exponential Family Conjugacy**: For exponential family likelihood:
$$p(x|\theta) = h(x) \exp(\eta(\theta)^T T(x) - A(\theta))$$

**Conjugate Prior**:
$$p(\theta) \propto \exp(\nu^T \eta(\theta) - \tau A(\theta))$$

**Posterior**:
$$p(\theta|x) \propto \exp((\nu + T(x))^T \eta(\theta) - (\tau + 1) A(\theta))$$

**Common Conjugate Pairs**:
1. **Beta-Binomial**: $\text{Beta}(\alpha, \beta)$ prior for $\text{Binomial}(n, p)$ likelihood
2. **Gamma-Poisson**: $\text{Gamma}(\alpha, \beta)$ prior for $\text{Poisson}(\lambda)$ likelihood
3. **Normal-Normal**: $N(\mu_0, \sigma_0^2)$ prior for $N(\mu, \sigma^2)$ likelihood (known variance)
4. **Inverse-Gamma-Normal**: For unknown variance in normal model

### Bayesian Decision Theory

**Definition 2.2** (Bayesian Decision Rule): Choose action $a$ that minimizes posterior expected loss:
$$\delta^*(x) = \arg\min_a \int L(\theta, a) p(\theta|x) d\theta$$

**Loss Functions**:
- **Squared Error Loss**: $L(\theta, a) = (\theta - a)^2$ gives posterior mean
- **Absolute Error Loss**: $L(\theta, a) = |\theta - a|$ gives posterior median
- **0-1 Loss**: $L(\theta, a) = \mathbf{1}_{\theta \neq a}$ gives posterior mode

**Credible Intervals**: $(1-\alpha)$ credible interval $[L, U]$ satisfies:
$$P(L \leq \theta \leq U | x) = 1 - \alpha$$

**Highest Posterior Density (HPD)**: Credible interval with shortest length.

### Computational Bayesian Methods

**Markov Chain Monte Carlo (MCMC)**:

**Metropolis-Hastings Algorithm**:
1. Given current state $\theta^{(t)}$, propose $\theta^* \sim q(\cdot | \theta^{(t)})$
2. Compute acceptance probability:
   $$\alpha = \min\left(1, \frac{p(\theta^*|x) q(\theta^{(t)} | \theta^*)}{p(\theta^{(t)}|x) q(\theta^* | \theta^{(t)})}\right)$$
3. Accept $\theta^{(t+1)} = \theta^*$ with probability $\alpha$, otherwise $\theta^{(t+1)} = \theta^{(t)}$

**Gibbs Sampling**: For multivariate parameter $\theta = (\theta_1, \ldots, \theta_k)$:
1. Sample $\theta_1^{(t+1)} \sim p(\theta_1 | \theta_2^{(t)}, \ldots, \theta_k^{(t)}, x)$
2. Sample $\theta_2^{(t+1)} \sim p(\theta_2 | \theta_1^{(t+1)}, \theta_3^{(t)}, \ldots, \theta_k^{(t)}, x)$
3. Continue for all components

**Hamiltonian Monte Carlo**: Uses gradient information for efficient sampling:
$$H(\theta, r) = U(\theta) + K(r) = -\log p(\theta|x) + \frac{1}{2}r^T M^{-1} r$$

### Variational Inference

**Definition 2.3** (Variational Inference): Approximate intractable posterior $p(\theta|x)$ with tractable distribution $q(\theta)$ by minimizing KL divergence:
$$\min_q D_{KL}(q(\theta) || p(\theta|x))$$

**Evidence Lower Bound (ELBO)**:
$$\log p(x) = \mathbb{E}_q[\log p(x, \theta)] - \mathbb{E}_q[\log q(\theta)] + D_{KL}(q(\theta) || p(\theta|x))$$
$$\geq \mathbb{E}_q[\log p(x, \theta)] - \mathbb{E}_q[\log q(\theta)] =: \mathcal{L}(q)$$

**Mean Field Variational Inference**: Assume factorization $q(\theta) = \prod_i q_i(\theta_i)$

**Coordinate Ascent Updates**:
$$\log q_j^*(\theta_j) = \mathbb{E}_{-j}[\log p(x, \theta)] + \text{constant}$$

### Bayesian Model Selection and Comparison

**Marginal Likelihood**: $p(x|M_k) = \int p(x|\theta_k, M_k) p(\theta_k|M_k) d\theta_k$

**Bayes Factor**: For models $M_1$ and $M_2$:
$$BF_{12} = \frac{p(x|M_1)}{p(x|M_2)}$$

**Interpretation**:
- $BF_{12} > 10$: Strong evidence for $M_1$
- $3 < BF_{12} < 10$: Moderate evidence for $M_1$
- $1 < BF_{12} < 3$: Weak evidence for $M_1$

**Information Criteria**:
- **AIC**: $-2 \log p(x|\hat{\theta}) + 2k$
- **BIC**: $-2 \log p(x|\hat{\theta}) + k \log n$
- **DIC**: $-2 \log p(x|\bar{\theta}) + 2p_D$ where $p_D$ is effective number of parameters

### Hierarchical Bayesian Models

**Definition 2.4** (Hierarchical Model): Multi-level model with parameters having their own distributions:
$$\begin{align}
x_i | \theta_i &\sim p(x_i|\theta_i) \\
\theta_i | \phi &\sim p(\theta_i|\phi) \\
\phi &\sim p(\phi)
\end{align}$$

**Benefits**:
- **Borrowing Strength**: Information sharing across groups
- **Shrinkage**: Extreme estimates pulled toward group mean
- **Uncertainty Propagation**: Full accounting of parameter uncertainty

**Example - Random Effects Model**:
$$y_{ij} = \mu + \alpha_i + \epsilon_{ij}$$
where $\alpha_i \sim N(0, \sigma_\alpha^2)$ and $\epsilon_{ij} \sim N(0, \sigma^2)$

### Bayesian Nonparametrics

**Dirichlet Process**: Random probability measure $G \sim DP(\alpha, G_0)$ where:
- $\alpha > 0$ is concentration parameter
- $G_0$ is base measure

**Stick-Breaking Construction**:
$$\begin{align}
\beta_k &\sim \text{Beta}(1, \alpha) \\
\pi_k &= \beta_k \prod_{l=1}^{k-1} (1 - \beta_l) \\
\theta_k &\sim G_0 \\
G &= \sum_{k=1}^\infty \pi_k \delta_{\theta_k}
\end{align}$$

**Dirichlet Process Mixture Model**:
$$\begin{align}
G &\sim DP(\alpha, G_0) \\
\theta_i &\sim G \\
x_i &\sim F(\theta_i)
\end{align}$$

**Chinese Restaurant Process**: Clustering interpretation of DP mixture.

**Gaussian Processes**: Infinite-dimensional extension of multivariate normal:
$$f \sim GP(m(x), k(x, x'))$$
where $m(x)$ is mean function and $k(x, x')$ is covariance function.

### Bayesian Deep Learning

**Bayesian Neural Networks**: Place priors on network weights:
$$p(w) = \prod_l p(w_l)$$

**Posterior Inference**: Typically intractable, require approximations:
- **Variational Inference**: Assume factorized posterior
- **Monte Carlo Dropout**: Approximate Bayesian inference
- **Ensemble Methods**: Multiple models for uncertainty

**Epistemic vs Aleatoric Uncertainty**:
- **Epistemic**: Model uncertainty (reducible with more data)
- **Aleatoric**: Data uncertainty (irreducible noise)

## 3. Nonparametric Statistics and Kernel Methods {#nonparametric}

Nonparametric statistics provides distribution-free methods and flexible modeling approaches that make minimal assumptions about the underlying data distribution. These methods are fundamental to modern machine learning and provide robust alternatives to parametric approaches.

### Distribution-Free Methods

**Definition 3.1** (Distribution-Free): A statistical method is **distribution-free** if its validity does not depend on the specific form of the population distribution.

**Advantages**:
- Robust to distributional assumptions
- Applicable to ordinal data
- Exact finite-sample validity
- Less sensitive to outliers

**Sign Test**: For median $m$, test statistic is:
$$S = \sum_{i=1}^n \mathbf{1}_{X_i > m_0}$$
Under $H_0: m = m_0$, $S \sim \text{Binomial}(n, 1/2)$

**Wilcoxon Signed-Rank Test**: Uses both magnitude and sign of differences:
$$W^+ = \sum_{i=1}^n R_i^+ \mathbf{1}_{X_i > m_0}$$
where $R_i^+$ is rank of $|X_i - m_0|$.

**Mann-Whitney U Test**: For two-sample location problem:
$$U = \sum_{i=1}^{n_1} \sum_{j=1}^{n_2} \mathbf{1}_{X_i > Y_j}$$

### Kernel Density Estimation

**Definition 3.2** (Kernel Density Estimator): For data $X_1, \ldots, X_n$ and kernel $K$:
$$\hat{f}_h(x) = \frac{1}{nh} \sum_{i=1}^n K\left(\frac{x - X_i}{h}\right)$$

**Common Kernels**:
- **Gaussian**: $K(u) = \frac{1}{\sqrt{2\pi}} e^{-u^2/2}$
- **Epanechnikov**: $K(u) = \frac{3}{4}(1 - u^2) \mathbf{1}_{|u| \leq 1}$
- **Uniform**: $K(u) = \frac{1}{2} \mathbf{1}_{|u| \leq 1}$

**Asymptotic Properties**: Under regularity conditions:
$$\text{MSE}(\hat{f}_h(x)) = \frac{R(K)}{nh} + \frac{h^4}{4} \mu_2(K)^2 [f''(x)]^2 + o(h^4 + (nh)^{-1})$$

**Optimal Bandwidth**: Minimizes AMISE (Asymptotic Mean Integrated Squared Error):
$$h_{AMISE} = \left(\frac{R(K)}{\mu_2(K)^2 \int [f''(x)]^2 dx}\right)^{1/5} n^{-1/5}$$

**Bandwidth Selection Methods**:
- **Cross-Validation**: $CV(h) = \int \hat{f}_h(x)^2 dx - \frac{2}{n} \sum_{i=1}^n \hat{f}_{h,-i}(X_i)$
- **Plug-in Methods**: Estimate $\int [f''(x)]^2 dx$ from data
- **Sheather-Jones**: Sophisticated plug-in estimator

### Kernel Regression and Smoothing

**Nadaraya-Watson Estimator**:
$$\hat{m}_h(x) = \frac{\sum_{i=1}^n K\left(\frac{x - X_i}{h}\right) Y_i}{\sum_{i=1}^n K\left(\frac{x - X_i}{h}\right)}$$

**Local Linear Regression**: Minimize locally weighted least squares:
$$\min_{\alpha, \beta} \sum_{i=1}^n K\left(\frac{x - X_i}{h}\right) [Y_i - \alpha - \beta(X_i - x)]^2$$

**Advantages of Local Linear**:
- Automatic boundary correction
- Better bias properties
- Minimax optimal

**Local Polynomial Regression**: Extension to degree $p$:
$$\min_{\boldsymbol{\beta}} \sum_{i=1}^n K\left(\frac{x - X_i}{h}\right) \left[Y_i - \sum_{j=0}^p \beta_j (X_i - x)^j\right]^2$$

### Reproducing Kernel Hilbert Spaces

**Definition 3.3** (RKHS): Hilbert space $\mathcal{H}$ of functions on $\mathcal{X}$ with reproducing kernel $k: \mathcal{X} \times \mathcal{X} \to \mathbb{R}$ such that:
1. $k(\cdot, x) \in \mathcal{H}$ for all $x \in \mathcal{X}$
2. $\langle f, k(\cdot, x) \rangle_\mathcal{H} = f(x)$ for all $f \in \mathcal{H}$ (reproducing property)

**Moore-Aronszajn Theorem**: Every positive definite kernel defines a unique RKHS.

**Mercer's Theorem**: For positive definite kernel on compact domain:
$$k(x, x') = \sum_{i=1}^\infty \lambda_i \phi_i(x) \phi_i(x')$$
where $\lambda_i \geq 0$ and $\{\phi_i\}$ are orthonormal eigenfunctions.

**Common Kernels**:
- **Linear**: $k(x, x') = x^T x'$
- **Polynomial**: $k(x, x') = (x^T x' + c)^d$
- **RBF**: $k(x, x') = \exp(-\gamma ||x - x'||^2)$
- **Matérn**: $k(x, x') = \frac{2^{1-\nu}}{\Gamma(\nu)} \left(\sqrt{2\nu} \frac{||x - x'||}{\ell}\right)^\nu K_\nu\left(\sqrt{2\nu} \frac{||x - x'||}{\ell}\right)$

### Support Vector Machines

**Primal Problem**: For separable case:
$$\min_{w, b} \frac{1}{2} ||w||^2 \quad \text{s.t.} \quad y_i(w^T x_i + b) \geq 1$$

**Dual Problem**:
$$\max_\alpha \sum_{i=1}^n \alpha_i - \frac{1}{2} \sum_{i,j=1}^n \alpha_i \alpha_j y_i y_j x_i^T x_j$$
$$\text{s.t.} \quad \sum_{i=1}^n \alpha_i y_i = 0, \quad \alpha_i \geq 0$$

**Kernel Trick**: Replace $x_i^T x_j$ with $k(x_i, x_j)$:
$$f(x) = \sum_{i=1}^n \alpha_i y_i k(x_i, x) + b$$

**Soft Margin SVM**: For non-separable case with slack variables $\xi_i$:
$$\min_{w,b,\xi} \frac{1}{2}||w||^2 + C\sum_{i=1}^n \xi_i$$
$$\text{s.t.} \quad y_i(w^T x_i + b) \geq 1 - \xi_i, \quad \xi_i \geq 0$$

### Gaussian Process Regression

**Model**: $f \sim GP(m(x), k(x, x'))$ with observations $y_i = f(x_i) + \epsilon_i$

**Posterior Prediction**: For test point $x_*$:
$$\begin{align}
f_* | X, y, x_* &\sim N(\mu_*, \sigma_*^2) \\
\mu_* &= k_*^T (K + \sigma_n^2 I)^{-1} y \\
\sigma_*^2 &= k_{**} - k_*^T (K + \sigma_n^2 I)^{-1} k_*
\end{align}$$

where $k_* = [k(x_*, x_1), \ldots, k(x_*, x_n)]^T$ and $K_{ij} = k(x_i, x_j)$.

**Hyperparameter Learning**: Maximize marginal likelihood:
$$\log p(y|X, \theta) = -\frac{1}{2} y^T (K + \sigma_n^2 I)^{-1} y - \frac{1}{2} \log |K + \sigma_n^2 I| - \frac{n}{2} \log 2\pi$$

### Empirical Process Theory

**Definition 3.4** (Empirical Process): For function class $\mathcal{F}$ and empirical measure $\mathbb{P}_n$:
$$G_n(f) = \sqrt{n}(\mathbb{P}_n f - P f) = \frac{1}{\sqrt{n}} \sum_{i=1}^n [f(X_i) - P f]$$

**Donsker Property**: Class $\mathcal{F}$ is **Donsker** if $G_n \Rightarrow G$ in $\ell^\infty(\mathcal{F})$ where $G$ is Gaussian process.

**Glivenko-Cantelli Theorem**: For empirical distribution function:
$$\sup_x |\mathbb{F}_n(x) - F(x)| \to 0 \text{ a.s.}$$

**Dvoretzky-Kiefer-Wolfowitz Inequality**:
$$P\left(\sup_x |\mathbb{F}_n(x) - F(x)| > \epsilon\right) \leq 2e^{-2n\epsilon^2}$$

**Rademacher Complexity**: For function class $\mathcal{F}$:
$$\mathcal{R}_n(\mathcal{F}) = \mathbb{E}\left[\sup_{f \in \mathcal{F}} \frac{1}{n} \sum_{i=1}^n \sigma_i f(X_i)\right]$$
where $\sigma_i$ are Rademacher random variables.

### Wavelets and Multiresolution Analysis

**Definition 3.5** (Multiresolution Analysis): Sequence of nested subspaces $\{V_j\}$ satisfying:
1. $V_j \subset V_{j+1}$ for all $j$
2. $\bigcup_j V_j$ is dense in $L^2(\mathbb{R})$
3. $\bigcap_j V_j = \{0\}$
4. $f(x) \in V_j \iff f(2x) \in V_{j+1}$
5. There exists $\phi$ such that $\{\phi(x - k)\}_{k \in \mathbb{Z}}$ is orthonormal basis for $V_0$

**Wavelet Decomposition**:
$$f(x) = \sum_k c_{j_0,k} \phi_{j_0,k}(x) + \sum_{j \geq j_0} \sum_k d_{j,k} \psi_{j,k}(x)$$

**Wavelet Denoising**: For noisy observations $Y_i = f(i/n) + \epsilon_i$:
1. Transform to wavelet domain: $\mathbf{d} = W \mathbf{Y}$
2. Apply threshold: $\tilde{d}_j = \eta_\lambda(d_j)$
3. Inverse transform: $\hat{\mathbf{f}} = W^{-1} \tilde{\mathbf{d}}$

**Thresholding Rules**:
- **Hard**: $\eta_\lambda^H(x) = x \mathbf{1}_{|x| > \lambda}$
- **Soft**: $\eta_\lambda^S(x) = \text{sign}(x) (|x| - \lambda)_+$
```

            sigma_U = np.sqrt(n1 * n2 * (n1 + n2 + 1) / 12)

            p_value = 2 * stats.norm.cdf(z_score)
            
            return {'U1': U1, 'U2': U2, 'z_score': z_score, 'p_value': p_value}

            group_sizes = np.array([np.sum(groups == g) for g in np.unique(groups)])
            
            # Within-group sum of squares (W)
            W = np.zeros((p, p))
            for g in np.unique(groups):
                X_g = X[groups == g]
                X_g_centered = X_g - group_means[g]
                W += X_g_centered.T @ X_g_centered

                # Roy's largest root

            except:

            return {
                'wilks_lambda': wilks_lambda,
                'pillai_trace': pillai_trace,
                'hotelling_trace': hotelling_trace,

    def demonstrate_time_series_analysis(self):
        """
        Time series analysis: ARIMA, state-space models, spectral analysis
        """
        print("\n=== TIME SERIES ANALYSIS ===")
        
        # Generate synthetic time series
        n_obs = 300
        
        # AR(2) process: x_t = 0.6*x_{t-1} - 0.2*x_{t-2} + ε_t
        ar_coeffs = np.array([0.6, -0.2])
        noise_var = 1.0
        
        # Generate AR process
        x_ar = np.zeros(n_obs)
        noise = np.random.normal(0, np.sqrt(noise_var), n_obs)
        
        for t in range(2, n_obs):
            x_ar[t] = np.dot(ar_coeffs, x_ar[t-2:t][::-1]) + noise[t]
        
        # Add trend and seasonality
        t = np.arange(n_obs)
        trend = 0.02 * t
        seasonal = 2 * np.sin(2 * np.pi * t / 12) + np.sin(2 * np.pi * t / 6)
        x_full = x_ar + trend + seasonal
        
        # Time series decomposition
        def stl_decomposition(x, period=12):
            """Simplified STL decomposition"""
            n = len(x)
            
            # Seasonal component (simple moving average approach)
            seasonal = np.zeros(n)
            for i in range(n):
                # Find indices for same season
                season_indices = np.arange(i % period, n, period)
                seasonal[i] = np.median(x[season_indices])
            
            # Remove seasonality to get trend + residual
            deseasonalized = x - seasonal
            
            # Trend component (moving average)
            trend = np.convolve(deseasonalized, np.ones(13)/13, mode='same')
            
            # Residual
            residual = x - trend - seasonal
            
            return trend, seasonal, residual
        
        trend_est, seasonal_est, residual_est = stl_decomposition(x_full)
        
        # Autocorrelation and Partial Autocorrelation
        def autocorrelation(x, max_lag=20):
            """Compute sample autocorrelation function"""
            n = len(x)
            x_centered = x - np.mean(x)
            
            autocorr = np.zeros(max_lag + 1)
            for lag in range(max_lag + 1):
                if lag == 0:
                    autocorr[lag] = 1.0
                else:
                    c_lag = np.mean(x_centered[:-lag] * x_centered[lag:])
                    c_0 = np.mean(x_centered**2)
                    autocorr[lag] = c_lag / c_0
            
            return autocorr
        
        def partial_autocorr(x, max_lag=20):
            """Compute partial autocorrelation function using Yule-Walker equations"""
            autocorr = autocorrelation(x, max_lag)
            pacf = np.zeros(max_lag + 1)
            pacf[0] = 1.0
            
            for k in range(1, max_lag + 1):
                # Set up Yule-Walker equations
                R = linalg.toeplitz(autocorr[:k])
                r = autocorr[1:k+1]
                
                try:
                    phi = linalg.solve(R, r)
                    pacf[k] = phi[-1]
                except:
                    pacf[k] = 0
            
            return pacf
        
        acf = autocorrelation(residual_est)
        pacf = partial_autocorr(residual_est)
        
        # ARIMA model fitting
        def fit_ar_model(x, order):
            """Fit AR model using Yule-Walker equations"""
            acf_vals = autocorrelation(x, order)
            
            if order == 0:
                return np.array([]), np.var(x), x
            
            # Yule-Walker equations
            R = linalg.toeplitz(acf_vals[:order])
            r = acf_vals[1:order+1]
            
            try:
                phi = linalg.solve(R, r)
                
                # Compute residuals
                residuals = np.zeros(len(x))
                for t in range(order, len(x)):
                    prediction = np.dot(phi, x[t-order:t][::-1])
                    residuals[t] = x[t] - prediction
                
                # Estimate noise variance
                sigma2 = np.var(residuals[order:])
                
                return phi, sigma2, residuals
            except:
                return np.array([]), np.var(x), x - np.mean(x)
        
        # Fit AR models of different orders
        ar_models = {}
        max_order = 6
        
        for p in range(max_order + 1):
            phi, sigma2, residuals = fit_ar_model(residual_est, p)
            
            # AIC and BIC
            if len(phi) > 0:
                loglik = -0.5 * (len(residuals) - p) * (np.log(2*np.pi) + np.log(sigma2) + 1)
            else:
                loglik = -0.5 * len(residuals) * (np.log(2*np.pi) + np.log(sigma2) + 1)
            
            aic = -2 * loglik + 2 * p
            bic = -2 * loglik + np.log(len(residuals)) * p
            
            ar_models[p] = {
                'coefficients': phi,
                'noise_variance': sigma2,
                'residuals': residuals,
                'aic': aic,
                'bic': bic,
                'loglik': loglik
            }
        
        # Select best model
        best_order = min(ar_models.keys(), key=lambda p: ar_models[p]['aic'])
        
        # Spectral analysis
        def periodogram(x):
            """Compute periodogram"""
            n = len(x)
            x_centered = x - np.mean(x)
            
            # FFT
            X = np.fft.fft(x_centered)
            frequencies = np.fft.fftfreq(n)
            
            # Power spectral density
            psd = np.abs(X)**2 / n
            
            # Keep only positive frequencies
            pos_freq_idx = frequencies >= 0
            return frequencies[pos_freq_idx], psd[pos_freq_idx]
        
        freq, psd = periodogram(x_full)
        
        # State space model (local level model)
        def kalman_filter(y, Q, R, initial_state=None, initial_cov=None):
            """Kalman filter for local level model: y_t = μ_t + ε_t, μ_t = μ_{t-1} + η_t"""
            n = len(y)
            
            # Initialize
            if initial_state is None:
                mu = np.zeros(n + 1)
                mu[0] = y[0]
            else:
                mu = np.zeros(n + 1)
                mu[0] = initial_state
            
            if initial_cov is None:
                P = np.zeros(n + 1)
                P[0] = 1.0
            else:
                P = np.zeros(n + 1)
                P[0] = initial_cov
            
            # Storage for filtered estimates
            mu_filtered = np.zeros(n)
            P_filtered = np.zeros(n)
            loglik = 0
            
            for t in range(n):
                # Prediction step
                mu_pred = mu[t]  # State transition: μ_t|t-1 = μ_{t-1}
                P_pred = P[t] + Q  # P_t|t-1 = P_{t-1} + Q
                
                # Update step
                innovation = y[t] - mu_pred  # y_t - μ_t|t-1
                innovation_var = P_pred + R  # S_t = P_t|t-1 + R
                kalman_gain = P_pred / innovation_var  # K_t = P_t|t-1 / S_t
                
                mu[t + 1] = mu_pred + kalman_gain * innovation
                P[t + 1] = P_pred * (1 - kalman_gain)
                
                mu_filtered[t] = mu[t + 1]
                P_filtered[t] = P[t + 1]
                
                # Log-likelihood
                loglik += -0.5 * (np.log(2*np.pi) + np.log(innovation_var) + innovation**2 / innovation_var)
            
            return mu_filtered, P_filtered, loglik
        
        # Fit local level model with MLE
        def fit_local_level(y):
            def neg_loglik(params):
                Q, R = np.exp(params)  # Ensure positivity
                _, _, loglik = kalman_filter(y, Q, R)
                return -loglik
            
            # Initial guess
            initial_params = np.log([np.var(np.diff(y)), np.var(y)])
            
            result = optimize.minimize(neg_loglik, initial_params, method='BFGS')
            Q_opt, R_opt = np.exp(result.x)
            
            mu_filtered, P_filtered, loglik = kalman_filter(y, Q_opt, R_opt)
            
            return {
                'Q': Q_opt,
                'R': R_opt,
                'mu_filtered': mu_filtered,
                'P_filtered': P_filtered,
                'loglik': loglik,
                'aic': -2 * loglik + 4,  # 2 parameters
                'bic': -2 * loglik + np.log(len(y)) * 2
            }
        
        state_space_result = fit_local_level(x_full)
        
        # Forecasting
        def ar_forecast(x, phi, sigma2, h=10):
            """Forecast using AR model"""
            forecasts = np.zeros(h)
            forecast_vars = np.zeros(h)
            
            # Extend series with forecasts

            for i in range(h):
                t = len(x) + i
                if len(phi) > 0:
                    forecasts[i] = np.dot(phi, x_extended[t-len(phi):t][::-1])
                else:
                    forecasts[i] = np.mean(x)
                
                x_extended[t] = forecasts[i]
                
                # Forecast variance (simplified)
                forecast_vars[i] = sigma2 * (i + 1)
            
            return forecasts, forecast_vars
        
        best_phi = ar_models[best_order]['coefficients']
        best_sigma2 = ar_models[best_order]['noise_variance']
        forecasts, forecast_vars = ar_forecast(residual_est, best_phi, best_sigma2)
        
        results = {
            'original_series': x_full,
            'components': {
                'ar_component': x_ar,
                'trend': trend,
                'seasonal': seasonal,
                'noise': noise
            },
            'decomposition': {
                'trend': trend_est,
                'seasonal': seasonal_est,
                'residual': residual_est
            },
            'correlation_analysis': {
                'acf': acf,

            },
            'ar_models': ar_models,
            'best_ar_order': best_order,
            'spectral_analysis': {
                'frequencies': freq,
                'psd': psd
            },
            'state_space': state_space_result,
            'forecasting': {
                'forecasts': forecasts,
                'forecast_vars': forecast_vars
            }
        }
        
        print(f"Best AR order: {best_order}")
        print(f"True AR coefficients: {ar_coeffs}")
        print(f"Estimated AR coefficients: {best_phi}")
        print(f"State space model AIC: {state_space_result['aic']:.2f}")

    def demonstrate_statistical_learning_theory(self):
        """

        """

        # Generate classification data
        n_samples = 1000
        n_features = 2
        
        # Generate data from two classes
        class_0 = np.random.multivariate_normal([0, 0], [[1, 0.3], [0.3, 1]], n_samples//2)
        class_1 = np.random.multivariate_normal([2, 2], [[1, -0.3], [-0.3, 1]], n_samples//2)
        
        X = np.vstack([class_0, class_1])
        y = np.hstack([np.zeros(n_samples//2), np.ones(n_samples//2)])
        
        # Empirical Risk Minimization
        def empirical_risk(predictions, true_labels):
            """0-1 loss"""
            return np.mean(predictions != true_labels)
        
        # Linear classifier
        def linear_classifier(X, w, b):
            return (X @ w + b) > 0
        
        # Fit linear classifier using logistic regression
        from sklearn.linear_model import LogisticRegression
        from sklearn.model_selection import train_test_split
        
        X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

        clf.fit(X_train, y_train)
        
        # Predictions
        y_train_pred = clf.predict(X_train)
        y_test_pred = clf.predict(X_test)
        
        # Empirical risks
        train_risk = empirical_risk(y_train_pred, y_train)
        test_risk = empirical_risk(y_test_pred, y_test)
        generalization_gap = test_risk - train_risk
        
        # Concentration inequalities
        def hoeffding_bound(n, delta):
            """Hoeffding's inequality for bounded random variables in [0,1]"""

            """Holm's step-down procedure"""
            n = len(p_values)
            sorted_indices = np.argsort(p_values)
            rejected = np.zeros(n, dtype=bool)
            
            for i, idx in enumerate(sorted_indices):
                if p_values[idx] <= alpha / (n - i):
                    rejected[idx] = True
                else:
                    break
            
            return rejected

            """Benjamini-Hochberg FDR control"""
            n = len(p_values)
            sorted_indices = np.argsort(p_values)
            rejected = np.zeros(n, dtype=bool)
            
            for i in range(n-1, -1, -1):
                idx = sorted_indices[i]
                if p_values[idx] <= (i + 1) * alpha / n:
                    # Reject this and all smaller p-values
                    for j in range(i + 1):
                        rejected[sorted_indices[j]] = True
                    break
            
            return rejected
