## 1. Arithmetic Operations and Properties

### The Foundation of Mathematical Computation

Arithmetic operations form the absolute foundation of all mathematical computation in machine learning and artificial intelligence. While these operations might seem elementary, their properties and behaviors become critically important when scaled to the massive computations required in modern AI systems.

### 1.1 The Four Fundamental Operations

#### Addition (+): The Combining Operation
Addition represents the fundamental operation of combining quantities. In machine learning contexts:
- **Vector addition**: Combining feature vectors or gradients
- **Matrix addition**: Combining transformations or updating parameters
- **Probability addition**: Computing joint probabilities of disjoint events

**Mathematical Definition**: For real numbers $a$ and $b$, addition is a binary operation $+: \mathbb{R} \times \mathbb{R} \to \mathbb{R}$ such that $a + b$ represents the result of combining $a$ and $b$.

#### Subtraction (-): The Difference Operation
Subtraction represents finding the difference between quantities:
- **Error computation**: $\text{error} = \text{predicted} - \text{actual}$
- **Gradient computation**: Rate of change calculations
- **Distance metrics**: Measuring dissimilarity between data points

#### Multiplication (×): The Scaling Operation
Multiplication represents repeated addition or scaling:
- **Dot products**: Core operation in neural networks
- **Matrix multiplication**: Linear transformations and feature mapping
- **Probability multiplication**: Independent event probability computation

#### Division (÷): The Partitioning Operation
Division represents partitioning into equal parts:
- **Normalization**: Converting values to probabilities or unit scale
- **Average computation**: Central tendency measures
- **Learning rate scaling**: Controlling optimization step sizes

### 1.2 Fundamental Algebraic Properties

These properties are not just mathematical curiosities—they enable computational optimizations and ensure algorithmic correctness in machine learning systems.

#### Commutative Property: Order Independence
**Definition**: An operation $\circ$ is commutative if $a \circ b = b \circ a$ for all valid $a, b$.

**Addition**: $a + b = b + a$
- **ML Application**: Gradient accumulation order doesn't matter
- **Example**: $(w_1 \Delta w_1) + (w_2 \Delta w_2) = (w_2 \Delta w_2) + (w_1 \Delta w_1)$

**Multiplication**: $a \times b = b \times a$
- **ML Application**: Feature scaling can be applied in any order
- **Example**: $(\text{feature} \times \text{weight}) = (\text{weight} \times \text{feature})$

**Non-commutative Operations**: 
- Matrix multiplication: $AB \neq BA$ in general
- Subtraction: $a - b \neq b - a$ in general
- Division: $a \div b \neq b \div a$ in general

#### Associative Property: Grouping Independence
**Definition**: An operation $\circ$ is associative if $(a \circ b) \circ c = a \circ (b \circ c)$ for all valid $a, b, c$.

**Addition**: $(a + b) + c = a + (b + c)$
- **ML Application**: Enables parallel computation in distributed training
- **Computational advantage**: Can group operations for efficiency

**Multiplication**: $(a \times b) \times c = a \times (b \times c)$
- **ML Application**: Chain rule in backpropagation relies on this property
- **Neural networks**: $(W_1 \times W_2) \times x = W_1 \times (W_2 \times x)$

#### Distributive Property: Distribution of Operations
**Definition**: Multiplication distributes over addition: $a \times (b + c) = (a \times b) + (a \times c)$

**ML Applications**:
- **Linear transformations**: $W(x_1 + x_2) = Wx_1 + Wx_2$
- **Batch processing**: Operations can be distributed across batch elements
- **Feature engineering**: $w(f_1 + f_2) = wf_1 + wf_2$

### 1.3 Order of Operations: PEMDAS/BODMAS

The order of operations is crucial for unambiguous mathematical expression evaluation:

1. **Parentheses/Brackets** (innermost first)
2. **Exponents/Orders** (right to left)
3. **Multiplication and Division** (left to right, equal precedence)
4. **Addition and Subtraction** (left to right, equal precedence)

**Critical in ML**: Loss function computation, gradient calculations, and neural network forward passes all rely on correct order of operations.

### 1.4 Advanced Arithmetic Concepts

#### Modular Arithmetic
**Definition**: For integers $a, b, n$ with $n > 0$, we say $a \equiv b \pmod{n}$ if $n$ divides $(a - b)$.

**ML Applications**:
- **Hashing**: Feature hashing for dimensionality reduction
- **Cryptography**: Secure multi-party computation in federated learning
- **Indexing**: Circular buffer implementations in streaming algorithms

#### Floating-Point Arithmetic Considerations
Real numbers in computers are approximated using floating-point representation, leading to:

**Precision limitations**:
- **Single precision (32-bit)**: ~7 decimal digits accuracy
- **Double precision (64-bit)**: ~15 decimal digits accuracy
- **Half precision (16-bit)**: ~3 decimal digits accuracy (common in deep learning)

**Numerical stability issues**:
- **Catastrophic cancellation**: Subtracting nearly equal numbers
- **Overflow/Underflow**: Numbers too large or too small to represent
- **Accumulation errors**: Small errors compound in iterative algorithms

**ML Implications**:
- **Training stability**: Numerical precision affects gradient computation
- **Model accuracy**: Accumulated errors can degrade performance
- **Hardware optimization**: Lower precision (FP16, INT8) for speed vs. accuracy trade-offs

### Important Properties for Machine Learning

These arithmetic properties enable key optimizations in ML systems:

1. **Parallelization**: Commutative and associative properties allow operations to be reordered for parallel computation
2. **Numerical stability**: Understanding precision limitations helps design robust algorithms
3. **Computational efficiency**: Distributive property enables optimized implementations
4. **Algorithmic correctness**: Proper order of operations ensures reproducible results

```python
class AdvancedArithmeticFramework:
    """Advanced framework for demonstrating deep arithmetic properties and their ML applications"""

    def __init__(self):
        self.precision_types = {
            'float16': np.float16,
            'float32': np.float32,
            'float64': np.float64
        }

    def demonstrate_algebraic_structures(self):
        """Demonstrate advanced algebraic structure properties"""
        print("=== ALGEBRAIC STRUCTURE ANALYSIS ===")

        # Group theory demonstration with modular arithmetic
        n = 7  # Prime modulus for field structure
        elements = list(range(n))

        print(f"\nField Z_{n} (Galois Field GF({n})):")
        print("Addition table (mod 7):")

        # Create addition table
        add_table = np.zeros((n, n), dtype=int)
        for i in range(n):
            for j in range(n):
                add_table[i][j] = (i + j) % n

        print(add_table)

        # Verify group properties
        print("\nGroup Property Verification:")

        # Identity element
        identity = 0
        print(f"Identity element: {identity}")

        # Inverse elements
        inverses = {}
        for a in elements:
            for b in elements:
                if (a + b) % n == identity:
                    inverses[a] = b
                    break
        print(f"Additive inverses: {inverses}")

        # Demonstrate ring structure with multiplication
        print("\nMultiplication table (mod 7):")
        mult_table = np.zeros((n, n), dtype=int)
        for i in range(n):
            for j in range(n):
                mult_table[i][j] = (i * j) % n
        print(mult_table)

        return add_table, mult_table

    def analyze_numerical_precision(self):
        """Advanced analysis of numerical precision and floating-point arithmetic"""
        print("\n=== NUMERICAL PRECISION ANALYSIS ===")

        # Machine epsilon for different precisions
        for name, dtype in self.precision_types.items():
            eps = np.finfo(dtype).eps
            print(f"{name} machine epsilon: {eps:.2e}")

        # Catastrophic cancellation demonstration
        print("\nCatastrophic Cancellation Example:")
        x = np.float32(1.0)
        y = np.float32(1.0) + np.float32(1e-7)

        # Direct computation
        direct = x**2 - y**2
        # Algebraically equivalent but numerically stable
        stable = (x - y) * (x + y)

        print(f"Direct computation: x² - y² = {direct:.2e}")
        print(f"Stable computation: (x-y)(x+y) = {stable:.2e}")
        print(f"Relative error: {abs(direct - stable) / abs(stable):.2e}")

        return eps, direct, stable

    def demonstrate_advanced_operations(self):
        """Demonstrate advanced arithmetic operations in ML contexts"""
        print("\n=== ADVANCED ARITHMETIC OPERATIONS ===")

        # Kahan summation for numerical stability
        def kahan_sum(numbers):
            """Numerically stable summation using Kahan algorithm"""
            sum_val = 0.0
            compensation = 0.0

            for num in numbers:
                y = num - compensation
                temp = sum_val + y
                compensation = (temp - sum_val) - y
                sum_val = temp

            return sum_val

        # Test with poorly conditioned sum
        large_numbers = [1e16] * 1000
        small_numbers = [1.0] * 1000
        test_numbers = large_numbers + small_numbers

        naive_sum = sum(test_numbers)
        stable_sum = kahan_sum(test_numbers)
        true_sum = 1e16 * 1000 + 1000

        print(f"Naive sum: {naive_sum:.2e}")
        print(f"Kahan sum: {stable_sum:.2e}")
        print(f"True sum: {true_sum:.2e}")
        print(f"Naive error: {abs(naive_sum - true_sum):.2e}")
        print(f"Kahan error: {abs(stable_sum - true_sum):.2e}")

        return naive_sum, stable_sum, true_sum

    def matrix_arithmetic_properties(self):
        """Demonstrate arithmetic properties in matrix algebra"""
        print("\n=== MATRIX ARITHMETIC PROPERTIES ===")

        # Non-commutative multiplication example
        A = np.array([[1, 2], [3, 4]])
        B = np.array([[5, 6], [7, 8]])

        AB = A @ B
        BA = B @ A

        print(f"Matrix A:\n{A}")
        print(f"Matrix B:\n{B}")
        print(f"A × B:\n{AB}")
        print(f"B × A:\n{BA}")
        print(f"A × B = B × A? {np.allclose(AB, BA)}")

        # Demonstrate associativity
        C = np.array([[1, 0], [0, 2]])
        ABC1 = (A @ B) @ C
        ABC2 = A @ (B @ C)

        print(f"\n(A × B) × C:\n{ABC1}")
        print(f"A × (B × C):\n{ABC2}")
        print(f"Associative? {np.allclose(ABC1, ABC2)}")

        return A, B, C, AB, BA

# Initialize and run advanced arithmetic demonstrations
advanced_arithmetic = AdvancedArithmeticFramework()
add_table, mult_table = advanced_arithmetic.demonstrate_algebraic_structures()
eps, direct, stable = advanced_arithmetic.analyze_numerical_precision()
naive_sum, stable_sum, true_sum = advanced_arithmetic.demonstrate_advanced_operations()
A, B, C, AB, BA = advanced_arithmetic.matrix_arithmetic_properties()
```

### ML Application: Data Preprocessing

These basic operations are used extensively in data preprocessing:

```python
class AdvancedDataNormalizationFramework:
    """Advanced normalization techniques with mathematical rigor"""

    def __init__(self):
        self.normalization_methods = {
            'z_score': self._z_score_normalize,
            'min_max': self._min_max_normalize,
            'robust': self._robust_normalize,
            'unit_vector': self._unit_vector_normalize,
            'quantile_uniform': self._quantile_uniform_normalize
        }

    def _z_score_normalize(self, data):
        """Z-score normalization: (x - μ) / σ"""
        # Use Bessel's correction for unbiased standard deviation
        mean = np.mean(data)
        std = np.std(data, ddof=1)  # Unbiased estimator
        return (data - mean) / std, {'mean': mean, 'std': std}

    def _min_max_normalize(self, data):
        """Min-max normalization: (x - min) / (max - min)"""
        min_val = np.min(data)
        max_val = np.max(data)
        return (data - min_val) / (max_val - min_val), {'min': min_val, 'max': max_val}

    def _robust_normalize(self, data):
        """Robust normalization using median and IQR"""
        median = np.median(data)
        q75, q25 = np.percentile(data, [75, 25])
        iqr = q75 - q25
        return (data - median) / iqr, {'median': median, 'iqr': iqr}

    def _unit_vector_normalize(self, data):
        """L2 normalization: x / ||x||_2"""
        norm = np.linalg.norm(data)
        return data / norm, {'norm': norm}

    def _quantile_uniform_normalize(self, data):
        """Quantile normalization to uniform distribution"""
        sorted_indices = np.argsort(data)
        ranks = np.empty_like(sorted_indices)
        ranks[sorted_indices] = np.arange(len(data))
        uniform_values = ranks / (len(data) - 1)
        return uniform_values, {'ranks': ranks}

    def demonstrate_advanced_normalization(self):
        """Comprehensive demonstration of normalization techniques"""
        print("=== ADVANCED DATA NORMALIZATION ANALYSIS ===")

        # Generate complex synthetic data with outliers
        np.random.seed(42)
        n_samples = 1000

        # Multi-modal distribution with outliers
        normal_data = np.random.normal(50, 10, int(0.8 * n_samples))
        outliers = np.random.uniform(100, 150, int(0.1 * n_samples))
        second_mode = np.random.normal(20, 5, int(0.1 * n_samples))

        data = np.concatenate([normal_data, outliers, second_mode])
        np.random.shuffle(data)

        print(f"Original data statistics:")
        print(f"Mean: {np.mean(data):.3f}")
        print(f"Std: {np.std(data, ddof=1):.3f}")
        print(f"Median: {np.median(data):.3f}")
        print(f"IQR: {np.percentile(data, 75) - np.percentile(data, 25):.3f}")
        print(f"Skewness: {self._compute_skewness(data):.3f}")
        print(f"Kurtosis: {self._compute_kurtosis(data):.3f}")

        # Apply all normalization methods
        normalized_results = {}
        for name, method in self.normalization_methods.items():
            norm_data, params = method(data)
            normalized_results[name] = {
                'data': norm_data,
                'params': params,
                'robustness_score': self._compute_robustness_score(norm_data)
            }

        # Visualize normalization results
        self._plot_normalization_comparison(data, normalized_results)

        return data, normalized_results

    def _compute_skewness(self, data):
        """Compute sample skewness"""
        n = len(data)
        mean = np.mean(data)
        std = np.std(data, ddof=1)
        skew = (n / ((n-1) * (n-2))) * np.sum(((data - mean) / std) ** 3)
        return skew

    def _compute_kurtosis(self, data):
        """Compute sample kurtosis (excess kurtosis)"""
        n = len(data)
        mean = np.mean(data)
        std = np.std(data, ddof=1)
        kurt = (n * (n+1) / ((n-1) * (n-2) * (n-3))) * np.sum(((data - mean) / std) ** 4)
        kurt -= 3 * (n-1)**2 / ((n-2) * (n-3))
        return kurt

    def _compute_robustness_score(self, data):
        """Compute robustness score based on outlier resistance"""
        # Use median absolute deviation as robustness measure
        median = np.median(data)
        mad = np.median(np.abs(data - median))
        return 1 / (1 + mad)  # Higher score = more robust

    def _plot_normalization_comparison(self, original_data, normalized_results):
        """Create comprehensive visualization of normalization methods"""
        fig, axes = plt.subplots(2, 3, figsize=(18, 12))
        axes = axes.flatten()

        # Plot original data
        axes[0].hist(original_data, bins=50, alpha=0.7, color='blue', edgecolor='black')
        axes[0].set_title('Original Data Distribution', fontsize=12, fontweight='bold')
        axes[0].set_xlabel('Value')
        axes[0].set_ylabel('Frequency')
        axes[0].grid(True, alpha=0.3)

        # Plot each normalization method
        colors = ['red', 'green', 'orange', 'purple', 'brown']
        for idx, (name, result) in enumerate(normalized_results.items()):
            if idx < 5:  # Plot first 5 methods
                axes[idx + 1].hist(result['data'], bins=50, alpha=0.7,
                                 color=colors[idx], edgecolor='black')
                axes[idx + 1].set_title(f'{name.replace("_", " ").title()} Normalization\n'
                                       f'Robustness: {result["robustness_score"]:.3f}',
                                       fontsize=12, fontweight='bold')
                axes[idx + 1].set_xlabel('Normalized Value')
                axes[idx + 1].set_ylabel('Frequency')
                axes[idx + 1].grid(True, alpha=0.3)

        plt.tight_layout()
        plt.show()

        # Statistical comparison table
        print("\n=== NORMALIZATION COMPARISON TABLE ===")
        print(f"{'Method':<15} {'Mean':<8} {'Std':<8} {'Median':<8} {'IQR':<8} {'Robustness':<10}")
        print("-" * 67)

        for name, result in normalized_results.items():
            data = result['data']
            mean = np.mean(data)
            std = np.std(data, ddof=1)
            median = np.median(data)
            iqr = np.percentile(data, 75) - np.percentile(data, 25)
            robustness = result['robustness_score']

            print(f"{name:<15} {mean:<8.3f} {std:<8.3f} {median:<8.3f} {iqr:<8.3f} {robustness:<10.3f}")

# Initialize and run advanced normalization analysis
normalization_framework = AdvancedDataNormalizationFramework()
original_data, norm_results = normalization_framework.demonstrate_advanced_normalization()
```

## 3. Exponents and Logarithms: The Mathematical Foundation of Growth and Information

### 3.1 The Exponential Function: Modeling Growth and Decay

Exponential functions are among the most important mathematical concepts in machine learning and artificial intelligence. They appear everywhere from activation functions to probability distributions, from optimization algorithms to information theory.

#### Definition and Properties

An **exponential function** has the form $f(x) = a^x$ where $a > 0$ and $a \neq 1$ is called the base.

**Key Properties**:
1. **Domain**: $(-\infty, \infty)$ - defined for all real numbers
2. **Range**: $(0, \infty)$ - always positive, never zero
3. **Monotonicity**: 
   - If $a > 1$: strictly increasing (exponential growth)
   - If $0 < a < 1$: strictly decreasing (exponential decay)
4. **Base case**: $a^0 = 1$ for any $a \neq 0$
5. **Continuity**: Exponential functions are continuous everywhere

#### The Natural Exponential Function: $e^x$

The most important exponential function uses **Euler's number** $e \approx 2.71828$ as the base. This arises naturally in:

- **Continuous compound interest**: $\lim_{n \to \infty}(1 + \frac{1}{n})^n = e$
- **Calculus**: The unique function that is its own derivative: $\frac{d}{dx}e^x = e^x$
- **Probability**: The rate parameter in exponential and normal distributions

**Mathematical Significance**: The natural exponential function $e^x$ has the remarkable property that its rate of change at any point equals its value at that point, making it fundamental to differential equations and continuous-time processes.

#### Exponential Laws and Rules

These algebraic rules are essential for manipulating expressions in machine learning:

1. **Product Rule**: $a^m \cdot a^n = a^{m+n}$
   - **ML Application**: Combining probabilities in independent events
   
2. **Quotient Rule**: $\frac{a^m}{a^n} = a^{m-n}$
   - **ML Application**: Normalizing probabilities, computing odds ratios
   
3. **Power Rule**: $(a^m)^n = a^{mn}$
   - **ML Application**: Nested transformations in deep networks
   
4. **Product to Power**: $(ab)^n = a^n b^n$
   - **ML Application**: Scaling transformations
   
5. **Zero Exponent**: $a^0 = 1$ (for $a \neq 0$)
   - **ML Application**: Bias terms, initialization strategies

### 3.2 Logarithmic Functions: The Inverse of Exponential Growth

The **logarithm** is the inverse function of the exponential. If $y = a^x$, then $x = \log_a(y)$.

#### Definition and Fundamental Properties

**Definition**: $\log_a(x) = y$ if and only if $a^y = x$, where $a > 0, a \neq 1, x > 0$.

**Key Properties**:
1. **Domain**: $(0, \infty)$ - only defined for positive numbers
2. **Range**: $(-\infty, \infty)$ - can take any real value
3. **Base cases**: 
   - $\log_a(1) = 0$ for any base $a$
   - $\log_a(a) = 1$ for any base $a$
4. **Monotonicity**: 
   - If $a > 1$: strictly increasing
   - If $0 < a < 1$: strictly decreasing

#### Common Logarithm Bases in Machine Learning

1. **Natural Logarithm**: $\ln(x) = \log_e(x)$
   - **Applications**: Information theory, entropy, log-likelihood
   - **Derivatives**: Simple derivative $\frac{d}{dx}\ln(x) = \frac{1}{x}$
   
2. **Common Logarithm**: $\log_{10}(x)$
   - **Applications**: Orders of magnitude, scientific notation
   
3. **Binary Logarithm**: $\log_2(x)$
   - **Applications**: Information theory, bits of information, algorithm complexity

#### Logarithmic Laws and Properties

These rules are fundamental for working with logarithms in ML applications:

1. **Product Rule**: $\log_a(xy) = \log_a(x) + \log_a(y)$
   - **ML Application**: Converting products to sums (e.g., likelihood to log-likelihood)
   - **Why it matters**: Products can cause numerical overflow; sums are numerically stable
   
2. **Quotient Rule**: $\log_a(\frac{x}{y}) = \log_a(x) - \log_a(y)$
   - **ML Application**: Log-odds ratios, relative entropy calculations
   
3. **Power Rule**: $\log_a(x^n) = n\log_a(x)$
   - **ML Application**: Handling exponential terms in optimization
   
4. **Change of Base Formula**: $\log_a(x) = \frac{\log_b(x)}{\log_b(a)}$
   - **ML Application**: Converting between different information units (bits, nats, etc.)

### 3.3 Critical Applications in Machine Learning

#### 1. Activation Functions: The Sigmoid Function

The **sigmoid function** $\sigma(x) = \frac{1}{1 + e^{-x}}$ is fundamental in neural networks:

**Mathematical Properties**:
- **Range**: $(0, 1)$ - perfect for representing probabilities
- **Derivative**: $\sigma'(x) = \sigma(x)(1 - \sigma(x))$ - convenient for backpropagation
- **Symmetry**: $\sigma(-x) = 1 - \sigma(x)$
- **Asymptotic behavior**: 
  - As $x \to \infty$: $\sigma(x) \to 1$
  - As $x \to -\infty$: $\sigma(x) \to 0$

**ML Applications**:
- **Binary classification**: Output layer activation for probability prediction
- **Gating mechanisms**: LSTM and GRU architectures use sigmoid gates
- **Attention mechanisms**: Normalizing attention weights

#### 2. Loss Functions: Cross-Entropy

The **cross-entropy loss** for binary classification:
$$\mathcal{L}(y, \hat{y}) = -[y\log(\hat{y}) + (1-y)\log(1-\hat{y})]$$

**Why logarithms?**:
- **Numerical stability**: Log probabilities prevent underflow
- **Convexity**: Log-likelihood creates convex optimization problems
- **Information theory**: Cross-entropy measures information distance

#### 3. Probability Distributions

Many important distributions involve exponential functions:

**Normal Distribution**: $f(x) = \frac{1}{\sqrt{2\pi\sigma^2}}e^{-\frac{(x-\mu)^2}{2\sigma^2}}$

**Exponential Distribution**: $f(x) = \lambda e^{-\lambda x}$ for $x \geq 0$

**Softmax Distribution**: $p_i = \frac{e^{z_i}}{\sum_{j=1}^K e^{z_j}}$

### 3.4 Information Theory Connections

#### Entropy: The Logarithmic Measure of Information

**Shannon Entropy**: $H(X) = -\sum_{i} p_i \log_2(p_i)$

**Interpretation**: Average number of bits needed to encode a message from distribution $X$.

**Key Properties**:
- **Maximum entropy**: Uniform distribution maximizes entropy
- **Minimum entropy**: Deterministic distribution (entropy = 0)
- **Additivity**: $H(X,Y) = H(X) + H(Y|X)$ for dependent variables

#### Kullback-Leibler Divergence

**Definition**: $D_{KL}(P||Q) = \sum_i P(i) \log\left(\frac{P(i)}{Q(i)}\right)$

**ML Applications**:
- **Variational inference**: Measuring distance between distributions
- **Model selection**: Comparing model fits
- **Regularization**: Encouraging specific parameter distributions

### 3.5 Computational Considerations

#### Numerical Stability with Logarithms

**The Log-Sum-Exp Trick**: Computing $\log(\sum_i e^{x_i})$ numerically stable:

$$\log\left(\sum_{i=1}^n e^{x_i}\right) = x_{\max} + \log\left(\sum_{i=1}^n e^{x_i - x_{\max}}\right)$$

where $x_{\max} = \max_i x_i$.

**Why this works**:
- Prevents overflow by subtracting the maximum
- Maintains numerical precision
- Essential for softmax computation

#### Gradient Computation

Understanding derivatives of exponential and logarithmic functions:

- $\frac{d}{dx}e^x = e^x$
- $\frac{d}{dx}\log(x) = \frac{1}{x}$
- $\frac{d}{dx}a^x = a^x \ln(a)$
- $\frac{d}{dx}\log_a(x) = \frac{1}{x \ln(a)}$

These derivatives are crucial for:
- **Backpropagation**: Computing gradients through activation functions
- **Optimization**: Understanding loss function landscapes
- **Automatic differentiation**: Building computational graphs

### 3.6 Advanced Topics

#### Matrix Exponentials

For square matrix $A$: $e^A = \sum_{k=0}^{\infty} \frac{A^k}{k!}$

**Applications**:
- **Linear dynamical systems**: Solution to $\frac{dx}{dt} = Ax$
- **Lie groups**: Exponential map from algebra to group
- **Neural ODEs**: Continuous-depth neural networks

#### Logarithms of Matrices

Matrix logarithm is the inverse of matrix exponential: $\log(e^A) = A$

**Applications**:
- **Riemannian optimization**: Optimization on manifolds
- **Covariance estimation**: Working with positive definite matrices
- **Geometric deep learning**: Operations on matrix manifolds

1. **Fundamental Mathematical Tools**: Enable modeling of growth, decay, and scaling phenomena
2. **Numerical Stability**: Logarithms convert unstable products to stable sums
3. **Information Processing**: Quantify information content and uncertainty

5. **Probability Theory**: Foundation for most probability distributions
6. **Activation Functions**: Enable non-linear transformations in neural networks

These concepts are not just mathematical abstractions—they are the computational building blocks that make modern AI systems possible.

```python
class AdvancedExponentialLogarithmicFramework:
    """Advanced analysis of exponential and logarithmic functions with ML applications"""

    def __init__(self):
        self.bases = [2, math.e, 10]
        self.activation_functions = {
            'sigmoid': self._sigmoid,
            'tanh': self._tanh,
            'softplus': self._softplus,
            'swish': self._swish,
            'gelu': self._gelu
        }

    def _sigmoid(self, x):
        """Sigmoid: σ(x) = 1/(1 + e^(-x))"""
        return 1 / (1 + np.exp(-np.clip(x, -500, 500)))

    def _tanh(self, x):
        """Hyperbolic tangent: tanh(x) = (e^x - e^(-x))/(e^x + e^(-x))"""
        return np.tanh(x)

    def _softplus(self, x):
        """Softplus: softplus(x) = log(1 + e^x)"""
        return np.log1p(np.exp(np.clip(x, -500, 500)))

    def _swish(self, x):
        """Swish: swish(x) = x * sigmoid(x)"""
        return x * self._sigmoid(x)

    def _gelu(self, x):
        """GELU: gelu(x) = 0.5x(1 + tanh(√(2/π)(x + 0.044715x³)))"""
        return 0.5 * x * (1 + np.tanh(np.sqrt(2/np.pi) * (x + 0.044715 * x**3)))

    def demonstrate_exponential_properties(self):
        """Demonstrate advanced exponential function properties"""
        print("=== EXPONENTIAL FUNCTION ANALYSIS ===")

        # Demonstrate exponential growth rates
        x = np.linspace(0, 5, 1000)
        
        print("Growth rate comparison:")
        for base in self.bases:
            y = base ** x
            derivative = np.gradient(y, x)
            
            # Growth rate at x=1
            growth_rate = derivative[200]  # Approximate index for x=1
            print(f"Base {base:.3f}: f'(1) ≈ {growth_rate:.3f}")

        # Compound interest demonstration (continuous compounding)
        print("\nContinuous Compound Interest Analysis:")
        principal = 1000
        rate = 0.05  # 5% annual rate
        time_years = np.linspace(0, 20, 100)
        
        # Discrete vs continuous compounding
        discrete_monthly = principal * (1 + rate/12)**(12 * time_years)
        continuous = principal * np.exp(rate * time_years)
        
        difference = continuous - discrete_monthly
        max_diff = np.max(difference)
        print(f"Maximum difference (continuous vs monthly): ${max_diff:.2f}")
        
        return x, time_years, discrete_monthly, continuous

    def demonstrate_logarithmic_properties(self):
        """Demonstrate advanced logarithmic properties and applications"""
        print("\n=== LOGARITHMIC FUNCTION ANALYSIS ===")

        # Information theory applications
        def entropy(probabilities):
            """Calculate Shannon entropy"""
            p = np.array(probabilities)
            p = p[p > 0]  # Remove zero probabilities
            return -np.sum(p * np.log2(p))

        def cross_entropy(true_probs, pred_probs):
            """Calculate cross-entropy"""
            p = np.array(true_probs)
            q = np.array(pred_probs)
            q = np.clip(q, 1e-15, 1 - 1e-15)  # Avoid log(0)
            return -np.sum(p * np.log(q))

        def kl_divergence(p, q):
            """Calculate Kullback-Leibler divergence"""
            p = np.array(p)
            q = np.array(q)
            q = np.clip(q, 1e-15, 1 - 1e-15)
            return np.sum(p * np.log(p / q))

        # Information theory examples
        print("Information Theory Applications:")
        
        # Uniform distribution (maximum entropy)
        uniform_probs = [0.25, 0.25, 0.25, 0.25]
        uniform_entropy = entropy(uniform_probs)
        print(f"Uniform distribution entropy: {uniform_entropy:.3f} bits")
        
        # Skewed distribution (lower entropy)
        skewed_probs = [0.7, 0.2, 0.05, 0.05]
        skewed_entropy = entropy(skewed_probs)
        print(f"Skewed distribution entropy: {skewed_entropy:.3f} bits")
        
        # Cross-entropy loss example
        true_dist = [1, 0, 0, 0]  # One-hot true distribution
        pred_dist = [0.8, 0.1, 0.05, 0.05]  # Model prediction
        ce_loss = cross_entropy(true_dist, pred_dist)
        print(f"Cross-entropy loss: {ce_loss:.3f}")
        
        # KL divergence
        kl_div = kl_divergence(uniform_probs, skewed_probs)
        print(f"KL divergence (uniform || skewed): {kl_div:.3f}")

        return uniform_entropy, skewed_entropy, ce_loss, kl_div

    def demonstrate_activation_functions(self):
        """Comprehensive analysis of activation functions"""
        print("\n=== ACTIVATION FUNCTION ANALYSIS ===")

        x = np.linspace(-5, 5, 1000)
        
        # Calculate activation functions and their derivatives
        activations = {}
        derivatives = {}
        
        for name, func in self.activation_functions.items():
            y = func(x)
            dy = np.gradient(y, x)
            
            activations[name] = y
            derivatives[name] = dy
            
            # Analyze function properties
            max_val = np.max(y)
            min_val = np.min(y)
            range_val = max_val - min_val
            
            # Saturation analysis (where derivative < 0.01)
            saturation_points = np.sum(np.abs(dy) < 0.01)
            saturation_percentage = (saturation_points / len(x)) * 100
            
            print(f"{name.upper()}:")
            print(f"  Range: [{min_val:.3f}, {max_val:.3f}]")
            print(f"  Saturation: {saturation_percentage:.1f}% of domain")
            print(f"  Max gradient: {np.max(dy):.3f}")

        # Visualize activation functions
        self._plot_activation_functions(x, activations, derivatives)
        
        return x, activations, derivatives

    def _plot_activation_functions(self, x, activations, derivatives):
        """Plot activation functions and their derivatives"""
        fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(2, 2, figsize=(16, 12))
        
        colors = ['blue', 'red', 'green', 'orange', 'purple']
        
        # Plot activation functions
        for i, (name, y) in enumerate(activations.items()):
            ax1.plot(x, y, color=colors[i], linewidth=2, label=name, alpha=0.8)
        
        ax1.set_title('Activation Functions', fontsize=14, fontweight='bold')
        ax1.set_xlabel('Input (x)')
        ax1.set_ylabel('Output f(x)')
        ax1.legend()
        ax1.grid(True, alpha=0.3)
        ax1.set_ylim(-2, 2)
        
        # Plot derivatives
        for i, (name, dy) in enumerate(derivatives.items()):
            ax2.plot(x, dy, color=colors[i], linewidth=2, label=f"{name} derivative", alpha=0.8)
        
        ax2.set_title('Activation Function Derivatives', fontsize=14, fontweight='bold')
        ax2.set_xlabel('Input (x)')
        ax2.set_ylabel("f'(x)")
        ax2.legend()
        ax2.grid(True, alpha=0.3)
        
        # Compare sigmoid and tanh
        sigmoid_y = activations['sigmoid']
        tanh_y = activations['tanh']
        
        ax3.plot(x, sigmoid_y, 'b-', linewidth=2, label='Sigmoid')
        ax3.plot(x, tanh_y, 'r-', linewidth=2, label='Tanh')
        ax3.set_title('Sigmoid vs Tanh Comparison', fontsize=14, fontweight='bold')
        ax3.set_xlabel('Input (x)')
        ax3.set_ylabel('Output')
        ax3.legend()
        ax3.grid(True, alpha=0.3)
        ax3.axhline(y=0, color='black', linestyle='--', alpha=0.5)
        ax3.axhline(y=0.5, color='black', linestyle='--', alpha=0.5)
        
        # Modern activation functions
        swish_y = activations['swish']
        gelu_y = activations['gelu']
        
        ax4.plot(x, swish_y, 'g-', linewidth=2, label='Swish')
        ax4.plot(x, gelu_y, 'm-', linewidth=2, label='GELU')
        ax4.plot(x, np.maximum(0, x), 'k--', linewidth=2, label='ReLU', alpha=0.7)
        ax4.set_title('Modern Activation Functions', fontsize=14, fontweight='bold')
        ax4.set_xlabel('Input (x)')
        ax4.set_ylabel('Output')
        ax4.legend()
        ax4.grid(True, alpha=0.3)
        ax4.axhline(y=0, color='black', linestyle='--', alpha=0.5)
        
        plt.tight_layout()
        plt.show()

    def demonstrate_numerical_stability(self):
        """Demonstrate numerical stability techniques with exponentials and logarithms"""
        print("\n=== NUMERICAL STABILITY ANALYSIS ===")

        # Log-sum-exp trick
        def log_sum_exp_naive(x):
            """Naive implementation (can overflow)"""
            return np.log(np.sum(np.exp(x)))

        def log_sum_exp_stable(x):
            """Numerically stable implementation"""
            x_max = np.max(x)
            return x_max + np.log(np.sum(np.exp(x - x_max)))

        # Test with large values
        large_values = np.array([700, 701, 702])  # Would overflow in naive version
        
        print("Log-Sum-Exp Comparison:")
        print(f"Input values: {large_values}")
        
        try:
            naive_result = log_sum_exp_naive(large_values)
            print(f"Naive result: {naive_result}")
        except OverflowError:
            print("Naive result: OVERFLOW ERROR")
        
        stable_result = log_sum_exp_stable(large_values)
        print(f"Stable result: {stable_result:.6f}")

        # Softmax stability
        def softmax_naive(x):
            """Naive softmax implementation"""
            exp_x = np.exp(x)
            return exp_x / np.sum(exp_x)

        def softmax_stable(x):
            """Numerically stable softmax"""
            x_shifted = x - np.max(x)
            exp_x = np.exp(x_shifted)
            return exp_x / np.sum(exp_x)

        print("\nSoftmax Stability Comparison:")
        test_logits = np.array([700, 701, 702])
        
        try:
            naive_softmax = softmax_naive(test_logits)
            print(f"Naive softmax: {naive_softmax}")
        except (OverflowError, RuntimeWarning):
            print("Naive softmax: NUMERICAL ISSUES")
        
        stable_softmax = softmax_stable(test_logits)
        print(f"Stable softmax: {stable_softmax}")
        print(f"Sum of probabilities: {np.sum(stable_softmax):.10f}")

        return stable_result, stable_softmax

# Initialize and run comprehensive exponential/logarithmic analysis
exp_log_framework = AdvancedExponentialLogarithmicFramework()
x_exp, time_years, discrete_monthly, continuous = exp_log_framework.demonstrate_exponential_properties()
uniform_entropy, skewed_entropy, ce_loss, kl_div = exp_log_framework.demonstrate_logarithmic_properties()
x_act, activations, derivatives = exp_log_framework.demonstrate_activation_functions()
stable_result, stable_softmax = exp_log_framework.demonstrate_numerical_stability()
```

## 5. Functions and Their Properties

Functions are fundamental to ML - they represent mappings from inputs to outputs, which is exactly what ML models do.

```python
# Define various types of functions
def linear_function(x):
    """Linear function: f(x) = 2x + 1"""
    return 2 * x + 1

def quadratic_function(x):
    """Quadratic function: f(x) = x² - 4x + 3"""
    return x**2 - 4*x + 3

def exponential_function(x):
    """Exponential function: f(x) = 2^x"""
    return 2**x

def logarithmic_function(x):
    """Logarithmic function: f(x) = log₂(x)"""
    return np.log2(x) if x > 0 else float('nan')

# Generate data points
x_linear = np.linspace(-5, 5, 100)
x_exp = np.linspace(-3, 3, 100)
x_log = np.linspace(0.1, 10, 100)

# Calculate function values
y_linear = [linear_function(x) for x in x_linear]
y_quadratic = [quadratic_function(x) for x in x_linear]
y_exp = [exponential_function(x) for x in x_exp]
y_log = [logarithmic_function(x) for x in x_log]

# Plot all functions
fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(2, 2, figsize=(12, 10))

ax1.plot(x_linear, y_linear, 'b-', linewidth=2)
ax1.set_title('Linear Function: f(x) = 2x + 1')
ax1.grid(True, alpha=0.3)
ax1.set_xlabel('x')
ax1.set_ylabel('f(x)')

ax2.plot(x_linear, y_quadratic, 'r-', linewidth=2)
ax2.set_title('Quadratic Function: f(x) = x² - 4x + 3')
ax2.grid(True, alpha=0.3)
ax2.set_xlabel('x')
ax2.set_ylabel('f(x)')

ax3.plot(x_exp, y_exp, 'g-', linewidth=2)
ax3.set_title('Exponential Function: f(x) = 2^x')
ax3.grid(True, alpha=0.3)
ax3.set_xlabel('x')
ax3.set_ylabel('f(x)')

ax4.plot(x_log, y_log, 'm-', linewidth=2)
ax4.set_title('Logarithmic Function: f(x) = log₂(x)')
ax4.grid(True, alpha=0.3)
ax4.set_xlabel('x')
ax4.set_ylabel('f(x)')

plt.tight_layout()
plt.show()
```

### Function Properties Important in ML

```python
# Function properties demonstration
def check_function_properties(func, x_values, func_name):
    """Check various properties of a function"""
    y_values = [func(x) for x in x_values]
    
    print(f"\n{func_name} Properties:")
    
    # Domain and Range (for given values)
    valid_pairs = [(x, y) for x, y in zip(x_values, y_values) if not np.isnan(y)]
    if valid_pairs:
        x_domain = [pair[0] for pair in valid_pairs]
        y_range = [pair[1] for pair in valid_pairs]
        print(f"Domain (sample): [{min(x_domain):.2f}, {max(x_domain):.2f}]")
        print(f"Range (sample): [{min(y_range):.2f}, {max(y_range):.2f}]")
    
    # Monotonicity check
    differences = [y_values[i+1] - y_values[i] for i in range(len(y_values)-1) if not np.isnan(y_values[i]) and not np.isnan(y_values[i+1])]
    if differences:
        if all(d >= 0 for d in differences):
            print("Monotonicity: Non-decreasing")
        elif all(d <= 0 for d in differences):
            print("Monotonicity: Non-increasing")
        else:
            print("Monotonicity: Neither monotonic")

# Test function properties
test_x = np.linspace(1, 10, 20)

check_function_properties(linear_function, test_x, "Linear Function")
check_function_properties(exponential_function, test_x, "Exponential Function")
check_function_properties(logarithmic_function, test_x, "Logarithmic Function")
```

### ML Application: Loss Functions

Functions are central to ML as loss functions that measure prediction errors:

```python
class AdvancedFunctionAnalysisFramework:
    """Advanced function analysis with applications to machine learning"""

    def __init__(self):
        self.function_types = {
            'linear': lambda x: 2*x + 1,
            'quadratic': lambda x: x**2 - 4*x + 3,
            'exponential': lambda x: 2**x,
            'logarithmic': lambda x: np.log2(np.maximum(x, 1e-10)),
            'sigmoid': lambda x: 1 / (1 + np.exp(-np.clip(x, -500, 500))),
            'relu': lambda x: np.maximum(0, x),
            'sine': lambda x: np.sin(x),
            'cosine': lambda x: np.cos(x)
        }

    def analyze_function_properties(self):
        """Comprehensive analysis of function properties"""
        print("=== ADVANCED FUNCTION PROPERTIES ANALYSIS ===")

        x = np.linspace(-5, 5, 1000)
        
        for name, func in self.function_types.items():
            y = func(x)
            
            # Calculate numerical derivatives
            dy_dx = np.gradient(y, x)
            d2y_dx2 = np.gradient(dy_dx, x)
            
            # Analyze properties
            properties = self._analyze_function_properties(x, y, dy_dx, d2y_dx2)
            
            print(f"\n{name.upper()} FUNCTION:")
            print(f"  Domain: [{x[0]:.2f}, {x[-1]:.2f}] (sample)")
            print(f"  Range: [{np.min(y):.3f}, {np.max(y):.3f}]")
            print(f"  Monotonicity: {properties['monotonicity']}")
            print(f"  Concavity: {properties['concavity']}")
            print(f"  Critical points: {properties['critical_points']}")
            print(f"  Inflection points: {properties['inflection_points']}")
            print(f"  Lipschitz constant: {properties['lipschitz']:.3f}")

        return x, self.function_types

    def _analyze_function_properties(self, x, y, dy_dx, d2y_dx2):
        """Analyze mathematical properties of a function"""
        properties = {}
        
        # Monotonicity analysis
        increasing = np.all(dy_dx >= -1e-6)  # Allow for numerical errors
        decreasing = np.all(dy_dx <= 1e-6)
        
        if increasing:
            properties['monotonicity'] = 'Increasing'
        elif decreasing:
            properties['monotonicity'] = 'Decreasing'
        else:
            properties['monotonicity'] = 'Mixed'
        
        # Concavity analysis
        concave_up = np.all(d2y_dx2 >= -1e-6)
        concave_down = np.all(d2y_dx2 <= 1e-6)
        
        if concave_up:
            properties['concavity'] = 'Concave up'
        elif concave_down:
            properties['concavity'] = 'Concave down'
        else:
            properties['concavity'] = 'Mixed'
        
        # Critical points (where derivative ≈ 0)
        critical_indices = np.where(np.abs(dy_dx) < 0.01)[0]
        properties['critical_points'] = len(critical_indices)
        
        # Inflection points (where second derivative changes sign)
        sign_changes = np.where(np.diff(np.sign(d2y_dx2)))[0]
        properties['inflection_points'] = len(sign_changes)
        
        # Lipschitz constant (maximum derivative magnitude)
        properties['lipschitz'] = np.max(np.abs(dy_dx))
        
        return properties

    def demonstrate_loss_functions(self):
        """Advanced analysis of loss functions in machine learning"""
        print("\n=== ADVANCED LOSS FUNCTION ANALYSIS ===")

        # Define various loss functions
        def mean_squared_error(y_true, y_pred):
            return np.mean((y_true - y_pred)**2)

        def mean_absolute_error(y_true, y_pred):
            return np.mean(np.abs(y_true - y_pred))

        def huber_loss(y_true, y_pred, delta=1.0):
            error = y_true - y_pred
            condition = np.abs(error) <= delta
            squared_loss = 0.5 * error**2
            linear_loss = delta * np.abs(error) - 0.5 * delta**2
            return np.where(condition, squared_loss, linear_loss)

        def quantile_loss(y_true, y_pred, tau=0.5):
            """Quantile regression loss"""
            error = y_true - y_pred
            return np.maximum(tau * error, (tau - 1) * error)

        def focal_loss(y_true, y_pred, alpha=1.0, gamma=2.0):
            """Focal loss for classification"""
            epsilon = 1e-15
            y_pred = np.clip(y_pred, epsilon, 1 - epsilon)
            pt = np.where(y_true == 1, y_pred, 1 - y_pred)
            return -alpha * (1 - pt)**gamma * np.log(pt)

        # Analyze loss function properties
        errors = np.linspace(-3, 3, 1000)
        y_true_dummy = np.zeros_like(errors)
        y_pred_dummy = errors

        loss_functions = {
            'MSE': lambda e: 0.5 * e**2,
            'MAE': lambda e: np.abs(e),
            'Huber': lambda e: huber_loss(0, -e, delta=1.0),
            'Quantile (τ=0.5)': lambda e: quantile_loss(0, -e, tau=0.5),
            'Quantile (τ=0.9)': lambda e: quantile_loss(0, -e, tau=0.9)
        }

        print("Loss Function Properties Analysis:")
        for name, loss_func in loss_functions.items():
            losses = loss_func(errors)
            
            # Robustness analysis
            robustness_score = self._compute_loss_robustness(errors, losses)
            
            # Smoothness analysis
            derivatives = np.gradient(losses, errors)
            smoothness = np.std(derivatives)
            
            print(f"\n{name}:")
            print(f"  Robustness score: {robustness_score:.3f}")
            print(f"  Smoothness (std of derivative): {smoothness:.3f}")
            print(f"  Minimum value: {np.min(losses):.3f}")
            print(f"  Symmetry: {self._check_symmetry(errors, losses)}")

        # Visualize loss functions
        self._plot_loss_functions(errors, loss_functions)

        return errors, loss_functions

    def _compute_loss_robustness(self, errors, losses):
        """Compute robustness score for loss function"""
        # Robustness measured by how loss grows with large errors
        large_error_idx = np.abs(errors) > 2
        if np.any(large_error_idx):
            large_error_losses = losses[large_error_idx]
            robustness = 1 / (1 + np.mean(large_error_losses))
            return robustness
        return 1.0

    def _check_symmetry(self, errors, losses):
        """Check if loss function is symmetric around zero"""
        mid_idx = len(errors) // 2
        left_losses = losses[:mid_idx]
        right_losses = losses[mid_idx+1:][::-1]  # Reverse right side
        
        if len(left_losses) == len(right_losses):
            symmetry_error = np.mean(np.abs(left_losses - right_losses))
            return "Symmetric" if symmetry_error < 0.01 else "Asymmetric"
        return "Unknown"

    def _plot_loss_functions(self, errors, loss_functions):
        """Create comprehensive visualization of loss functions"""
        fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(2, 2, figsize=(16, 12))
        
        colors = ['blue', 'red', 'green', 'orange', 'purple']
        
        # Plot all loss functions
        for i, (name, loss_func) in enumerate(loss_functions.items()):
            losses = loss_func(errors)
            ax1.plot(errors, losses, color=colors[i], linewidth=2, 
                    label=name, alpha=0.8)
        
        ax1.set_title('Loss Functions Comparison', fontsize=14, fontweight='bold')
        ax1.set_xlabel('Prediction Error')
        ax1.set_ylabel('Loss Value')
        ax1.legend()
        ax1.grid(True, alpha=0.3)
        ax1.set_ylim(0, 5)
        
        # Plot derivatives (gradients)
        for i, (name, loss_func) in enumerate(loss_functions.items()):
            losses = loss_func(errors)
            gradients = np.gradient(losses, errors)
            ax2.plot(errors, gradients, color=colors[i], linewidth=2, 
                    label=f'{name} gradient', alpha=0.8)
        
        ax2.set_title('Loss Function Gradients', fontsize=14, fontweight='bold')
        ax2.set_xlabel('Prediction Error')
        ax2.set_ylabel('Gradient')
        ax2.legend()
        ax2.grid(True, alpha=0.3)
        ax2.axhline(y=0, color='black', linestyle='--', alpha=0.5)
        
        # Zoom in on robust vs non-robust losses
        robust_region = np.abs(errors) <= 2
        ax3.plot(errors[robust_region], 
                loss_functions['MSE'](errors[robust_region]), 
                'b-', linewidth=2, label='MSE')
        ax3.plot(errors[robust_region], 
                loss_functions['Huber'](errors[robust_region]), 
                'r-', linewidth=2, label='Huber')
        ax3.plot(errors[robust_region], 
                loss_functions['MAE'](errors[robust_region]), 
                'g-', linewidth=2, label='MAE')
        
        ax3.set_title('Robust vs Non-Robust Losses (Zoomed)', fontsize=14, fontweight='bold')
        ax3.set_xlabel('Prediction Error')
        ax3.set_ylabel('Loss Value')
        ax3.legend()
        ax3.grid(True, alpha=0.3)
        
        # Quantile loss comparison
        tau_values = [0.1, 0.5, 0.9]
        for tau in tau_values:
            quantile_losses = quantile_loss(0, -errors, tau=tau)
            ax4.plot(errors, quantile_losses, linewidth=2, 
                    label=f'τ = {tau}', alpha=0.8)
        
        ax4.set_title('Quantile Loss for Different τ Values', fontsize=14, fontweight='bold')
        ax4.set_xlabel('Prediction Error')
        ax4.set_ylabel('Loss Value')
        ax4.legend()
        ax4.grid(True, alpha=0.3)
        ax4.set_ylim(0, 3)
        
        plt.tight_layout()
        plt.show()

    def demonstrate_function_composition(self):
        """Demonstrate function composition and its ML applications"""
        print("\n=== FUNCTION COMPOSITION ANALYSIS ===")

        # Neural network layer composition
        def linear_layer(x, w, b):
            """Linear transformation: f(x) = Wx + b"""
            return w * x + b

        def activation_function(x, func_type='relu'):
            """Apply activation function"""
            if func_type == 'relu':
                return np.maximum(0, x)
            elif func_type == 'sigmoid':
                return 1 / (1 + np.exp(-np.clip(x, -500, 500)))
            elif func_type == 'tanh':
                return np.tanh(x)
            return x

        # Compose a simple 2-layer network
        def two_layer_network(x, w1=2, b1=-1, w2=1.5, b2=0.5, activation='relu'):
            """Composition: f(g(x)) where g is first layer, f is second layer"""
            # First layer: g(x) = w1*x + b1
            layer1_output = linear_layer(x, w1, b1)
            activated1 = activation_function(layer1_output, activation)
            
            # Second layer: f(z) = w2*z + b2
            layer2_output = linear_layer(activated1, w2, b2)
            
            return layer2_output, layer1_output, activated1

        # Analyze composition
        x = np.linspace(-3, 3, 1000)
        
        activations = ['relu', 'sigmoid', 'tanh']
        composition_results = {}
        
        for activation in activations:
            final_output, layer1, activated1 = two_layer_network(x, activation=activation)
            composition_results[activation] = {
                'final': final_output,
                'layer1': layer1,
                'activated': activated1
            }
        
        # Visualize composition
        self._plot_function_composition(x, composition_results)
        
        # Analyze composition properties
        print("Function Composition Properties:")
        for activation, results in composition_results.items():
            final_range = np.max(results['final']) - np.min(results['final'])
            expressivity = np.std(results['final'])  # Measure of expressivity
            
            print(f"\n{activation.upper()} Network:")
            print(f"  Output range: {final_range:.3f}")
            print(f"  Expressivity (std): {expressivity:.3f}")
            print(f"  Non-linearity: {'High' if activation != 'linear' else 'None'}")

        return x, composition_results

    def _plot_function_composition(self, x, composition_results):
        """Visualize function composition in neural networks"""
        fig, axes = plt.subplots(len(composition_results), 3, figsize=(18, 12))
        
        for i, (activation, results) in enumerate(composition_results.items()):
            # Plot layer 1 output (before activation)
            axes[i, 0].plot(x, results['layer1'], 'b-', linewidth=2)
            axes[i, 0].set_title(f'Layer 1 Output ({activation})', fontweight='bold')
            axes[i, 0].set_xlabel('Input x')
            axes[i, 0].set_ylabel('Linear Output')
            axes[i, 0].grid(True, alpha=0.3)
            axes[i, 0].axhline(y=0, color='black', linestyle='--', alpha=0.5)
            
            # Plot after activation
            axes[i, 1].plot(x, results['activated'], 'r-', linewidth=2)
            axes[i, 1].set_title(f'After {activation} Activation', fontweight='bold')
            axes[i, 1].set_xlabel('Input x')
            axes[i, 1].set_ylabel('Activated Output')
            axes[i, 1].grid(True, alpha=0.3)
            axes[i, 1].axhline(y=0, color='black', linestyle='--', alpha=0.5)
            
            # Plot final network output
            axes[i, 2].plot(x, results['final'], 'g-', linewidth=2)
            axes[i, 2].set_title(f'Final Network Output ({activation})', fontweight='bold')
            axes[i, 2].set_xlabel('Input x')
            axes[i, 2].set_ylabel('Final Output')
            axes[i, 2].grid(True, alpha=0.3)
            axes[i, 2].axhline(y=0, color='black', linestyle='--', alpha=0.5)
        
        plt.tight_layout()
        plt.show()

# Initialize and run advanced function analysis
function_framework = AdvancedFunctionAnalysisFramework()
x_func, function_types = function_framework.analyze_function_properties()
errors, loss_functions = function_framework.demonstrate_loss_functions()
x_comp, composition_results = function_framework.demonstrate_function_composition()
```
