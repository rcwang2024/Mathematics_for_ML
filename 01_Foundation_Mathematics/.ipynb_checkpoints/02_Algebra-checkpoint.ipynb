## 1. Linear Equations and Inequalities

Linear equations form the backbone of many ML algorithms, from linear regression to support vector machines.

### Linear Equations

A linear equation in one variable has the form: **ax + b = 0**

A linear equation in multiple variables: **a₁x₁ + a₂x₂ + ... + aₙxₙ = b**

```python
# Solving linear equations symbolically
x = symbols('x')

# Example 1: Simple linear equation
eq1 = Eq(2*x + 3, 7)
solution1 = solve(eq1, x)
print(f"Equation: {eq1}")
print(f"Solution: x = {solution1[0]}")

# Example 2: More complex linear equation
eq2 = Eq(5*x - 2, 3*x + 8)
solution2 = solve(eq2, x)
print(f"\nEquation: {eq2}")
print(f"Solution: x = {solution2[0]}")

# Verification
x_val = solution2[0]
left_side = 5*x_val - 2
right_side = 3*x_val + 8
print(f"Verification: {left_side} = {right_side}")
```

### Linear Inequalities

Linear inequalities are crucial for understanding constraints in optimization problems:

```python
# Linear inequalities
from sympy import *

x = symbols('x')

# Example: 2x + 3 > 7
inequality = 2*x + 3 > 7
solution = solve(inequality, x)
print(f"Inequality: {inequality}")
print(f"Solution: x {solution}")

# Visualizing linear equations and inequalities
x_vals = np.linspace(-5, 5, 100)
y1 = 2*x_vals + 3  # Line: y = 2x + 3
y2 = 7 * np.ones_like(x_vals)  # Horizontal line: y = 7

plt.figure(figsize=(10, 6))
plt.plot(x_vals, y1, 'b-', linewidth=2, label='y = 2x + 3')
plt.plot(x_vals, y2, 'r--', linewidth=2, label='y = 7')

# Highlight the solution region for 2x + 3 > 7
intersection_x = 2  # From solving 2x + 3 = 7
plt.axvline(x=intersection_x, color='green', linestyle=':', alpha=0.7)
plt.fill_between(x_vals[x_vals > intersection_x], 
                 np.minimum(y1[x_vals > intersection_x], 15), 
                 15, alpha=0.3, color='lightgreen', 
                 label='Region where 2x + 3 > 7')

plt.xlabel('x')
plt.ylabel('y')
plt.title('Linear Equation and Inequality Visualization')
plt.legend()
plt.grid(True, alpha=0.3)
plt.xlim(-5, 5)
plt.ylim(-5, 15)
plt.show()
```

### ML Application: Linear Regression

Linear equations are fundamental to linear regression, one of the most basic ML algorithms:

```python
# Linear regression example
# Model: y = mx + b (where m is slope, b is intercept)

# Generate synthetic data
np.random.seed(42)
n_points = 50
x_data = np.linspace(0, 10, n_points)
true_slope = 2.5
true_intercept = 1.0
noise = np.random.normal(0, 1, n_points)
y_data = true_slope * x_data + true_intercept + noise

# Solve for best fit line using normal equation
# For y = mx + b, we solve the system:
# Σ(x²)m + Σ(x)b = Σ(xy)
# Σ(x)m + nb = Σ(y)

sum_x = np.sum(x_data)
sum_y = np.sum(y_data)
sum_xx = np.sum(x_data**2)
sum_xy = np.sum(x_data * y_data)
n = len(x_data)

# Solve using linear algebra
A = np.array([[sum_xx, sum_x],
              [sum_x, n]])
b = np.array([sum_xy, sum_y])
solution = np.linalg.solve(A, b)
fitted_slope, fitted_intercept = solution

print(f"True parameters: slope = {true_slope}, intercept = {true_intercept}")
print(f"Fitted parameters: slope = {fitted_slope:.3f}, intercept = {fitted_intercept:.3f}")

# Visualize results
y_fitted = fitted_slope * x_data + fitted_intercept

plt.figure(figsize=(10, 6))
plt.scatter(x_data, y_data, alpha=0.6, label='Data points')
plt.plot(x_data, y_fitted, 'r-', linewidth=2, label=f'Fitted line: y = {fitted_slope:.2f}x + {fitted_intercept:.2f}')
plt.plot(x_data, true_slope * x_data + true_intercept, 'g--', linewidth=2, label=f'True line: y = {true_slope}x + {true_intercept}')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Linear Regression: Solving for Best Fit Line')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

# Calculate R-squared
ss_res = np.sum((y_data - y_fitted)**2)
ss_tot = np.sum((y_data - np.mean(y_data))**2)
r_squared = 1 - (ss_res / ss_tot)
print(f"R-squared: {r_squared:.4f}")
```

## 2. Quadratic Equations

Quadratic equations appear in optimization problems and certain ML algorithms, particularly in kernel methods and neural network loss landscapes.

```python
# Quadratic equations: ax² + bx + c = 0
x = symbols('x')

# Example quadratic equation
a, b, c = 1, -5, 6  # x² - 5x + 6 = 0
quadratic_eq = Eq(a*x**2 + b*x + c, 0)
solutions = solve(quadratic_eq, x)

print(f"Quadratic equation: {quadratic_eq}")
print(f"Solutions: x = {solutions}")

# Using the quadratic formula
discriminant = b**2 - 4*a*c
print(f"\nDiscriminant: {discriminant}")

if discriminant > 0:
    x1 = (-b + np.sqrt(discriminant)) / (2*a)
    x2 = (-b - np.sqrt(discriminant)) / (2*a)
    print(f"Two real solutions: x₁ = {x1}, x₂ = {x2}")
elif discriminant == 0:
    x1 = -b / (2*a)
    print(f"One repeated solution: x = {x1}")
else:
    real_part = -b / (2*a)
    imag_part = np.sqrt(-discriminant) / (2*a)
    print(f"Complex solutions: x = {real_part} ± {imag_part}i")
```

### Visualizing Quadratic Functions

```python
# Visualize different quadratic functions
x_vals = np.linspace(-3, 8, 200)

# Different quadratic functions
def quadratic(x, a, b, c):
    return a*x**2 + b*x + c

functions = [
    (1, -5, 6, 'x² - 5x + 6'),      # Two real roots
    (1, -4, 4, 'x² - 4x + 4'),      # One repeated root
    (1, -2, 5, 'x² - 2x + 5'),      # No real roots
    (-0.5, 2, 1, '-0.5x² + 2x + 1') # Downward opening
]

plt.figure(figsize=(12, 8))
colors = ['blue', 'red', 'green', 'purple']

for i, (a, b, c, label) in enumerate(functions):
    y_vals = quadratic(x_vals, a, b, c)
    plt.plot(x_vals, y_vals, color=colors[i], linewidth=2, label=label)
    
    # Mark roots if they exist and are real
    discriminant = b**2 - 4*a*c
    if discriminant >= 0:
        if discriminant > 0:
            x1 = (-b + np.sqrt(discriminant)) / (2*a)
            x2 = (-b - np.sqrt(discriminant)) / (2*a)
            plt.plot([x1, x2], [0, 0], 'o', color=colors[i], markersize=8)
        else:
            x1 = -b / (2*a)
            plt.plot(x1, 0, 'o', color=colors[i], markersize=8)
    
    # Mark vertex
    vertex_x = -b / (2*a)
    vertex_y = quadratic(vertex_x, a, b, c)
    plt.plot(vertex_x, vertex_y, 's', color=colors[i], markersize=8)

plt.axhline(y=0, color='black', linestyle='-', alpha=0.3)
plt.axvline(x=0, color='black', linestyle='-', alpha=0.3)
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Quadratic Functions with Different Properties')
plt.legend()
plt.grid(True, alpha=0.3)
plt.xlim(-1, 7)
plt.ylim(-3, 8)
plt.show()
```

### ML Application: Parabolic Loss Functions

Quadratic functions appear in many ML loss functions, particularly in regularization:

```python
# Ridge regression loss function (includes quadratic regularization)
# Loss = MSE + α * ||w||²
# where ||w||² is the squared L2 norm (quadratic in weights)

def ridge_loss(w, X, y, alpha):
    """
    Ridge regression loss function
    w: weights
    X: features
    y: targets
    alpha: regularization parameter
    """
    mse = np.mean((X @ w - y)**2)
    l2_penalty = alpha * np.sum(w**2)  # Quadratic penalty
    return mse + l2_penalty

# Generate example data
np.random.seed(42)
n_samples, n_features = 50, 2
X = np.random.randn(n_samples, n_features)
true_weights = np.array([2.0, -1.5])
y = X @ true_weights + 0.1 * np.random.randn(n_samples)

# Visualize loss landscape for different regularization strengths
w1_range = np.linspace(0, 4, 50)
w2_range = np.linspace(-3, 0, 50)
W1, W2 = np.meshgrid(w1_range, w2_range)

fig, axes = plt.subplots(1, 3, figsize=(15, 4))
alphas = [0.0, 0.1, 1.0]

for i, alpha in enumerate(alphas):
    Z = np.zeros_like(W1)
    for j in range(W1.shape[0]):
        for k in range(W1.shape[1]):
            w = np.array([W1[j, k], W2[j, k]])
            Z[j, k] = ridge_loss(w, X, y, alpha)
    
    contour = axes[i].contour(W1, W2, Z, levels=20)
    axes[i].plot(true_weights[0], true_weights[1], 'r*', markersize=15, label='True weights')
    axes[i].set_xlabel('w₁')
    axes[i].set_ylabel('w₂')
    axes[i].set_title(f'Loss Landscape (α = {alpha})')
    axes[i].legend()
    axes[i].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print("Notice how increasing regularization (α) makes the loss function more quadratic")
print("and pushes the optimal weights toward zero.")
```

## 3. Polynomial Operations

Polynomials are important in ML for feature engineering, kernel methods, and understanding model complexity.

```python
# Polynomial operations
x = symbols('x')

# Define polynomials
p1 = x**3 + 2*x**2 - x + 1
p2 = 2*x**2 + 3*x - 2

print(f"P₁(x) = {p1}")
print(f"P₂(x) = {p2}")

# Polynomial arithmetic
print(f"\nPolynomial Operations:")
print(f"P₁ + P₂ = {p1 + p2}")
print(f"P₁ - P₂ = {p1 - p2}")
print(f"P₁ × P₂ = {expand(p1 * p2)}")

# Polynomial evaluation
x_val = 2
p1_at_2 = p1.subs(x, x_val)
p2_at_2 = p2.subs(x, x_val)
print(f"\nEvaluation at x = {x_val}:")
print(f"P₁({x_val}) = {p1_at_2}")
print(f"P₂({x_val}) = {p2_at_2}")

# Finding roots
roots_p2 = solve(p2, x)
print(f"\nRoots of P₂(x): {roots_p2}")
```

### Polynomial Degree and Complexity

```python
# Visualizing polynomials of different degrees
x_vals = np.linspace(-2, 2, 200)

# Define polynomials of different degrees
polynomials = {
    'Linear (degree 1)': lambda x: 2*x + 1,
    'Quadratic (degree 2)': lambda x: x**2 - x + 1,
    'Cubic (degree 3)': lambda x: x**3 - 2*x**2 + x + 1,
    'Quartic (degree 4)': lambda x: 0.5*x**4 - x**3 + 0.5*x**2 + x + 1,
    'Quintic (degree 5)': lambda x: 0.1*x**5 - 0.3*x**4 + 0.2*x**3 + 0.5*x**2 - x + 1
}

plt.figure(figsize=(12, 8))
colors = ['blue', 'red', 'green', 'purple', 'orange']

for i, (name, poly_func) in enumerate(polynomials.items()):
    y_vals = poly_func(x_vals)
    plt.plot(x_vals, y_vals, color=colors[i], linewidth=2, label=name)

plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Polynomials of Different Degrees')
plt.legend()
plt.grid(True, alpha=0.3)
plt.ylim(-3, 5)
plt.show()

print("Higher degree polynomials can fit more complex curves but may overfit to data.")
```

### ML Application: Polynomial Feature Engineering

Polynomial features allow linear models to capture non-linear relationships:

```python
# Polynomial feature engineering example
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import Pipeline

# Generate non-linear data
np.random.seed(42)
x_true = np.linspace(0, 1, 50)
y_true = 2 * x_true**3 - 3 * x_true**2 + x_true + 0.1 * np.random.randn(50)

X_true = x_true.reshape(-1, 1)

# Fit polynomial models of different degrees
degrees = [1, 2, 3, 5, 10]
x_test = np.linspace(0, 1, 100).reshape(-1, 1)

plt.figure(figsize=(15, 10))

for i, degree in enumerate(degrees):
    plt.subplot(2, 3, i + 1)
    
    # Create polynomial features and fit model
    poly_model = Pipeline([
        ('poly', PolynomialFeatures(degree=degree)),
        ('linear', LinearRegression())
    ])
    
    poly_model.fit(X_true, y_true)
    y_pred = poly_model.predict(x_test)
    
    # Calculate training error
    y_train_pred = poly_model.predict(X_true)
    mse = np.mean((y_true - y_train_pred)**2)
    
    # Plot results
    plt.scatter(x_true, y_true, alpha=0.6, label='Data')
    plt.plot(x_test, y_pred, 'r-', linewidth=2, label=f'Degree {degree} fit')
    plt.xlabel('x')
    plt.ylabel('y')
    plt.title(f'Polynomial Degree {degree}\nMSE: {mse:.4f}')
    plt.legend()
    plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# Show how polynomial features are created
X_example = np.array([[2], [3]])
print("Original features:", X_example.flatten())

for degree in [2, 3]:
    poly = PolynomialFeatures(degree=degree, include_bias=False)
    X_poly = poly.fit_transform(X_example)
    feature_names = poly.get_feature_names_out(['x'])
    print(f"\nDegree {degree} polynomial features:")
    for i, name in enumerate(feature_names):
        print(f"{name}: {X_poly[:, i]}")
```

## 4. Matrix Notation and Basic Operations

Matrices are fundamental to ML - they represent datasets, transformations, and model parameters.

```python
# Matrix creation and basic operations
# Using SymPy for symbolic computation
A = Matrix([[1, 2, 3],
            [4, 5, 6]])

B = Matrix([[7, 8],
            [9, 10],
            [11, 12]])

C = Matrix([[1, 2],
            [3, 4]])

print("Matrix A (2×3):")
print(A)
print(f"Shape: {A.shape}")

print("\nMatrix B (3×2):")
print(B)
print(f"Shape: {B.shape}")

print("\nMatrix C (2×2):")
print(C)
print(f"Shape: {C.shape}")
```

### Matrix Operations

```python
# Matrix multiplication
print("Matrix Multiplication:")
print(f"A × B = ")
AB = A * B
print(AB)
print(f"Result shape: {AB.shape}")

# Matrix addition (same dimensions)
D = Matrix([[1, 1],
            [1, 1]])
print(f"\nMatrix Addition:")
print(f"C + D = ")
print(C + D)

# Matrix transpose
print(f"\nMatrix Transpose:")
print(f"A^T = ")
print(A.T)

# Scalar multiplication
print(f"\nScalar Multiplication:")
print(f"2 × C = ")
print(2 * C)

# Matrix inverse (for square matrices)
print(f"\nMatrix Inverse:")
print(f"C^(-1) = ")
print(C.inv())

# Verify inverse
print(f"\nVerification C × C^(-1) = I:")
print(C * C.inv())
```

### Matrix Properties Important in ML

```python
# Important matrix properties
import numpy as np

# Convert to NumPy for numerical computations
A_np = np.array(A).astype(float)
C_np = np.array(C).astype(float)

print("Matrix Properties:")

# Determinant (square matrices only)
det_C = np.linalg.det(C_np)
print(f"Determinant of C: {det_C}")

# Trace (sum of diagonal elements)
trace_C = np.trace(C_np)
print(f"Trace of C: {trace_C}")

# Rank
rank_A = np.linalg.matrix_rank(A_np)
rank_C = np.linalg.matrix_rank(C_np)
print(f"Rank of A: {rank_A}")
print(f"Rank of C: {rank_C}")

# Eigenvalues and eigenvectors (square matrices)
eigenvals, eigenvecs = np.linalg.eig(C_np)
print(f"\nEigenvalues of C: {eigenvals}")
print(f"Eigenvectors of C:\n{eigenvecs}")

# Norm (Frobenius norm)
norm_C = np.linalg.norm(C_np, 'fro')
print(f"\nFrobenius norm of C: {norm_C:.4f}")
```

### ML Application: Data Representation

In ML, data is typically represented as matrices where rows are samples and columns are features:

```python
# Example: Housing price dataset representation
# Features: [square_feet, bedrooms, bathrooms, age]
# Each row represents one house

X = np.array([
    [1500, 3, 2, 10],  # House 1
    [2000, 4, 3, 5],   # House 2
    [1200, 2, 1, 15],  # House 3
    [1800, 3, 2, 8],   # House 4
    [2200, 4, 3, 3]    # House 5
])

y = np.array([300000, 450000, 250000, 380000, 520000])  # Prices

print("Feature Matrix X (5 houses × 4 features):")
print("Columns: [sq_feet, bedrooms, bathrooms, age]")
print(X)
print(f"Shape: {X.shape}")

print(f"\nTarget vector y (prices):")
print(y)
print(f"Shape: {y.shape}")

# Feature engineering: Add polynomial features
# For example, interaction between sq_feet and bedrooms
sq_feet_bedrooms = X[:, 0] * X[:, 1]  # Element-wise multiplication

# Add bias term (column of ones for intercept)
bias = np.ones((X.shape[0], 1))
X_extended = np.column_stack([bias, X, sq_feet_bedrooms])

print(f"\nExtended feature matrix (with bias and interaction):")
print("Columns: [bias, sq_feet, bedrooms, bathrooms, age, sq_feet*bedrooms]")
print(X_extended)
print(f"Shape: {X_extended.shape}")

# Demonstrate matrix operations for ML
print(f"\nMatrix Operations in ML:")

# Mean of each feature (column-wise mean)
feature_means = np.mean(X, axis=0)
print(f"Feature means: {feature_means}")

# Covariance matrix (important for PCA)
# Center the data first
X_centered = X - feature_means
cov_matrix = (X_centered.T @ X_centered) / (X.shape[0] - 1)
print(f"\nCovariance matrix:")
print(cov_matrix)

# Visualize the covariance matrix
plt.figure(figsize=(8, 6))
plt.imshow(cov_matrix, cmap='coolwarm', aspect='auto')
plt.colorbar(label='Covariance')
plt.title('Feature Covariance Matrix')
feature_names = ['sq_feet', 'bedrooms', 'bathrooms', 'age']
plt.xticks(range(len(feature_names)), feature_names)
plt.yticks(range(len(feature_names)), feature_names)
plt.show()
```

## 5. Systems of Linear Equations

Systems of linear equations are fundamental to many ML algorithms, from linear regression to neural networks.

```python
# Systems of linear equations
# Example system:
# 2x + 3y = 7
# x - y = 1

x, y = symbols('x y')

# Define the system
eq1 = Eq(2*x + 3*y, 7)
eq2 = Eq(x - y, 1)

print("System of equations:")
print(f"Equation 1: {eq1}")
print(f"Equation 2: {eq2}")

# Solve the system
solution = solve([eq1, eq2], [x, y])
print(f"\nSolution: {solution}")

# Verify the solution
x_val, y_val = solution[x], solution[y]
print(f"\nVerification:")
print(f"Equation 1: 2({x_val}) + 3({y_val}) = {2*x_val + 3*y_val}")
print(f"Equation 2: {x_val} - {y_val} = {x_val - y_val}")
```

### Matrix Form of Linear Systems

A system of linear equations can be written as **Ax = b**:

```python
# Matrix form: Ax = b
# 2x + 3y = 7  =>  [2  3] [x] = [7]
# x - y = 1    =>  [1 -1] [y]   [1]

A = np.array([[2, 3],
              [1, -1]])
b = np.array([7, 1])

print("Matrix form Ax = b:")
print(f"A = \n{A}")
print(f"b = {b}")

# Solve using numpy
x_solution = np.linalg.solve(A, b)
print(f"\nSolution x = {x_solution}")

# Verify
verification = A @ x_solution
print(f"Verification Ax = {verification}")
print(f"Should equal b = {b}")
print(f"Difference: {np.abs(verification - b)}")
```

### Different Types of Linear Systems

```python
# Different types of linear systems

print("Types of Linear Systems:")

# 1. Unique solution (square matrix, full rank)
A1 = np.array([[2, 1],
               [1, 3]])
b1 = np.array([5, 7])
det1 = np.linalg.det(A1)
print(f"\n1. Unique solution system:")
print(f"   Determinant: {det1} (non-zero)")
x1 = np.linalg.solve(A1, b1)
print(f"   Solution: {x1}")

# 2. No solution (inconsistent system)
A2 = np.array([[1, 2],
               [2, 4]])
b2 = np.array([3, 7])  # Inconsistent
det2 = np.linalg.det(A2)
print(f"\n2. No solution system:")
print(f"   Determinant: {det2} (zero - singular matrix)")
print(f"   Rank of A: {np.linalg.matrix_rank(A2)}")
print(f"   Rank of [A|b]: {np.linalg.matrix_rank(np.column_stack([A2, b2]))}")
print(f"   Since ranks differ, no solution exists")

# 3. Infinite solutions (underdetermined system)
A3 = np.array([[1, 2],
               [2, 4]])
b3 = np.array([3, 6])  # Consistent
print(f"\n3. Infinite solutions system:")
print(f"   Rank of A: {np.linalg.matrix_rank(A3)}")
print(f"   Rank of [A|b]: {np.linalg.matrix_rank(np.column_stack([A3, b3]))}")
print(f"   Since ranks are equal but less than number of variables, infinite solutions")

# For infinite solutions, we can use least squares
x3_lstsq = np.linalg.lstsq(A3, b3, rcond=None)[0]
print(f"   Least squares solution: {x3_lstsq}")
```

### ML Application: Normal Equations for Linear Regression

The normal equations provide a closed-form solution for linear regression:

```python
# Normal equations for linear regression
# Given: y = Xw + ε
# Solution: w = (X^T X)^(-1) X^T y

# Generate synthetic regression data
np.random.seed(42)
n_samples, n_features = 100, 3
X = np.random.randn(n_samples, n_features)
true_weights = np.array([2.5, -1.5, 0.8])
noise = 0.1 * np.random.randn(n_samples)
y = X @ true_weights + noise

# Add bias term
X_with_bias = np.column_stack([np.ones(n_samples), X])
true_weights_with_bias = np.array([0, 2.5, -1.5, 0.8])  # bias term = 0

print("Linear Regression using Normal Equations")
print(f"Data shape: X = {X_with_bias.shape}, y = {y.shape}")
print(f"True weights (with bias): {true_weights_with_bias}")

# Solve normal equations
# Method 1: Direct matrix inversion (can be numerically unstable)
XTX = X_with_bias.T @ X_with_bias
XTy = X_with_bias.T @ y
w_direct = np.linalg.inv(XTX) @ XTy

# Method 2: Using np.linalg.solve (more stable)
w_solve = np.linalg.solve(XTX, XTy)

# Method 3: Using least squares directly
w_lstsq = np.linalg.lstsq(X_with_bias, y, rcond=None)[0]

print(f"\nSolutions:")
print(f"Direct inversion:  {w_direct}")
print(f"Matrix solve:      {w_solve}")
print(f"Least squares:     {w_lstsq}")

# Calculate residuals and R-squared
y_pred = X_with_bias @ w_solve
residuals = y - y_pred
ss_res = np.sum(residuals**2)
ss_tot = np.sum((y - np.mean(y))**2)
r_squared = 1 - (ss_res / ss_tot)

print(f"\nModel Performance:")
print(f"R-squared: {r_squared:.4f}")
print(f"RMSE: {np.sqrt(np.mean(residuals**2)):.4f}")

# Visualize predictions vs actual (for first feature)
plt.figure(figsize=(12, 4))

plt.subplot(1, 2, 1)
plt.scatter(y, y_pred, alpha=0.6)
plt.plot([y.min(), y.max()], [y.min(), y.max()], 'r--', linewidth=2)
plt.xlabel('Actual values')
plt.ylabel('Predicted values')
plt.title('Predictions vs Actual')
plt.grid(True, alpha=0.3)

plt.subplot(1, 2, 2)
plt.hist(residuals, bins=20, alpha=0.7, edgecolor='black')
plt.xlabel('Residuals')
plt.ylabel('Frequency')
plt.title('Distribution of Residuals')
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# Show the condition number (numerical stability indicator)
condition_number = np.linalg.cond(XTX)
print(f"\nCondition number of X^T X: {condition_number:.2e}")
print(f"(Lower values indicate better numerical stability)")
```
